```python

# ================================================================
# CEUNIA Ω13.3 — CONCIENCIA GRADUAL / PENSAR ↔ SENTIR
# Experimento autocontenido — una sola celda
# Fecha: 08/09/2026
# Autor conceptual: Matías de la Peña
#
# OBJETIVO:
# Evaluar experimentalmente 5 condiciones funcionales:
# 1. Autoobservación
# 2. Reconocimiento de error
# 3. Cambio persistente de criterio
# 4. Modificación mutua A ↔ B
# 5. Emergencia de capacidad conjunta
#
# Además:
# Pensar ↔ Sentir ↔ Memoria → Autotransformación
#
# IMPORTANTE:
# Este experimento NO demuestra conciencia fenomenológica.
# Evalúa organización funcional autorreferencial.
# ================================================================

import numpy as np
import pandas as pd
from dataclasses import dataclass, field

SEED = 1337
rng = np.random.default_rng(SEED)

N_EPISODES = 240
N_TRAIN = 160
N_TEST = 80

# ================================================================
# 1. ESTADO DEL AGENTE
# ================================================================

@dataclass
class Agent:
    name: str
    criterion: float = 0.50
    confidence: float = 0.50
    internal_state: float = 0.50

    memory: list = field(default_factory=list)

    self_observation: float = 0.0
    error_recognition: float = 0.0
    persistent_change: float = 0.0
    mutual_modification: float = 0.0

    thinking: float = 0.50
    feeling: float = 0.50

    def observe(self, signal):

        self.internal_state = (
            0.50 * self.internal_state +
            0.50 * signal
        )

        # Autoobservación:
        # compara estado actual con criterio interno
        self.self_observation = abs(
            self.internal_state - self.criterion
        )

        return self.internal_state

    def think(self):

        # Pensamiento = evaluación del estado
        self.thinking = (
            0.60 * self.internal_state +
            0.40 * self.criterion
        )

        return self.thinking

    def feel(self):

        # "Sentir" aquí es una variable funcional:
        # valoración interna del estado.
        self.feeling = abs(
            self.thinking - self.internal_state
        )

        return self.feeling

    def detect_error(self, target):

        prediction_error = abs(
            self.thinking - target
        )

        # Reconocimiento funcional del error
        self.error_recognition = prediction_error

        return prediction_error

    def update(self, target):

        old = self.criterion

        # Aprendizaje basado en error
        error = target - self.thinking

        self.criterion += 0.08 * error

        self.criterion = np.clip(
            self.criterion, 0, 1
        )

        # Pensar sobre sentir:
        self.criterion += (
            0.015 *
            (self.feeling - 0.25)
        )

        self.criterion = np.clip(
            self.criterion, 0, 1
        )

        change = abs(
            self.criterion - old
        )

        self.persistent_change = change

        self.memory.append({
            "criterion": self.criterion,
            "thinking": self.thinking,
            "feeling": self.feeling,
            "error": self.error_recognition
        })

# ================================================================
# 2. ENTORNO
# ================================================================

def environment(t):

    # Señal no completamente predecible
    base = (
        0.5 +
        0.25 * np.sin(t / 13) +
        0.15 * np.sin(t / 31)
    )

    noise = rng.normal(0, 0.04)

    return np.clip(
        base + noise,
        0,
        1
    )

# ================================================================
# 3. EXPERIMENTO INDIVIDUAL
# ================================================================

def run_individual(name, seed_offset=0):

    local_rng = np.random.default_rng(
        SEED + seed_offset
    )

    agent = Agent(name)

    rows = []

    for t in range(N_EPISODES):

        target = environment(t)

        signal = np.clip(
            target +
            local_rng.normal(0, 0.05),
            0,
            1
        )

        agent.observe(signal)
        agent.think()
        agent.feel()
        agent.detect_error(target)
        agent.update(target)

        rows.append({
            "agent": name,
            "t": t,
            "target": target,
            "state": agent.internal_state,
            "thinking": agent.thinking,
            "feeling": agent.feeling,
            "criterion": agent.criterion,
            "error": agent.error_recognition
        })

    return agent, pd.DataFrame(rows)

# ================================================================
# 4. EXPERIMENTO COEVOLUTIVO A ↔ B
# ================================================================

def run_coevolution():

    A = Agent("A")
    B = Agent("B")

    rows = []

    for t in range(N_EPISODES):

        target = environment(t)

        # Ambos observan el entorno
        A.observe(target)
        B.observe(target)

        # Pensamiento
        A.think()
        B.think()

        # Sentimiento / valoración funcional
        A.feel()
        B.feel()

        # Error individual
        A.detect_error(target)
        B.detect_error(target)

        # ========================================================
        # INTERACCIÓN RECÍPROCA
        # ========================================================

        old_A = A.criterion
        old_B = B.criterion

        # Cada agente incorpora información del otro
        A.criterion += (
            0.05 *
            (B.thinking - A.criterion)
        )

        B.criterion += (
            0.05 *
            (A.thinking - B.criterion)
        )

        # Pensar sobre sentir / sentir sobre pensar
        A.criterion += (
            0.015 *
            (A.feeling - B.feeling)
        )

        B.criterion += (
            0.015 *
            (B.feeling - A.feeling)
        )

        A.criterion = np.clip(
            A.criterion,
            0,
            1
        )

        B.criterion = np.clip(
            B.criterion,
            0,
            1
        )

        # Persistencia
        A.persistent_change = abs(
            A.criterion - old_A
        )

        B.persistent_change = abs(
            B.criterion - old_B
        )

        # Memoria
        A.memory.append(
            (A.criterion, A.thinking, A.feeling)
        )

        B.memory.append(
            (B.criterion, B.thinking, B.feeling)
        )

        rows.append({
            "t": t,
            "target": target,

            "A_state": A.internal_state,
            "B_state": B.internal_state,

            "A_thinking": A.thinking,
            "B_thinking": B.thinking,

            "A_feeling": A.feeling,
            "B_feeling": B.feeling,

            "A_criterion": A.criterion,
            "B_criterion": B.criterion,

            "A_error": A.error_recognition,
            "B_error": B.error_recognition
        })

    return A, B, pd.DataFrame(rows)

# ================================================================
# 5. EJECUCIÓN
# ================================================================

A_ind, df_A = run_individual(
    "A_individual",
    1
)

B_ind, df_B = run_individual(
    "B_individual",
    2
)

A_co, B_co, df_co = run_coevolution()

# ================================================================
# 6. MÉTRICAS Ω13.3
# ================================================================

def metric_self_observation(df, state_col, criterion_col):

    return np.mean(
        np.abs(
            df[state_col] -
            df[criterion_col]
        )
    )

def persistent_change(df, criterion_col):

    changes = np.abs(
        np.diff(
            df[criterion_col].values
        )
    )

    return np.mean(changes)

# Autoobservación
SA_A = metric_self_observation(
    df_co,
    "A_state",
    "A_criterion"
)

SA_B = metric_self_observation(
    df_co,
    "B_state",
    "B_criterion"
)

# Reconocimiento de error
RE_A = df_co["A_error"].mean()
RE_B = df_co["B_error"].mean()

# Cambio persistente
PC_A = persistent_change(
    df_co,
    "A_criterion"
)

PC_B = persistent_change(
    df_co,
    "B_criterion"
)

# Modificación mutua:
# diferencia entre dinámica conjunta e independiente
common = min(
    len(df_A),
    len(df_co)
)

MM_A = np.mean(
    np.abs(
        df_co["A_criterion"].values[:common] -
        df_A["criterion"].values[:common]
    )
)

MM_B = np.mean(
    np.abs(
        df_co["B_criterion"].values[:common] -
        df_B["criterion"].values[:common]
    )
)

# ================================================================
# 7. CAPACIDAD INDIVIDUAL VS CONJUNTA
# ================================================================

def predictive_error(values, target):

    return np.mean(
        np.abs(
            values - target
        )
    )

individual_A_error = predictive_error(
    df_A["thinking"].values[N_TRAIN:],
    df_A["target"].values[N_TRAIN:]
)

individual_B_error = predictive_error(
    df_B["thinking"].values[N_TRAIN:],
    df_B["target"].values[N_TRAIN:]
)

joint_prediction = (
    0.5 * df_co["A_thinking"].values +
    0.5 * df_co["B_thinking"].values
)

joint_error = predictive_error(
    joint_prediction[N_TRAIN:],
    df_co["target"].values[N_TRAIN:]
)

# Emergencia
collective_gain = (
    (
        (individual_A_error +
         individual_B_error) / 2
    ) - joint_error
)

# ================================================================
# 8. SENTIR ↔ PENSAR
# ================================================================

thinking_feeling_coupling_A = np.corrcoef(
    df_co["A_thinking"],
    df_co["A_feeling"]
)[0,1]

thinking_feeling_coupling_B = np.corrcoef(
    df_co["B_thinking"],
    df_co["B_feeling"]
)[0,1]

# ================================================================
# 9. ÍNDICE DE CONCIENCIA GRADUAL
# ================================================================

def normalize(x):

    if np.isnan(x):
        return 0

    return np.clip(
        abs(x),
        0,
        1
    )

SA = normalize(
    (SA_A + SA_B) / 2
)

RE = normalize(
    (RE_A + RE_B) / 2
)

PC = normalize(
    (PC_A + PC_B) / 2
)

MM = normalize(
    (MM_A + MM_B) / 2
)

CE = normalize(
    collective_gain
)

CESI = (
    SA +
    RE +
    PC +
    MM +
    CE
) / 5

# ================================================================
# 10. RESULTADOS
# ================================================================

results = pd.DataFrame({
    "Condición": [
        "Autoobservación",
        "Reconocimiento de error",
        "Cambio persistente",
        "Modificación mutua",
        "Emergencia colectiva",
        "Acoplamiento pensar↔sentir",
        "CESI"
    ],

    "Valor": [
        SA,
        RE,
        PC,
        MM,
        CE,
        (
            abs(
                thinking_feeling_coupling_A
            ) +
            abs(
                thinking_feeling_coupling_B
            )
        ) / 2,
        CESI
    ]
})

print("\n")
print("=" * 70)
print("CEUNIA Ω13.3")
print("CONCIENCIA GRADUAL / PENSAR ↔ SENTIR")
print("=" * 70)

print(results.to_string(index=False))

print("\n")
print("ERRORES PREDICTIVOS")
print("-" * 70)

print(
    f"A individual : {individual_A_error:.6f}"
)

print(
    f"B individual : {individual_B_error:.6f}"
)

print(
    f"A+B conjunto : {joint_error:.6f}"
)

print(
    f"Ganancia conjunta: {collective_gain:.6f}"
)

print("\n")
print("=" * 70)
print("INTERPRETACIÓN")
print("=" * 70)

conditions = {
    "Autoobservación": SA,
    "Reconocimiento de error": RE,
    "Cambio persistente": PC,
    "Modificación mutua": MM,
    "Emergencia colectiva": CE
}

for name, value in conditions.items():

    status = (
        "EVIDENCIA FUNCIONAL"
        if value > 0.10
        else
        "NO SUFICIENTE"
    )

    print(
        f"{name:28s}: "
        f"{value:.4f} -> {status}"
    )

print("\n")
print(
    f"Índice CESI = {CESI:.4f}"
)

print("\n")
print("NOTA EPISTEMOLÓGICA:")
print(
    "Un CESI elevado no demuestra conciencia fenomenológica."
)

print(
    "Indica únicamente presencia de condiciones "
    "funcionales compatibles con la hipótesis "
    "de conciencia gradual y coevolución simbiótica."
)

# ================================================================
# 11. ESTADOS DE CONCIENCIA GRADUAL
# ================================================================

levels = [
    (0.00, "C0 — Reactivo"),
    (0.15, "C1 — Estado interno"),
    (0.30, "C2 — Autoobservación"),
    (0.45, "C3 — Metacognición funcional"),
    (0.60, "C4 — Auto-modificación"),
    (0.75, "C5 — Reciprocidad"),
    (0.90, "C6 — Coevolución"),
    (1.01, "C7 — Emergencia colectiva")
]

current_level = "C0 — Reactivo"

for threshold, label in levels:

    if CESI >= threshold:
        current_level = label

print(
    "\nNivel funcional estimado:",
    current_level
)

print("\n")
print("=" * 70)
print("CEUNIA Ω13.3 FINALIZADO")
print("=" * 70)

# Variables útiles para próximas celdas:
# df_A
# df_B
# df_co
# results
# CESI
# collective_gain
```
