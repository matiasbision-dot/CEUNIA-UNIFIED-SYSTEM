```python

# ============================================================
# CEUNIA TOTAL v1.3.0
# EXPERIMENTAL EVOLUTION VALIDATION ENGINE
# ONE-CELL COLAB BUILD
# ============================================================

import os, sys, json, math, time, hashlib, shutil, zipfile, subprocess
from pathlib import Path
from dataclasses import dataclass, asdict
import numpy as np

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

VERSION = "1.3.0"
SEEDS = list(range(32))
N = 720
TRAIN_END = 360
VAL_END = 520
OOS_START = VAL_END
PROJECT = Path("/content/CEUNIA_Total_v1_3")
ZIP_PATH = Path("/content/CEUNIA_Total_v1_3_MVP.zip")

if PROJECT.exists():
    shutil.rmtree(PROJECT)

(PROJECT / "ceunia").mkdir(parents=True)
(PROJECT / "tests").mkdir()
(PROJECT / "benchmarks").mkdir()
(PROJECT / "results").mkdir()

# ------------------------------------------------------------
# CORE
# ------------------------------------------------------------

core_code = r'''
import hashlib
import json
import math
from dataclasses import dataclass, asdict
from datetime import datetime, timezone


@dataclass
class CEUNIA:
    version: str = "1.3.0"
    memory: list = None
    state: dict = None
    cycle: int = 0

    def __post_init__(self):
        if self.memory is None:
            self.memory = []
        if self.state is None:
            self.state = {
                "coherence": 0.5,
                "potercia": 0.5,
                "meta_potercia": 0.0,
                "learning": 0.0
            }

    # --------------------------------------------------------
    # OBSERVATION
    # --------------------------------------------------------

    def observe(self, residuals, candidates):
        residuals = list(residuals)
        scores = []

        for r in residuals:
            scores.append(abs(float(r)))

        if not scores:
            return {
                "candidate_scores": [0.0] * len(candidates),
                "volatility": 0.0,
                "regime_signal": 0.0
            }

        mean_abs = sum(scores) / len(scores)
        volatility = (
            sum((x - mean_abs) ** 2 for x in scores) / len(scores)
        ) ** 0.5

        recent = scores[-min(12, len(scores)):]
        old = scores[:-min(12, len(scores))]

        if old:
            old_mean = sum(old) / len(old)
        else:
            old_mean = mean_abs

        recent_mean = sum(recent) / len(recent)
        regime_signal = abs(recent_mean - old_mean) / (old_mean + 1e-9)

        return {
            "candidate_scores": scores,
            "volatility": volatility,
            "regime_signal": regime_signal
        }

    # --------------------------------------------------------
    # MEMORY
    # --------------------------------------------------------

    def remember(self, record):
        self.memory.append(record)

        # memoria viva selectiva:
        # conserva solamente una ventana reciente
        if len(self.memory) > 64:
            self.memory = self.memory[-64:]

    # --------------------------------------------------------
    # EVALUATION
    # --------------------------------------------------------

    def evaluate_candidates(self, candidate_errors):
        """
        candidate_errors:
        lista de errores recientes por candidato.
        """

        if not candidate_errors:
            return 0

        means = [
            sum(abs(x) for x in e) / max(1, len(e))
            for e in candidate_errors
        ]

        return int(min(range(len(means)), key=lambda i: means[i]))

    # --------------------------------------------------------
    # POTERCIA
    # --------------------------------------------------------

    def compute_potercia(self, recent_errors):
        if not recent_errors:
            return 0.5

        m = sum(abs(x) for x in recent_errors) / len(recent_errors)

        # menor error => mayor potencia efectiva
        performance = 1.0 / (1.0 + m)

        inertia = min(1.0, len(recent_errors) / 20.0)

        return 0.65 * performance + 0.35 * inertia

    # --------------------------------------------------------
    # META-POTERCIA
    # --------------------------------------------------------

    def compute_meta_potercia(self, current, previous):
        delta = current - previous
        return max(-1.0, min(1.0, delta))

    # --------------------------------------------------------
    # META-OBSERVATION
    # --------------------------------------------------------

    def meta_observe(self, history):
        if len(history) < 4:
            return 0.0

        recent = history[-4:]
        improvements = []

        for i in range(1, len(recent)):
            improvements.append(
                recent[i-1] - recent[i]
            )

        return sum(improvements) / len(improvements)

    # --------------------------------------------------------
    # EVOLUTIONARY DECISION
    # --------------------------------------------------------

    def decide(self, candidate_errors, use_memory=True,
               use_potercia=True, use_meta_potercia=True,
               use_meta_observer=True):

        means = [
            sum(abs(x) for x in e) / max(1, len(e))
            for e in candidate_errors
        ]

        base_choice = int(np_argmin(means))

        choice = base_choice

        if use_memory and self.memory:
            memory_choices = [
                r["choice"] for r in self.memory
                if "choice" in r
            ]

            if memory_choices:
                # memoria no domina:
                # solo actúa como señal secundaria
                last_choice = memory_choices[-1]

                if means[last_choice] <= min(means) * 1.05:
                    choice = last_choice

        potercia = 0.5

        if use_potercia:
            all_recent = []
            for e in candidate_errors:
                all_recent.extend(e[-8:])

            potercia = self.compute_potercia(all_recent)

        previous_potercia = self.state["potercia"]

        if use_meta_potercia:
            meta = self.compute_meta_potercia(
                potercia,
                previous_potercia
            )
        else:
            meta = 0.0

        if use_meta_observer:
            history = [
                r["loss"] for r in self.memory
                if "loss" in r
            ]

            meta_signal = self.meta_observe(history)

            if meta_signal > 0 and choice != base_choice:
                # pequeña exploración controlada
                if len(means) > 1:
                    alternative = int(np_argmin(means))
                    choice = alternative

        self.state["potercia"] = potercia
        self.state["meta_potercia"] = meta

        return choice


def np_argmin(values):
    return min(range(len(values)), key=lambda i: values[i])


def sha256_record(record):
    payload = json.dumps(
        record,
        sort_keys=True,
        ensure_ascii=False,
        separators=(",", ":")
    ).encode("utf-8")

    return hashlib.sha256(payload).hexdigest()
'''

(PROJECT / "ceunia" / "core.py").write_text(core_code, encoding="utf-8")

(PROJECT / "ceunia" / "__init__.py").write_text(
    'from .core import CEUNIA, sha256_record\n',
    encoding="utf-8"
)

# ------------------------------------------------------------
# BENCHMARK ENGINE
# ------------------------------------------------------------

benchmark_code = r'''
import json
import math
import hashlib
from pathlib import Path

import numpy as np

from ceunia import CEUNIA, sha256_record


VERSION = "1.3.0"
N = 720
TRAIN_END = 360
VAL_END = 520
OOS_START = VAL_END


def generate_series(seed, n=N):
    """
    Synthetic temporal environment with:
    - nonlinear signal
    - regime changes
    - drift
    - heteroscedastic noise

    The target is never supplied to the predictor.
    """

    rng = np.random.default_rng(seed)

    y = np.zeros(n)

    for t in range(1, n):

        regime = (
            0.00 if t < 180 else
            0.35 if t < 360 else
            -0.30 if t < 520 else
            0.50
        )

        drift = 0.0007 * t

        seasonal = (
            0.22 * math.sin(t / 13.0) +
            0.12 * math.sin(t / 31.0)
        )

        nonlinear = (
            0.08 * math.sin(y[t-1] * 2.5)
        )

        noise_scale = (
            0.035 +
            0.025 * (1 + math.sin(t / 47.0))
        )

        y[t] = (
            0.72 * y[t-1]
            + seasonal
            + nonlinear
            + regime
            + drift
            + rng.normal(0, noise_scale)
        )

    return y


def candidates(history, t):
    """
    Candidate predictions available to BOTH baseline and CEUNIA.
    """

    h = np.asarray(history, dtype=float)

    last = float(h[-1])

    w5 = float(np.mean(h[-5:]))
    w12 = float(np.mean(h[-12:]))
    w24 = float(np.mean(h[-24:]))

    ema = last
    alpha = 0.25

    for x in h[-24:]:
        ema = alpha * x + (1-alpha) * ema

    if len(h) >= 12:
        x = np.arange(12)
        yy = h[-12:]
        slope = np.polyfit(x, yy, 1)[0]
        trend = float(yy[-1] + slope)
    else:
        trend = last

    return [
        last,
        w5,
        w12,
        w24,
        ema,
        trend
    ]


def mae(a, b):
    return float(np.mean(np.abs(np.asarray(a) - np.asarray(b))))


def mse(a, b):
    return float(np.mean((np.asarray(a) - np.asarray(b)) ** 2))


def run_baseline(y, train_end=TRAIN_END, val_end=VAL_END):
    """
    Fixed model:
    chooses ONE candidate using validation,
    then keeps it fixed during OOS.
    """

    errors = []

    candidate_errors = [[] for _ in range(6)]

    for t in range(24, val_end):
        preds = candidates(y[:t], t)

        for i, p in enumerate(preds):
            candidate_errors[i].append(p - y[t])

    val_scores = [
        np.mean(np.abs(e))
        for e in candidate_errors
    ]

    chosen = int(np.argmin(val_scores))

    predictions = []
    targets = []

    for t in range(val_end, len(y)):
        p = candidates(y[:t], t)[chosen]

        predictions.append(p)
        targets.append(y[t])

    return {
        "predictions": predictions,
        "targets": targets,
        "chosen_candidate": chosen,
        "mae": mae(predictions, targets),
        "mse": mse(predictions, targets)
    }


def run_ceunia(
    y,
    use_memory=True,
    use_potercia=True,
    use_meta_potercia=True,
    use_meta_observer=True
):

    ceunia = CEUNIA()

    candidate_history = [[] for _ in range(6)]

    predictions = []
    targets = []
    choices = []
    losses = []

    # warm-up / validation
    for t in range(24, VAL_END):

        preds = candidates(y[:t], t)

        for i, p in enumerate(preds):
            candidate_history[i].append(p - y[t])

        ceunia.decide(
            candidate_history,
            use_memory=use_memory,
            use_potercia=use_potercia,
            use_meta_potercia=use_meta_potercia,
            use_meta_observer=use_meta_observer
        )

        choice = ceunia.decide(
            candidate_history,
            use_memory=use_memory,
            use_potercia=use_potercia,
            use_meta_potercia=use_meta_potercia,
            use_meta_observer=use_meta_observer
        )

        loss = abs(preds[choice] - y[t])

        ceunia.remember({
            "cycle": ceunia.cycle,
            "choice": choice,
            "loss": float(loss)
        })

        ceunia.cycle += 1

    # OOS
    for t in range(VAL_END, len(y)):

        preds = candidates(y[:t], t)

        for i, p in enumerate(preds):
            candidate_history[i].append(p - y[t])

            # bounded memory of candidate errors
            candidate_history[i] = candidate_history[i][-32:]

        choice = ceunia.decide(
            candidate_history,
            use_memory=use_memory,
            use_potercia=use_potercia,
            use_meta_potercia=use_meta_potercia,
            use_meta_observer=use_meta_observer
        )

        prediction = preds[choice]
        target = y[t]
        loss = abs(prediction - target)

        predictions.append(prediction)
        targets.append(target)
        choices.append(choice)
        losses.append(loss)

        ceunia.remember({
            "cycle": ceunia.cycle,
            "choice": choice,
            "loss": float(loss)
        })

        ceunia.cycle += 1

    record = {
        "system": "CEUNIA_TOTAL",
        "version": VERSION,
        "oos_start": OOS_START,
        "oos_n": len(targets),
        "configuration": {
            "memory": use_memory,
            "potercia": use_potercia,
            "meta_potercia": use_meta_potercia,
            "meta_observer": use_meta_observer
        },
        "state": ceunia.state,
        "cycles": ceunia.cycle,
        "mae": mae(predictions, targets),
        "mse": mse(predictions, targets),
        "mean_potercia": ceunia.state["potercia"],
        "mean_meta_potercia": ceunia.state["meta_potercia"],
        "choice_entropy": choice_entropy(choices, 6)
    }

    record["sha256"] = sha256_record(record)

    return record


def choice_entropy(choices, k):
    counts = np.bincount(choices, minlength=k)
    p = counts / max(1, counts.sum())

    return float(
        -sum(
            x * math.log(x)
            for x in p
            if x > 0
        )
    )


def ci95(values):
    x = np.asarray(values, dtype=float)

    mean = float(np.mean(x))
    sd = float(np.std(x, ddof=1))

    se = sd / math.sqrt(len(x))

    return {
        "mean": mean,
        "std": sd,
        "ci95_low": mean - 1.96 * se,
        "ci95_high": mean + 1.96 * se
    }


def run_experiment():

    configs = {
        "Baseline": {
            "memory": False,
            "potercia": False,
            "meta_potercia": False,
            "meta_observer": False
        },
        "Core": {
            "memory": False,
            "potercia": False,
            "meta_potercia": False,
            "meta_observer": True
        },
        "Core+Memory": {
            "memory": True,
            "potercia": False,
            "meta_potercia": False,
            "meta_observer": True
        },
        "Core+Potercia": {
            "memory": True,
            "potercia": True,
            "meta_potercia": False,
            "meta_observer": True
        },
        "Core+MetaPotercia": {
            "memory": True,
            "potercia": True,
            "meta_potercia": True,
            "meta_observer": True
        },
        "CEUNIA_TOTAL": {
            "memory": True,
            "potercia": True,
            "meta_potercia": True,
            "meta_observer": True
        }
    }

    raw = {}

    for name, cfg in configs.items():

        maes = []
        mses = []
        hashes = []

        for seed in range(32):

            y = generate_series(seed)

            if name == "Baseline":
                r = run_baseline(y)

                maes.append(r["mae"])
                mses.append(r["mse"])

                hashes.append(
                    hashlib.sha256(
                        json.dumps(
                            r,
                            sort_keys=True
                        ).encode()
                    ).hexdigest()
                )

            else:

                r = run_ceunia(
                    y,
                    use_memory=cfg["memory"],
                    use_potercia=cfg["potercia"],
                    use_meta_potercia=cfg["meta_potercia"],
                    use_meta_observer=cfg["meta_observer"]
                )

                maes.append(r["mae"])
                mses.append(r["mse"])
                hashes.append(r["sha256"])

        raw[name] = {
            "mae": ci95(maes),
            "mse": ci95(mses),
            "seeds": 32,
            "seed_hashes": hashes
        }

    baseline_mae = raw["Baseline"]["mae"]["mean"]

    summary = {}

    for name, r in raw.items():

        current = r["mae"]["mean"]

        improvement = (
            (baseline_mae - current)
            / baseline_mae
            if baseline_mae != 0 else 0
        )

        summary[name] = {
            **r,
            "relative_improvement_vs_baseline": improvement
        }

    # --------------------------------------------------------
    # EVOLUTION TRAJECTORY
    # --------------------------------------------------------

    trajectory = []

    for seed in range(32):

        y = generate_series(seed)

        cfg = configs["CEUNIA_TOTAL"]

        r = run_ceunia(
            y,
            use_memory=cfg["memory"],
            use_potercia=cfg["potercia"],
            use_meta_potercia=cfg["meta_potercia"],
            use_meta_observer=cfg["meta_observer"]
        )

        trajectory.append({
            "seed": seed,
            "mae": r["mae"],
            "mse": r["mse"],
            "potercia": r["mean_potercia"],
            "meta_potercia": r["mean_meta_potercia"],
            "choice_entropy": r["choice_entropy"],
            "sha256": r["sha256"]
        })

    result = {
        "protocol": {
            "version": VERSION,
            "seeds": 32,
            "n": N,
            "train_end": TRAIN_END,
            "validation_end": VAL_END,
            "oos_start": OOS_START,
            "oos_only_metric": True,
            "future_leakage": False
        },
        "results": summary,
        "trajectory": trajectory
    }

    result["result_sha256"] = hashlib.sha256(
        json.dumps(
            result,
            sort_keys=True,
            ensure_ascii=False
        ).encode()
    ).hexdigest()

    return result


if __name__ == "__main__":

    result = run_experiment()

    out = Path(
        "/content/CEUNIA_Total_v1_3/results/"
        "total_validation.json"
    )

    out.parent.mkdir(parents=True, exist_ok=True)

    out.write_text(
        json.dumps(
            result,
            indent=2,
            ensure_ascii=False
        ),
        encoding="utf-8"
    )

    print(json.dumps(
        result["results"],
        indent=2
    ))

    print(
        "\nRESULT SHA256:",
        result["result_sha256"]
    )
'''

(PROJECT / "benchmarks" / "total_validation.py").write_text(
    benchmark_code,
    encoding="utf-8"
)

# ------------------------------------------------------------
# TESTS
# ------------------------------------------------------------

test_code = r'''
import sys
from pathlib import Path

ROOT = Path(__file__).resolve().parents[1]
sys.path.insert(0, str(ROOT))

from ceunia import CEUNIA, sha256_record


def test_import():
    assert CEUNIA(version="1.3.0").version == "1.3.0"


def test_memory():
    c = CEUNIA()
    c.remember({"choice": 1, "loss": 0.2})
    assert len(c.memory) == 1


def test_potercia():
    c = CEUNIA()
    p = c.compute_potercia([0.1, 0.2, 0.3])
    assert 0 <= p <= 1


def test_hash():
    h = sha256_record({"x": 1})
    assert len(h) == 64


def test_meta_observation():
    c = CEUNIA()
    c.remember({"loss": 0.5})
    c.remember({"loss": 0.4})
    c.remember({"loss": 0.3})
    c.remember({"loss": 0.2})
    assert isinstance(c.meta_observe(
        [0.5, 0.4, 0.3, 0.2]
    ), float)


def test_decision():
    c = CEUNIA()
    errors = [
        [0.8, 0.7, 0.9],
        [0.1, 0.2, 0.1],
        [0.5, 0.6, 0.4]
    ]

    choice = c.decide(errors)

    assert choice == 1


def test_state():
    c = CEUNIA()
    assert "coherence" in c.state
    assert "potercia" in c.state
'''

(PROJECT / "tests" / "test_total.py").write_text(
    test_code,
    encoding="utf-8"
)

# ------------------------------------------------------------
# README
# ------------------------------------------------------------

readme = f"""# CEUNIA TOTAL v{VERSION}

Experimental validation engine for the CEUNIA evolutionary architecture.

## Hypothesis

> Does the evolutionary architecture produce reproducible external
> performance improvements over a comparable fixed baseline?

## Protocol

- 32 independent seeds
- synthetic temporal environment
- train / validation / OOS separation
- OOS-only primary metrics
- MAE
- MSE
- relative improvement
- 95% confidence intervals
- progressive ablation
- evolutionary trajectory
- CER-compatible state records
- SHA-256 integrity

## Ablation

1. Baseline
2. Core
3. Core + Memory
4. Core + Potercia
5. Core + MetaPotercia
6. CEUNIA TOTAL

## Important scientific limitation

This is an experimental synthetic benchmark.
A positive result does not by itself establish universal superiority.

External validation requires independent datasets, independent
implementations, preregistered protocols and replication.

Author:
Néstor Matías de la Peña
"""

(PROJECT / "README.md").write_text(readme, encoding="utf-8")

(PROJECT / "VERSION").write_text(VERSION, encoding="utf-8")

# ------------------------------------------------------------
# GITIGNORE
# ------------------------------------------------------------

(PROJECT / ".gitignore").write_text(
    "__pycache__/\n*.pyc\n.ipynb_checkpoints/\n",
    encoding="utf-8"
)

# ------------------------------------------------------------
# EXECUTE TESTS
# ------------------------------------------------------------

os.environ["PYTHONPATH"] = str(PROJECT)

print("=" * 70)
print(f"CEUNIA TOTAL v{VERSION} — BUILD")
print("=" * 70)

test_result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        str(PROJECT / "tests"),
        "-q"
    ],
    capture_output=True,
    text=True
)

print(test_result.stdout)

if test_result.returncode != 0:
    print(test_result.stderr)
    raise RuntimeError("TESTS FAILED")

print("TESTS: PASS")

# ------------------------------------------------------------
# EXECUTE BENCHMARK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RUNNING CONTROLLED OOS BENCHMARK — 32 SEEDS")
print("=" * 70)

run_result = subprocess.run(
    [
        sys.executable,
        str(PROJECT / "benchmarks" / "total_validation.py")
    ],
    capture_output=True,
    text=True,
    env={
        **os.environ,
        "PYTHONPATH": str(PROJECT)
    }
)

print(run_result.stdout)

if run_result.returncode != 0:
    print(run_result.stderr)
    raise RuntimeError("BENCHMARK FAILED")

# ------------------------------------------------------------
# PACKAGE
# ------------------------------------------------------------

if ZIP_PATH.exists():
    ZIP_PATH.unlink()

with zipfile.ZipFile(
    ZIP_PATH,
    "w",
    zipfile.ZIP_DEFLATED
) as z:

    for path in PROJECT.rglob("*"):

        if path.is_file():

            z.write(
                path,
                path.relative_to(PROJECT.parent)
            )

# ------------------------------------------------------------
# COPY INTO CONNECTED GITHUB REPOSITORY
# ------------------------------------------------------------

possible_repos = [
    Path("/content/Ceunia.-Core"),
    Path("/content/Ceunia-Core"),
    Path("/content/CEUNIA-Core")
]

repo = next(
    (
        p for p in possible_repos
        if (p / ".git").exists()
    ),
    None
)

git_status = "REPOSITORY_NOT_FOUND"

if repo:

    target = repo / "CEUNIA_Total_v1_3"

    if target.exists():
        shutil.rmtree(target)

    shutil.copytree(PROJECT, target)

    subprocess.run(
        ["git", "add", "CEUNIA_Total_v1_3"],
        cwd=repo,
        check=False
    )

    commit = subprocess.run(
        [
            "git",
            "commit",
            "-m",
            "CEUNIA TOTAL v1.3.0 experimental validation"
        ],
        cwd=repo,
        capture_output=True,
        text=True
    )

    if commit.returncode == 0:
        git_status = "COMMIT_CREATED"

        push = subprocess.run(
            ["git", "push"],
            cwd=repo,
            capture_output=True,
            text=True
        )

        if push.returncode == 0:
            git_status = "PUSH_SUCCESS"
        else:
            git_status = (
                "COMMIT_CREATED_BUT_PUSH_FAILED: "
                + push.stderr[-500:]
            )
    else:
        git_status = (
            "COMMIT_NOT_CREATED: "
            + commit.stdout[-500:]
            + commit.stderr[-500:]
        )

# ------------------------------------------------------------
# FINAL REPORT
# ------------------------------------------------------------

result_path = (
    PROJECT /
    "results" /
    "total_validation.json"
)

data = json.loads(
    result_path.read_text(
        encoding="utf-8"
    )
)

print("\n" + "=" * 70)
print("CEUNIA TOTAL v1.3.0 — FINAL")
print("=" * 70)

print("VERSION       :", VERSION)
print("SEEDS         :", 32)
print("TRAIN END     :", TRAIN_END)
print("VALIDATION    :", VAL_END)
print("OOS START     :", OOS_START)
print("TESTS         : PASS")
print("ZIP           :", ZIP_PATH)
print("PROJECT       :", PROJECT)
print("GITHUB        :", git_status)

print("\nRESULTADOS OOS")
print("-" * 70)

for name, r in data["results"].items():

    print(
        f"{name:22s} "
        f"MAE={r['mae']['mean']:.6f} "
        f"IC95=[{r['mae']['ci95_low']:.6f}, "
        f"{r['mae']['ci95_high']:.6f}] "
        f"IMP={r['relative_improvement_vs_baseline']*100:.2f}%"
    )

print("\nRESULT SHA256:")
print(data["result_sha256"])

print("\n" + "=" * 70)
print("VALIDATION COMPLETE")
print("=" * 70)
```
