```python

# ================================================================
# CEUNIA Ω-REAL-1 — SPY REAL DATA / ONE CELL
# ================================================================
# DATOS REALES:
#   SPY diario vía Yahoo Finance
#
# OBJETIVO:
#   Predecir retorno de t+1 usando SOLO información disponible
#   hasta t.
#
# DISEÑO:
#   - datos reales
#   - 32 seeds bootstrap/sensibilidad
#   - TRAIN / VAL / OOS temporal
#   - baseline causal
#   - CEUNIA causal
#   - cero features construidas con t+1 o posteriores
#   - bootstrap IC95
#   - permutation test
#   - placebo temporal
#   - régimen definido en TRAIN y congelado
#   - ranking
#   - ZIP
#   - SHA256
# ================================================================

!pip -q install yfinance

import os, math, json, hashlib, zipfile, tempfile
from pathlib import Path

import numpy as np
import pandas as pd
import yfinance as yf


# ================================================================
# CONFIG
# ================================================================

TICKER = "SPY"

START = "2000-01-01"
END   = "2026-09-14"

SEEDS = list(range(32))

TRAIN_FRAC = 0.60
VAL_FRAC   = 0.20
OOS_FRAC   = 0.20

BOOTSTRAP = 5000
PERMUTATIONS = 5000

VERSION = "CEUNIA Ω-REAL-1"


# ================================================================
# DESCARGA DATOS REALES
# ================================================================

raw = yf.download(
    TICKER,
    start=START,
    end=END,
    auto_adjust=True,
    progress=False
)

if raw is None or len(raw) < 1000:
    raise RuntimeError(
        "No se descargaron suficientes datos reales de SPY."
    )

if isinstance(raw.columns, pd.MultiIndex):
    raw.columns = raw.columns.get_level_values(0)

raw = raw[
    ["Open", "High", "Low", "Close", "Volume"]
].dropna().copy()

raw = raw.sort_index()

if len(raw) < 1000:
    raise RuntimeError(
        f"Datos insuficientes: {len(raw)} filas."
    )


# ================================================================
# FEATURES CAUSALES
# TODAS USAN t O ANTERIORES
# ================================================================

df = raw.copy()

df["ret1"] = df["Close"].pct_change(1)
df["ret2"] = df["Close"].pct_change(2)
df["ret5"] = df["Close"].pct_change(5)
df["ret10"] = df["Close"].pct_change(10)
df["ret20"] = df["Close"].pct_change(20)

df["range"] = (
    df["High"] - df["Low"]
) / df["Close"]

df["gap"] = (
    df["Open"] / df["Close"].shift(1)
) - 1.0

df["vol_chg"] = (
    df["Volume"]
    .pct_change()
    .replace([np.inf, -np.inf], np.nan)
)

# Volatilidad histórica: SOLO pasado.
df["vol5"] = (
    df["ret1"]
    .rolling(5)
    .std()
)

df["vol10"] = (
    df["ret1"]
    .rolling(10)
    .std()
)

df["vol20"] = (
    df["ret1"]
    .rolling(20)
    .std()
)

# Tendencia histórica.
df["ma5"] = (
    df["Close"] /
    df["Close"].rolling(5).mean()
) - 1

df["ma20"] = (
    df["Close"] /
    df["Close"].rolling(20).mean()
) - 1

df["ma50"] = (
    df["Close"] /
    df["Close"].rolling(50).mean()
) - 1


# ================================================================
# VARIABLES CEUNIA CAUSALES
# ================================================================

# Potencia:
# magnitud del movimiento disponible hasta t.
df["potency"] = (
    df["ret1"].abs()
    + 0.5 * df["range"].abs()
)

# Inercia:
# persistencia histórica de magnitud.
df["inertia"] = (
    df["potency"]
    .ewm(
        span=20,
        adjust=False
    )
    .mean()
)

# Potercia:
df["potercia"] = (
    df["potency"]
    * (1.0 + df["inertia"])
)

# Energía/magnitud espectral aproximada:
# variación histórica, sin mirar futuro.
df["spectral_power"] = (
    df["ret1"]
    .rolling(20)
    .apply(
        lambda x: float(
            np.mean(
                np.abs(
                    np.fft.rfft(
                        x - np.mean(x)
                    )
                ) ** 2
            )
        ),
        raw=True
    )
)

# Meta:
df["meta"] = (
    0.40 * df["spectral_power"]
    + 0.35 * df["potercia"]
    + 0.25 * df["vol20"]
)

# Φ:
df["phi"] = (
    1.0 /
    (
        1.0
        + (
            df["meta"]
            - df["meta"].rolling(50).mean()
        ).abs()
    )
)

# ================================================================
# OBJETIVO
# RETORNO DEL DÍA SIGUIENTE
#
# IMPORTANTE:
# target usa t+1.
# Ninguna feature anterior usa t+1.
# ================================================================

df["target"] = (
    df["Close"].shift(-1)
    / df["Close"]
) - 1.0


# ================================================================
# LIMPIEZA FINAL
# ================================================================

feature_base = [
    "ret1",
    "ret2",
    "ret5",
    "ret10",
    "ret20",
    "range",
    "gap",
    "vol_chg",
    "vol5",
    "vol10",
    "vol20",
    "ma5",
    "ma20",
    "ma50"
]

feature_ceunia = feature_base + [
    "potency",
    "inertia",
    "potercia",
    "spectral_power",
    "meta",
    "phi"
]

required = (
    feature_ceunia
    + ["target"]
)

df = df.dropna(
    subset=required
).copy()

df = df.replace(
    [np.inf, -np.inf],
    np.nan
).dropna(
    subset=required
)

if len(df) < 1000:
    raise RuntimeError(
        f"Observaciones utilizables insuficientes: {len(df)}"
    )


# ================================================================
# SPLIT TEMPORAL
# ================================================================

n = len(df)

n_train = int(
    n * TRAIN_FRAC
)

n_val = int(
    n * VAL_FRAC
)

train_end = n_train
val_end = n_train + n_val

train = df.iloc[:train_end].copy()
val   = df.iloc[train_end:val_end].copy()
oos   = df.iloc[val_end:].copy()

if len(oos) < 100:
    raise RuntimeError(
        "OOS demasiado pequeño."
    )


# ================================================================
# NORMALIZACIÓN:
# PARÁMETROS APRENDIDOS SOLO EN TRAIN
# ================================================================

def normalize_train(
    train_df,
    other_df,
    cols
):
    mu = train_df[cols].mean()
    sd = train_df[cols].std()

    sd = sd.replace(
        0,
        1.0
    )

    A = (
        (train_df[cols] - mu)
        / sd
    )

    B = (
        (other_df[cols] - mu)
        / sd
    )

    return A, B, mu, sd


Xb_train, Xb_oos, _, _ = normalize_train(
    train,
    oos,
    feature_base
)

Xc_train, Xc_oos, _, _ = normalize_train(
    train,
    oos,
    feature_ceunia
)

y_train = train["target"].to_numpy()
y_oos = oos["target"].to_numpy()


# ================================================================
# REGRESIÓN LINEAL ESTABLE
# ================================================================

def fit(X, y):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float)

    X1 = np.column_stack([
        np.ones(len(X)),
        X
    ])

    return np.linalg.pinv(
        X1
    ) @ y


def predict(X, beta):
    X = np.asarray(X, dtype=float)

    X1 = np.column_stack([
        np.ones(len(X)),
        X
    ])

    return X1 @ beta


def mse(y, p):
    return float(
        np.mean(
            (y - p) ** 2
        )
    )


# ================================================================
# MODELOS
# ================================================================

beta_base = fit(
    Xb_train,
    y_train
)

beta_ceunia = fit(
    Xc_train,
    y_train
)

pred_base = predict(
    Xb_oos,
    beta_base
)

pred_ceunia = predict(
    Xc_oos,
    beta_ceunia
)

baseline_mse = mse(
    y_oos,
    pred_base
)

ceunia_mse = mse(
    y_oos,
    pred_ceunia
)

gain = (
    baseline_mse
    - ceunia_mse
)


# ================================================================
# BOOTSTRAP
# ================================================================

def bootstrap_ci(
    values,
    seed=735,
    n_boot=5000
):
    rng = np.random.default_rng(seed)

    values = np.asarray(
        values,
        dtype=float
    )

    idx = rng.integers(
        0,
        len(values),
        size=(
            n_boot,
            len(values)
        )
    )

    means = values[idx].mean(
        axis=1
    )

    return (
        float(np.quantile(means, 0.025)),
        float(np.quantile(means, 0.975))
    )


# Ganancia por observación OOS.
sq_base = (
    y_oos - pred_base
) ** 2

sq_ceunia = (
    y_oos - pred_ceunia
) ** 2

pointwise_gain = (
    sq_base
    - sq_ceunia
)

ci_low, ci_high = bootstrap_ci(
    pointwise_gain
)


# ================================================================
# PERMUTATION TEST
# ================================================================

rng_perm = np.random.default_rng(
    736
)

observed = abs(
    np.mean(pointwise_gain)
)

null = np.empty(
    PERMUTATIONS
)

for i in range(
    PERMUTATIONS
):
    signs = rng_perm.choice(
        [-1.0, 1.0],
        size=len(pointwise_gain)
    )

    null[i] = abs(
        np.mean(
            pointwise_gain
            * signs
        )
    )

permutation_p = float(
    (
        np.sum(
            null >= observed
        ) + 1
    )
    /
    (
        PERMUTATIONS + 1
    )
)


# ================================================================
# WIN RATE
# ================================================================

win_rate = float(
    np.mean(
        pointwise_gain > 0
    )
)


# ================================================================
# PLACEBO TEMPORAL
#
# Destruye correspondencia entre Xc(t) y target(t+1)
# ================================================================

rng_placebo = np.random.default_rng(
    100000
)

perm = rng_placebo.permutation(
    len(Xc_oos)
)

Xc_placebo = Xc_oos.iloc[
    perm
] if isinstance(
    Xc_oos,
    pd.DataFrame
) else Xc_oos[perm]

beta_placebo = fit(
    Xc_train,
    y_train
)

pred_placebo = predict(
    Xc_placebo,
    beta_placebo
)

placebo_mse = mse(
    y_oos,
    pred_placebo
)

placebo_gain = (
    baseline_mse
    - placebo_mse
)


# ================================================================
# RÉGIMEN
#
# Umbral SOLO desde TRAIN.
# ================================================================

threshold = float(
    train[
        "spectral_power"
    ].median()
)

low_mask = (
    oos[
        "spectral_power"
    ].to_numpy()
    <= threshold
)

high_mask = (
    oos[
        "spectral_power"
    ].to_numpy()
    > threshold
)

gain_low = float(
    np.mean(
        pointwise_gain[
            low_mask
        ]
    )
)

gain_high = float(
    np.mean(
        pointwise_gain[
            high_mask
        ]
    )
)


# ================================================================
# SENSIBILIDAD POR SEED
#
# No reentrenamos sobre OOS.
# Las seeds solamente re-muestrean OOS.
# ================================================================

seed_results = []

for seed in SEEDS:

    rng = np.random.default_rng(
        seed
    )

    idx = rng.integers(
        0,
        len(pointwise_gain),
        size=len(pointwise_gain)
    )

    g = float(
        np.mean(
            pointwise_gain[idx]
        )
    )

    seed_results.append({
        "seed": seed,
        "gain": g,
        "baseline_mse": baseline_mse,
        "ceunia_mse": ceunia_mse
    })


seed_results = sorted(
    seed_results,
    key=lambda x: x["gain"],
    reverse=True
)


# ================================================================
# DIAGNÓSTICO
# ================================================================

if (
    gain > 0
    and ci_low > 0
    and permutation_p < 0.05
):
    diagnostic = (
        "POSITIVE REAL-DATA OOS EVIDENCE"
    )
elif gain > 0:
    diagnostic = (
        "POSITIVE BUT INCONCLUSIVE"
    )
else:
    diagnostic = (
        "NO OOS ADVANTAGE"
    )


warnings = []

warnings.append(
    "Features CEUNIA restringidas a información disponible en t."
)

warnings.append(
    "Target = retorno de t+1."
)

warnings.append(
    "Split temporal TRAIN / VAL / OOS."
)

warnings.append(
    "Umbral de régimen calculado únicamente en TRAIN."
)

if ci_low <= 0 <= ci_high:
    warnings.append(
        "IC95 cruza cero."
    )

if permutation_p >= 0.05:
    warnings.append(
        "Permutation p >= 0.05."
    )

if placebo_gain < gain:
    warnings.append(
        "El placebo temporal destruye la ventaja."
    )


# ================================================================
# EXPORTACIÓN
# ================================================================

output_dir = Path(
    tempfile.mkdtemp(
        prefix="ceunia_omega_real1_"
    )
)

results_json = (
    output_dir
    / "CEUNIA_OMEGA_REAL1_RESULTS.json"
)

payload = {
    "version": VERSION,
    "ticker": TICKER,
    "start": START,
    "end": END,
    "rows": int(len(df)),
    "train_rows": int(len(train)),
    "val_rows": int(len(val)),
    "oos_rows": int(len(oos)),
    "baseline_oos_mse": baseline_mse,
    "ceunia_oos_mse": ceunia_mse,
    "gain": gain,
    "win_rate": win_rate,
    "bootstrap_ic95": [
        ci_low,
        ci_high
    ],
    "permutation_p": permutation_p,
    "placebo_gain": placebo_gain,
    "regime_low_gain": gain_low,
    "regime_high_gain": gain_high,
    "diagnostic": diagnostic,
    "warnings": warnings,
    "ranking": seed_results
}

with open(
    results_json,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        payload,
        f,
        ensure_ascii=False,
        indent=2
    )


zip_path = (
    output_dir
    / "CEUNIA_OMEGA_REAL1.zip"
)

with zipfile.ZipFile(
    zip_path,
    "w",
    compression=zipfile.ZIP_DEFLATED
) as z:

    z.write(
        results_json,
        arcname=results_json.name
    )


# ================================================================
# SHA256
# ================================================================

sha = hashlib.sha256()

with open(
    zip_path,
    "rb"
) as f:

    for block in iter(
        lambda: f.read(
            1024 * 1024
        ),
        b""
    ):
        sha.update(block)

sha256 = sha.hexdigest()


# ================================================================
# SALIDA FINAL
# ================================================================

print("=" * 76)
print("CEUNIA Ω-REAL-1 — SPY REAL DATA")
print("STATUS             : COMPLETED")
print("=" * 76)

print(
    f"DATASET            : {TICKER}"
)

print(
    f"ROWS TOTAL         : {len(df)}"
)

print(
    f"TRAIN / VAL / OOS  : "
    f"{len(train)} / {len(val)} / {len(oos)}"
)

print(
    f"DATE START         : {df.index.min().date()}"
)

print(
    f"DATE END           : {df.index.max().date()}"
)

print("-" * 76)

print(
    f"BASELINE OOS MSE   : "
    f"{baseline_mse:.12f}"
)

print(
    f"CEUNIA OOS MSE     : "
    f"{ceunia_mse:.12f}"
)

print(
    f"GAIN               : "
    f"{gain:.12f}"
)

print(
    f"GAIN %             : "
    f"{100 * gain / baseline_mse:.4f}%"
)

print(
    f"WIN RATE           : "
    f"{100 * win_rate:.2f}%"
)

print(
    f"BOOTSTRAP IC95     : "
    f"[{ci_low:.12f}, {ci_high:.12f}]"
)

print(
    f"PERMUTATION P      : "
    f"{permutation_p:.6f}"
)

print("-" * 76)

print(
    f"PLACEBO GAIN       : "
    f"{placebo_gain:.12f}"
)

print(
    f"REGIME LOW GAIN    : "
    f"{gain_low:.12f}"
)

print(
    f"REGIME HIGH GAIN   : "
    f"{gain_high:.12f}"
)

print("-" * 76)

print(
    f"DIAGNOSTIC         : "
    f"{diagnostic}"
)

print("WARNINGS")

for w in warnings:
    print(
        f"  - {w}"
    )

print("-" * 76)

print("RANKING")

for r in seed_results:
    print(
        f"{r['seed']:02d}  "
        f"gain={r['gain']:+.12f}"
    )

print("-" * 76)

print(
    f"ZIP                : "
    f"{zip_path}"
)

print(
    f"SHA256             : "
    f"{sha256}"
)

print("=" * 76)
```
