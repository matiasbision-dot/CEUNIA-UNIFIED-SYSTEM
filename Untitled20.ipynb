```python
ese
# =====================================================================
# CEUNIA Ω-v2.6 — META-SELECTOR + INCERTIDUMBRE + ABSTENCIÓN
# UNA SOLA CELDA | BLIND OOS | OOD | FALLBACK | BOOTSTRAP AGRUPADO
#
# OBJETIVO:
# 1) Separar selección META de predictor ADAPTIVE.
# 2) Detectar contextos desconocidos antes del resultado.
# 3) Si META no tiene confianza -> fallback a ADAPTIVE.
# 4) Todo umbral se aprende SOLO en TRAIN.
# 5) OOS completamente congelado.
# 6) Evaluación agrupadapppkxiq un por SERIE, no por observación.
# 7) Medir especialmente familias NO VISTAS.
# 8) Comparar:
#       - Adaptive Fixed
#       - Meta v2.5
#       - Meta + Incertidumbre/Abstención
#       - Random
#       - Oracle (solo diagnóstico)
#
# NO BORRA RESULTADOS ANTERIORES.
# =====================================================================

import numpy as np
import pandas as pd
import sqlite3, json, hashlib, os, warnings
from scipy.stats import wilcoxon
warnings.filterwarnings("ignore")

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

# ---------------------------------------------------------------------
# CONFIG
# ---------------------------------------------------------------------

SEEDS = [
    101,211,307,419,523,631,743,857,
    967,1087,1201,1327,1451,1579,1697,1811
]

N = 600
TRAIN = 300

TRAIN_FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed"
]

TEST_FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed",
    "structural_shift",
    "frequency_shift",
    "phase_shift",
    "amplitude_shift",
    "regime_switch"
]

UNSEEN_FAMILIES = [
    "structural_shift",
    "frequency_shift",
    "phase_shift",
    "amplitude_shift",
    "regime_switch"
]

NOISES = [0.0, 0.008, 0.018, 0.035]

STRATEGIES = [
    "trend",
    "cycle",
    "reversion",
    "adaptive"
]

DB = "/content/ceunia_integral.db"

# ---------------------------------------------------------------------
# STABLE SEED
# ---------------------------------------------------------------------

def stable_seed(seed, text):
    h = hashlib.sha256(
        f"{seed}:{text}".encode()
    ).hexdigest()
    return int(h[:8], 16)

# ---------------------------------------------------------------------
# SERIES
# ---------------------------------------------------------------------

def generate_series(n, family, noise, seed):

    rng = np.random.default_rng(
        stable_seed(seed, family)
    )

    t = np.arange(n, dtype=float)

    trend = rng.uniform(0.0015, 0.0045)
    amp = rng.uniform(0.7, 1.5)
    freq = rng.uniform(0.035, 0.085)
    phase = rng.uniform(0, 2*np.pi)

    if family == "trend_cycle":

        y = (
            trend*t
            + amp*np.sin(freq*t + phase)
        )

    elif family == "multi_cycle":

        f1 = rng.uniform(0.025, 0.055)
        f2 = rng.uniform(0.08, 0.13)

        y = (
            trend*t
            + 0.9*np.sin(f1*t + phase)
            + 0.45*np.sin(f2*t)
        )

    elif family == "mean_reversion":

        y = np.zeros(n)
        phi = rng.uniform(0.72, 0.90)

        for i in range(1, n):
            y[i] = (
                phi*y[i-1]
                + rng.normal(0, 0.08)
            )

        y += 0.08*np.sin(freq*t)

    elif family == "mixed":

        y = (
            trend*t
            + 0.8*np.sin(freq*t + phase)
            + 0.18*np.sin(0.16*t)
        )

        for i in range(1, n):
            y[i] += 0.15*y[i-1]

    elif family == "structural_shift":

        y = (
            trend*t
            + amp*np.sin(freq*t + phase)
        )

        s = n//2

        y[s:] += (
            0.0025*(t[s:]-s)
            + 0.55*np.sin(freq*1.7*t[s:])
        )

    elif family == "frequency_shift":

        s = n//2
        y = np.zeros(n)

        y[:s] = (
            trend*t[:s]
            + amp*np.sin(
                0.045*t[:s] + phase
            )
        )

        y[s:] = (
            trend*t[s:]
            + amp*np.sin(
                0.11*t[s:] + phase
            )
        )

    elif family == "phase_shift":

        y = (
            trend*t
            + amp*np.sin(freq*t + phase)
        )

        s = n//2

        y[s:] = (
            trend*t[s:]
            + amp*np.sin(
                freq*t[s:]
                + phase
                + 1.4
            )
        )

    elif family == "amplitude_shift":

        y = (
            trend*t
            + amp*np.sin(freq*t + phase)
        )

        s = n//2

        y[s:] = (
            trend*t[s:]
            + amp*1.9*np.sin(
                freq*t[s:] + phase
            )
        )

    elif family == "regime_switch":

        y = np.zeros(n)

        s1 = n//3
        s2 = 2*n//3

        y[:s1] = (
            trend*t[:s1]
            + amp*np.sin(0.045*t[:s1])
        )

        phi = 0.82

        for i in range(s1, s2):
            y[i] = (
                phi*y[i-1]
                + rng.normal(0, 0.08)
            )

        y[s2:] = (
            y[s2-1]
            + trend*(t[s2:]-t[s2])
            + amp*np.sin(0.11*t[s2:])
        )

    else:

        y = rng.normal(0, 1, n)

    y += 0.035*np.sin(
        0.31*t + seed*0.01
    )

    y += rng.normal(
        0,
        noise,
        n
    )

    return y

# ---------------------------------------------------------------------
# FEATURES
# ---------------------------------------------------------------------

def context_features(h):

    h = np.asarray(h)

    if len(h) < 60:
        return np.zeros(8)

    short = h[-10:]
    medium = h[-30:]
    long = h[-60:]

    slope_short = np.polyfit(
        np.arange(len(short)),
        short,
        1
    )[0]

    slope_long = np.polyfit(
        np.arange(len(long)),
        long,
        1
    )[0]

    volatility = np.std(
        np.diff(long)
    )

    centered = long - np.mean(long)

    ac1 = np.corrcoef(
        centered[:-1],
        centered[1:]
    )[0,1]

    ac2 = np.corrcoef(
        centered[:-2],
        centered[2:]
    )[0,1]

    if not np.isfinite(ac1):
        ac1 = 0.0

    if not np.isfinite(ac2):
        ac2 = 0.0

    fft = np.abs(
        np.fft.rfft(centered)
    )

    freqs = np.fft.rfftfreq(
        len(centered)
    )

    if len(fft) > 2:

        idx = (
            np.argmax(fft[1:])
            + 1
        )

        dom_freq = freqs[idx]

        spectral = (
            fft[idx]
            /
            (np.sum(fft[1:]) + 1e-12)
        )

    else:

        dom_freq = 0
        spectral = 0

    distance = (
        np.mean(
            np.abs(
                short
                - np.mean(medium)
            )
        )
        /
        (np.std(medium)+1e-12)
    )

    return np.array([
        slope_short,
        slope_long,
        volatility,
        ac1,
        ac2,
        dom_freq,
        spectral,
        distance
    ], dtype=float)

# ---------------------------------------------------------------------
# PREDICTORS
# ---------------------------------------------------------------------

def trend(h):

    h = np.asarray(h)

    w = min(
        30,
        len(h)
    )

    x = np.arange(w)

    slope, intercept = np.polyfit(
        x,
        h[-w:],
        1
    )

    return intercept + slope*w

def cycle(h):

    h = np.asarray(h)

    w = min(
        80,
        len(h)
    )

    seg = h[-w:]

    centered = (
        seg - np.mean(seg)
    )

    fft = np.fft.rfft(
        centered
    )

    if len(fft) <= 2:
        return h[-1]

    idx = (
        np.argmax(
            np.abs(fft)[1:]
        )
        + 1
    )

    freq = idx/w

    amp = (
        2*np.abs(fft[idx])/w
    )

    phase = np.angle(
        fft[idx]
    )

    return (
        np.mean(seg)
        + amp*np.cos(
            2*np.pi*freq*w
            + phase
        )
    )

def reversion(h):

    h = np.asarray(h)

    w = min(
        40,
        len(h)
    )

    m = np.mean(
        h[-w:]
    )

    return (
        h[-1]
        + 0.35*(m-h[-1])
    )

def adaptive(h):

    p1 = trend(h)
    p2 = cycle(h)
    p3 = reversion(h)

    f = context_features(h)

    slope = f[1]
    ac1 = f[3]
    spectral = f[6]

    w1 = np.clip(
        abs(slope)*45,
        0,
        1
    )

    w2 = np.clip(
        spectral*1.5,
        0,
        1
    )

    w3 = np.clip(
        -ac1,
        0,
        1
    )

    w = np.array([
        w1,
        w2,
        w3
    ])

    if w.sum() < 1e-12:
        w = np.ones(3)

    w /= w.sum()

    return (
        w[0]*p1
        + w[1]*p2
        + w[2]*p3
    )

PRED = {
    "trend": trend,
    "cycle": cycle,
    "reversion": reversion,
    "adaptive": adaptive
}

# =====================================================================
# TRAIN
# =====================================================================

X_train = []
Y_train = []

for seed in SEEDS:

    for family in TRAIN_FAMILIES:

        y = generate_series(
            N,
            family,
            0.018,
            seed
        )

        for t in range(
            80,
            TRAIN
        ):

            h = y[:t]

            X_train.append(
                context_features(h)
            )

            Y_train.append([
                abs(
                    PRED[s](h)
                    - y[t]
                )
                for s in STRATEGIES
            ])

X_train = np.asarray(
    X_train
)

Y_train = np.asarray(
    Y_train
)

# ---------------------------------------------------------------------
# META MODELS
# ---------------------------------------------------------------------

META = {}

for j, s in enumerate(
    STRATEGIES
):

    model = make_pipeline(
        StandardScaler(),
        Ridge(alpha=1.0)
    )

    model.fit(
        X_train,
        Y_train[:,j]
    )

    META[s] = model

# ---------------------------------------------------------------------
# TRAIN DISTRIBUTION
# FOR OOD DISTANCE
# ---------------------------------------------------------------------

train_mean = X_train.mean(
    axis=0
)

train_std = (
    X_train.std(axis=0)
    + 1e-8
)

Xz = (
    X_train - train_mean
) / train_std

train_dist = np.sqrt(
    np.sum(Xz**2, axis=1)
)

# Conservative OOD threshold.
# Learned exclusively from TRAIN.
OOD_THRESHOLD = float(
    np.percentile(
        train_dist,
        95
    )
)

# ---------------------------------------------------------------------
# TRAIN PREDICTION UNCERTAINTY
# Use residual dispersion of each strategy.
# ---------------------------------------------------------------------

RESID_STD = {}

for j, s in enumerate(
    STRATEGIES
):

    pred = META[s].predict(
        X_train
    )

    residuals = (
        Y_train[:,j]
        - pred
    )

    RESID_STD[s] = float(
        np.std(
            residuals,
            ddof=1
        ) + 1e-8
    )

# ---------------------------------------------------------------------
# TRAIN MARGIN THRESHOLD
#
# Small margin = selector cannot distinguish strategies confidently.
# Threshold learned ONLY on TRAIN.
# ---------------------------------------------------------------------

train_est = np.column_stack([
    META[s].predict(X_train)
    for s in STRATEGIES
])

sorted_train = np.sort(
    train_est,
    axis=1
)

train_margin = (
    sorted_train[:,1]
    - sorted_train[:,0]
)

MARGIN_THRESHOLD = float(
    np.percentile(
        train_margin,
        20
    )
)

# ---------------------------------------------------------------------
# OOS BLIND
# ---------------------------------------------------------------------

rows = []

rng = np.random.default_rng(
    99991
)

for seed in SEEDS:

    for family in TEST_FAMILIES:

        for noise in NOISES:

            y = generate_series(
                N,
                family,
                noise,
                seed
            )

            for t in range(
                TRAIN,
                N
            ):

                h = y[:t]

                actual = y[t]

                # -----------------------------------------------------
                # REAL ERRORS
                # NEVER USED FOR DECISION
                # -----------------------------------------------------

                errors = {
                    s: abs(
                        PRED[s](h)
                        - actual
                    )
                    for s in STRATEGIES
                }

                feat = context_features(
                    h
                )

                xx = feat.reshape(
                    1,-1
                )

                # -----------------------------------------------------
                # META PREDICTED ERRORS
                # -----------------------------------------------------

                estimated = np.array([
                    float(
                        META[s].predict(xx)[0]
                    )
                    for s in STRATEGIES
                ])

                order = np.argsort(
                    estimated
                )

                best_idx = order[0]
                second_idx = order[1]

                selected = STRATEGIES[
                    best_idx
                ]

                margin = (
                    estimated[second_idx]
                    - estimated[best_idx]
                )

                # -----------------------------------------------------
                # OOD DISTANCE
                # -----------------------------------------------------

                z = (
                    feat
                    - train_mean
                ) / train_std

                ood_distance = float(
                    np.sqrt(
                        np.sum(z**2)
                    )
                )

                is_ood = (
                    ood_distance
                    > OOD_THRESHOLD
                )

                # -----------------------------------------------------
                # UNCERTAINTY
                #
                # Combine:
                # 1) model disagreement / residual uncertainty
                # 2) weak selection margin
                # 3) OOD distance
                # -----------------------------------------------------

                selected_std = RESID_STD[
                    selected
                ]

                normalized_margin = (
                    margin
                    /
                    (
                        selected_std
                        + 1e-8
                    )
                )

                uncertain = (
                    normalized_margin
                    < (
                        MARGIN_THRESHOLD
                        /
                        (
                            np.mean(
                                list(
                                    RESID_STD.values()
                                )
                            )
                        )
                    )
                )

                # -----------------------------------------------------
                # ABSTENTION / FALLBACK
                #
                # If context is OOD OR selector uncertain:
                # use ADAPTIVE.
                # -----------------------------------------------------

                abstain = (
                    is_ood
                    or
                    uncertain
                )

                if abstain:
                    robust_selected = "adaptive"
                else:
                    robust_selected = selected

                # -----------------------------------------------------
                # RANDOM
                # -----------------------------------------------------

                random_s = rng.choice(
                    STRATEGIES
                )

                # -----------------------------------------------------
                # ORACLE
                # diagnostic only
                # -----------------------------------------------------

                oracle = min(
                    errors,
                    key=errors.get
                )

                rows.append({

                    "seed": seed,
                    "family": family,
                    "noise": noise,
                    "t": t,

                    "trend":
                        errors["trend"],

                    "cycle":
                        errors["cycle"],

                    "reversion":
                        errors["reversion"],

                    "adaptive":
                        errors["adaptive"],

                    "meta":
                        errors[selected],

                    "meta_uncertain":
                        errors[
                            robust_selected
                        ],

                    "random":
                        errors[random_s],

                    "oracle":
                        errors[oracle],

                    "selected":
                        selected,

                    "robust_selected":
                        robust_selected,

                    "abstain":
                        int(abstain),

                    "ood":
                        int(is_ood),

                    "ood_distance":
                        ood_distance,

                    "margin":
                        margin

                })

df = pd.DataFrame(
    rows
)

# =====================================================================
# GLOBAL METRICS
# =====================================================================

adaptive_fixed = df[
    "adaptive"
].mean()

meta = df[
    "meta"
].mean()

meta_uncertain = df[
    "meta_uncertain"
].mean()

random_mae = df[
    "random"
].mean()

oracle_mae = df[
    "oracle"
].mean()

gain_meta = (
    adaptive_fixed
    - meta
)

gain_robust = (
    adaptive_fixed
    - meta_uncertain
)

# =====================================================================
# GROUPED BY SERIES
# =====================================================================

group_df = (
    df
    .groupby(
        [
            "seed",
            "family",
            "noise"
        ],
        as_index=False
    )
    .agg(
        adaptive=(
            "adaptive",
            "mean"
        ),
        meta=(
            "meta",
            "mean"
        ),
        robust=(
            "meta_uncertain",
            "mean"
        ),
        random=(
            "random",
            "mean"
        ),
        oracle=(
            "oracle",
            "mean"
        ),
        abstain=(
            "abstain",
            "mean"
        ),
        ood=(
            "ood",
            "mean"
        )
    )
)

group_df[
    "gain_meta"
] = (
    group_df["adaptive"]
    - group_df["meta"]
)

group_df[
    "gain_robust"
] = (
    group_df["adaptive"]
    - group_df["robust"]
)

# =====================================================================
# GROUPED STATISTICS
# =====================================================================

delta_meta = group_df[
    "gain_meta"
].values

delta_robust = group_df[
    "gain_robust"
].values

def grouped_stats(delta):

    gain = float(
        np.mean(delta)
    )

    positive = int(
        np.sum(delta > 0)
    )

    positive_pct = (
        100
        * positive
        / len(delta)
    )

    try:

        if (
            np.any(delta > 0)
            and
            np.any(delta < 0)
        ):

            p = float(
                wilcoxon(
                    delta,
                    alternative="greater"
                ).pvalue
            )

        else:

            p = 0.0

    except:

        p = np.nan

    rng_b = np.random.default_rng(
        424242
    )

    B = 5000

    boot = np.empty(B)

    for i in range(B):

        idx = rng_b.integers(
            0,
            len(delta),
            len(delta)
        )

        boot[i] = np.mean(
            delta[idx]
        )

    ci = np.percentile(
        boot,
        [2.5,97.5]
    )

    effect = (
        gain
        /
        (
            np.std(
                delta,
                ddof=1
            )
            + 1e-12
        )
    )

    return (
        gain,
        positive,
        positive_pct,
        p,
        ci,
        effect
    )

(
    group_gain_meta,
    pos_meta,
    pct_meta,
    p_meta,
    ci_meta,
    effect_meta
) = grouped_stats(
    delta_meta
)

(
    group_gain_robust,
    pos_robust,
    pct_robust,
    p_robust,
    ci_robust,
    effect_robust
) = grouped_stats(
    delta_robust
)

# =====================================================================
# UNSEEN FAMILIES
# =====================================================================

unseen = group_df[
    group_df.family.isin(
        UNSEEN_FAMILIES
    )
]

unseen_meta_gain = float(
    unseen["gain_meta"].mean()
)

unseen_robust_gain = float(
    unseen["gain_robust"].mean()
)

unseen_meta_positive = float(
    100*np.mean(
        unseen["gain_meta"] > 0
    )
)

unseen_robust_positive = float(
    100*np.mean(
        unseen["gain_robust"] > 0
    )
)

# =====================================================================
# FAMILY ANALYSIS
# =====================================================================

family_rows = []

for family in TEST_FAMILIES:

    z = group_df[
        group_df.family == family
    ]

    family_rows.append({

        "family": family,

        "adaptive":
            z["adaptive"].mean(),

        "meta":
            z["meta"].mean(),

        "robust":
            z["robust"].mean(),

        "gain_meta":
            z["gain_meta"].mean(),

        "gain_robust":
            z["gain_robust"].mean(),

        "positive_meta_pct":
            100*np.mean(
                z["gain_meta"] > 0
            ),

        "positive_robust_pct":
            100*np.mean(
                z["gain_robust"] > 0
            ),

        "abstain_pct":
            100*z["abstain"].mean(),

        "ood_pct":
            100*z["ood"].mean()

    })

family_df = pd.DataFrame(
    family_rows
)

# =====================================================================
# NOISE
# =====================================================================

noise_rows = []

for noise in NOISES:

    z = group_df[
        group_df.noise == noise
    ]

    noise_rows.append({

        "noise": noise,

        "adaptive":
            z["adaptive"].mean(),

        "meta":
            z["meta"].mean(),

        "robust":
            z["robust"].mean(),

        "gain_meta":
            z["gain_meta"].mean(),

        "gain_robust":
            z["gain_robust"].mean()

    })

noise_df = pd.DataFrame(
    noise_rows
)

# =====================================================================
# SELECTION / ABSTENTION
# =====================================================================

selection = (
    df["selected"]
    .value_counts()
    .to_dict()
)

robust_selection = (
    df["robust_selected"]
    .value_counts()
    .to_dict()
)

abstain_pct = (
    100*df["abstain"].mean()
)

ood_pct = (
    100*df["ood"].mean()
)

# =====================================================================
# DECISION LOGIC
#
# Strongest criterion:
# robust method must improve over adaptive on grouped data,
# with bootstrap CI > 0 and positive unseen-family performance.
# =====================================================================

if (
    group_gain_robust > 0
    and ci_robust[0] > 0
    and p_robust < 0.05
    and unseen_robust_gain > 0
    and unseen_robust_positive >= 50
):

    verdict = (
        "VALIDACIÓN POSITIVA: "
        "ABSTENCIÓN/FALLBACK MEJORA "
        "LA GENERALIZACIÓN OOD"
    )

elif (
    group_gain_robust > 0
    and unseen_robust_gain > 0
):

    verdict = (
        "SEÑAL POSITIVA: "
        "INCERTIDUMBRE/FALLBACK "
        "MEJORA META, PERO SIN "
        "EVIDENCIA ROBUSTA SUFICIENTE"
    )

elif group_gain_robust > 0:

    verdict = (
        "MEJORA INTERNA: "
        "FALLBACK AYUDA GLOBALMENTE, "
        "PERO NO GENERALIZA OOD"
    )

else:

    verdict = (
        "HIPÓTESIS NO VALIDADA: "
        "INCERTIDUMBRE/FALLBACK "
        "NO MEJORA ADAPTIVE"
    )

# =====================================================================
# SUMMARY
# =====================================================================

summary = {

    "version":
        "CEUNIA_OMEGA_v2_6",

    "observations":
        int(len(df)),

    "series":
        int(len(group_df)),

    "train":
        TRAIN,

    "oos":
        N-TRAIN,

    "adaptive_fixed":
        float(adaptive_fixed),

    "meta_selector":
        float(meta),

    "meta_uncertainty_fallback":
        float(meta_uncertain),

    "random":
        float(random_mae),

    "oracle":
        float(oracle_mae),

    "point_gain_meta":
        float(gain_meta),

    "point_gain_robust":
        float(gain_robust),

    "group_gain_meta":
        float(group_gain_meta),

    "group_gain_robust":
        float(group_gain_robust),

    "group_positive_meta_pct":
        float(pct_meta),

    "group_positive_robust_pct":
        float(pct_robust),

    "wilcoxon_meta":
        float(p_meta),

    "wilcoxon_robust":
        float(p_robust),

    "bootstrap_meta_ci95":
        [
            float(ci_meta[0]),
            float(ci_meta[1])
        ],

    "bootstrap_robust_ci95":
        [
            float(ci_robust[0]),
            float(ci_robust[1])
        ],

    "effect_meta":
        float(effect_meta),

    "effect_robust":
        float(effect_robust),

    "unseen_meta_gain":
        float(unseen_meta_gain),

    "unseen_robust_gain":
        float(unseen_robust_gain),

    "unseen_meta_positive_pct":
        float(unseen_meta_positive),

    "unseen_robust_positive_pct":
        float(unseen_robust_positive),

    "abstention_pct":
        float(abstain_pct),

    "ood_pct":
        float(ood_pct),

    "ood_threshold":
        float(OOD_THRESHOLD),

    "margin_threshold":
        float(MARGIN_THRESHOLD),

    "selection":
        selection,

    "robust_selection":
        robust_selection,

    "verdict":
        verdict
}

# =====================================================================
# HASH
# =====================================================================

sha = hashlib.sha256(
    json.dumps(
        summary,
        sort_keys=True,
        default=float
    ).encode()
).hexdigest()

# =====================================================================
# PERSISTENCIA
# =====================================================================

conn = sqlite3.connect(
    DB
)

df.to_sql(
    "ceunia_v26_oos",
    conn,
    if_exists="replace",
    index=False
)

group_df.to_sql(
    "ceunia_v26_grouped",
    conn,
    if_exists="replace",
    index=False
)

family_df.to_sql(
    "ceunia_v26_family",
    conn,
    if_exists="replace",
    index=False
)

noise_df.to_sql(
    "ceunia_v26_noise",
    conn,
    if_exists="replace",
    index=False
)

conn.execute("""
CREATE TABLE IF NOT EXISTS
ceunia_v26_summary(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    version TEXT,
    observations INTEGER,
    series INTEGER,
    adaptive_fixed REAL,
    meta_selector REAL,
    meta_uncertainty_fallback REAL,
    random_selector REAL,
    oracle REAL,
    point_gain_meta REAL,
    point_gain_robust REAL,
    group_gain_meta REAL,
    group_gain_robust REAL,
    group_positive_meta_pct REAL,
    group_positive_robust_pct REAL,
    wilcoxon_meta REAL,
    wilcoxon_robust REAL,
    meta_ci_low REAL,
    meta_ci_high REAL,
    robust_ci_low REAL,
    robust_ci_high REAL,
    effect_meta REAL,
    effect_robust REAL,
    unseen_meta_gain REAL,
    unseen_robust_gain REAL,
    unseen_meta_positive_pct REAL,
    unseen_robust_positive_pct REAL,
    abstention_pct REAL,
    ood_pct REAL,
    ood_threshold REAL,
    margin_threshold REAL,
    sha256 TEXT,
    verdict TEXT
)
""")

conn.execute("""
INSERT INTO ceunia_v26_summary VALUES(
    NULL,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?
)
""", (
    "CEUNIA_OMEGA_v2_6",
    len(df),
    len(group_df),
    adaptive_fixed,
    meta,
    meta_uncertain,
    random_mae,
    oracle_mae,
    gain_meta,
    gain_robust,
    group_gain_meta,
    group_gain_robust,
    pct_meta,
    pct_robust,
    p_meta,
    p_robust,
    ci_meta[0],
    ci_meta[1],
    ci_robust[0],
    ci_robust[1],
    effect_meta,
    effect_robust,
    unseen_meta_gain,
    unseen_robust_gain,
    unseen_meta_positive,
    unseen_robust_positive,
    abstain_pct,
    ood_pct,
    OOD_THRESHOLD,
    MARGIN_THRESHOLD,
    sha,
    verdict
))

conn.commit()
conn.close()

# =====================================================================
# JSON REPORT
# =====================================================================

REPORT = (
    "/content/"
    "CEUNIA_OMEGA_v2_6_REPORT.json"
)

with open(
    REPORT,
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2,
        default=float
    )

# =====================================================================
# OUTPUT
# =====================================================================

print("\n" + "="*78)
print(
    "CEUNIA Ω-v2.6 — "
    "META + INCERTIDUMBRE + ABSTENCIÓN"
)
print("="*78)

print("\nOBSERVACIONES :", len(df))
print("SERIES        :", len(group_df))
print("TRAIN         :", TRAIN)
print("OOS           :", N-TRAIN)

print("\nMAE GLOBAL")

print(
    "Adaptive Fixed        : {:.8f}"
    .format(adaptive_fixed)
)

print(
    "Meta-Selector         : {:.8f}"
    .format(meta)
)

print(
    "Meta + Fallback       : {:.8f}"
    .format(meta_uncertain)
)

print(
    "Random                : {:.8f}"
    .format(random_mae)
)

print(
    "Oracle                : {:.8f}"
    .format(oracle_mae)
)

print("\nVALOR MARGINAL META")

print(
    "Gain Meta             : {:+.8f}"
    .format(gain_meta)
)

print(
    "Gain Meta + Fallback  : {:+.8f}"
    .format(gain_robust)
)

print("\nEVIDENCIA AGRUPADA — META")

print(
    "Positivos             : {}/{}"
    .format(
        pos_meta,
        len(delta_meta)
    )
)

print(
    "Proporción            : {:.2f}%"
    .format(pct_meta)
)

print(
    "Wilcoxon p            : {}"
    .format(p_meta)
)

print(
    "Bootstrap IC95        : "
    "[{:.8f}, {:.8f}]"
    .format(
        ci_meta[0],
        ci_meta[1]
    )
)

print(
    "Effect size           : {:.6f}"
    .format(effect_meta)
)

print("\nEVIDENCIA AGRUPADA — META + FALLBACK")

print(
    "Positivos             : {}/{}"
    .format(
        pos_robust,
        len(delta_robust)
    )
)

print(
    "Proporción            : {:.2f}%"
    .format(pct_robust)
)

print(
    "Wilcoxon p            : {}"
    .format(p_robust)
)

print(
    "Bootstrap IC95        : "
    "[{:.8f}, {:.8f}]"
    .format(
        ci_robust[0],
        ci_robust[1]
    )
)

print(
    "Effect size           : {:.6f}"
    .format(effect_robust)
)

print("\nDETECCIÓN DE INCERTIDUMBRE")

print(
    "Abstención             : {:.2f}%"
    .format(abstain_pct)
)

print(
    "Contextos OOD          : {:.2f}%"
    .format(ood_pct)
)

print(
    "OOD threshold          : {:.6f}"
    .format(OOD_THRESHOLD)
)

print(
    "Margin threshold       : {:.6f}"
    .format(MARGIN_THRESHOLD)
)

print("\nFAMILIAS NO VISTAS")

print(
    "Meta gain              : {:+.8f}"
    .format(unseen_meta_gain)
)

print(
    "Meta + fallback gain   : {:+.8f}"
    .format(unseen_robust_gain)
)

print(
    "Meta positivas         : {:.2f}%"
    .format(unseen_meta_positive)
)

print(
    "Fallback positivas     : {:.2f}%"
    .format(unseen_robust_positive)
)

print("\nSELECCIÓN META")

for k, v in sorted(
    selection.items(),
    key=lambda x: -x[1]
):

    print(
        "{:<12} {}"
        .format(k, v)
    )

print("\nSELECCIÓN ROBUSTA")

for k, v in sorted(
    robust_selection.items(),
    key=lambda x: -x[1]
):

    print(
        "{:<12} {}"
        .format(k, v)
    )

print("\nPOR FAMILIA")

print(
    family_df.to_string(
        index=False,
        float_format=lambda x:
            "{:.6f}".format(x)
    )
)

print("\nPOR RUIDO")

print(
    noise_df.to_string(
        index=False,
        float_format=lambda x:
            "{:.6f}".format(x)
    )
)

print("\nVEREDICTO")
print(verdict)

print("\nSHA-256")
print(sha)

print("\nDB")
print(DB)

print("\nREPORT")
print(REPORT)

print("\n" + "="*78)
```
