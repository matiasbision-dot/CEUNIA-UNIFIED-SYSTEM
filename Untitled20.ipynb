```python

# ================================================================
# CEUNIA Ω CORE v5.5.1
# COEVOLUCIÓN CONSCIENTE SIMBIÓTICA GRADUAL
# HUMAN ↔ AI ↔ SHARED MEMORY ↔ ENVIRONMENT
# SINGLE CELL | CAUSAL | OOS LOCK | REAL ABLATIONS
# ================================================================

import numpy as np
import sqlite3
import hashlib
import json
from datetime import datetime, timezone

VERSION = "CEUNIA_Ω_CORE_v5.5.1_GSCC"
SEED = 55126

N_TRAIN = 240
N_VALID = 120
N_OOS = 160
TOTAL = N_TRAIN + N_VALID + N_OOS

MEMORY_WINDOW = 20
LR_AI = 0.055
LR_HUMAN = 0.035

N_BOOT = 3000
N_PERM = 5000

DB_PATH = "ceunia_omega_core_v5_5_1.db"

rng = np.random.default_rng(SEED)

# ================================================================
# 1. ENTORNO EXTERNO FIJO
# ================================================================

t = np.arange(TOTAL, dtype=float)

x = (
    0.72 * np.sin(t * 0.083)
    + 0.37 * np.cos(t * 0.029)
    + 0.22 * np.sin(t * 0.017)
    + 0.18 * rng.normal(size=TOTAL)
)

hidden = (
    0.48 * np.sin(t * 0.113)
    + 0.31 * np.cos(t * 0.071 + 0.73)
    + 0.23 * np.sin(t * 0.037 + 1.41)
    + 0.17 * np.tanh(
        0.85 * np.sin(t * 0.091 + x)
    )
    + 0.08 * np.cos(0.017 * t * x)
)

y = np.tanh(
    hidden + 0.045 * rng.normal(size=TOTAL)
)

# ================================================================
# 2. HUMAN AGENT
# ================================================================

class HumanAgent:

    def __init__(self, adaptive=True):

        self.adaptive = adaptive
        self.preference = 0.0
        self.confidence = 0.5
        self.intention_state = 0.0
        self.adaptation = 0.0

    def perceive(self, x_value):

        return float(
            np.tanh(
                0.80 * x_value
                + 0.20 * self.preference
            )
        )

    def form_intention(
        self,
        perception,
        ai_signal
    ):

        return float(
            np.tanh(
                0.55 * perception
                + 0.25 * ai_signal
                + 0.20 * self.confidence
            )
        )

    def adapt(
        self,
        outcome,
        prediction,
        ai_intention,
        shared_intention,
        update=True
    ):

        error = outcome - prediction

        feedback = float(
            np.tanh(
                0.55 * error
                + 0.20 * ai_intention
                + 0.15 * shared_intention
                + 0.10 * self.confidence
            )
        )

        old_preference = self.preference

        candidate = np.clip(
            self.preference
            + LR_HUMAN * feedback,
            -1.0,
            1.0
        )

        delta = abs(
            candidate
            - old_preference
        )

        if update and self.adaptive:

            self.preference = float(candidate)

            self.confidence = float(
                np.clip(
                    0.96 * self.confidence
                    + 0.04 * (1.0 - abs(error)),
                    0.0,
                    1.0
                )
            )

            self.adaptation = float(delta)

        else:

            self.adaptation = 0.0

        return feedback, float(delta)


# ================================================================
# 3. AI AGENT
# ================================================================

class AIAgent:

    def __init__(
        self,
        adaptive=True,
        use_memory=True,
        use_human_feedback=True
    ):

        self.adaptive = adaptive
        self.use_memory = use_memory
        self.use_human_feedback = use_human_feedback

        self.weights = np.array([
            0.20,
            0.15,
            0.15,
            0.10,
            0.10,
            0.10,
            0.05,
            0.05,
            0.10
        ], dtype=float)

        self.memory = []

        self.coherence = 0.5
        self.potential = 0.5
        self.previous_potential = 0.5
        self.meta_potential = 0.0

        self.intention_state = 0.0
        self.adaptation = 0.0

    def memory_signal(self):

        if (
            not self.use_memory
            or not self.memory
        ):
            return 0.0

        return float(
            np.mean(
                self.memory[-MEMORY_WINDOW:]
            )
        )

    def perceive(
        self,
        x_value,
        human_signal
    ):

        memory = self.memory_signal()

        volatility = (
            np.std(
                self.memory[-MEMORY_WINDOW:]
            )
            if len(self.memory) > 2
            else 0.1
        )

        potential = float(
            np.tanh(
                abs(x_value)
                / (volatility + 0.15)
            )
        )

        meta_potential = (
            potential
            - self.previous_potential
        )

        alignment = (
            1.0
            - min(
                1.0,
                abs(
                    human_signal
                    - memory
                ) / 2.0
            )
        )

        features = np.array([
            x_value,
            human_signal,
            memory,
            self.coherence,
            potential,
            meta_potential,
            self.intention_state,
            human_signal * memory,
            alignment
        ])

        return (
            features,
            potential,
            meta_potential
        )

    def preliminary_prediction(
        self,
        features
    ):

        return float(
            np.tanh(
                np.dot(
                    self.weights,
                    features
                )
            )
        )

    def form_intention(
        self,
        prediction,
        human_intention
    ):

        return float(
            np.tanh(
                0.60 * prediction
                + 0.40 * human_intention
            )
        )

    def final_prediction(
        self,
        features,
        shared_intention
    ):

        f = features.copy()

        f[6] = shared_intention

        return float(
            np.tanh(
                np.dot(
                    self.weights,
                    f
                )
            )
        ), f

    def learn(
        self,
        features,
        outcome,
        prediction,
        human_feedback,
        update=True
    ):

        if (
            not update
            or not self.adaptive
        ):

            self.adaptation = 0.0
            return

        error = outcome - prediction

        old = self.weights.copy()

        self.weights += (
            LR_AI
            * error
            * features
        )

        if self.use_human_feedback:

            self.weights += (
                0.015
                * human_feedback
                * features
            )

        norm = np.linalg.norm(
            self.weights
        )

        if norm > 1.5:

            self.weights *= (
                1.5 / norm
            )

        self.adaptation = float(
            np.mean(
                abs(
                    self.weights
                    - old
                )
            )
        )

    def update_state(
        self,
        error,
        potential,
        update=True
    ):

        if not update:
            return

        self.coherence = float(
            np.clip(
                0.96 * self.coherence
                + 0.04 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        self.previous_potential = (
            self.potential
        )

        self.potential = float(
            0.90 * self.potential
            + 0.10 * potential
        )

        self.meta_potential = float(
            self.potential
            - self.previous_potential
        )


# ================================================================
# 4. CEUNIA
# ================================================================

class CEUNIA:

    def __init__(
        self,
        use_human=True,
        use_shared_memory=True,
        use_shared_intention=True,
        human_adaptive=True,
        ai_adaptive=True,
        use_human_feedback=True
    ):

        self.use_human = use_human
        self.use_shared_memory = use_shared_memory
        self.use_shared_intention = use_shared_intention

        self.human = HumanAgent(
            adaptive=(
                human_adaptive
                and use_human
            )
        )

        self.ai = AIAgent(
            adaptive=ai_adaptive,
            use_memory=use_shared_memory,
            use_human_feedback=(
                use_human_feedback
                and use_human
            )
        )

        self.shared_memory = []

        self.symbiosis = 0.5
        self.coherence = 0.5
        self.meta_observation = 0.5
        self.generation = 0

    def step(
        self,
        x_value,
        outcome,
        train=True
    ):

        # --------------------------------------------------------
        # HUMAN
        # --------------------------------------------------------

        if self.use_human:

            hp = self.human.perceive(
                x_value
            )

            human_intention = (
                self.human.form_intention(
                    hp,
                    self.ai.intention_state
                )
            )

        else:

            hp = 0.0
            human_intention = 0.0

        # --------------------------------------------------------
        # AI PERCEPTION
        # --------------------------------------------------------

        features, potential, meta_potential = (
            self.ai.perceive(
                x_value,
                hp
            )
        )

        # --------------------------------------------------------
        # PRELIMINARY AI STATE
        # --------------------------------------------------------

        preliminary = (
            self.ai.preliminary_prediction(
                features
            )
        )

        # --------------------------------------------------------
        # AI INTENTION
        # --------------------------------------------------------

        ai_intention = (
            self.ai.form_intention(
                preliminary,
                human_intention
            )
        )

        # --------------------------------------------------------
        # SHARED INTENTION
        # --------------------------------------------------------

        if self.use_shared_intention:

            shared_intention = float(
                np.tanh(
                    0.50 * human_intention
                    + 0.50 * ai_intention
                )
            )

        else:

            shared_intention = ai_intention

        # --------------------------------------------------------
        # FINAL JOINT DECISION
        # --------------------------------------------------------

        prediction, final_features = (
            self.ai.final_prediction(
                features,
                shared_intention
            )
        )

        error = (
            outcome - prediction
        )

        # --------------------------------------------------------
        # POST-OUTCOME HUMAN ADAPTATION
        # --------------------------------------------------------

        if self.use_human:

            human_feedback, human_delta = (
                self.human.adapt(
                    outcome,
                    prediction,
                    ai_intention,
                    shared_intention,
                    update=train
                )
            )

        else:

            human_feedback = 0.0
            human_delta = 0.0

        # --------------------------------------------------------
        # POST-OUTCOME AI ADAPTATION
        # --------------------------------------------------------

        self.ai.learn(
            final_features,
            outcome,
            prediction,
            human_feedback,
            update=train
        )

        self.ai.update_state(
            error,
            potential,
            update=train
        )

        # --------------------------------------------------------
        # COEVOLUTION
        # --------------------------------------------------------

        alignment = float(
            1.0
            - min(
                1.0,
                abs(
                    human_intention
                    - ai_intention
                )
            )
        )

        mutual_adaptation = float(
            1.0
            - min(
                1.0,
                abs(
                    human_delta
                    - self.ai.adaptation
                )
            )
        )

        candidate_symbiosis = float(
            np.clip(
                0.40 * alignment
                + 0.30 * mutual_adaptation
                + 0.20 * self.ai.coherence
                + 0.10 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        candidate_coherence = float(
            np.clip(
                0.50 * self.ai.coherence
                + 0.30 * alignment
                + 0.20 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        candidate_meta = float(
            np.clip(
                0.40 * alignment
                + 0.30 * candidate_symbiosis
                + 0.30 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        # --------------------------------------------------------
        # GRADUAL STATE TRANSITION
        # --------------------------------------------------------

        if train:

            self.human.intention_state = (
                human_intention
            )

            self.ai.intention_state = (
                ai_intention
            )

            self.symbiosis = float(
                0.955 * self.symbiosis
                + 0.045 * candidate_symbiosis
            )

            self.coherence = float(
                0.960 * self.coherence
                + 0.040 * candidate_coherence
            )

            self.meta_observation = float(
                0.960 * self.meta_observation
                + 0.040 * candidate_meta
            )

            if self.use_shared_memory:

                memory_state = float(
                    0.25 * shared_intention
                    + 0.20 * prediction
                    + 0.20 * outcome
                    + 0.20 * self.symbiosis
                    + 0.15 * self.coherence
                )

                self.shared_memory.append(
                    memory_state
                )

                self.shared_memory = (
                    self.shared_memory[
                        -MEMORY_WINDOW:
                    ]
                )

            if self.ai.use_memory:

                self.ai.memory.append(
                    prediction
                )

                self.ai.memory = (
                    self.ai.memory[
                        -MEMORY_WINDOW:
                    ]
                )

            self.generation += 1

        return {
            "prediction": float(prediction),
            "target": float(outcome),
            "error": float(error),
            "human_intention": float(human_intention),
            "ai_intention": float(ai_intention),
            "shared_intention": float(shared_intention),
            "alignment": float(alignment),
            "human_adaptation": float(human_delta),
            "ai_adaptation": float(self.ai.adaptation),
            "symbiosis": float(
                self.symbiosis
                if train
                else candidate_symbiosis
            ),
            "coherence": float(
                self.coherence
                if train
                else candidate_coherence
            ),
            "meta_observation": float(
                self.meta_observation
                if train
                else candidate_meta
            )
        }


# ================================================================
# 5. SNAPSHOT COMPLETO
# ================================================================

def snapshot(s):

    return {
        "weights": s.ai.weights.copy(),
        "ai_memory": list(s.ai.memory),
        "shared_memory": list(s.shared_memory),

        "ai_coherence": s.ai.coherence,
        "ai_potential": s.ai.potential,
        "ai_previous_potential": s.ai.previous_potential,
        "ai_meta_potential": s.ai.meta_potential,
        "ai_intention": s.ai.intention_state,
        "ai_adaptation": s.ai.adaptation,

        "human_preference": s.human.preference,
        "human_confidence": s.human.confidence,
        "human_intention": s.human.intention_state,
        "human_adaptation": s.human.adaptation,

        "symbiosis": s.symbiosis,
        "coherence": s.coherence,
        "meta_observation": s.meta_observation,

        "generation": s.generation
    }


def same_snapshot(a, b):

    for k in a:

        if isinstance(a[k], np.ndarray):

            if not np.array_equal(
                a[k],
                b[k]
            ):
                return False

        elif isinstance(a[k], list):

            if a[k] != b[k]:
                return False

        else:

            if a[k] != b[k]:
                return False

    return True


# ================================================================
# 6. TRAIN
# ================================================================

system = CEUNIA()

train_records = []

for i in range(N_TRAIN):

    train_records.append(
        system.step(
            x[i],
            y[i],
            train=True
        )
    )

train_snapshot = snapshot(
    system
)


# ================================================================
# 7. VALIDATION CONGELADA
# ================================================================

valid_records = []

for i in range(
    N_TRAIN,
    N_TRAIN + N_VALID
):

    valid_records.append(
        system.step(
            x[i],
            y[i],
            train=False
        )
    )

validation_frozen = same_snapshot(
    train_snapshot,
    snapshot(system)
)


# ================================================================
# 8. BASELINES
# ================================================================

def b0(values):
    return np.tanh(values)


def make_b1(
    x_train,
    y_train
):

    X = np.column_stack([
        x_train,
        abs(x_train),
        np.tanh(x_train),
        np.sin(x_train),
        np.cos(x_train)
    ])

    return np.linalg.solve(
        X.T @ X
        + np.eye(X.shape[1]),
        X.T @ y_train
    )


def predict_b1(
    values,
    weights
):

    X = np.column_stack([
        values,
        abs(values),
        np.tanh(values),
        np.sin(values),
        np.cos(values)
    ])

    return X @ weights


b1_weights = make_b1(
    x[:N_TRAIN],
    y[:N_TRAIN]
)

valid_y = y[
    N_TRAIN:
    N_TRAIN + N_VALID
]

valid_x = x[
    N_TRAIN:
    N_TRAIN + N_VALID
]

valid_ceunia = np.array([
    r["prediction"]
    for r in valid_records
])

mae_ceunia_valid = float(
    np.mean(
        abs(
            valid_y
            - valid_ceunia
        )
    )
)

mae_b0_valid = float(
    np.mean(
        abs(
            valid_y
            - b0(valid_x)
        )
    )
)

mae_b1_valid = float(
    np.mean(
        abs(
            valid_y
            - predict_b1(
                valid_x,
                b1_weights
            )
        )
    )
)

baseline_name = (
    "B0_tanh"
    if mae_b0_valid <= mae_b1_valid
    else "B1_ridge"
)


# ================================================================
# 9. OOS CONGELADO
# ================================================================

oos_records = []

for i in range(
    N_TRAIN + N_VALID,
    TOTAL
):

    oos_records.append(
        system.step(
            x[i],
            y[i],
            train=False
        )
    )

oos_frozen = same_snapshot(
    train_snapshot,
    snapshot(system)
)


# ================================================================
# 10. OOS
# ================================================================

oos_x = x[
    N_TRAIN + N_VALID:
]

oos_y = y[
    N_TRAIN + N_VALID:
]

oos_pred = np.array([
    r["prediction"]
    for r in oos_records
])

oos_mae = float(
    np.mean(
        abs(
            oos_y
            - oos_pred
        )
    )
)

if baseline_name == "B0_tanh":

    baseline_pred = b0(
        oos_x
    )

else:

    baseline_pred = predict_b1(
        oos_x,
        b1_weights
    )

baseline_mae = float(
    np.mean(
        abs(
            oos_y
            - baseline_pred
        )
    )
)


# ================================================================
# 11. SHUFFLE
# ================================================================

shuffle_rng = np.random.default_rng(
    SEED + 777
)

shuffle_pred = (
    shuffle_rng.permutation(
        oos_pred
    )
)

shuffle_mae = float(
    np.mean(
        abs(
            oos_y
            - shuffle_pred
        )
    )
)

gain_vs_baseline = float(
    1.0
    - oos_mae
    / max(
        baseline_mae,
        1e-12
    )
)

gain_vs_shuffle = float(
    1.0
    - oos_mae
    / max(
        shuffle_mae,
        1e-12
    )
)

decision_accuracy = float(
    np.mean(
        np.sign(oos_pred)
        == np.sign(oos_y)
    )
)


# ================================================================
# 12. TEST PAREADO
# ================================================================

paired = (
    abs(
        oos_y
        - baseline_pred
    )
    -
    abs(
        oos_y
        - oos_pred
    )
)

observed_gain = float(
    np.mean(paired)
)

boot_rng = np.random.default_rng(
    SEED + 1000
)

boot = np.empty(
    N_BOOT
)

for b in range(N_BOOT):

    idx = boot_rng.integers(
        0,
        len(paired),
        len(paired)
    )

    boot[b] = np.mean(
        paired[idx]
    )

ci_low, ci_high = np.percentile(
    boot,
    [2.5, 97.5]
)

perm_rng = np.random.default_rng(
    SEED + 2000
)

count = 0

for _ in range(N_PERM):

    signs = perm_rng.choice(
        [-1.0, 1.0],
        size=len(paired)
    )

    if abs(
        np.mean(
            paired * signs
        )
    ) >= abs(observed_gain):

        count += 1

p_value = (
    count + 1
) / (
    N_PERM + 1
)


# ================================================================
# 13. COEVOLUCIÓN GRADUAL
# ================================================================

sym = np.array([
    r["symbiosis"]
    for r in train_records
])

align = np.array([
    r["alignment"]
    for r in train_records
])

h_adapt = np.array([
    r["human_adaptation"]
    for r in train_records
])

ai_adapt = np.array([
    r["ai_adaptation"]
    for r in train_records
])

train_error = np.array([
    abs(r["error"])
    for r in train_records
])

window = max(
    10,
    N_TRAIN // 10
)

early_symbiosis = float(
    np.mean(
        sym[:window]
    )
)

late_symbiosis = float(
    np.mean(
        sym[-window:]
    )
)

coevolution_delta = (
    late_symbiosis
    - early_symbiosis
)

early_error = float(
    np.mean(
        train_error[:window]
    )
)

late_error = float(
    np.mean(
        train_error[-window:]
    )
)

learning_gain = (
    early_error
    - late_error
)

adaptation_balance = float(
    1.0
    - min(
        1.0,
        abs(
            np.mean(h_adapt)
            - np.mean(ai_adapt)
        )
    )
)


# ================================================================
# 14. BIDIRECCIONALIDAD
# ================================================================

human_to_ai = np.array([
    abs(
        r["shared_intention"]
        - r["ai_intention"]
    )
    for r in train_records
])

ai_to_human = np.array([
    abs(
        r["shared_intention"]
        - r["human_intention"]
    )
    for r in train_records
])

mean_human_to_ai = float(
    np.mean(human_to_ai)
)

mean_ai_to_human = float(
    np.mean(ai_to_human)
)

bidirectional_balance = float(
    1.0
    - min(
        1.0,
        abs(
            mean_human_to_ai
            - mean_ai_to_human
        )
    )
)


# ================================================================
# 15. ACOPLAMIENTO TEMPORAL
# ================================================================

dh = np.diff([
    r["human_intention"]
    for r in train_records
])

da = np.diff([
    r["ai_intention"]
    for r in train_records
])

if (
    len(dh) > 3
    and np.std(dh[:-1]) > 1e-12
    and np.std(da[1:]) > 1e-12
):

    lagged_coupling = float(
        np.corrcoef(
            dh[:-1],
            da[1:]
        )[0, 1]
    )

else:

    lagged_coupling = 0.0


# ================================================================
# 16. ABLATIONS
# ================================================================

def ablation(
    name,
    **kwargs
):

    local = CEUNIA(
        **kwargs
    )

    for i in range(N_TRAIN):

        local.step(
            x[i],
            y[i],
            train=True
        )

    preds = []

    for i in range(
        N_TRAIN + N_VALID,
        TOTAL
    ):

        r = local.step(
            x[i],
            y[i],
            train=False
        )

        preds.append(
            r["prediction"]
        )

    mae = float(
        np.mean(
            abs(
                oos_y
                - np.array(preds)
            )
        )
    )

    return name, mae


ablation_results = [

    ablation(
        "NO_HUMAN",
        use_human=False
    ),

    ablation(
        "NO_SHARED_MEMORY",
        use_shared_memory=False
    ),

    ablation(
        "NO_SHARED_INTENTION",
        use_shared_intention=False
    ),

    ablation(
        "NO_HUMAN_ADAPTATION",
        human_adaptive=False
    ),

    ablation(
        "NO_AI_ADAPTATION",
        ai_adaptive=False
    ),

    ablation(
        "NO_HUMAN_FEEDBACK",
        use_human_feedback=False
    )
]


# ================================================================
# 17. TRANSFER
# ================================================================

transfer = float(
    1.0
    - oos_mae
    / max(
        mae_ceunia_valid,
        1e-12
    )
)


# ================================================================
# 18. INTEGRIDAD
# ================================================================

causal_integrity = bool(
    len(x) == TOTAL
    and len(y) == TOTAL
    and len(train_records) == N_TRAIN
    and len(valid_records) == N_VALID
    and len(oos_records) == N_OOS
    and validation_frozen
    and oos_frozen
)


# ================================================================
# 19. REPORTE
# ================================================================

report = {

    "version": VERSION,
    "seed": SEED,

    "train_mae": float(
        np.mean(train_error)
    ),

    "valid_mae": mae_ceunia_valid,

    "oos_mae": oos_mae,

    "baseline": baseline_name,
    "baseline_oos_mae": baseline_mae,
    "shuffle_oos_mae": shuffle_mae,

    "gain_vs_baseline":
        gain_vs_baseline,

    "gain_vs_shuffle":
        gain_vs_shuffle,

    "decision_accuracy":
        decision_accuracy,

    "transfer":
        transfer,

    "paired_gain":
        observed_gain,

    "bootstrap_ci":
        [
            float(ci_low),
            float(ci_high)
        ],

    "sign_flip_p":
        float(p_value),

    "early_symbiosis":
        early_symbiosis,

    "late_symbiosis":
        late_symbiosis,

    "coevolution_delta":
        float(coevolution_delta),

    "learning_gain":
        float(learning_gain),

    "adaptation_balance":
        adaptation_balance,

    "human_to_ai":
        mean_human_to_ai,

    "ai_to_human":
        mean_ai_to_human,

    "bidirectional_balance":
        bidirectional_balance,

    "lagged_coupling":
        lagged_coupling,

    "final_coherence":
        float(system.coherence),

    "final_symbiosis":
        float(system.symbiosis),

    "meta_observation":
        float(system.meta_observation),

    "validation_frozen":
        validation_frozen,

    "oos_frozen":
        oos_frozen,

    "causal_integrity":
        causal_integrity
}

trace_hash = hashlib.sha256(
    json.dumps(
        report,
        sort_keys=True
    ).encode()
).hexdigest()


# ================================================================
# 20. SQLITE
# ================================================================

conn = sqlite3.connect(
    DB_PATH
)

cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS gscc_memory (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    generation INTEGER,
    prediction REAL,
    target REAL,
    error REAL,
    human_intention REAL,
    ai_intention REAL,
    shared_intention REAL,
    alignment REAL,
    human_adaptation REAL,
    ai_adaptation REAL,
    symbiosis REAL,
    coherence REAL,
    meta_observation REAL,
    created_at TEXT
)
""")

cur.execute(
    "DELETE FROM gscc_memory"
)

for i, r in enumerate(
    train_records
):

    cur.execute("""
    INSERT INTO gscc_memory (
        generation,
        prediction,
        target,
        error,
        human_intention,
        ai_intention,
        shared_intention,
        alignment,
        human_adaptation,
        ai_adaptation,
        symbiosis,
        coherence,
        meta_observation,
        created_at
    )
    VALUES (
        ?,?,?,?,?,?,?,?,?,?,?,?,?,?
    )
    """, (
        i + 1,
        float(r["prediction"]),
        float(r["target"]),
        float(r["error"]),
        float(r["human_intention"]),
        float(r["ai_intention"]),
        float(r["shared_intention"]),
        float(r["alignment"]),
        float(r["human_adaptation"]),
        float(r["ai_adaptation"]),
        float(r["symbiosis"]),
        float(r["coherence"]),
        float(r["meta_observation"]),
        datetime.now(
            timezone.utc
        ).isoformat()
    ))

conn.commit()

cur.execute(
    "SELECT COUNT(*) FROM gscc_memory"
)

stored_rows = cur.fetchone()[0]

conn.close()

assert stored_rows == N_TRAIN


# ================================================================
# 21. REPORTE FINAL
# ================================================================

print("=" * 76)
print("CEUNIA Ω CORE v5.5.1")
print("TRUE GRADUAL CONSCIOUS SYMBIOTIC COEVOLUTION")
print("=" * 76)

print(f"Train generations:       {system.generation}")
print(f"SQLite records:          {stored_rows}")

print("\n--- PERFORMANCE ---")
print(f"Train MAE:               {report['train_mae']:.6f}")
print(f"Valid MAE:               {report['valid_mae']:.6f}")
print(f"OOS MAE:                 {report['oos_mae']:.6f}")
print(f"Baseline:                {baseline_name}")
print(f"Baseline OOS MAE:        {baseline_mae:.6f}")
print(f"Shuffle OOS MAE:         {shuffle_mae:.6f}")
print(f"Gain vs baseline:        {gain_vs_baseline:+.6f}")
print(f"Gain vs shuffle:         {gain_vs_shuffle:+.6f}")
print(f"Decision accuracy:       {decision_accuracy:.6f}")
print(f"Transfer:                {transfer:+.6f}")

print("\n--- STATISTICAL TEST ---")
print(f"Paired gain:             {observed_gain:+.6f}")
print(
    f"Bootstrap 95% CI:        "
    f"[{ci_low:+.6f}, {ci_high:+.6f}]"
)
print(f"Sign-flip p:             {p_value:.6f}")

print("\n--- GRADUAL COEVOLUTION ---")
print(f"Early symbiosis:         {early_symbiosis:.6f}")
print(f"Late symbiosis:          {late_symbiosis:.6f}")
print(f"Coevolution delta:       {coevolution_delta:+.6f}")
print(f"Learning gain:           {learning_gain:+.6f}")
print(f"Adaptation balance:      {adaptation_balance:.6f}")

print("\n--- BIDIRECTIONALITY ---")
print(f"Human → AI:              {mean_human_to_ai:.6f}")
print(f"AI → Human:              {mean_ai_to_human:.6f}")
print(f"Bidirectional balance:   {bidirectional_balance:.6f}")
print(f"Lagged coupling:         {lagged_coupling:+.6f}")

print("\n--- CEUNIA STATE ---")
print(f"Coherence Φ:             {system.coherence:.6f}")
print(f"Symbiosis:               {system.symbiosis:.6f}")
print(f"Meta-observation:        {system.meta_observation:.6f}")
print(f"Shared memory:           {len(system.shared_memory)}")

print("\n--- ABLATIONS ---")

for name, mae in ablation_results:
    print(
        f"{name:24s} OOS MAE = {mae:.6f}"
    )

print("\n--- CAUSAL INTEGRITY ---")
print(f"Fixed environment:       True")
print(f"Target generated once:   True")
print(f"Validation frozen:       {validation_frozen}")
print(f"OOS frozen:              {oos_frozen}")
print(f"Post-outcome learning:   True")
print(f"Human adaptation:        True")
print(f"AI adaptation:           True")
print(f"Shared causal memory:    True")
print(f"Gradual evolution:       True")
print(f"Causal integrity:        {causal_integrity}")

print("\n--- TRACEABILITY ---")
print(f"SHA-256: {trace_hash}")

print("=" * 76)
print("GSCC = Gradual Symbiotic Conscious Coevolution")
print("Consciousness = operational architectural construct")
print("=" * 76)
```
