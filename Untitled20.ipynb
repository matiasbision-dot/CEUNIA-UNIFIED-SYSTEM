```python

# =============================================================================
# CEUNIA Ω-v2.3 — BLIND OOS / GENERALIZACIÓN DEL META-SELECTOR
# Selector entrenado en TRAIN y COMPLETAMENTE CONGELADO en OOS
# Multi-seed + familias no vistas + bootstrap + Wilcoxon + ablaciones
# =============================================================================

import numpy as np
import sqlite3, hashlib, json, os
from datetime import datetime, timezone

try:
    from scipy.stats import wilcoxon
    SCIPY_OK = True
except Exception:
    SCIPY_OK = False

# -----------------------------------------------------------------------------
# CONFIG
# -----------------------------------------------------------------------------

SEEDS = [101, 211, 307, 419, 523, 631, 743, 857,
         967, 1087, 1201, 1327, 1451, 1579, 1697, 1811]

TRAIN = 300
N = 600
OOS_START = TRAIN

TRAIN_FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed"
]

TEST_FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed",
    "structural_shift",
    "frequency_shift",
    "phase_shift",
    "amplitude_shift",
    "regime_switch"
]

NOISES = [0.00, 0.008, 0.018, 0.035]

DB = "/content/ceunia_integral.db"

# -----------------------------------------------------------------------------
# MÉTRICAS
# -----------------------------------------------------------------------------

def mae(a, b):
    return float(np.mean(np.abs(np.asarray(a) - np.asarray(b))))

def paired_gain(base_err, ceunia_err):
    return np.asarray(base_err) - np.asarray(ceunia_err)

# -----------------------------------------------------------------------------
# GENERADORES
# -----------------------------------------------------------------------------

def make_series(seed, regime, noise):
    rng = np.random.default_rng(seed)
    t = np.arange(N, dtype=float)

    if regime == "trend_cycle":
        y = (
            0.0025*t
            + 0.8*np.sin(2*np.pi*t/42)
            + 0.25*np.sin(2*np.pi*t/17)
        )

    elif regime == "multi_cycle":
        y = (
            0.0015*t
            + 0.55*np.sin(2*np.pi*t/29)
            + 0.42*np.sin(2*np.pi*t/67)
            + 0.18*np.sin(2*np.pi*t/11)
        )

    elif regime == "mean_reversion":
        y = np.zeros(N)
        for i in range(1, N):
            shock = rng.normal(0, 0.08)
            y[i] = (
                0.86*y[i-1]
                + shock
                - 0.015*np.sign(y[i-1])
            )

    elif regime == "mixed":
        y = (
            0.0018*t
            + 0.55*np.sin(2*np.pi*t/36)
            + 0.30*np.sin(2*np.pi*t/83)
            + 0.15*np.sin(2*np.pi*t/9)
        )

    elif regime == "structural_shift":
        y = (
            0.0015*t
            + 0.55*np.sin(2*np.pi*t/38)
        )
        y[TRAIN:] += (
            0.55
            + 0.0025*(t[TRAIN:] - TRAIN)
            + 0.35*np.sin(2*np.pi*t[TRAIN:]/61)
        )

    elif regime == "frequency_shift":
        y = 0.0015*t + 0.65*np.sin(2*np.pi*t/31)
        y[TRAIN:] = (
            0.0015*t[TRAIN:]
            + 0.65*np.sin(2*np.pi*t[TRAIN:]/57)
        )

    elif regime == "phase_shift":
        y = 0.0015*t + 0.65*np.sin(2*np.pi*t/39)
        y[TRAIN:] = (
            0.0015*t[TRAIN:]
            + 0.65*np.sin(2*np.pi*t[TRAIN:]/39 + np.pi/2)
        )

    elif regime == "amplitude_shift":
        y = 0.0015*t + 0.45*np.sin(2*np.pi*t/41)
        y[TRAIN:] = (
            0.0015*t[TRAIN:]
            + 0.95*np.sin(2*np.pi*t[TRAIN:]/41)
        )

    elif regime == "regime_switch":
        y = np.zeros(N)
        for i in range(1, N):
            if i < TRAIN:
                phi = 0.92
                shock_sd = 0.07
            else:
                phi = 0.55
                shock_sd = 0.13
            y[i] = phi*y[i-1] + rng.normal(0, shock_sd)

    else:
        raise ValueError(regime)

    if noise > 0:
        y += rng.normal(0, noise, N)

    return y

# -----------------------------------------------------------------------------
# ESTRATEGIAS
# -----------------------------------------------------------------------------

STRATEGIES = ["trend", "cycle", "reversion", "adaptive"]

def trend(y, i):
    w = min(25, i)
    if w < 3:
        return float(y[i-1])
    z = y[i-w:i]
    x = np.arange(w)
    slope, intercept = np.polyfit(x, z, 1)
    return float(intercept + slope*w)

def cycle(y, i):
    w = min(80, i)
    if w < 10:
        return float(y[i-1])

    z = y[i-w:i]
    candidates = [11,17,23,29,31,36,42,48,57,61,67]

    best_p = 11
    best_err = np.inf

    for p in candidates:
        if p*2 >= w:
            continue

        pred = np.roll(z, p)
        err = np.mean(np.abs(z[p:] - pred[p:]))

        if err < best_err:
            best_err = err
            best_p = p

    return float(y[i-best_p])

def reversion(y, i):
    w = min(30, i)
    z = y[i-w:i]
    m = np.mean(z)
    return float(y[i-1] + 0.35*(m-y[i-1]))

def adaptive(y, i):
    w = min(18, i)
    if w < 4:
        return float(y[i-1])

    z = y[i-w:i]
    slope = np.polyfit(np.arange(w), z, 1)[0]

    return float(y[i-1] + 0.35*slope)

def strategy_predictions(y, i):
    return {
        "trend": trend(y, i),
        "cycle": cycle(y, i),
        "reversion": reversion(y, i),
        "adaptive": adaptive(y, i)
    }

# -----------------------------------------------------------------------------
# BASELINE
# -----------------------------------------------------------------------------

def baseline_prediction(y, i):
    w = min(20, i)
    z = y[i-w:i]

    if len(z) < 3:
        return float(y[i-1])

    x = np.arange(len(z))
    slope, intercept = np.polyfit(x, z, 1)

    return float(intercept + slope*len(z))

# -----------------------------------------------------------------------------
# ENTRENAMIENTO DEL META-SELECTOR
# IMPORTANTE: SOLO TRAIN
# -----------------------------------------------------------------------------

def train_selector(series_list):
    errors = {s: [] for s in STRATEGIES}

    for y in series_list:

        for i in range(30, TRAIN):

            cand = strategy_predictions(y, i)

            for s in STRATEGIES:
                errors[s].append(
                    abs(cand[s] - y[i])
                )

    learned = {}

    for s in STRATEGIES:
        learned[s] = float(
            np.mean(errors[s][-300:])
        )

    return learned

# -----------------------------------------------------------------------------
# SELECTOR CONGELADO
# NO APRENDE NADA DURANTE OOS
# -----------------------------------------------------------------------------

def frozen_selector(y, start, learned_scores):

    preds = []
    actuals = []
    selected = []

    # Scores congelados desde TRAIN
    scores = dict(learned_scores)

    for i in range(start, N):

        cand = strategy_predictions(y, i)

        # Contexto permitido, pero sin aprendizaje de parámetros.
        recent = y[max(0, i-20):i]

        if len(recent) >= 5:
            local_mean = np.mean(recent)
            local_std = np.std(recent)
        else:
            local_mean = y[i-1]
            local_std = 0.0

        adjusted = {}

        for s in STRATEGIES:

            score = scores[s]

            if s == "reversion":
                distance = abs(y[i-1] - local_mean)
                score -= 0.04*distance

            if s == "trend":
                score -= 0.015*local_std

            adjusted[s] = score

        choice = min(adjusted, key=adjusted.get)

        preds.append(cand[choice])
        actuals.append(y[i])
        selected.append(choice)

    return (
        np.asarray(preds),
        np.asarray(actuals),
        selected
    )

# -----------------------------------------------------------------------------
# SELECTOR SIN CONTRAFACTUAL
# BASELINE INTERNO: ESTRATEGIA FIJA APRENDIDA EN TRAIN
# -----------------------------------------------------------------------------

def fixed_best_strategy(y_train_list):

    scores = {s: [] for s in STRATEGIES}

    for y in y_train_list:

        for i in range(30, TRAIN):

            cand = strategy_predictions(y, i)

            for s in STRATEGIES:
                scores[s].append(
                    abs(cand[s] - y[i])
                )

    return min(
        scores,
        key=lambda s: np.mean(scores[s])
    )

# -----------------------------------------------------------------------------
# OOS
# -----------------------------------------------------------------------------

def evaluate(seed, regime, noise, learned_scores, fixed_strategy):

    y = make_series(seed, regime, noise)

    bp = []
    ba = []

    for i in range(OOS_START, N):
        bp.append(baseline_prediction(y, i))
        ba.append(y[i])

    bp = np.asarray(bp)
    ba = np.asarray(ba)

    cp, ca, choices = frozen_selector(
        y,
        OOS_START,
        learned_scores
    )

    fixed_preds = []

    for i in range(OOS_START, N):
        cand = strategy_predictions(y, i)
        fixed_preds.append(cand[fixed_strategy])

    fixed_preds = np.asarray(fixed_preds)

    e_base = np.abs(bp-ba)
    e_ceunia = np.abs(cp-ca)
    e_fixed = np.abs(fixed_preds-ba)

    gain = e_base - e_ceunia

    return {
        "seed": seed,
        "regime": regime,
        "noise": noise,
        "baseline": mae(bp, ba),
        "ceunia": mae(cp, ca),
        "fixed": mae(fixed_preds, ba),
        "gain": float(np.mean(gain)),
        "win_rate": float(np.mean(gain > 0)),
        "paired_gain": gain.tolist(),
        "selection_counts": {
            s: int(choices.count(s))
            for s in STRATEGIES
        }
    }

# -----------------------------------------------------------------------------
# ENTRENAMIENTO GLOBAL
# -----------------------------------------------------------------------------

print("="*78)
print("CEUNIA Ω-v2.3 — BLIND OOS / GENERALIZACIÓN")
print("="*78)

train_series = []

for seed in SEEDS:
    for regime in TRAIN_FAMILIES:
        train_series.append(
            make_series(seed, regime, 0.018)
        )

learned_scores = train_selector(train_series)
fixed_strategy = fixed_best_strategy(train_series)

print("\nMETA-SELECTOR ENTRENADO SOLO EN TRAIN")
for s, v in learned_scores.items():
    print("{:<12}: {:.8f}".format(s, v))

print("\nESTRATEGIA FIJA DE CONTROL:", fixed_strategy)

# -----------------------------------------------------------------------------
# EVALUACIÓN
# -----------------------------------------------------------------------------

results = []

for seed in SEEDS:
    for regime in TEST_FAMILIES:
        for noise in NOISES:

            r = evaluate(
                seed,
                regime,
                noise,
                learned_scores,
                fixed_strategy
            )

            results.append(r)

print("\nExperimentos OOS:", len(results))

# -----------------------------------------------------------------------------
# AGREGACIÓN
# -----------------------------------------------------------------------------

base = np.mean([r["baseline"] for r in results])
ceunia = np.mean([r["ceunia"] for r in results])
fixed = np.mean([r["fixed"] for r in results])

gain = base - ceunia
relative = 100*gain/base if base else 0

positive = sum(r["gain"] > 0 for r in results)
positive_ratio = positive/len(results)

win_rate = np.mean(
    [r["win_rate"] for r in results]
)

# -----------------------------------------------------------------------------
# DISTRIBUCIÓN PAREADA
# -----------------------------------------------------------------------------

all_paired = []

for r in results:
    all_paired.extend(r["paired_gain"])

all_paired = np.asarray(all_paired)

mean_paired = float(np.mean(all_paired))
median_paired = float(np.median(all_paired))

# bootstrap IC95 de la ganancia media
rng_boot = np.random.default_rng(424242)

boot_means = []

for _ in range(5000):
    sample = rng_boot.choice(
        all_paired,
        size=len(all_paired),
        replace=True
    )
    boot_means.append(np.mean(sample))

ci_low, ci_high = np.percentile(
    boot_means,
    [2.5, 97.5]
)

# -----------------------------------------------------------------------------
# WILCOXON
# -----------------------------------------------------------------------------

if SCIPY_OK:

    try:
        stat, pvalue = wilcoxon(
            all_paired,
            alternative="greater"
        )
    except Exception:
        stat = np.nan
        pvalue = np.nan

else:
    stat = np.nan
    pvalue = np.nan

# -----------------------------------------------------------------------------
# TAMAÑO DE EFECTO
# -----------------------------------------------------------------------------

sd = float(np.std(all_paired, ddof=1))

if sd > 0:
    effect_size = mean_paired/sd
else:
    effect_size = 0.0

# -----------------------------------------------------------------------------
# POR RÉGIMEN
# -----------------------------------------------------------------------------

by_regime = {}

for regime in TEST_FAMILIES:

    sub = [
        r for r in results
        if r["regime"] == regime
    ]

    b = np.mean([r["baseline"] for r in sub])
    c = np.mean([r["ceunia"] for r in sub])

    by_regime[regime] = {
        "baseline": float(b),
        "ceunia": float(c),
        "gain": float(b-c),
        "relative_pct": float(
            100*(b-c)/b if b else 0
        ),
        "positive_ratio": float(
            np.mean([r["gain"] > 0 for r in sub])
        )
    }

# -----------------------------------------------------------------------------
# POR RUIDO
# -----------------------------------------------------------------------------

by_noise = {}

for noise in NOISES:

    sub = [
        r for r in results
        if r["noise"] == noise
    ]

    b = np.mean([r["baseline"] for r in sub])
    c = np.mean([r["ceunia"] for r in sub])

    by_noise[str(noise)] = {
        "baseline": float(b),
        "ceunia": float(c),
        "gain": float(b-c),
        "relative_pct": float(
            100*(b-c)/b if b else 0
        )
    }

# -----------------------------------------------------------------------------
# GENERALIZACIÓN A FAMILIAS NO VISTAS
# -----------------------------------------------------------------------------

unseen = [
    "structural_shift",
    "frequency_shift",
    "phase_shift",
    "amplitude_shift",
    "regime_switch"
]

unseen_results = [
    r for r in results
    if r["regime"] in unseen
]

unseen_base = np.mean(
    [r["baseline"] for r in unseen_results]
)

unseen_ceunia = np.mean(
    [r["ceunia"] for r in unseen_results]
)

unseen_gain = unseen_base - unseen_ceunia

unseen_positive = np.mean(
    [r["gain"] > 0 for r in unseen_results]
)

# -----------------------------------------------------------------------------
# VEREDICTO
# -----------------------------------------------------------------------------

# Criterio fuerte:
# 1) ganancia media > 0
# 2) IC95 completamente > 0
# 3) >=60% experimentos positivos
# 4) win rate >=50%
# 5) familias no vistas también positivas

robust_positive = (
    gain > 0
    and ci_low > 0
    and positive_ratio >= 0.60
    and win_rate >= 0.50
    and unseen_gain > 0
    and unseen_positive >= 0.60
)

if robust_positive:
    verdict = (
        "VALIDACIÓN POSITIVA DE GENERALIZACIÓN BLIND OOS"
    )
elif gain > 0 and ci_low > 0:
    verdict = (
        "MEJORA ESTADÍSTICAMENTE CONSISTENTE, "
        "PERO GENERALIZACIÓN INSUFICIENTE"
    )
elif gain > 0:
    verdict = (
        "MEJORA DETECTADA, "
        "SIN EVIDENCIA ESTADÍSTICA SUFICIENTE"
    )
else:
    verdict = (
        "HIPÓTESIS NO VALIDADA EN Ω-v2.3"
    )

# -----------------------------------------------------------------------------
# HASH
# -----------------------------------------------------------------------------

hash_payload = {
    "version": "CEUNIA_OMEGA_v2.3",
    "seeds": SEEDS,
    "train_families": TRAIN_FAMILIES,
    "test_families": TEST_FAMILIES,
    "noise": NOISES,
    "N": N,
    "TRAIN": TRAIN,
    "learned_scores": learned_scores,
    "fixed_strategy": fixed_strategy,
    "results": results,
    "aggregate": {
        "baseline": float(base),
        "ceunia": float(ceunia),
        "fixed": float(fixed),
        "gain": float(gain),
        "relative_pct": float(relative),
        "positive_ratio": float(positive_ratio),
        "win_rate": float(win_rate),
        "bootstrap_ci95": [
            float(ci_low),
            float(ci_high)
        ],
        "wilcoxon_stat": float(stat)
            if np.isfinite(stat) else None,
        "wilcoxon_p": float(pvalue)
            if np.isfinite(pvalue) else None,
        "effect_size": float(effect_size),
        "unseen_gain": float(unseen_gain),
        "unseen_positive_ratio": float(unseen_positive)
    }
}

sha = hashlib.sha256(
    json.dumps(
        hash_payload,
        sort_keys=True
    ).encode()
).hexdigest()

# -----------------------------------------------------------------------------
# SQLITE
# -----------------------------------------------------------------------------

conn = sqlite3.connect(DB)

conn.execute("""
CREATE TABLE IF NOT EXISTS ceunia_v23_blind_oos (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    timestamp TEXT,
    version TEXT,
    seed INTEGER,
    regime TEXT,
    noise REAL,
    baseline_mae REAL,
    ceunia_mae REAL,
    fixed_mae REAL,
    gain REAL,
    win_rate REAL,
    sha256 TEXT
)
""")

timestamp = datetime.now(
    timezone.utc
).isoformat()

for r in results:

    conn.execute("""
    INSERT INTO ceunia_v23_blind_oos
    (
        timestamp,
        version,
        seed,
        regime,
        noise,
        baseline_mae,
        ceunia_mae,
        fixed_mae,
        gain,
        win_rate,
        sha256
    )
    VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
    """, (
        timestamp,
        "2.3",
        r["seed"],
        r["regime"],
        r["noise"],
        r["baseline"],
        r["ceunia"],
        r["fixed"],
        r["gain"],
        r["win_rate"],
        sha
    ))

conn.commit()

# -----------------------------------------------------------------------------
# REPORTE JSON
# -----------------------------------------------------------------------------

report = {
    "version": "CEUNIA Ω-v2.3",
    "experiments": len(results),
    "seeds": len(SEEDS),
    "train_families": TRAIN_FAMILIES,
    "test_families": TEST_FAMILIES,
    "noise_levels": NOISES,
    "N": N,
    "TRAIN": TRAIN,
    "baseline_mae": float(base),
    "ceunia_mae": float(ceunia),
    "fixed_mae": float(fixed),
    "absolute_gain": float(gain),
    "relative_gain_pct": float(relative),
    "positive_cases": int(positive),
    "positive_ratio": float(positive_ratio),
    "win_rate": float(win_rate),
    "paired_mean_gain": mean_paired,
    "paired_median_gain": median_paired,
    "bootstrap_ci95": [
        float(ci_low),
        float(ci_high)
    ],
    "wilcoxon_available": SCIPY_OK,
    "wilcoxon_stat": (
        float(stat) if np.isfinite(stat) else None
    ),
    "wilcoxon_p": (
        float(pvalue) if np.isfinite(pvalue) else None
    ),
    "effect_size": float(effect_size),
    "unseen_gain": float(unseen_gain),
    "unseen_positive_ratio": float(unseen_positive),
    "by_regime": by_regime,
    "by_noise": by_noise,
    "verdict": verdict,
    "sha256": sha
}

REPORT = "/content/CEUNIA_OMEGA_v2_3_REPORT.json"

with open(
    REPORT,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        report,
        f,
        indent=2,
        ensure_ascii=False
    )

# -----------------------------------------------------------------------------
# SALIDA
# -----------------------------------------------------------------------------

print("\n" + "="*78)
print("RESULTADOS CEUNIA Ω-v2.3")
print("="*78)

print("Experimentos OOS      :", len(results))
print("Semillas              :", len(SEEDS))
print("Familias TRAIN        :", len(TRAIN_FAMILIES))
print("Familias TEST         :", len(TEST_FAMILIES))
print("Niveles de ruido      :", len(NOISES))

print("\nMAE")
print("Baseline              :", f"{base:.8f}")
print("CEUNIA BLIND OOS      :", f"{ceunia:.8f}")
print("Mejor fija TRAIN      :", f"{fixed:.8f}")

print("\nGANANCIA")
print("Absoluta              :", f"{gain:+.8f}")
print("Relativa              :", f"{relative:+.4f}%")

print("\nROBUSTEZ")
print("Casos positivos       :", f"{positive}/{len(results)}")
print("Proporción positiva   :", f"{positive_ratio*100:.2f}%")
print("Win rate              :", f"{win_rate*100:.2f}%")

print("\nESTADÍSTICA PAREADA")
print("Ganancia media        :", f"{mean_paired:+.8f}")
print("Mediana               :", f"{median_paired:+.8f}")
print("Bootstrap IC95        :",
      f"[{ci_low:+.8f}, {ci_high:+.8f}]")

if SCIPY_OK:
    print("Wilcoxon p            :", f"{pvalue:.8g}")
else:
    print("Wilcoxon              : scipy no disponible")

print("Effect size           :", f"{effect_size:.6f}")

print("\nFAMILIAS NO VISTAS")
print("Gain unseen           :", f"{unseen_gain:+.8f}")
print("Positivos unseen      :", f"{unseen_positive*100:.2f}%")

print("\nPOR RÉGIMEN")
for regime, x in by_regime.items():
    print(
        "{:<20} Base={:.6f}  CEUNIA={:.6f}  Gain={:+.6f}  Pos={:.1f}%".format(
            regime,
            x["baseline"],
            x["ceunia"],
            x["gain"],
            x["positive_ratio"]*100
        )
    )

print("\nPOR RUIDO")
for noise, x in by_noise.items():
    print(
        "noise={:<6} Base={:.6f}  CEUNIA={:.6f}  Gain={:+.6f}".format(
            noise,
            x["baseline"],
            x["ceunia"],
            x["gain"]
        )
    )

print("\nVEREDICTO:")
print(verdict)

print("\nSHA-256:")
print(sha)

print("\nDB:")
print(DB)

print("\nREPORTE:")
print(REPORT)

conn.close()

print("\n" + "="*78)
print("Ω-v2.3 FINALIZADO")
print("="*78)
```
