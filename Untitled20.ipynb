```python

# ================================================================
# CEUNIA Ω-v2.4 — META-SELECTOR MARGINAL / BLIND OOS
# Una sola celda | No borra resultados anteriores
# ================================================================

import numpy as np, pandas as pd, sqlite3, json, hashlib, os, warnings
from scipy.stats import wilcoxon

warnings.filterwarnings("ignore")

# ---------------- CONFIG ----------------

SEEDS = [
    101,211,307,419,523,631,743,857,
    967,1087,1201,1327,1451,1579,1697,1811
]

N = 600
TRAIN = 300

TRAIN_FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed"
]

TEST_FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed",
    "structural_shift",
    "frequency_shift",
    "phase_shift",
    "amplitude_shift",
    "regime_switch"
]

NOISES = [0.0, 0.008, 0.018, 0.035]

STRATEGIES = [
    "trend",
    "cycle",
    "reversion",
    "adaptive"
]

DB = "/content/ceunia_integral.db"

# ================================================================
# UTILIDADES
# ================================================================

def stable_seed(seed, text):
    h = hashlib.sha256(
        f"{seed}:{text}".encode()
    ).hexdigest()
    return int(h[:8], 16)

def generate_series(n, family, noise, seed):

    rng = np.random.default_rng(
        stable_seed(seed, family)
    )

    t = np.arange(n, dtype=float)

    trend = rng.uniform(0.0015, 0.0045)
    amp = rng.uniform(0.7, 1.5)
    freq = rng.uniform(0.035, 0.085)
    phase = rng.uniform(0, 2*np.pi)

    if family == "trend_cycle":

        y = trend*t + amp*np.sin(freq*t + phase)

    elif family == "multi_cycle":

        f1 = rng.uniform(0.025, 0.055)
        f2 = rng.uniform(0.08, 0.13)

        y = (
            trend*t
            + 0.9*np.sin(f1*t + phase)
            + 0.45*np.sin(f2*t)
        )

    elif family == "mean_reversion":

        y = np.zeros(n)
        phi = rng.uniform(0.72, 0.90)

        for i in range(1,n):
            y[i] = (
                phi*y[i-1]
                + rng.normal(0,0.08)
            )

        y += 0.08*np.sin(freq*t)

    elif family == "mixed":

        y = (
            trend*t
            + 0.8*np.sin(freq*t + phase)
            + 0.18*np.sin(0.16*t)
        )

        for i in range(1,n):
            y[i] += 0.15*y[i-1]

    elif family == "structural_shift":

        y = trend*t + amp*np.sin(freq*t + phase)

        s = n//2

        y[s:] += (
            0.0025*(t[s:]-s)
            + 0.55*np.sin(freq*1.7*t[s:])
        )

    elif family == "frequency_shift":

        s = n//2

        y = np.zeros(n)

        y[:s] = (
            trend*t[:s]
            + amp*np.sin(0.045*t[:s] + phase)
        )

        y[s:] = (
            trend*t[s:]
            + amp*np.sin(0.11*t[s:] + phase)
        )

    elif family == "phase_shift":

        y = trend*t + amp*np.sin(freq*t + phase)

        s = n//2

        y[s:] = (
            trend*t[s:]
            + amp*np.sin(
                freq*t[s:] + phase + 1.4
            )
        )

    elif family == "amplitude_shift":

        y = trend*t + amp*np.sin(freq*t + phase)

        s = n//2

        y[s:] = (
            trend*t[s:]
            + amp*1.9*np.sin(
                freq*t[s:] + phase
            )
        )

    elif family == "regime_switch":

        y = np.zeros(n)

        s1 = n//3
        s2 = 2*n//3

        y[:s1] = (
            trend*t[:s1]
            + amp*np.sin(0.045*t[:s1])
        )

        phi = 0.82

        for i in range(s1,s2):
            y[i] = (
                phi*y[i-1]
                + rng.normal(0,0.08)
            )

        y[s2:] = (
            y[s2-1]
            + trend*(t[s2:]-t[s2])
            + amp*np.sin(0.11*t[s2:])
        )

    else:
        y = rng.normal(0,1,n)

    y += 0.035*np.sin(0.31*t + seed*0.01)
    y += rng.normal(0,noise,n)

    return y


# ================================================================
# FEATURES
# ================================================================

def context_features(h):

    h = np.asarray(h)

    if len(h) < 60:
        return np.zeros(8)

    short = h[-10:]
    medium = h[-30:]
    long = h[-60:]

    slope_short = np.polyfit(
        np.arange(len(short)),
        short,
        1
    )[0]

    slope_long = np.polyfit(
        np.arange(len(long)),
        long,
        1
    )[0]

    volatility = np.std(np.diff(long))

    centered = long - np.mean(long)

    ac1 = np.corrcoef(
        centered[:-1],
        centered[1:]
    )[0,1]

    ac2 = np.corrcoef(
        centered[:-2],
        centered[2:]
    )[0,1]

    fft = np.abs(np.fft.rfft(centered))
    freqs = np.fft.rfftfreq(len(centered))

    if len(fft) > 2:

        idx = np.argmax(fft[1:]) + 1

        dom_freq = freqs[idx]

        spectral = (
            fft[idx]
            / (np.sum(fft[1:])+1e-12)
        )

    else:

        dom_freq = 0
        spectral = 0

    distance = (
        np.mean(
            np.abs(
                short-np.mean(medium)
            )
        )
        /(np.std(medium)+1e-12)
    )

    return np.array([
        slope_short,
        slope_long,
        volatility,
        ac1,
        ac2,
        dom_freq,
        spectral,
        distance
    ])


# ================================================================
# PREDICTORES
# ================================================================

def trend(h):

    h = np.asarray(h)
    w = min(30,len(h))

    x = np.arange(w)

    slope,intercept = np.polyfit(
        x,
        h[-w:],
        1
    )

    return intercept+slope*w


def cycle(h):

    h = np.asarray(h)

    w = min(80,len(h))

    seg = h[-w:]
    centered = seg-np.mean(seg)

    fft = np.fft.rfft(centered)

    if len(fft)<=2:
        return h[-1]

    idx = np.argmax(
        np.abs(fft)[1:]
    )+1

    freq = idx/w

    amp = 2*np.abs(fft[idx])/w
    phase = np.angle(fft[idx])

    return (
        np.mean(seg)
        + amp*np.cos(
            2*np.pi*freq*w+phase
        )
    )


def reversion(h):

    h = np.asarray(h)

    w = min(40,len(h))

    m = np.mean(h[-w:])

    return h[-1]+0.35*(m-h[-1])


def adaptive(h):

    p1 = trend(h)
    p2 = cycle(h)
    p3 = reversion(h)

    f = context_features(h)

    slope = f[1]
    ac1 = f[3]
    spectral = f[6]

    w1 = np.clip(abs(slope)*45,0,1)
    w2 = np.clip(spectral*1.5,0,1)
    w3 = np.clip(-ac1,0,1)

    w = np.array([w1,w2,w3])

    if w.sum()<1e-12:
        w=np.ones(3)

    w/=w.sum()

    return (
        w[0]*p1
        +w[1]*p2
        +w[2]*p3
    )


PRED = {
    "trend":trend,
    "cycle":cycle,
    "reversion":reversion,
    "adaptive":adaptive
}


# ================================================================
# TRAIN DEL META-SELECTOR
# SOLO TRAIN
# ================================================================

X_train = []
Y_train = []

for seed in SEEDS:

    for family in TRAIN_FAMILIES:

        y = generate_series(
            N,
            family,
            0.018,
            seed
        )

        for t in range(80,TRAIN):

            h = y[:t]

            X_train.append(
                context_features(h)
            )

            Y_train.append([
                abs(PRED[s](h)-y[t])
                for s in STRATEGIES
            ])

X_train = np.asarray(X_train)
Y_train = np.asarray(Y_train)


# ================================================================
# META-SELECTOR
# Predice error futuro de cada estrategia
# ================================================================

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

META = {}

for j,s in enumerate(STRATEGIES):

    model = make_pipeline(
        StandardScaler(),
        Ridge(alpha=1.0)
    )

    model.fit(
        X_train,
        Y_train[:,j]
    )

    META[s]=model


# ================================================================
# OOS BLIND
# ================================================================

rows=[]

rng = np.random.default_rng(99991)

for seed in SEEDS:

    for family in TEST_FAMILIES:

        for noise in NOISES:

            y = generate_series(
                N,
                family,
                noise,
                seed
            )

            for t in range(TRAIN,N):

                h = y[:t]
                actual = y[t]

                errors = {
                    s:abs(
                        PRED[s](h)-actual
                    )
                    for s in STRATEGIES
                }

                # -------- META --------

                feat = context_features(h)
                xx = feat.reshape(1,-1)

                estimated = {
                    s:float(
                        META[s].predict(xx)[0]
                    )
                    for s in STRATEGIES
                }

                selected = min(
                    estimated,
                    key=estimated.get
                )

                # -------- RANDOM --------

                random_s = rng.choice(
                    STRATEGIES
                )

                rows.append({
                    "seed":seed,
                    "family":family,
                    "noise":noise,
                    "t":t,
                    "trend":errors["trend"],
                    "cycle":errors["cycle"],
                    "reversion":errors["reversion"],
                    "adaptive":errors["adaptive"],
                    "meta":errors[selected],
                    "random":errors[random_s],
                    "selected":selected
                })


df = pd.DataFrame(rows)


# ================================================================
# MÉTRICAS
# ================================================================

baseline = df["trend"].mean()
trend_fixed = df["trend"].mean()
cycle_fixed = df["cycle"].mean()
reversion_fixed = df["reversion"].mean()
adaptive_fixed = df["adaptive"].mean()
meta = df["meta"].mean()
random = df["random"].mean()

delta = (
    df["adaptive"].values
    - df["meta"].values
)

gain = delta.mean()

relative = (
    100*gain/(adaptive_fixed+1e-12)
)

positive = np.sum(delta>0)

# Wilcoxon
try:
    if np.any(delta>0) and np.any(delta<0):
        p = wilcoxon(
            delta,
            alternative="greater"
        ).pvalue
    else:
        p = 0.0
except:
    p = np.nan

# Bootstrap
rng_boot = np.random.default_rng(424242)

B=2000
boot=np.empty(B)

for i in range(B):

    idx=rng_boot.integers(
        0,len(delta),len(delta)
    )

    boot[i]=delta[idx].mean()

ci=np.percentile(
    boot,[2.5,97.5]
)

effect = (
    gain/
    (np.std(delta,ddof=1)+1e-12)
)


# ================================================================
# POR FAMILIA
# ================================================================

family_rows=[]

for family in TEST_FAMILIES:

    z=df[df.family==family]

    a=z["adaptive"].mean()
    m=z["meta"].mean()

    family_rows.append({
        "family":family,
        "adaptive":a,
        "meta":m,
        "gain":a-m,
        "positive_pct":
            100*np.mean(
                z["adaptive"]>
                z["meta"]
            )
    })

family_df=pd.DataFrame(
    family_rows
)


# ================================================================
# POR RUIDO
# ================================================================

noise_rows=[]

for noise in NOISES:

    z=df[df.noise==noise]

    noise_rows.append({
        "noise":noise,
        "adaptive":z["adaptive"].mean(),
        "meta":z["meta"].mean(),
        "gain":
            z["adaptive"].mean()
            -z["meta"].mean()
    })

noise_df=pd.DataFrame(
    noise_rows
)


# ================================================================
# SELECCIÓN
# ================================================================

selection = (
    df["selected"]
    .value_counts()
    .to_dict()
)


# ================================================================
# VEREDICTO
# ================================================================

if gain>0 and ci[0]>0 and p<0.05:

    verdict = (
        "VALIDACIÓN POSITIVA: "
        "META-SELECTOR APORTA VALOR MARGINAL "
        "SOBRE ADAPTIVE FIXED"
    )

elif gain>0:

    verdict = (
        "SEÑAL POSITIVA NO SIGNIFICATIVA: "
        "META > ADAPTIVE, "
        "PERO SIN EVIDENCIA ESTADÍSTICA SUFICIENTE"
    )

else:

    verdict = (
        "HIPÓTESIS NO VALIDADA: "
        "META-SELECTOR NO SUPERA "
        "ADAPTIVE FIXED"
    )


# ================================================================
# SHA
# ================================================================

summary={
    "version":"CEUNIA_OMEGA_v2_4",
    "experiments":len(df),
    "seeds":SEEDS,
    "train_families":TRAIN_FAMILIES,
    "test_families":TEST_FAMILIES,
    "noise":NOISES,
    "train":TRAIN,
    "oos":N-TRAIN,
    "baseline":baseline,
    "trend_fixed":trend_fixed,
    "cycle_fixed":cycle_fixed,
    "reversion_fixed":reversion_fixed,
    "adaptive_fixed":adaptive_fixed,
    "meta_selector":meta,
    "random":random,
    "gain_meta_vs_adaptive":gain,
    "relative_gain_pct":relative,
    "positive_cases":int(positive),
    "positive_pct":100*positive/len(delta),
    "wilcoxon_p":float(p),
    "bootstrap_ci95":[
        float(ci[0]),
        float(ci[1])
    ],
    "effect_size":float(effect),
    "selection":selection,
    "verdict":verdict
}

sha=hashlib.sha256(
    json.dumps(
        summary,
        sort_keys=True,
        default=float
    ).encode()
).hexdigest()


# ================================================================
# PERSISTENCIA SIN BORRAR NADA
# ================================================================

conn=sqlite3.connect(DB)

df.to_sql(
    "ceunia_v24_oos",
    conn,
    if_exists="replace",
    index=False
)

family_df.to_sql(
    "ceunia_v24_family",
    conn,
    if_exists="replace",
    index=False
)

noise_df.to_sql(
    "ceunia_v24_noise",
    conn,
    if_exists="replace",
    index=False
)

conn.execute("""
CREATE TABLE IF NOT EXISTS ceunia_v24_summary(
id INTEGER PRIMARY KEY AUTOINCREMENT,
version TEXT,
baseline REAL,
trend_fixed REAL,
cycle_fixed REAL,
reversion_fixed REAL,
adaptive_fixed REAL,
meta_selector REAL,
random_selector REAL,
gain_meta_vs_adaptive REAL,
relative_gain_pct REAL,
positive_cases INTEGER,
positive_pct REAL,
wilcoxon_p REAL,
ci_low REAL,
ci_high REAL,
effect_size REAL,
sha256 TEXT,
verdict TEXT
)
""")

conn.execute("""
INSERT INTO ceunia_v24_summary VALUES(
NULL,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?
)
""",(
"CEUNIA_OMEGA_v2_4",
baseline,
trend_fixed,
cycle_fixed,
reversion_fixed,
adaptive_fixed,
meta,
random,
gain,
relative,
int(positive),
100*positive/len(delta),
float(p),
float(ci[0]),
float(ci[1]),
float(effect),
sha,
verdict
))

conn.commit()
conn.close()


# ================================================================
# REPORTE
# ================================================================

REPORT="/content/CEUNIA_OMEGA_v2_4_REPORT.json"

with open(REPORT,"w") as f:
    json.dump(
        summary,
        f,
        indent=2,
        default=float
    )


# ================================================================
# SALIDA
# ================================================================

print("\n"+"="*72)
print("CEUNIA Ω-v2.4 — META-SELECTOR MARGINAL")
print("="*72)

print("\nEXPERIMENTOS OOS :",len(df))
print("TRAIN            :",TRAIN)
print("OOS              :",N-TRAIN)

print("\nMAE")
print("Baseline         : {:.8f}".format(baseline))
print("Trend Fixed      : {:.8f}".format(trend_fixed))
print("Cycle Fixed      : {:.8f}".format(cycle_fixed))
print("Reversion Fixed  : {:.8f}".format(reversion_fixed))
print("Adaptive Fixed   : {:.8f}".format(adaptive_fixed))
print("Meta-Selector    : {:.8f}".format(meta))
print("Random           : {:.8f}".format(random))

print("\nVALOR MARGINAL DEL META")
print("Gain vs Adaptive : {:+.8f}".format(gain))
print("Relativo         : {:+.4f}%".format(relative))

print("\nEVIDENCIA")
print("Casos positivos  : {}/{}".format(
    positive,len(delta)
))
print("Proporción       : {:.2f}%".format(
    100*positive/len(delta)
))
print("Wilcoxon p       :",p)
print("Bootstrap IC95   : [{:.8f}, {:.8f}]".format(
    ci[0],ci[1]
))
print("Effect size      : {:.6f}".format(effect))

print("\nSELECCIÓN")
for k,v in sorted(
    selection.items(),
    key=lambda x:-x[1]
):
    print("{:<12} {}".format(k,v))

print("\nPOR FAMILIA")
print(
    family_df.to_string(
        index=False,
        float_format=lambda x:"{:.6f}".format(x)
    )
)

print("\nPOR RUIDO")
print(
    noise_df.to_string(
        index=False,
        float_format=lambda x:"{:.6f}".format(x)
    )
)

print("\nVEREDICTO")
print(verdict)

print("\nSHA-256")
print(sha)

print("\nDB")
print(DB)

print("\nREPORT")
print(REPORT)

print("\n"+"="*72)
```
