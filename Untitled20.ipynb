```python

# ==============================================================================
# CEUNIA Ω-v2.7 — METACOGNICIÓN OPERACIONAL / DETECCIÓN DE FALLA
# ------------------------------------------------------------------------------
# Objetivo:
#   Evaluar si el mecanismo de incertidumbre/abstención de Ω-v2.6 realmente
#   identifica los casos donde Meta-Selector pierde frente a Adaptive.
#
# Diseño:
#   - TRAIN / OOS estrictamente separados.
#   - Umbrales aprendidos SOLO en TRAIN.
#   - OOS completamente congelado.
#   - Diagnóstico posterior NO retroalimenta decisiones.
#   - Evaluación agrupada por serie: seed + family + noise.
#
# Métricas:
#   1) MAE Adaptive
#   2) MAE Meta
#   3) MAE Meta+Fallback
#   4) Risk score
#   5) Cobertura
#   6) Selective risk
#   7) Failure rate de Meta
#   8) Precisión de abstención
#   9) Recall de fallos
#  10) Enrichment: cuánto aumenta la concentración de fallos entre abstenciones
#  11) Relación riesgo → error
#  12) Bootstrap agrupado
#  13) Seen / unseen families
# ==============================================================================

import numpy as np
import pandas as pd
import sqlite3, json, hashlib, os, math
from collections import defaultdict

# ---------------- CONFIG ----------------

SEEDS = [
    101,211,307,419,523,631,743,857,
    967,1087,1201,1327,1451,1579,1697,1811
]

N = 600
TRAIN = 300
OOS = N - TRAIN

NOISES = [0.000, 0.008, 0.018, 0.035]

FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed",
    "structural_shift",
    "frequency_shift",
    "phase_shift",
    "amplitude_shift",
    "regime_switch"
]

TRAIN_FAMILIES = [
    "trend_cycle",
    "multi_cycle",
    "mean_reversion",
    "mixed"
]

TEST_FAMILIES = FAMILIES

STRATEGIES = ["trend","cycle","reversion","adaptive"]

DB = "/content/ceunia_integral.db"
REPORT = "/content/CEUNIA_OMEGA_v2_7_REPORT.json"

# ==============================================================================
# GENERADOR SINTÉTICO
# ==============================================================================

def generate_series(n, family, seed, noise):
    rng = np.random.default_rng(seed)

    t = np.arange(n, dtype=float)

    if family == "trend_cycle":
        y = (
            0.015*t
            + 0.85*np.sin(2*np.pi*t/42)
            + 0.25*np.sin(2*np.pi*t/17)
        )

    elif family == "multi_cycle":
        y = (
            0.55*np.sin(2*np.pi*t/21)
            + 0.40*np.sin(2*np.pi*t/57)
            + 0.25*np.sin(2*np.pi*t/91)
            + 0.004*t
        )

    elif family == "mean_reversion":
        y = np.zeros(n)
        for i in range(1,n):
            y[i] = (
                0.78*y[i-1]
                + 0.30*np.sin(2*np.pi*i/31)
                + rng.normal(0,0.07)
            )

    elif family == "mixed":
        y = (
            0.010*t
            + 0.55*np.sin(2*np.pi*t/29)
            + 0.30*np.sin(2*np.pi*t/73)
            + 0.10*np.cos(2*np.pi*t/13)
        )

    elif family == "structural_shift":
        shift = np.where(t < n*0.62, 0.0, 1.8)
        slope = np.where(t < n*0.62, 0.012, -0.006)
        y = (
            slope*t
            + shift
            + 0.65*np.sin(2*np.pi*t/41)
        )

    elif family == "frequency_shift":
        freq = np.where(t < n*0.58, 34.0, 13.0)
        phase = np.cumsum(2*np.pi/freq)
        y = 0.95*np.sin(phase) + 0.012*t

    elif family == "phase_shift":
        phase_shift = np.where(t < n*0.58, 0.0, np.pi/2)
        y = (
            0.012*t
            + 0.85*np.sin(2*np.pi*t/37 + phase_shift)
        )

    elif family == "amplitude_shift":
        amp = np.where(t < n*0.58, 0.45, 1.20)
        y = (
            0.009*t
            + amp*np.sin(2*np.pi*t/39)
        )

    elif family == "regime_switch":
        y = np.zeros(n)
        for i in range(1,n):
            if i < n*0.35:
                y[i] = (
                    0.88*y[i-1]
                    + 0.20*np.sin(2*np.pi*i/27)
                )
            elif i < n*0.70:
                y[i] = (
                    y[i-1]
                    + 0.018
                    + 0.18*np.sin(2*np.pi*i/51)
                )
            else:
                y[i] = (
                    0.58*y[i-1]
                    + 0.50*np.sin(2*np.pi*i/17)
                )

    else:
        raise ValueError(f"Familia desconocida: {family}")

    y = y + rng.normal(0, noise, n)

    # normalización para estabilidad
    y = (y - np.mean(y)) / (np.std(y) + 1e-9)

    return y


# ==============================================================================
# PREDICTORES
# ==============================================================================

def pred_trend(hist):
    w = min(25, len(hist)-1)
    x = np.arange(w)
    yy = hist[-w:]
    coef = np.polyfit(x, yy, 1)
    return yy[-1] + coef[0]


def pred_cycle(hist):
    if len(hist) < 30:
        return hist[-1]

    periods = [13,17,21,27,31,37,41,51,57,73,91]

    best_p = periods[0]
    best_corr = -1e9

    h = np.asarray(hist)

    for p in periods:
        if len(h) <= p:
            continue

        a = h[-p:]
        b = h[-2*p:-p]

        if np.std(a) < 1e-9 or np.std(b) < 1e-9:
            continue

        c = np.corrcoef(a,b)[0,1]

        if np.isfinite(c) and c > best_corr:
            best_corr = c
            best_p = p

    return h[-best_p]


def pred_reversion(hist):
    w = min(35, len(hist))
    h = np.asarray(hist[-w:])
    mean = np.mean(h)
    return 0.72*h[-1] + 0.28*mean


def pred_adaptive(hist):
    h = np.asarray(hist)

    if len(h) < 25:
        return h[-1]

    # combinación adaptativa basada en:
    # tendencia local + persistencia + reversión
    w = min(30, len(h))
    local = h[-w:]

    x = np.arange(w)

    try:
        slope = np.polyfit(x, local, 1)[0]
    except:
        slope = 0.0

    persistence = h[-1]
    mean = np.mean(local)

    # predictor deliberadamente estable
    p = (
        0.42*persistence
        + 0.38*(persistence + 4*slope)
        + 0.20*mean
    )

    return p


def predictors(hist):
    return {
        "trend": pred_trend(hist),
        "cycle": pred_cycle(hist),
        "reversion": pred_reversion(hist),
        "adaptive": pred_adaptive(hist)
    }


# ==============================================================================
# FEATURES DE CONTEXTO
# ==============================================================================

def context_features(hist):
    h = np.asarray(hist)

    def safe_corr(a,b):
        if len(a) < 3 or np.std(a) < 1e-10 or np.std(b) < 1e-10:
            return 0.0
        c = np.corrcoef(a,b)[0,1]
        return float(c) if np.isfinite(c) else 0.0

    def slope(x):
        if len(x) < 3:
            return 0.0
        return float(np.polyfit(np.arange(len(x)), x, 1)[0])

    w20 = h[-20:]
    w40 = h[-40:] if len(h) >= 40 else h

    d1 = np.diff(w20) if len(w20) > 1 else np.array([0.0])

    ac1 = safe_corr(w40[1:], w40[:-1])

    if len(w40) >= 12:
        ac5 = safe_corr(w40[5:], w40[:-5])
    else:
        ac5 = 0.0

    return np.array([
        np.mean(w20),
        np.std(w20),
        slope(w20),
        slope(w40),
        np.mean(np.abs(d1)),
        ac1,
        ac5,
        h[-1],
        h[-1] - np.mean(w20),
        np.max(w20)-np.min(w20)
    ], dtype=float)


# ==============================================================================
# TRAIN: APRENDIZAJE DE ERROR CONTRAFACTUAL
# ==============================================================================

train_records = []

for seed in SEEDS:
    for family in TRAIN_FAMILIES:

        y = generate_series(N, family, seed, 0.018)

        for t in range(40, TRAIN):

            hist = y[:t]
            true = y[t]

            ps = predictors(hist)
            x = context_features(hist)

            for strategy in STRATEGIES:

                err = abs(ps[strategy] - true)

                train_records.append({
                    "seed":seed,
                    "family":family,
                    "t":t,
                    "strategy":strategy,
                    "error":err,
                    "x":x
                })

# ==============================================================================
# REGRESIÓN RIDGE SIMPLE IMPLEMENTADA LOCALMENTE
# ==============================================================================

X = np.vstack([r["x"] for r in train_records])
yerr = np.array([r["error"] for r in train_records])

mu = X.mean(axis=0)
sd = X.std(axis=0) + 1e-9

Xs = (X - mu) / sd

lam = 5.0

A = Xs.T @ Xs + lam*np.eye(Xs.shape[1])
B = Xs.T @ yerr

beta_global = np.linalg.solve(A,B)

# residual std global
pred_train = Xs @ beta_global
resid_std = float(np.std(yerr - pred_train) + 1e-9)

# ==============================================================================
# DISTRIBUCIÓN TRAIN PARA OOD
# ==============================================================================

train_contexts = []

for seed in SEEDS:
    for family in TRAIN_FAMILIES:
        y = generate_series(N, family, seed, 0.018)

        for t in range(40, TRAIN):
            train_contexts.append(context_features(y[:t]))

train_contexts = np.vstack(train_contexts)

ctx_mu = train_contexts.mean(axis=0)
ctx_sd = train_contexts.std(axis=0) + 1e-9

train_z = (train_contexts - ctx_mu) / ctx_sd
train_dist = np.sqrt(np.sum(train_z**2,axis=1))

OOD_THRESHOLD = float(np.quantile(train_dist,0.95))

# ==============================================================================
# CALIBRACIÓN DE MARGEN EN TRAIN
# ==============================================================================

train_margins = []

for seed in SEEDS:
    for family in TRAIN_FAMILIES:

        y = generate_series(N, family, seed, 0.018)

        for t in range(40, TRAIN):

            hist = y[:t]
            x = context_features(hist)

            xs = (x-mu)/sd

            preds = {
                s: float(xs @ beta_global)
                for s in STRATEGIES
            }

            vals = sorted(preds.values())

            margin = vals[1] - vals[0] if len(vals) > 1 else 0.0

            train_margins.append(margin)

MARGIN_THRESHOLD = float(np.quantile(train_margins,0.20))

# ==============================================================================
# FUNCIÓN META
# ==============================================================================

def meta_decision(hist):

    x = context_features(hist)

    xs = (x-mu)/sd

    predicted_errors = {
        s: float(xs @ beta_global)
        for s in STRATEGIES
    }

    selected = min(
        predicted_errors,
        key=predicted_errors.get
    )

    vals = sorted(predicted_errors.values())

    margin = vals[1]-vals[0] if len(vals)>1 else 0.0

    z = (x-ctx_mu)/ctx_sd

    ood_distance = float(np.sqrt(np.sum(z*z)))

    ood = ood_distance > OOD_THRESHOLD

    low_margin = margin < MARGIN_THRESHOLD

    # risk score continuo.
    # Ambos componentes están definidos por TRAIN.
    ood_component = ood_distance / (OOD_THRESHOLD + 1e-9)

    margin_component = (
        MARGIN_THRESHOLD / (margin + 1e-9)
    )

    risk_score = 0.5*ood_component + 0.5*margin_component

    abstain = bool(ood or low_margin)

    return {
        "selected":selected,
        "predicted_errors":predicted_errors,
        "margin":margin,
        "ood_distance":ood_distance,
        "ood":ood,
        "low_margin":low_margin,
        "risk_score":risk_score,
        "abstain":abstain
    }


# ==============================================================================
# OOS
# ==============================================================================

rows = []

for seed in SEEDS:
    for family in TEST_FAMILIES:
        for noise in NOISES:

            y = generate_series(N, family, seed, noise)

            for t in range(TRAIN,N):

                hist = y[:t]
                true = y[t]

                ps = predictors(hist)

                meta = meta_decision(hist)

                meta_strategy = meta["selected"]

                adaptive_err = abs(ps["adaptive"]-true)
                meta_err = abs(ps[meta_strategy]-true)

                robust_strategy = (
                    "adaptive"
                    if meta["abstain"]
                    else meta_strategy
                )

                robust_err = abs(
                    ps[robust_strategy]-true
                )

                # failure de META respecto a Adaptive
                failure = bool(meta_err > adaptive_err)

                # éxito de abstención:
                # abstener cuando Meta habría perdido.
                true_positive = bool(
                    meta["abstain"] and failure
                )

                false_positive = bool(
                    meta["abstain"] and not failure
                )

                false_negative = bool(
                    (not meta["abstain"]) and failure
                )

                true_negative = bool(
                    (not meta["abstain"]) and not failure
                )

                rows.append({
                    "seed":seed,
                    "family":family,
                    "noise":noise,
                    "t":t,
                    "adaptive_err":adaptive_err,
                    "meta_err":meta_err,
                    "robust_err":robust_err,
                    "gain_meta":adaptive_err-meta_err,
                    "gain_robust":adaptive_err-robust_err,
                    "failure":int(failure),
                    "abstain":int(meta["abstain"]),
                    "ood":int(meta["ood"]),
                    "low_margin":int(meta["low_margin"]),
                    "risk_score":meta["risk_score"],
                    "ood_distance":meta["ood_distance"],
                    "margin":meta["margin"],
                    "selected":meta_strategy
                })

df = pd.DataFrame(rows)

# ==============================================================================
# DIAGNÓSTICO DE DETECCIÓN
# ==============================================================================

tp = int(((df.abstain==1)&(df.failure==1)).sum())
fp = int(((df.abstain==1)&(df.failure==0)).sum())
fn = int(((df.abstain==0)&(df.failure==1)).sum())
tn = int(((df.abstain==0)&(df.failure==0)).sum())

precision = tp/(tp+fp) if tp+fp else 0.0
recall = tp/(tp+fn) if tp+fn else 0.0
specificity = tn/(tn+fp) if tn+fp else 0.0

failure_abstain = (
    df.loc[df.abstain==1,"failure"].mean()
    if (df.abstain==1).any() else 0.0
)

failure_continue = (
    df.loc[df.abstain==0,"failure"].mean()
    if (df.abstain==0).any() else 0.0
)

enrichment = (
    failure_abstain/(failure_continue+1e-12)
)

coverage = float((df.abstain==0).mean())

selective_risk = (
    df.loc[df.abstain==0,"meta_err"].mean()
    if (df.abstain==0).any() else np.nan
)

overall_meta = df.meta_err.mean()
overall_robust = df.robust_err.mean()
overall_adaptive = df.adaptive_err.mean()

gain_meta = overall_adaptive-overall_meta
gain_robust = overall_adaptive-overall_robust

# ==============================================================================
# RIESGO → ERROR
# ==============================================================================

# bins por cuantiles aprendidos únicamente de OOS para diagnóstico descriptivo.
# NO intervienen en ninguna decisión.
try:
    df["risk_bin"] = pd.qcut(
        df["risk_score"],
        q=10,
        duplicates="drop"
    )
    risk_table = (
        df.groupby("risk_bin",observed=True)
        .agg(
            n=("meta_err","size"),
            meta_error=("meta_err","mean"),
            adaptive_error=("adaptive_err","mean"),
            failure_rate=("failure","mean"),
            abstention_rate=("abstain","mean")
        )
        .reset_index()
    )
except:
    risk_table = pd.DataFrame()

# correlación diagnóstica
if df["risk_score"].std() > 1e-12:
    risk_corr = float(
        np.corrcoef(
            df["risk_score"],
            df["meta_err"]
        )[0,1]
    )
else:
    risk_corr = 0.0

# ==============================================================================
# AGRUPACIÓN POR SERIE
# ==============================================================================

group_cols = ["seed","family","noise"]

grouped = (
    df.groupby(group_cols)
    .agg(
        adaptive=("adaptive_err","mean"),
        meta=("meta_err","mean"),
        robust=("robust_err","mean"),
        gain_meta=("gain_meta","mean"),
        gain_robust=("gain_robust","mean"),
        failure_rate=("failure","mean"),
        abstain_rate=("abstain","mean"),
        ood_rate=("ood","mean"),
        low_margin_rate=("low_margin","mean")
    )
    .reset_index()
)

grouped["meta_positive"] = (
    grouped["gain_meta"] > 0
).astype(int)

grouped["robust_positive"] = (
    grouped["gain_robust"] > 0
).astype(int)

# ==============================================================================
# SEEN / UNSEEN
# ==============================================================================

grouped["seen"] = grouped.family.isin(
    TRAIN_FAMILIES
)

seen = grouped[grouped.seen]
unseen = grouped[~grouped.seen]

def safe_mean(x):
    return float(np.mean(x)) if len(x) else np.nan

seen_gain = safe_mean(seen.gain_robust)
unseen_gain = safe_mean(unseen.gain_robust)

seen_positive = (
    float(seen.robust_positive.mean()*100)
    if len(seen) else np.nan
)

unseen_positive = (
    float(unseen.robust_positive.mean()*100)
    if len(unseen) else np.nan
)

# ==============================================================================
# BOOTSTRAP AGRUPADO
# ==============================================================================

rng = np.random.default_rng(2027)

group_gain = grouped.gain_robust.values

B = 3000
boot = []

if len(group_gain):
    for _ in range(B):
        sample = rng.choice(
            group_gain,
            size=len(group_gain),
            replace=True
        )
        boot.append(np.mean(sample))

    boot = np.array(boot)

    ci_low = float(np.quantile(boot,0.025))
    ci_high = float(np.quantile(boot,0.975))
else:
    ci_low = np.nan
    ci_high = np.nan

# ==============================================================================
# WILCOXON SIN SCIPY OBLIGATORIO
# ==============================================================================

try:
    from scipy.stats import wilcoxon

    if np.any(group_gain != 0):
        w_stat, p_value = wilcoxon(
            group_gain,
            alternative="greater"
        )
        p_value = float(p_value)
        w_stat = float(w_stat)
    else:
        w_stat = 0.0
        p_value = 1.0

except Exception:
    w_stat = np.nan
    p_value = np.nan

# ==============================================================================
# EFFECT SIZE AGRUPADO
# ==============================================================================

gmean = np.mean(group_gain)
gstd = np.std(group_gain,ddof=1) + 1e-12

effect_size = float(gmean/gstd)

# ==============================================================================
# FAMILIAS
# ==============================================================================

family_table = (
    df.groupby("family")
    .agg(
        adaptive=("adaptive_err","mean"),
        meta=("meta_err","mean"),
        robust=("robust_err","mean"),
        gain_meta=("gain_meta","mean"),
        gain_robust=("gain_robust","mean"),
        failure_rate=("failure","mean"),
        abstain_rate=("abstain","mean"),
        ood_rate=("ood","mean"),
        positive_robust=("gain_robust",lambda x: np.mean(x>0)*100)
    )
    .reset_index()
)

# ==============================================================================
# RUIDO
# ==============================================================================

noise_table = (
    df.groupby("noise")
    .agg(
        adaptive=("adaptive_err","mean"),
        meta=("meta_err","mean"),
        robust=("robust_err","mean"),
        gain_meta=("gain_meta","mean"),
        gain_robust=("gain_robust","mean"),
        failure_rate=("failure","mean"),
        abstain_rate=("abstain","mean")
    )
    .reset_index()
)

# ==============================================================================
# DECISIÓN CIENTÍFICA
# ==============================================================================

positive_grouped = int(
    (grouped.robust_positive==1).sum()
)

n_groups = len(grouped)

positive_pct = (
    100*positive_grouped/n_groups
    if n_groups else 0.0
)

# criterio deliberadamente exigente:
# - ganancia media > 0
# - IC bootstrap totalmente > 0
# - mayoría de series positivas
# - recall de fallos > 0.50
# - enrichment > 1
#
# No significa "metacognición humana".
# Significa que el mecanismo operacional de abstención detecta
# condiciones asociadas a fallos del selector.

validated = (
    gmean > 0
    and ci_low > 0
    and positive_pct > 50
    and recall > 0.50
    and enrichment > 1.0
)

if validated:
    verdict = (
        "VALIDACIÓN POSITIVA: LA ABSTENCIÓN DETECTA "
        "RIESGO Y MEJORA EL CONTROL OOS"
    )
else:
    verdict = (
        "HIPÓTESIS NO VALIDADA COMPLETAMENTE: "
        "LA ABSTENCIÓN MEJORA O NO MEJORA, PERO "
        "LA DETECCIÓN DE FALLOS AÚN NO ES SUFICIENTE"
    )

# ==============================================================================
# RESUMEN
# ==============================================================================

summary = {
    "version":"CEUNIA Ω-v2.7",
    "observations":int(len(df)),
    "series":int(len(grouped)),
    "train":TRAIN,
    "oos":OOS,

    "mae_adaptive":float(overall_adaptive),
    "mae_meta":float(overall_meta),
    "mae_robust":float(overall_robust),

    "gain_meta":float(gain_meta),
    "gain_robust":float(gain_robust),

    "coverage":float(coverage),
    "abstention_rate":float(df.abstain.mean()),
    "ood_rate":float(df.ood.mean()),
    "low_margin_rate":float(df.low_margin.mean()),

    "tp":tp,
    "fp":fp,
    "fn":fn,
    "tn":tn,

    "abstention_precision":float(precision),
    "abstention_recall":float(recall),
    "specificity":float(specificity),

    "failure_rate_abstain":float(failure_abstain),
    "failure_rate_continue":float(failure_continue),
    "failure_enrichment":float(enrichment),

    "risk_error_correlation":float(risk_corr),

    "selective_risk":float(selective_risk),

    "group_positive":positive_grouped,
    "group_total":n_groups,
    "group_positive_pct":float(positive_pct),

    "group_mean_gain":float(gmean),
    "bootstrap_ci95_low":float(ci_low),
    "bootstrap_ci95_high":float(ci_high),

    "wilcoxon_stat":w_stat,
    "wilcoxon_p":p_value,
    "effect_size":effect_size,

    "seen_gain":seen_gain,
    "unseen_gain":unseen_gain,
    "seen_positive_pct":seen_positive,
    "unseen_positive_pct":unseen_positive,

    "ood_threshold":OOD_THRESHOLD,
    "margin_threshold":MARGIN_THRESHOLD,

    "verdict":verdict
}

# ==============================================================================
# SHA
# ==============================================================================

canonical = json.dumps(
    summary,
    sort_keys=True,
    separators=(",",":"),
    default=str
)

sha = hashlib.sha256(
    canonical.encode("utf-8")
).hexdigest()

summary["sha256"] = sha

# ==============================================================================
# SQLITE
# ==============================================================================

con = sqlite3.connect(DB)

df.to_sql(
    "ceunia_v27_oos",
    con,
    if_exists="replace",
    index=False
)

grouped.to_sql(
    "ceunia_v27_grouped",
    con,
    if_exists="replace",
    index=False
)

family_table.to_sql(
    "ceunia_v27_family",
    con,
    if_exists="replace",
    index=False
)

noise_table.to_sql(
    "ceunia_v27_noise",
    con,
    if_exists="replace",
    index=False
)

if len(risk_table):
    risk_table.to_sql(
        "ceunia_v27_risk_calibration",
        con,
        if_exists="replace",
        index=False
    )

pd.DataFrame([summary]).to_sql(
    "ceunia_v27_summary",
    con,
    if_exists="replace",
    index=False
)

con.commit()
con.close()

# ==============================================================================
# JSON
# ==============================================================================

report = {
    "summary":summary,
    "family":family_table.to_dict(orient="records"),
    "noise":noise_table.to_dict(orient="records"),
    "risk_calibration":risk_table.to_dict(orient="records")
}

with open(REPORT,"w") as f:
    json.dump(
        report,
        f,
        indent=2,
        default=str
    )

# ==============================================================================
# OUTPUT
# ==============================================================================

print("="*78)
print("CEUNIA Ω-v2.7 — METACOGNICIÓN OPERACIONAL / DETECCIÓN DE FALLA")
print("="*78)

print()
print("OBSERVACIONES :",len(df))
print("SERIES        :",len(grouped))
print("TRAIN         :",TRAIN)
print("OOS           :",OOS)

print()
print("MAE")
print("Adaptive Fixed        : {:.8f}".format(overall_adaptive))
print("Meta-Selector         : {:.8f}".format(overall_meta))
print("Meta + Fallback       : {:.8f}".format(overall_robust))

print()
print("GANANCIA")
print("Meta vs Adaptive      : {:+.8f}".format(gain_meta))
print("Robust vs Adaptive    : {:+.8f}".format(gain_robust))

print()
print("COBERTURA / ABSTENCIÓN")
print("Cobertura              : {:.2f}%".format(coverage*100))
print("Abstención             : {:.2f}%".format(df.abstain.mean()*100))
print("OOD                    : {:.2f}%".format(df.ood.mean()*100))
print("Low margin             : {:.2f}%".format(df.low_margin.mean()*100))

print()
print("DETECCIÓN DE FALLOS")
print("TP                     :",tp)
print("FP                     :",fp)
print("FN                     :",fn)
print("TN                     :",tn)
print("Precisión abstención   : {:.4f}".format(precision))
print("Recall de fallos       : {:.4f}".format(recall))
print("Specificity            : {:.4f}".format(specificity))

print()
print("RIESGO CONDICIONAL")
print("Fallo | abstención     : {:.4f}".format(failure_abstain))
print("Fallo | continúa       : {:.4f}".format(failure_continue))
print("Enrichment             : {:.4f}x".format(enrichment))
print("Corr riesgo → error    : {:.4f}".format(risk_corr))

print()
print("EVIDENCIA AGRUPADA")
print("Series positivas       : {}/{} ({:.2f}%)".format(
    positive_grouped,
    n_groups,
    positive_pct
))
print("Gain medio agrupado    : {:+.8f}".format(gmean))
print("Bootstrap IC95         : [{:.8f}, {:.8f}]".format(
    ci_low,ci_high
))
print("Wilcoxon p             :",p_value)
print("Effect size            : {:.6f}".format(effect_size))

print()
print("SEEN / UNSEEN")
print("Seen gain              : {:+.8f}".format(seen_gain))
print("Unseen gain            : {:+.8f}".format(unseen_gain))
print("Seen positive          : {:.2f}%".format(seen_positive))
print("Unseen positive        : {:.2f}%".format(unseen_positive))

print()
print("UMBRALES ENTRENADOS")
print("OOD threshold          : {:.8f}".format(OOD_THRESHOLD))
print("Margin threshold       : {:.8f}".format(MARGIN_THRESHOLD))

print()
print("POR FAMILIA")
print(
    family_table[
        [
            "family",
            "adaptive",
            "meta",
            "robust",
            "gain_meta",
            "gain_robust",
            "failure_rate",
            "abstain_rate",
            "ood_rate",
            "positive_robust"
        ]
    ].to_string(index=False)
)

print()
print("POR RUIDO")
print(
    noise_table.to_string(index=False)
)

print()
print("VEREDICTO")
print(verdict)

print()
print("SHA-256")
print(sha)

print()
print("DB")
print(DB)

print()
print("REPORT")
print(REPORT)

print("="*78)
```
