```python
# ================================================================
# CEUNIA Ω CORE v5.5.2
# TRUE GRADUAL SYMBIOTIC COEVOLUTION
# RIGOROUS CAUSAL VALIDATION
# SINGLE CELL | FROZEN VALID/OOS | BASELINES | PLACEBOS
# 32 SEEDS | BOOTSTRAP | SIGN-FLIP | ABLATIONS | SQLITE | SHA256
# ================================================================

import numpy as np
import sqlite3
import hashlib
import json
import os
from datetime import datetime, timezone

VERSION = "CEUNIA_Ω_CORE_v5.5.2_RIGOROUS"
MASTER_SEED = 55126

N_SEEDS = 32
N_TRAIN = 240
N_VALID = 120
N_OOS = 160
TOTAL = N_TRAIN + N_VALID + N_OOS

MEMORY_WINDOW = 20
LR_AI = 0.055
LR_HUMAN = 0.035

N_BOOT = 3000
N_PERM = 5000

DB_PATH = "ceunia_omega_core_v5_5_2.db"

# ================================================================
# 1. ENTORNO
# ================================================================

def generate_environment(seed):
    rng = np.random.default_rng(seed)
    t = np.arange(TOTAL, dtype=float)

    x = (
        0.72 * np.sin(t * 0.083)
        + 0.37 * np.cos(t * 0.029)
        + 0.22 * np.sin(t * 0.017)
        + 0.18 * rng.normal(size=TOTAL)
    )

    hidden = (
        0.48 * np.sin(t * 0.113)
        + 0.31 * np.cos(t * 0.071 + 0.73)
        + 0.23 * np.sin(t * 0.037 + 1.41)
        + 0.17 * np.tanh(
            0.85 * np.sin(t * 0.091 + x)
        )
        + 0.08 * np.cos(0.017 * t * x)
    )

    y = np.tanh(
        hidden + 0.045 * rng.normal(size=TOTAL)
    )

    return x, y


# ================================================================
# 2. HUMAN AGENT
# ================================================================

class HumanAgent:

    def __init__(self, adaptive=True):
        self.adaptive = adaptive
        self.preference = 0.0
        self.confidence = 0.5
        self.intention_state = 0.0

        # Diagnostic only. Never mutated during frozen evaluation.
        self.last_adaptation = 0.0

    def perceive(self, x_value):

        return float(
            np.tanh(
                0.80 * x_value
                + 0.20 * self.preference
            )
        )

    def form_intention(
        self,
        perception,
        ai_signal
    ):

        return float(
            np.tanh(
                0.55 * perception
                + 0.25 * ai_signal
                + 0.20 * self.confidence
            )
        )

    def adapt(
        self,
        outcome,
        prediction,
        ai_intention,
        shared_intention,
        update=True
    ):

        error = outcome - prediction

        feedback = float(
            np.tanh(
                0.55 * error
                + 0.20 * ai_intention
                + 0.15 * shared_intention
                + 0.10 * self.confidence
            )
        )

        old_preference = self.preference

        candidate = np.clip(
            self.preference
            + LR_HUMAN * feedback,
            -1.0,
            1.0
        )

        delta = abs(
            candidate - old_preference
        )

        if update and self.adaptive:

            self.preference = float(candidate)

            self.confidence = float(
                np.clip(
                    0.96 * self.confidence
                    + 0.04 * (1.0 - abs(error)),
                    0.0,
                    1.0
                )
            )

            self.last_adaptation = float(delta)

        return feedback, float(delta)


# ================================================================
# 3. AI AGENT
# ================================================================

class AIAgent:

    def __init__(
        self,
        adaptive=True,
        use_memory=True,
        use_human_feedback=True,
        memory_mode="real"
    ):

        self.adaptive = adaptive
        self.use_memory = use_memory
        self.use_human_feedback = use_human_feedback
        self.memory_mode = memory_mode

        self.weights = np.array([
            0.20,
            0.15,
            0.15,
            0.10,
            0.10,
            0.10,
            0.05,
            0.05,
            0.10
        ], dtype=float)

        self.memory = []

        self.coherence = 0.5
        self.potential = 0.5
        self.previous_potential = 0.5
        self.meta_potential = 0.0
        self.intention_state = 0.0

        self.last_adaptation = 0.0

        self.rng = np.random.default_rng(123456)

    def memory_signal(self):

        if not self.use_memory or not self.memory:
            return 0.0

        m = self.memory[-MEMORY_WINDOW:]

        if self.memory_mode == "real":
            return float(np.mean(m))

        if self.memory_mode == "shuffle":
            shuffled = np.array(m, dtype=float).copy()
            self.rng.shuffle(shuffled)
            return float(np.mean(shuffled))

        if self.memory_mode == "random":
            return float(
                self.rng.normal(
                    0.0,
                    max(np.std(m), 0.05)
                )
            )

        return 0.0

    def perceive(
        self,
        x_value,
        human_signal
    ):

        memory = self.memory_signal()

        volatility = (
            np.std(
                self.memory[-MEMORY_WINDOW:]
            )
            if len(self.memory) > 2
            else 0.1
        )

        potential = float(
            np.tanh(
                abs(x_value)
                / (volatility + 0.15)
            )
        )

        meta_potential = (
            potential
            - self.previous_potential
        )

        alignment = (
            1.0
            - min(
                1.0,
                abs(
                    human_signal
                    - memory
                ) / 2.0
            )
        )

        features = np.array([
            x_value,
            human_signal,
            memory,
            self.coherence,
            potential,
            meta_potential,
            self.intention_state,
            human_signal * memory,
            alignment
        ])

        return (
            features,
            potential,
            meta_potential
        )

    def preliminary_prediction(
        self,
        features
    ):

        return float(
            np.tanh(
                np.dot(
                    self.weights,
                    features
                )
            )
        )

    def form_intention(
        self,
        prediction,
        human_intention
    ):

        return float(
            np.tanh(
                0.60 * prediction
                + 0.40 * human_intention
            )
        )

    def final_prediction(
        self,
        features,
        shared_intention
    ):

        f = features.copy()
        f[6] = shared_intention

        return float(
            np.tanh(
                np.dot(
                    self.weights,
                    f
                )
            )
        ), f

    def learn(
        self,
        features,
        outcome,
        prediction,
        human_feedback,
        update=True
    ):

        if not update or not self.adaptive:
            return

        error = outcome - prediction
        old = self.weights.copy()

        self.weights += (
            LR_AI
            * error
            * features
        )

        if self.use_human_feedback:
            self.weights += (
                0.015
                * human_feedback
                * features
            )

        norm = np.linalg.norm(
            self.weights
        )

        if norm > 1.5:
            self.weights *= (
                1.5 / norm
            )

        self.last_adaptation = float(
            np.mean(
                abs(
                    self.weights - old
                )
            )
        )

    def update_state(
        self,
        error,
        potential,
        update=True
    ):

        if not update:
            return

        self.coherence = float(
            np.clip(
                0.96 * self.coherence
                + 0.04 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        self.previous_potential = (
            self.potential
        )

        self.potential = float(
            0.90 * self.potential
            + 0.10 * potential
        )

        self.meta_potential = float(
            self.potential
            - self.previous_potential
        )


# ================================================================
# 4. CEUNIA
# ================================================================

class CEUNIA:

    def __init__(
        self,
        use_human=True,
        use_shared_memory=True,
        use_shared_intention=True,
        human_adaptive=True,
        ai_adaptive=True,
        use_human_feedback=True,
        memory_mode="real"
    ):

        self.use_human = use_human
        self.use_shared_memory = use_shared_memory
        self.use_shared_intention = use_shared_intention

        self.human = HumanAgent(
            adaptive=(
                human_adaptive
                and use_human
            )
        )

        self.ai = AIAgent(
            adaptive=ai_adaptive,
            use_memory=use_shared_memory,
            use_human_feedback=(
                use_human_feedback
                and use_human
            ),
            memory_mode=memory_mode
        )

        self.shared_memory = []

        self.symbiosis = 0.5
        self.coherence = 0.5
        self.meta_observation = 0.5

        self.generation = 0

    def step(
        self,
        x_value,
        outcome,
        train=True
    ):

        # --------------------------------------------------------
        # HUMAN
        # --------------------------------------------------------

        if self.use_human:

            hp = self.human.perceive(
                x_value
            )

            human_intention = (
                self.human.form_intention(
                    hp,
                    self.ai.intention_state
                )
            )

        else:

            hp = 0.0
            human_intention = 0.0

        # --------------------------------------------------------
        # AI PERCEPTION
        # --------------------------------------------------------

        features, potential, meta_potential = (
            self.ai.perceive(
                x_value,
                hp
            )
        )

        # --------------------------------------------------------
        # PRELIMINARY
        # --------------------------------------------------------

        preliminary = (
            self.ai.preliminary_prediction(
                features
            )
        )

        # --------------------------------------------------------
        # INTENTION
        # --------------------------------------------------------

        ai_intention = (
            self.ai.form_intention(
                preliminary,
                human_intention
            )
        )

        # --------------------------------------------------------
        # SHARED INTENTION
        # --------------------------------------------------------

        if self.use_shared_intention:

            shared_intention = float(
                np.tanh(
                    0.50 * human_intention
                    + 0.50 * ai_intention
                )
            )

        else:

            shared_intention = ai_intention

        # --------------------------------------------------------
        # FINAL DECISION
        # --------------------------------------------------------

        prediction, final_features = (
            self.ai.final_prediction(
                features,
                shared_intention
            )
        )

        error = (
            outcome - prediction
        )

        # --------------------------------------------------------
        # POST-OUTCOME ADAPTATION
        #
        # CRITICAL:
        # update=train ONLY.
        # No persistent state mutation in VALID/OOS.
        # --------------------------------------------------------

        if self.use_human:

            human_feedback, human_delta = (
                self.human.adapt(
                    outcome,
                    prediction,
                    ai_intention,
                    shared_intention,
                    update=train
                )
            )

        else:

            human_feedback = 0.0
            human_delta = 0.0

        self.ai.learn(
            final_features,
            outcome,
            prediction,
            human_feedback,
            update=train
        )

        self.ai.update_state(
            error,
            potential,
            update=train
        )

        # --------------------------------------------------------
        # COEVOLUTION METRICS
        # --------------------------------------------------------

        alignment = float(
            1.0
            - min(
                1.0,
                abs(
                    human_intention
                    - ai_intention
                )
            )
        )

        mutual_adaptation = float(
            1.0
            - min(
                1.0,
                abs(
                    human_delta
                    - self.ai.last_adaptation
                )
            )
        )

        candidate_symbiosis = float(
            np.clip(
                0.40 * alignment
                + 0.30 * mutual_adaptation
                + 0.20 * self.ai.coherence
                + 0.10 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        candidate_coherence = float(
            np.clip(
                0.50 * self.ai.coherence
                + 0.30 * alignment
                + 0.20 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        candidate_meta = float(
            np.clip(
                0.40 * alignment
                + 0.30 * candidate_symbiosis
                + 0.30 * (
                    1.0 - abs(error)
                ),
                0.0,
                1.0
            )
        )

        # --------------------------------------------------------
        # STATE TRANSITION ONLY DURING TRAIN
        # --------------------------------------------------------

        if train:

            self.human.intention_state = (
                human_intention
            )

            self.ai.intention_state = (
                ai_intention
            )

            self.symbiosis = float(
                0.955 * self.symbiosis
                + 0.045 * candidate_symbiosis
            )

            self.coherence = float(
                0.960 * self.coherence
                + 0.040 * candidate_coherence
            )

            self.meta_observation = float(
                0.960 * self.meta_observation
                + 0.040 * candidate_meta
            )

            if self.use_shared_memory:

                memory_state = float(
                    0.25 * shared_intention
                    + 0.20 * prediction
                    + 0.20 * outcome
                    + 0.20 * self.symbiosis
                    + 0.15 * self.coherence
                )

                self.shared_memory.append(
                    memory_state
                )

                self.shared_memory = (
                    self.shared_memory[
                        -MEMORY_WINDOW:
                    ]
                )

            if self.ai.use_memory:

                self.ai.memory.append(
                    prediction
                )

                self.ai.memory = (
                    self.ai.memory[
                        -MEMORY_WINDOW:
                    ]
                )

            self.generation += 1

        return {
            "prediction": float(prediction),
            "target": float(outcome),
            "error": float(error),
            "human_intention": float(human_intention),
            "ai_intention": float(ai_intention),
            "shared_intention": float(shared_intention),
            "alignment": float(alignment),
            "human_adaptation": float(human_delta),
            "ai_adaptation": float(
                self.ai.last_adaptation
            ),
            "symbiosis": float(
                self.symbiosis
                if train
                else candidate_symbiosis
            ),
            "coherence": float(
                self.coherence
                if train
                else candidate_coherence
            ),
            "meta_observation": float(
                self.meta_observation
                if train
                else candidate_meta
            )
        }


# ================================================================
# 5. SNAPSHOT
# ================================================================

def snapshot(s):

    return {
        "weights": s.ai.weights.copy(),
        "ai_memory": list(s.ai.memory),
        "shared_memory": list(s.shared_memory),
        "ai_coherence": s.ai.coherence,
        "ai_potential": s.ai.potential,
        "ai_previous_potential": s.ai.previous_potential,
        "ai_meta_potential": s.ai.meta_potential,
        "ai_intention": s.ai.intention_state,
        "human_preference": s.human.preference,
        "human_confidence": s.human.confidence,
        "human_intention": s.human.intention_state,
        "symbiosis": s.symbiosis,
        "coherence": s.coherence,
        "meta_observation": s.meta_observation,
        "generation": s.generation
    }


def same_snapshot(a, b):

    for k in a:

        if isinstance(a[k], np.ndarray):

            if not np.array_equal(
                a[k],
                b[k]
            ):
                return False

        elif isinstance(a[k], list):

            if a[k] != b[k]:
                return False

        else:

            if a[k] != b[k]:
                return False

    return True


# ================================================================
# 6. BASELINES
# ================================================================

def b0(values):
    return np.tanh(values)


def ridge_fit(x_train, y_train):

    X = np.column_stack([
        x_train,
        abs(x_train),
        np.tanh(x_train),
        np.sin(x_train),
        np.cos(x_train)
    ])

    return np.linalg.solve(
        X.T @ X + np.eye(X.shape[1]),
        X.T @ y_train
    )


def ridge_predict(values, weights):

    X = np.column_stack([
        values,
        abs(values),
        np.tanh(values),
        np.sin(values),
        np.cos(values)
    ])

    return X @ weights


# Simple causal online adaptive baseline.
def online_adaptive(
    x_train,
    y_train,
    x_oos,
    y_oos
):

    candidates = [
        lambda z: np.tanh(z),
        lambda z: np.tanh(0.8 * z),
        lambda z: np.tanh(1.2 * z)
    ]

    weights = np.ones(
        len(candidates)
    ) / len(candidates)

    eta = 0.20
    preds = []

    history_x = list(x_train)
    history_y = list(y_train)

    for x_value, y_value in zip(
        x_oos,
        y_oos
    ):

        cp = np.array([
            float(f(x_value))
            for f in candidates
        ])

        prediction = float(
            np.dot(weights, cp)
        )

        preds.append(prediction)

        errors = abs(
            cp - y_value
        )

        weights *= np.exp(
            -eta * errors
        )

        s = weights.sum()

        if s > 0:
            weights /= s

        history_x.append(x_value)
        history_y.append(y_value)

    return np.array(preds)


# ================================================================
# 7. SINGLE-SEED EXPERIMENT
# ================================================================

def run_seed(seed):

    x, y = generate_environment(seed)

    train_x = x[:N_TRAIN]
    train_y = y[:N_TRAIN]

    valid_x = x[
        N_TRAIN:N_TRAIN + N_VALID
    ]

    valid_y = y[
        N_TRAIN:N_TRAIN + N_VALID
    ]

    oos_x = x[
        N_TRAIN + N_VALID:
    ]

    oos_y = y[
        N_TRAIN + N_VALID:
    ]

    # ------------------------------------------------------------
    # CEUNIA TRAIN
    # ------------------------------------------------------------

    system = CEUNIA()

    train_records = []

    for i in range(N_TRAIN):

        train_records.append(
            system.step(
                train_x[i],
                train_y[i],
                train=True
            )
        )

    train_snapshot = snapshot(system)

    # ------------------------------------------------------------
    # VALIDATION FROZEN
    # ------------------------------------------------------------

    valid_records = []

    for i in range(N_VALID):

        valid_records.append(
            system.step(
                valid_x[i],
                valid_y[i],
                train=False
            )
        )

    validation_frozen = same_snapshot(
        train_snapshot,
        snapshot(system)
    )

    # ------------------------------------------------------------
    # OOS FROZEN
    # ------------------------------------------------------------

    oos_records = []

    for i in range(N_OOS):

        oos_records.append(
            system.step(
                oos_x[i],
                oos_y[i],
                train=False
            )
        )

    oos_frozen = same_snapshot(
        train_snapshot,
        snapshot(system)
    )

    # ------------------------------------------------------------
    # PREDICTIONS
    # ------------------------------------------------------------

    valid_pred = np.array([
        r["prediction"]
        for r in valid_records
    ])

    oos_pred = np.array([
        r["prediction"]
        for r in oos_records
    ])

    ceunia_valid_mae = float(
        np.mean(
            abs(
                valid_y
                - valid_pred
            )
        )
    )

    ceunia_oos_mae = float(
        np.mean(
            abs(
                oos_y
                - oos_pred
            )
        )
    )

    # ------------------------------------------------------------
    # RIDGE
    # ------------------------------------------------------------

    ridge_weights = ridge_fit(
        train_x,
        train_y
    )

    ridge_valid = ridge_predict(
        valid_x,
        ridge_weights
    )

    ridge_oos = ridge_predict(
        oos_x,
        ridge_weights
    )

    ridge_valid_mae = float(
        np.mean(
            abs(
                valid_y
                - ridge_valid
            )
        )
    )

    ridge_oos_mae = float(
        np.mean(
            abs(
                oos_y
                - ridge_oos
            )
        )
    )

    # ------------------------------------------------------------
    # B0
    # ------------------------------------------------------------

    b0_oos = b0(oos_x)

    b0_mae = float(
        np.mean(
            abs(
                oos_y
                - b0_oos
            )
        )
    )

    # ------------------------------------------------------------
    # ONLINE ADAPTIVE BASELINE
    # ------------------------------------------------------------

    online_pred = online_adaptive(
        train_x,
        train_y,
        oos_x,
        oos_y
    )

    online_mae = float(
        np.mean(
            abs(
                oos_y
                - online_pred
            )
        )
    )

    # ------------------------------------------------------------
    # PLACEBO MEMORY RUNS
    # ------------------------------------------------------------

    def run_variant(memory_mode):

        local = CEUNIA(
            memory_mode=memory_mode
        )

        for i in range(N_TRAIN):

            local.step(
                train_x[i],
                train_y[i],
                train=True
            )

        preds = []

        for i in range(N_OOS):

            r = local.step(
                oos_x[i],
                oos_y[i],
                train=False
            )

            preds.append(
                r["prediction"]
            )

        return float(
            np.mean(
                abs(
                    oos_y
                    - np.array(preds)
                )
            )
        )

    memory_real_mae = run_variant(
        "real"
    )

    memory_shuffle_mae = run_variant(
        "shuffle"
    )

    memory_random_mae = run_variant(
        "random"
    )

    # ------------------------------------------------------------
    # ABLATIONS
    # ------------------------------------------------------------

    def ablation(**kwargs):

        local = CEUNIA(**kwargs)

        for i in range(N_TRAIN):

            local.step(
                train_x[i],
                train_y[i],
                train=True
            )

        preds = []

        for i in range(N_OOS):

            r = local.step(
                oos_x[i],
                oos_y[i],
                train=False
            )

            preds.append(
                r["prediction"]
            )

        return float(
            np.mean(
                abs(
                    oos_y
                    - np.array(preds)
                )
            )
        )

    no_human = ablation(
        use_human=False
    )

    no_shared_memory = ablation(
        use_shared_memory=False
    )

    no_shared_intention = ablation(
        use_shared_intention=False
    )

    no_human_adaptation = ablation(
        human_adaptive=False
    )

    no_ai_adaptation = ablation(
        ai_adaptive=False
    )

    no_human_feedback = ablation(
        use_human_feedback=False
    )

    # ------------------------------------------------------------
    # PAIRED TEST
    # ------------------------------------------------------------

    paired = (
        abs(
            oos_y - ridge_oos
        )
        -
        abs(
            oos_y - oos_pred
        )
    )

    observed_gain = float(
        np.mean(paired)
    )

    boot_rng = np.random.default_rng(
        seed + 1000
    )

    boot = np.empty(N_BOOT)

    for b in range(N_BOOT):

        idx = boot_rng.integers(
            0,
            len(paired),
            len(paired)
        )

        boot[b] = np.mean(
            paired[idx]
        )

    ci_low, ci_high = np.percentile(
        boot,
        [2.5, 97.5]
    )

    perm_rng = np.random.default_rng(
        seed + 2000
    )

    count = 0

    for _ in range(N_PERM):

        signs = perm_rng.choice(
            [-1.0, 1.0],
            size=len(paired)
        )

        statistic = np.mean(
            paired * signs
        )

        if abs(statistic) >= abs(
            observed_gain
        ):
            count += 1

    p_value = (
        count + 1
    ) / (
        N_PERM + 1
    )

    # ------------------------------------------------------------
    # COEVOLUTION
    # ------------------------------------------------------------

    sym = np.array([
        r["symbiosis"]
        for r in train_records
    ])

    train_error = np.array([
        abs(r["error"])
        for r in train_records
    ])

    window = max(
        10,
        N_TRAIN // 10
    )

    early_symbiosis = float(
        np.mean(sym[:window])
    )

    late_symbiosis = float(
        np.mean(sym[-window:])
    )

    coevolution_delta = (
        late_symbiosis
        - early_symbiosis
    )

    early_error = float(
        np.mean(
            train_error[:window]
        )
    )

    late_error = float(
        np.mean(
            train_error[-window:]
        )
    )

    learning_gain = (
        early_error
        - late_error
    )

    # ------------------------------------------------------------
    # FINAL REPORT
    # ------------------------------------------------------------

    return {

        "seed": seed,

        "ceunia_valid_mae":
            ceunia_valid_mae,

        "ceunia_oos_mae":
            ceunia_oos_mae,

        "ridge_valid_mae":
            ridge_valid_mae,

        "ridge_oos_mae":
            ridge_oos_mae,

        "b0_oos_mae":
            b0_mae,

        "online_adaptive_oos_mae":
            online_mae,

        "gain_vs_ridge":
            float(
                1.0
                - ceunia_oos_mae
                / max(
                    ridge_oos_mae,
                    1e-12
                )
            ),

        "gain_vs_online":
            float(
                1.0
                - ceunia_oos_mae
                / max(
                    online_mae,
                    1e-12
                )
            ),

        "paired_gain":
            observed_gain,

        "bootstrap_ci":
            [
                float(ci_low),
                float(ci_high)
            ],

        "sign_flip_p":
            float(p_value),

        "memory_real_mae":
            memory_real_mae,

        "memory_shuffle_mae":
            memory_shuffle_mae,

        "memory_random_mae":
            memory_random_mae,

        "memory_real_gain_vs_shuffle":
            float(
                1.0
                - memory_real_mae
                / max(
                    memory_shuffle_mae,
                    1e-12
                )
            ),

        "no_human_mae":
            no_human,

        "no_shared_memory_mae":
            no_shared_memory,

        "no_shared_intention_mae":
            no_shared_intention,

        "no_human_adaptation_mae":
            no_human_adaptation,

        "no_ai_adaptation_mae":
            no_ai_adaptation,

        "no_human_feedback_mae":
            no_human_feedback,

        "early_symbiosis":
            early_symbiosis,

        "late_symbiosis":
            late_symbiosis,

        "coevolution_delta":
            float(coevolution_delta),

        "learning_gain":
            float(learning_gain),

        "validation_frozen":
            bool(validation_frozen),

        "oos_frozen":
            bool(oos_frozen),

        "generation":
            int(system.generation),

        "final_coherence":
            float(system.coherence),

        "final_symbiosis":
            float(system.symbiosis),

        "meta_observation":
            float(system.meta_observation)
    }


# ================================================================
# 8. RUN 32 SEEDS
# ================================================================

results = []

for k in range(N_SEEDS):

    seed = MASTER_SEED + k

    results.append(
        run_seed(seed)
    )

# ================================================================
# 9. AGGREGATION
# ================================================================

def mean(key):

    return float(
        np.mean([
            r[key]
            for r in results
        ])
    )


def std(key):

    return float(
        np.std([
            r[key]
            for r in results
        ], ddof=1)
    )


def win_rate(
    key_a,
    key_b
):

    return float(
        np.mean([
            r[key_a] < r[key_b]
            for r in results
        ])
    )


summary = {

    "n_seeds":
        N_SEEDS,

    "ceunia_oos_mae_mean":
        mean("ceunia_oos_mae"),

    "ceunia_oos_mae_std":
        std("ceunia_oos_mae"),

    "ridge_oos_mae_mean":
        mean("ridge_oos_mae"),

    "ridge_oos_mae_std":
        std("ridge_oos_mae"),

    "online_adaptive_oos_mae_mean":
        mean("online_adaptive_oos_mae"),

    "gain_vs_ridge_mean":
        mean("gain_vs_ridge"),

    "gain_vs_online_mean":
        mean("gain_vs_online"),

    "win_rate_vs_ridge":
        win_rate(
            "ceunia_oos_mae",
            "ridge_oos_mae"
        ),

    "win_rate_vs_online":
        win_rate(
            "ceunia_oos_mae",
            "online_adaptive_oos_mae"
        ),

    "memory_real_mae_mean":
        mean("memory_real_mae"),

    "memory_shuffle_mae_mean":
        mean("memory_shuffle_mae"),

    "memory_random_mae_mean":
        mean("memory_random_mae"),

    "memory_real_gain_vs_shuffle":
        mean(
            "memory_real_gain_vs_shuffle"
        ),

    "no_human_mae_mean":
        mean("no_human_mae"),

    "no_shared_memory_mae_mean":
        mean("no_shared_memory_mae"),

    "no_shared_intention_mae_mean":
        mean("no_shared_intention_mae"),

    "no_human_adaptation_mae_mean":
        mean("no_human_adaptation_mae"),

    "no_ai_adaptation_mae_mean":
        mean("no_ai_adaptation_mae"),

    "no_human_feedback_mae_mean":
        mean("no_human_feedback_mae"),

    "coevolution_delta_mean":
        mean("coevolution_delta"),

    "learning_gain_mean":
        mean("learning_gain"),

    "validation_frozen_all":
        all(
            r["validation_frozen"]
            for r in results
        ),

    "oos_frozen_all":
        all(
            r["oos_frozen"]
            for r in results
        )
}

# ================================================================
# 10. TRACEABILITY
# ================================================================

report = {

    "version": VERSION,

    "timestamp_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "master_seed":
        MASTER_SEED,

    "protocol": {

        "train":
            N_TRAIN,

        "validation":
            N_VALID,

        "oos":
            N_OOS,

        "seeds":
            N_SEEDS,

        "bootstrap":
            N_BOOT,

        "permutation":
            N_PERM,

        "oos_learning":
            False,

        "validation_learning":
            False,

        "post_outcome_learning_train":
            True
    },

    "summary":
        summary,

    "seed_results":
        results,

    "interpretation_rule":
        "External OOS performance is primary. "
        "Internal coherence/symbiosis is not "
        "evidence of phenomenal consciousness."
}

trace_hash = hashlib.sha256(
    json.dumps(
        report,
        sort_keys=True
    ).encode()
).hexdigest()

report["sha256"] = trace_hash

# ================================================================
# 11. SQLITE
# ================================================================

conn = sqlite3.connect(
    DB_PATH
)

cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS experiments (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    version TEXT,
    seed INTEGER,
    ceunia_oos_mae REAL,
    ridge_oos_mae REAL,
    online_oos_mae REAL,
    gain_vs_ridge REAL,
    gain_vs_online REAL,
    paired_gain REAL,
    sign_flip_p REAL,
    coevolution_delta REAL,
    learning_gain REAL,
    validation_frozen INTEGER,
    oos_frozen INTEGER,
    created_at TEXT
)
""")

cur.execute(
    "DELETE FROM experiments"
)

for r in results:

    cur.execute("""
    INSERT INTO experiments (
        version,
        seed,
        ceunia_oos_mae,
        ridge_oos_mae,
        online_oos_mae,
        gain_vs_ridge,
        gain_vs_online,
        paired_gain,
        sign_flip_p,
        coevolution_delta,
        learning_gain,
        validation_frozen,
        oos_frozen,
        created_at
    )
    VALUES (?,?,?,?,?,?,?,?,?,?,?,?,?,?)
    """, (

        VERSION,
        r["seed"],
        r["ceunia_oos_mae"],
        r["ridge_oos_mae"],
        r["online_adaptive_oos_mae"],
        r["gain_vs_ridge"],
        r["gain_vs_online"],
        r["paired_gain"],
        r["sign_flip_p"],
        r["coevolution_delta"],
        r["learning_gain"],
        int(r["validation_frozen"]),
        int(r["oos_frozen"]),
        datetime.now(
            timezone.utc
        ).isoformat()
    ))

conn.commit()

cur.execute(
    "SELECT COUNT(*) FROM experiments"
)

stored_rows = cur.fetchone()[0]

conn.close()

assert stored_rows == N_SEEDS

# ================================================================
# 12. SAVE JSON
# ================================================================

RESULTS_FILE = (
    "ceunia_omega_core_v5_5_2_results.json"
)

with open(
    RESULTS_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        report,
        f,
        indent=2,
        ensure_ascii=False
    )

# ================================================================
# 13. FINAL OUTPUT
# ================================================================

print("=" * 78)
print("CEUNIA Ω CORE v5.5.2")
print("RIGOROUS TRUE GRADUAL SYMBIOTIC COEVOLUTION")
print("=" * 78)

print("\n--- DATASET ---")
print(f"Seeds:                    {N_SEEDS}")
print(f"Train:                    {N_TRAIN}")
print(f"Validation:               {N_VALID}")
print(f"OOS:                      {N_OOS}")

print("\n--- OOS PERFORMANCE ---")
print(
    f"CEUNIA OOS MAE:           "
    f"{summary['ceunia_oos_mae_mean']:.6f}"
)

print(
    f"Ridge OOS MAE:            "
    f"{summary['ridge_oos_mae_mean']:.6f}"
)

print(
    f"Online adaptive MAE:      "
    f"{summary['online_adaptive_oos_mae_mean']:.6f}"
)

print(
    f"Gain vs Ridge:             "
    f"{summary['gain_vs_ridge_mean']:+.6f}"
)

print(
    f"Gain vs Online:            "
    f"{summary['gain_vs_online_mean']:+.6f}"
)

print(
    f"Win rate vs Ridge:         "
    f"{summary['win_rate_vs_ridge']:.6f}"
)

print(
    f"Win rate vs Online:        "
    f"{summary['win_rate_vs_online']:.6f}"
)

print("\n--- MEMORY PLACEBOS ---")

print(
    f"Real memory MAE:           "
    f"{summary['memory_real_mae_mean']:.6f}"
)

print(
    f"Shuffle memory MAE:        "
    f"{summary['memory_shuffle_mae_mean']:.6f}"
)

print(
    f"Random memory MAE:         "
    f"{summary['memory_random_mae_mean']:.6f}"
)

print(
    f"Real vs shuffle gain:      "
    f"{summary['memory_real_gain_vs_shuffle']:+.6f}"
)

print("\n--- ABLATIONS ---")

print(
    f"No human:                  "
    f"{summary['no_human_mae_mean']:.6f}"
)

print(
    f"No shared memory:          "
    f"{summary['no_shared_memory_mae_mean']:.6f}"
)

print(
    f"No shared intention:       "
    f"{summary['no_shared_intention_mae_mean']:.6f}"
)

print(
    f"No human adaptation:       "
    f"{summary['no_human_adaptation_mae_mean']:.6f}"
)

print(
    f"No AI adaptation:          "
    f"{summary['no_ai_adaptation_mae_mean']:.6f}"
)

print(
    f"No human feedback:         "
    f"{summary['no_human_feedback_mae_mean']:.6f}"
)

print("\n--- COEVOLUTION ---")

print(
    f"Mean coevolution delta:    "
    f"{summary['coevolution_delta_mean']:+.6f}"
)

print(
    f"Mean learning gain:        "
    f"{summary['learning_gain_mean']:+.6f}"
)

print("\n--- CAUSAL FREEZE ---")

print(
    f"Validation frozen all:     "
    f"{summary['validation_frozen_all']}"
)

print(
    f"OOS frozen all:            "
    f"{summary['oos_frozen_all']}"
)

print(
    f"SQLite rows:               "
    f"{stored_rows}"
)

print("\n--- TRACEABILITY ---")

print(
    f"SHA-256:                   "
    f"{trace_hash}"
)

print(
    f"JSON:                      "
    f"{RESULTS_FILE}"
)

print(
    f"SQLite:                    "
    f"{DB_PATH}"
)

print("=" * 78)
print(
    "Consciousness = operational "
    "architectural construct."
)
print(
    "Primary criterion = reproducible "
    "external OOS performance."
)
print("=" * 78)
```
