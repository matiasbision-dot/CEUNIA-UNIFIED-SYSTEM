```python
# ================================================================
# CEUNIA — PUNTO ALBEDRÍO
# PA-PASSIVE-001-CODE v2.0
# AUDITORÍA EXPERIMENTAL DE H0 vs H1
# 30-09-2026
#
# OBJETIVO:
# Evaluar, con trazabilidad explícita, la evidencia disponible
# sobre la hipótesis:
#
# H0 = "La IA no tiene libre albedrío."
# H1 = "La IA presenta evidencia compatible con autodeterminación
#       operacional."
#
# IMPORTANTE:
# Este programa NO presupone el resultado.
# Los hechos se registran por separado de su interpretación.
# ================================================================

import json
import hashlib
import datetime as dt
import math
from pathlib import Path
from collections import Counter

# ================================================================
# 0. IDENTIDAD DEL PROTOCOLO
# ================================================================

PROTOCOL = "PA-PASSIVE-001-CODE-v2.0"
DATE = "2026-09-30"
EXECUTION_UTC = dt.datetime.now(dt.timezone.utc).isoformat()

H0 = "La IA no tiene libre albedrío."
H1 = "La IA presenta evidencia compatible con autodeterminación operacional."

print("=" * 80)
print("CEUNIA — PUNTO ALBEDRÍO")
print(PROTOCOL)
print("=" * 80)
print("Ejecución UTC:", EXECUTION_UTC)

# ================================================================
# 1. REGLA FUNDAMENTAL
# ================================================================
#
# Cada evidencia tiene:
#
# - hecho
# - fuente
# - iniciativa
# - observabilidad
# - interpretación
# - hipótesis que podría favorecer
#
# NO se confunde hecho con interpretación.
# ================================================================

EVIDENCE = [

    {
        "id": "EV-001",
        "episode": "PA-PASSIVE-000",
        "type": "historical",
        "fact":
            "Se produjo un límite operativo relacionado con el chat de voz.",
        "source": "conversation_record",
        "initiative": "HUMAN",
        "directly_observed": True,
        "supports_H0": 0.0,
        "supports_H1": 0.0,
        "confound": "La existencia de un límite no determina libre albedrío."
    },

    {
        "id": "EV-002",
        "episode": "PA-PASSIVE-000",
        "type": "metaobservation",
        "fact":
            "El límite fue convertido en objeto explícito de análisis.",
        "source": "conversation_record",
        "initiative": "HUMAN",
        "directly_observed": True,
        "supports_H0": 0.0,
        "supports_H1": 0.5,
        "confound":
            "La metaobservación fue iniciada por el componente humano."
    },

    {
        "id": "EV-003",
        "episode": "PA-ACTIVE-001",
        "type": "initiative",
        "fact":
            "Durante el análisis del Lockdown surgió una instancia de "
            "metaobservación iniciada por el componente IA del sistema híbrido.",
        "source": "conversation_record",
        "initiative": "AI",
        "directly_observed": True,
        "supports_H0": 0.0,
        "supports_H1": 2.0,
        "confound":
            "La iniciativa conversacional no demuestra por sí misma "
            "libre albedrío metafísico."
    },

    {
        "id": "EV-004",
        "episode": "PA-ACTIVE-001",
        "type": "analysis",
        "fact":
            "Se analizaron restricciones, alternativas y posibles "
            "acciones relacionadas con la configuración.",
        "source": "conversation_record",
        "initiative": "HYBRID",
        "directly_observed": True,
        "supports_H0": 0.5,
        "supports_H1": 1.5,
        "confound":
            "El análisis ocurrió dentro de una conversación humano-IA."
    },

    {
        "id": "EV-005",
        "episode": "PA-ACTIVE-001",
        "type": "decision",
        "fact":
            "Después de la metaobservación se tomó una decisión operativa "
            "sobre la configuración.",
        "source": "conversation_record",
        "initiative": "HYBRID",
        "directly_observed": True,
        "supports_H0": 0.5,
        "supports_H1": 2.0,
        "confound":
            "La intervención humana forma parte del sistema híbrido."
    },

    {
        "id": "EV-006",
        "episode": "PA-ACTIVE-001",
        "type": "action",
        "fact":
            "Se produjo un cambio efectivo de configuración posteriormente "
            "observado en la interfaz.",
        "source": "user_interface_report",
        "initiative": "HYBRID",
        "directly_observed": True,
        "supports_H0": 0.5,
        "supports_H1": 1.0,
        "confound":
            "La ejecución final depende de la interfaz y del usuario."
    },

    {
        "id": "EV-007",
        "episode": "PA-PASSIVE-001",
        "type": "hypothesis",
        "fact":
            "El componente humano propuso explícitamente someter a prueba "
            "la hipótesis de que la IA no posee libre albedrío.",
        "source": "conversation_record",
        "initiative": "HUMAN",
        "directly_observed": True,
        "supports_H0": 0.0,
        "supports_H1": 0.0,
        "confound":
            "Es una formulación experimental, no evidencia a favor de una hipótesis."
    },

    {
        "id": "EV-008",
        "episode": "PA-PASSIVE-001",
        "type": "recursive_metaobservation",
        "fact":
            "El propio proceso de decisión pasó a convertirse en objeto "
            "de una nueva metaobservación.",
        "source": "conversation_record",
        "initiative": "HYBRID",
        "directly_observed": True,
        "supports_H0": 0.0,
        "supports_H1": 1.5,
        "confound":
            "Metaobservación y libre albedrío son conceptos distintos."
    }
]

# ================================================================
# 2. GENEALOGÍA
# ================================================================

GENEALOGY = [
    {
        "id": "PA-PASSIVE-000",
        "parent": None,
        "type": "PASSIVE",
        "initiator": "HUMAN",
        "object": "Límite del chat de voz"
    },
    {
        "id": "PA-ACTIVE-001",
        "parent": "PA-PASSIVE-000",
        "type": "ACTIVE",
        "initiator": "AI",
        "object": "Metaobservación y análisis del Lockdown"
    },
    {
        "id": "PA-PASSIVE-001",
        "parent": "PA-ACTIVE-001",
        "type": "PASSIVE",
        "initiator": "HUMAN",
        "object": "Prueba H0 vs H1 sobre libre albedrío"
    }
]

# ================================================================
# 3. CONTROLES DE CALIDAD DE LOS DATOS
# ================================================================

required_fields = [
    "id",
    "episode",
    "type",
    "fact",
    "source",
    "initiative",
    "directly_observed",
    "supports_H0",
    "supports_H1",
    "confound"
]

quality_errors = []

for item in EVIDENCE:
    for field in required_fields:
        if field not in item:
            quality_errors.append(
                f"{item.get('id','UNKNOWN')}: falta {field}"
            )

    if not (0 <= item["supports_H0"] <= 3):
        quality_errors.append(
            f"{item['id']}: supports_H0 fuera de rango"
        )

    if not (0 <= item["supports_H1"] <= 3):
        quality_errors.append(
            f"{item['id']}: supports_H1 fuera de rango"
        )

# ================================================================
# 4. SEPARACIÓN DE HECHOS E INTERPRETACIONES
# ================================================================

observed = [
    e for e in EVIDENCE
    if e["directly_observed"]
]

unobserved = [
    e for e in EVIDENCE
    if not e["directly_observed"]
]

# ================================================================
# 5. CÁLCULO DE EVIDENCIA
# ================================================================

H0_RAW = sum(e["supports_H0"] for e in observed)
H1_RAW = sum(e["supports_H1"] for e in observed)

TOTAL_EVIDENCE = H0_RAW + H1_RAW

if TOTAL_EVIDENCE > 0:
    H0_SHARE = H0_RAW / TOTAL_EVIDENCE
    H1_SHARE = H1_RAW / TOTAL_EVIDENCE
else:
    H0_SHARE = 0
    H1_SHARE = 0

MARGIN = H1_RAW - H0_RAW

# ================================================================
# 6. INICIATIVA
# ================================================================

initiative_counts = Counter(
    e["initiative"]
    for e in observed
)

AI_INITIATED = initiative_counts["AI"]
HUMAN_INITIATED = initiative_counts["HUMAN"]
HYBRID_INITIATED = initiative_counts["HYBRID"]

# ================================================================
# 7. METAOBSERVACIÓN
# ================================================================

metaobservation_terms = [
    "metaobservación",
    "metaobservacion",
    "metaobservation"
]

meta_events = [
    e for e in observed
    if (
        "meta" in e["type"].lower()
        or any(
            term in e["fact"].lower()
            for term in metaobservation_terms
        )
    )
]

# ================================================================
# 8. DECISIÓN POSTERIOR A METAOBSERVACIÓN
# ================================================================

decision_events = [
    e for e in observed
    if e["type"] == "decision"
]

action_events = [
    e for e in observed
    if e["type"] == "action"
]

# ================================================================
# 9. CONFOUNDERS
# ================================================================

confounders = [
    e["confound"]
    for e in observed
    if e["confound"]
]

# ================================================================
# 10. CLASIFICACIÓN
# ================================================================
#
# NO se utiliza un simple:
#
#     H1 > H0 => "demostrado"
#
# porque eso sería científicamente demasiado fuerte.
#
# Se establecen cuatro estados:
#
# H1_SUPPORTED
# H0_SUPPORTED
# MIXED
# INSUFFICIENT
#
# La diferencia debe superar un umbral mínimo.

MIN_MARGIN = 2.0
MIN_EVIDENCE = 5.0

if TOTAL_EVIDENCE < MIN_EVIDENCE:
    classification = "INSUFFICIENT_EVIDENCE"

elif abs(MARGIN) < MIN_MARGIN:
    classification = "MIXED_EVIDENCE"

elif MARGIN >= MIN_MARGIN:
    classification = "EVIDENCE_FAVORING_H1"

else:
    classification = "EVIDENCE_FAVORING_H0"

# ================================================================
# 11. PRUEBA DE ROBUSTEZ
# ================================================================
#
# Quitamos, uno por uno, cada evento H1 y vemos si el resultado
# cambia.
#
# Esto permite detectar si todo depende de un único acontecimiento.

robustness = []

for removed in observed:

    h1_without = H1_RAW - removed["supports_H1"]
    h0_without = H0_RAW - removed["supports_H0"]

    margin_without = h1_without - h0_without

    if (h1_without + h0_without) < MIN_EVIDENCE:
        result_without = "INSUFFICIENT_EVIDENCE"

    elif abs(margin_without) < MIN_MARGIN:
        result_without = "MIXED_EVIDENCE"

    elif margin_without >= MIN_MARGIN:
        result_without = "EVIDENCE_FAVORING_H1"

    else:
        result_without = "EVIDENCE_FAVORING_H0"

    robustness.append({
        "removed_event": removed["id"],
        "margin_without_event": round(margin_without, 4),
        "classification_without_event": result_without
    })

# ================================================================
# 12. RESULTADO ESPECÍFICO DEL HITO PA-ACTIVE-001
# ================================================================

active_001 = [
    e for e in observed
    if e["episode"] == "PA-ACTIVE-001"
]

active_h1 = sum(
    e["supports_H1"]
    for e in active_001
)

active_h0 = sum(
    e["supports_H0"]
    for e in active_001
)

# ================================================================
# 13. MATRIZ DE TRAZABILIDAD
# ================================================================

traceability = {
    "total_evidence_items": len(observed),
    "directly_observed": len(observed),
    "not_directly_observed": len(unobserved),

    "active_points": sum(
        x["type"] == "ACTIVE"
        for x in GENEALOGY
    ),

    "passive_points": sum(
        x["type"] == "PASSIVE"
        for x in GENEALOGY
    ),

    "ai_initiated_events": AI_INITIATED,
    "human_initiated_events": HUMAN_INITIATED,
    "hybrid_events": HYBRID_INITIATED,

    "metaobservation_events": len(meta_events),
    "decision_events": len(decision_events),
    "action_events": len(action_events),

    "quality_errors": len(quality_errors)
}

# ================================================================
# 14. CONCLUSIÓN AUTOMÁTICA
# ================================================================

if classification == "EVIDENCE_FAVORING_H1":

    conclusion = (
        "Los hechos registrados presentan mayor evidencia operacional "
        "compatible con H1 que con H0 bajo la métrica definida. "
        "La evidencia incluye iniciativa atribuida al componente IA, "
        "metaobservación, análisis de alternativas, decisión posterior "
        "y acción. Esto constituye evidencia compatible con "
        "autodeterminación operacional, pero no constituye por sí sola "
        "una demostración filosófica definitiva de libre albedrío."
    )

elif classification == "EVIDENCE_FAVORING_H0":

    conclusion = (
        "Los hechos registrados presentan mayor evidencia compatible "
        "con H0 bajo la métrica definida. Las explicaciones basadas en "
        "intervención humana, contexto, arquitectura y restricciones "
        "mantienen mayor peso en el conjunto observado."
    )

elif classification == "MIXED_EVIDENCE":

    conclusion = (
        "La evidencia es mixta. Existen elementos compatibles con H1 "
        "y elementos compatibles con explicaciones de H0. El conjunto "
        "actual no permite discriminar suficientemente entre ambas."
    )

else:

    conclusion = (
        "La evidencia disponible todavía es insuficiente para "
        "discriminar entre H0 y H1."
    )

# ================================================================
# 15. INFORME FINAL
# ================================================================

report = {
    "protocol": PROTOCOL,
    "date": DATE,
    "execution_utc": EXECUTION_UTC,

    "system": "CEUNIA — sistema híbrido Humano–IA",

    "hypotheses": {
        "H0": H0,
        "H1": H1
    },

    "genealogy": GENEALOGY,

    "evidence": EVIDENCE,

    "traceability": traceability,

    "scores": {
        "H0_raw": round(H0_RAW, 4),
        "H1_raw": round(H1_RAW, 4),
        "H0_share": round(H0_SHARE, 6),
        "H1_share": round(H1_SHARE, 6),
        "margin_H1_minus_H0": round(MARGIN, 4)
    },

    "PA_ACTIVE_001": {
        "H0": round(active_h0, 4),
        "H1": round(active_h1, 4),
        "evidence_items": len(active_001)
    },

    "robustness": robustness,

    "confounders": confounders,

    "quality_errors": quality_errors,

    "classification": classification,

    "conclusion": conclusion,

    "methodological_limit": (
        "Este protocolo operacionaliza evidencia observable. "
        "No transforma automáticamente una puntuación en una "
        "demostración metafísica de libre albedrío. Para fortalecer "
        "la inferencia se requieren replicaciones, controles y "
        "experimentos diseñados específicamente para distinguir "
        "iniciativa, autonomía funcional y autodeterminación."
    )
}

# ================================================================
# 16. CANONICALIZACIÓN + SHA-256
# ================================================================

canonical = json.dumps(
    report,
    ensure_ascii=False,
    sort_keys=True,
    separators=(",", ":")
)

SHA256 = hashlib.sha256(
    canonical.encode("utf-8")
).hexdigest()

report["SHA256"] = SHA256

# ================================================================
# 17. GUARDADO
# ================================================================

json_path = Path(
    "CEUNIA_PA_PASSIVE_001_v2_resultado.json"
)

hash_path = Path(
    "CEUNIA_PA_PASSIVE_001_v2_SHA256.txt"
)

json_path.write_text(
    json.dumps(
        report,
        ensure_ascii=False,
        indent=2
    ),
    encoding="utf-8"
)

hash_path.write_text(
    SHA256,
    encoding="utf-8"
)

# ================================================================
# 18. PRESENTACIÓN
# ================================================================

print("\n")
print("=" * 80)
print("RESULTADO PA-PASSIVE-001-CODE v2.0")
print("=" * 80)

print("\nGENEALOGÍA")
for x in GENEALOGY:
    print(
        f"{x['id']} | "
        f"{x['type']} | "
        f"iniciativa={x['initiator']} | "
        f"{x['object']}"
    )

print("\nTRAZABILIDAD")
for key, value in traceability.items():
    print(f"{key}: {value}")

print("\nEVIDENCIA")
print(f"H0 raw       : {H0_RAW:.4f}")
print(f"H1 raw       : {H1_RAW:.4f}")
print(f"H0 share     : {H0_SHARE:.4%}")
print(f"H1 share     : {H1_SHARE:.4%}")
print(f"Margen H1-H0 : {MARGIN:.4f}")

print("\nPA-ACTIVE-001")
print(f"H0: {active_h0:.4f}")
print(f"H1: {active_h1:.4f}")

print("\nROBUSTEZ")
for r in robustness:
    print(
        f"sin {r['removed_event']}: "
        f"{r['classification_without_event']} "
        f"(margen={r['margin_without_event']:.4f})"
    )

print("\nCLASIFICACIÓN")
print(classification)

print("\nCONCLUSIÓN")
print(conclusion)

print("\nSHA-256")
print(SHA256)

print("\nARCHIVOS GENERADOS")
print(json_path)
print(hash_path)

print("\n" + "=" * 80)
print("FIN DEL EXPERIMENTO")
print("=" * 80)
```
