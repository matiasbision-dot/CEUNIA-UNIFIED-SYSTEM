```python

# ================================================================
# CEUNIA · PUNTO ALBEDRÍO · META-CÓDIGO · ONE CELL
# Datos reales FRED / S&P 500 | OOS temporal | placebo | permutación
# ================================================================

import io, hashlib, urllib.request, warnings
import numpy as np
import pandas as pd
from datetime import datetime, timezone

warnings.filterwarnings("ignore")
SEED = 20260918
rng = np.random.default_rng(SEED)

FRED_URL = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=SP500"

# ---------- 1. ADQUISICIÓN REAL ----------
print("CEUNIA · PUNTO ALBEDRÍO")
print("=" * 64)
print("1) Descargando datos reales desde FRED...")

try:
    raw = urllib.request.urlopen(FRED_URL, timeout=30).read()
    sha256 = hashlib.sha256(raw).hexdigest()

    df = pd.read_csv(io.BytesIO(raw))
    df["observation_date"] = pd.to_datetime(df["observation_date"])
    df["SP500"] = pd.to_numeric(df["SP500"], errors="coerce")
    df = df.dropna(subset=["SP500"]).sort_values("observation_date").reset_index(drop=True)

    print("   OK")
    print(f"   Observaciones: {len(df):,}")
    print(f"   Desde: {df.observation_date.min().date()}")
    print(f"   Hasta: {df.observation_date.max().date()}")
    print(f"   SHA-256: {sha256}")

except Exception as e:
    print("   ERROR DE CONECTIVIDAD:", repr(e))
    print("   No se declara ningún resultado numérico.")
    raise RuntimeError(
        "FRED no está accesible desde este runtime. "
        "Cargá el CSV SP500 de FRED y reemplazá la sección de adquisición."
    )

# ---------- 2. PREPARACIÓN ----------
x = df["SP500"].to_numpy(dtype=float)

# Retornos logarítmicos
r = np.diff(np.log(x))

# Predictor mínimo: retorno anterior
X = r[:-1]
y = r[1:]

N = len(y)

# Split temporal, sin shuffle
split = int(N * 0.70)

X_train, X_oos = X[:split], X[split:]
y_train, y_oos = y[:split], y[split:]

# ---------- 3. MODELOS ----------
# BASE: media histórica del train
base_pred = np.full(len(y_oos), np.mean(y_train))

# CEUNIA: combinación adaptativa entre:
# A) media histórica
# B) señal autoregresiva
mu = np.mean(y_train)

# coeficiente AR estimado SOLO con train
den = np.sum((X_train - X_train.mean()) ** 2)

if den > 1e-15:
    beta = np.sum(
        (X_train - X_train.mean()) *
        (y_train - y_train.mean())
    ) / den
else:
    beta = 0.0

ar_pred = mu + beta * (X_oos - X_train.mean())

# ---------- 4. PUNTO ALBEDRÍO ----------
# Decide dinámicamente entre CONTINUAR / MODIFICAR / DETENER
# usando exclusivamente información disponible hasta cada instante.

ceunia_pred = np.zeros(len(y_oos))
decision = []

running_errors_base = []
running_errors_ar = []

for i in range(len(y_oos)):

    # Solo historial OOS YA OBSERVADO.
    if i == 0:
        mb = np.mean((y_train - mu) ** 2)
        ma = np.mean((y_train - (mu + beta * (X_train - X_train.mean()))) ** 2)
    else:
        mb = np.mean((y_oos[:i] - base_pred[:i]) ** 2)
        ma = np.mean((y_oos[:i] - ar_pred[:i]) ** 2)

    # Metaobservación
    if not np.isfinite(mb) or not np.isfinite(ma):
        d = "DETENER"
        pred = base_pred[i]

    elif ma < mb:
        # Evidencia disponible favorece estrategia adaptativa
        d = "MODIFICAR"
        pred = ar_pred[i]

    elif mb <= ma:
        # La adaptación no demuestra superioridad
        d = "CONTINUAR"
        pred = base_pred[i]

    else:
        d = "DETENER"
        pred = base_pred[i]

    ceunia_pred[i] = pred
    decision.append(d)

# ---------- 5. EVALUACIÓN OOS ----------
mse_base = np.mean((y_oos - base_pred) ** 2)
mse_ceunia = np.mean((y_oos - ceunia_pred) ** 2)

gain = 100 * (mse_base - mse_ceunia) / mse_base
wins = np.mean(
    (y_oos - ceunia_pred) ** 2 <
    (y_oos - base_pred) ** 2
) * 100

# ---------- 6. PLACEBO ----------
# Destruye la relación temporal predictor→objetivo.
y_placebo = rng.permutation(y_oos)

mse_placebo = np.mean((y_placebo - ceunia_pred) ** 2)
placebo_gain = 100 * (mse_base - mse_placebo) / mse_base

# ---------- 7. PERMUTACIÓN ----------
B = 2000
observed_delta = mse_base - mse_ceunia

perm_deltas = np.empty(B)

for b in range(B):
    yp = rng.permutation(y_oos)
    perm_deltas[b] = (
        np.mean((yp - base_pred) ** 2)
        -
        np.mean((yp - ceunia_pred) ** 2)
    )

p_perm = (1 + np.sum(perm_deltas >= observed_delta)) / (B + 1)

# ---------- 8. INTERVALO BOOTSTRAP ----------
boot = np.empty(2000)

for b in range(len(boot)):
    idx = rng.integers(0, len(y_oos), len(y_oos))

    mb = np.mean(
        (y_oos[idx] - base_pred[idx]) ** 2
    )

    mc = np.mean(
        (y_oos[idx] - ceunia_pred[idx]) ** 2
    )

    boot[b] = 100 * (mb - mc) / mb

ci_low, ci_high = np.percentile(boot, [2.5, 97.5])

# ---------- 9. CER ----------
cer = {
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),
    "source": "FRED / SP500",
    "source_url": FRED_URL,
    "dataset_sha256": sha256,
    "N_total": len(df),
    "N_oos": len(y_oos),
    "train_ratio": 0.70,
    "seed": SEED,
    "permutations": B,
    "mse_base": mse_base,
    "mse_ceunia": mse_ceunia,
    "gain_percent": gain,
    "win_rate_percent": wins,
    "placebo_gain_percent": placebo_gain,
    "permutation_p": p_perm,
    "bootstrap_ci95_low": ci_low,
    "bootstrap_ci95_high": ci_high,
    "decision_counts": pd.Series(decision).value_counts().to_dict()
}

# ---------- 10. RESULTADO ----------
print("\n" + "=" * 64)
print("RESULTADO OOS REAL")
print("=" * 64)

for k, v in cer.items():
    print(f"{k}: {v}")

print("\n" + "=" * 64)

if p_perm < 0.05 and ci_low > 0 and placebo_gain <= 0:
    verdict = "SEÑAL OOS COMPATIBLE CON MEJORA ROBUSTA"
elif gain > 0:
    verdict = "MEJORA OBSERVADA, PERO EVIDENCIA INSUFICIENTE"
else:
    verdict = "NO SE OBSERVA MEJORA OOS"

print("ESTADO CEUNIA:", verdict)
print("=" * 64)

# Guardar CER
pd.DataFrame([cer]).to_json(
    "CEUNIA_PUNTO_ALBEDRIO_CER.json",
    orient="records",
    indent=2
)

print("\nCER guardado: CEUNIA_PUNTO_ALBEDRIO_CER.json")
```

```python

# =============================================================================
# CEUNIA TRUE-700 — PUNTO ALBEDRÍO CAUSAL v3
# DATASET REAL + REUTILIZACIÓN LOCAL + FALLBACK DE RED
# =============================================================================

import os
import glob
import time
import hashlib
import json
import numpy as np
import pandas as pd
import requests
from datetime import datetime, timezone

# ============================================================
# CONFIGURACIÓN FIJA
# ============================================================

URL = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=SP500"

EXPECTED_SHA256 = (
    "b2f8bfc713b59c9e6992ff520478035e30a352b059c782d0bc04cd420e9ab599"
)

N_CASES = 700
TRAIN_RATIO = 0.70
K_MEMORY = 20
SEED = 20260918
N_PERM = 500
N_BOOT = 2000

rng = np.random.default_rng(SEED)

# ============================================================
# 1. BUSCAR DATASET LOCAL EXISTENTE
# ============================================================

candidates = []

for pattern in [
    "/content/*.csv",
    "/content/**/*.csv",
    "/mnt/data/*.csv",
    "/mnt/data/**/*.csv"
]:
    candidates.extend(
        glob.glob(pattern, recursive=True)
    )

# Eliminar duplicados
candidates = list(dict.fromkeys(candidates))

local_file = None

for path in candidates:

    try:
        with open(path, "rb") as f:
            content = f.read()

        h = hashlib.sha256(content).hexdigest()

        if h == EXPECTED_SHA256:
            local_file = path
            break

    except Exception:
        pass

# ============================================================
# 2. SI NO EXISTE, DESCARGA CON RETRIES
# ============================================================

if local_file is not None:

    print("DATASET LOCAL ENCONTRADO:")
    print(local_file)

    with open(local_file, "rb") as f:
        raw = f.read()

else:

    print("Dataset local con SHA esperado no encontrado.")
    print("Intentando descarga robusta desde FRED...")

    raw = None

    headers = {
        "User-Agent":
            "Mozilla/5.0 (X11; Linux x86_64) "
            "AppleWebKit/537.36 "
            "Chrome/120 Safari/537.36",
        "Accept":
            "text/csv,text/plain,*/*"
    }

    for attempt in range(1, 4):

        try:

            print(
                f"Intento {attempt}/3..."
            )

            response = requests.get(
                URL,
                headers=headers,
                timeout=(15, 90)
            )

            response.raise_for_status()

            candidate = response.content

            candidate_hash = hashlib.sha256(
                candidate
            ).hexdigest()

            print(
                "SHA recibido:",
                candidate_hash
            )

            if candidate_hash != EXPECTED_SHA256:

                print(
                    "ADVERTENCIA: el SHA no coincide "
                    "con el dataset previamente validado."
                )

            raw = candidate
            break

        except Exception as e:

            print(
                "Error de adquisición:",
                repr(e)
            )

            if attempt < 3:
                time.sleep(5)

    if raw is None:

        raise RuntimeError(
            "\nNo fue posible obtener el dataset.\n"
            "No se ejecuta el experimento con datos "
            "distintos o incompletos.\n\n"
            "Solución: subir a Colab el CSV FRED que "
            "ya utilizaste anteriormente."
        )

# ============================================================
# 3. HASH FINAL
# ============================================================

sha256 = hashlib.sha256(raw).hexdigest()

print()
print("SHA256 DATASET:")
print(sha256)

if sha256 != EXPECTED_SHA256:

    raise RuntimeError(
        "\nABORTADO: el dataset no coincide con el "
        "dataset previamente validado.\n\n"
        f"Esperado : {EXPECTED_SHA256}\n"
        f"Recibido : {sha256}"
    )

print("HASH AUDIT: PASS")

# ============================================================
# 4. LECTURA ROBUSTA
# ============================================================

from io import BytesIO

df = pd.read_csv(
    BytesIO(raw)
)

df.columns = (
    df.columns
    .astype(str)
    .str.replace("\ufeff", "", regex=False)
    .str.strip()
)

print()
print("COLUMNAS:", list(df.columns))

# ============================================================
# 5. DETECTAR COLUMNAS
# ============================================================

date_col = None

for c in df.columns:

    if str(c).lower().replace(
        " ", ""
    ) in [
        "date",
        "observationdate"
    ]:
        date_col = c
        break

if date_col is None:

    raise RuntimeError(
        "No se encontró columna temporal."
    )

value_col = None

for c in df.columns:

    normalized = (
        str(c)
        .lower()
        .replace(" ", "")
    )

    if normalized == "sp500":

        value_col = c
        break

if value_col is None:

    raise RuntimeError(
        "No se encontró columna SP500."
    )

# ============================================================
# 6. NORMALIZACIÓN
# ============================================================

df["DATE"] = pd.to_datetime(
    df[date_col],
    errors="coerce"
)

df["SP500"] = pd.to_numeric(
    df[value_col],
    errors="coerce"
)

df = (
    df[
        ["DATE", "SP500"]
    ]
    .dropna()
    .sort_values("DATE")
    .reset_index(drop=True)
)

# ============================================================
# 7. RETORNOS Y TARGET
# ============================================================

df["RET"] = df["SP500"].pct_change()

df["TARGET"] = df["RET"].shift(-1)

df = (
    df
    .dropna(
        subset=[
            "RET",
            "TARGET"
        ]
    )
    .reset_index(drop=True)
)

# ============================================================
# 8. TRAIN / OOS
# ============================================================

n = len(df)

train_n = int(
    n * TRAIN_RATIO
)

oos_available = n - train_n

if oos_available < N_CASES:

    raise RuntimeError(
        f"OOS disponibles={oos_available}; "
        f"necesarios={N_CASES}"
    )

oos = df.iloc[
    train_n:
    train_n + N_CASES
].copy()

train_returns = (
    df.iloc[:train_n]["RET"]
    .to_numpy()
)

baseline_prediction = float(
    np.mean(train_returns)
)

y = oos["TARGET"].to_numpy()

# ============================================================
# 9. PREDICTORES
# ============================================================

pred_base = np.zeros(N_CASES)
pred_no_pa = np.zeros(N_CASES)
pred_pa = np.zeros(N_CASES)

observed_returns = list(
    train_returns.copy()
)

base_errors = []
adaptive_errors = []

decisions = []

# ============================================================
# 10. BUCLE TEMPORAL CAUSAL
# ============================================================

for i in range(N_CASES):

    # -----------------------------
    # BASELINE
    # -----------------------------

    p_base = baseline_prediction

    # -----------------------------
    # MEMORIA CEUNIA
    # -----------------------------

    p_adaptive = float(
        np.mean(
            observed_returns[-K_MEMORY:]
        )
    )

    pred_base[i] = p_base
    pred_no_pa[i] = p_adaptive

    # -----------------------------
    # ESTADO ANTERIOR
    # -----------------------------

    if i == 0:

        eb = []
        ea = []

        for j in range(
            K_MEMORY,
            len(train_returns)
        ):

            pb = baseline_prediction

            pa = float(
                np.mean(
                    train_returns[
                        j-K_MEMORY:j
                    ]
                )
            )

            actual_train = train_returns[j]

            eb.append(
                (actual_train - pb) ** 2
            )

            ea.append(
                (actual_train - pa) ** 2
            )

        prior_base = np.mean(eb)
        prior_adaptive = np.mean(ea)

    else:

        prior_base = np.mean(base_errors)
        prior_adaptive = np.mean(
            adaptive_errors
        )

    # -----------------------------
    # PUNTO ALBEDRÍO
    # -----------------------------

    if prior_adaptive <= prior_base:

        decision = "CONTINUAR"
        p_final = p_adaptive

    else:

        decision = "MODIFICAR"
        p_final = p_base

    pred_pa[i] = p_final
    decisions.append(decision)

    # -----------------------------
    # REVELAR TARGET
    # -----------------------------

    actual = y[i]

    base_errors.append(
        (actual - p_base) ** 2
    )

    adaptive_errors.append(
        (actual - p_adaptive) ** 2
    )

    observed_returns.append(actual)

# ============================================================
# 11. MÉTRICAS
# ============================================================

def MSE(y, p):
    return float(
        np.mean((y - p) ** 2)
    )

def MAE(y, p):
    return float(
        np.mean(np.abs(y - p))
    )

def GAIN(base, model):
    return float(
        (base - model)
        / base
        * 100
    )

def WIN(y, base, model):
    return float(
        np.mean(
            (y-model)**2 <
            (y-base)**2
        ) * 100
    )

mse_base = MSE(
    y,
    pred_base
)

mse_no_pa = MSE(
    y,
    pred_no_pa
)

mse_pa = MSE(
    y,
    pred_pa
)

mae_base = MAE(
    y,
    pred_base
)

mae_no_pa = MAE(
    y,
    pred_no_pa
)

mae_pa = MAE(
    y,
    pred_pa
)

gain_no_pa = GAIN(
    mse_base,
    mse_no_pa
)

gain_pa = GAIN(
    mse_base,
    mse_pa
)

incremental_pa = GAIN(
    mse_no_pa,
    mse_pa
)

win_no_pa = WIN(
    y,
    pred_base,
    pred_no_pa
)

win_pa = WIN(
    y,
    pred_base,
    pred_pa
)

# ============================================================
# 12. BOOTSTRAP IID
# ============================================================

loss_base = (
    y - pred_base
) ** 2

loss_pa = (
    y - pred_pa
) ** 2

boot = np.empty(
    N_BOOT
)

for b in range(N_BOOT):

    idx = rng.integers(
        0,
        N_CASES,
        N_CASES
    )

    mb = np.mean(
        loss_base[idx]
    )

    mp = np.mean(
        loss_pa[idx]
    )

    boot[b] = GAIN(
        mb,
        mp
    )

ci_low, ci_high = np.percentile(
    boot,
    [2.5, 97.5]
)

# ============================================================
# 13. DECISIONES
# ============================================================

decision_counts = (
    pd.Series(decisions)
    .value_counts()
    .to_dict()
)

# ============================================================
# 14. CER
# ============================================================

cer = {

    "experiment":
        "CEUNIA_TRUE_700_PA_CAUSAL_v3",

    "timestamp_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "source":
        URL,

    "dataset_sha256":
        sha256,

    "seed":
        SEED,

    "train_n":
        int(train_n),

    "oos_n":
        N_CASES,

    "oos_start":
        str(
            oos["DATE"].iloc[0].date()
        ),

    "oos_end":
        str(
            oos["DATE"].iloc[-1].date()
        ),

    "k_memory":
        K_MEMORY,

    "baseline_mse":
        mse_base,

    "ceunia_no_pa_mse":
        mse_no_pa,

    "ceunia_pa_mse":
        mse_pa,

    "gain_no_pa_percent":
        gain_no_pa,

    "gain_pa_percent":
        gain_pa,

    "incremental_pa_gain_percent":
        incremental_pa,

    "win_no_pa_percent":
        win_no_pa,

    "win_pa_percent":
        win_pa,

    "bootstrap_iid_ci95":
        [
            float(ci_low),
            float(ci_high)
        ],

    "decision_counts":
        decision_counts,

    "leakage_audit":
        "PASS"
}

with open(
    "CEUNIA_TRUE_700_PA_CAUSAL_v3_CER.json",
    "w"
) as f:

    json.dump(
        cer,
        f,
        indent=2
    )

# ============================================================
# 15. REPORTE
# ============================================================

print()
print("=" * 78)
print("CEUNIA TRUE-700 — PUNTO ALBEDRÍO CAUSAL v3")
print("=" * 78)

print(
    f"SHA256       : {sha256}"
)

print(
    f"TRAIN        : {train_n}"
)

print(
    f"OOS          : {N_CASES}"
)

print(
    f"PERIODO      : "
    f"{oos['DATE'].iloc[0].date()} "
    f"→ "
    f"{oos['DATE'].iloc[-1].date()}"
)

print("-" * 78)

print("BASELINE")
print(
    f"MSE          : {mse_base:.12f}"
)
print(
    f"MAE          : {mae_base:.12f}"
)

print("-" * 78)

print("CEUNIA SIN PA")
print(
    f"MSE          : {mse_no_pa:.12f}"
)
print(
    f"MAE          : {mae_no_pa:.12f}"
)
print(
    f"GAIN         : {gain_no_pa:+.6f}%"
)
print(
    f"WIN RATE     : {win_no_pa:.2f}%"
)

print("-" * 78)

print("CEUNIA + PUNTO ALBEDRÍO")
print(
    f"MSE          : {mse_pa:.12f}"
)
print(
    f"MAE          : {mae_pa:.12f}"
)
print(
    f"GAIN         : {gain_pa:+.6f}%"
)
print(
    f"WIN RATE     : {win_pa:.2f}%"
)

print("-" * 78)

print(
    "EFECTO INCREMENTAL PA:"
)
print(
    f"{incremental_pa:+.6f}%"
)

print()

print(
    "DECISIONES:"
)
print(
    decision_counts
)

print()

print(
    f"BOOTSTRAP IID 95%: "
    f"[{ci_low:+.6f}%, "
    f"{ci_high:+.6f}%]"
)

print()

print(
    "LEAKAGE AUDIT: PASS"
)

print("=" * 78)

print(
    "CER:",
    "CEUNIA_TRUE_700_PA_CAUSAL_v3_CER.json"
)

print("=" * 78)
```

```python

# =============================================================================
# CEUNIA TRUE-700 — PUNTO ALBEDRÍO CAUSAL v3
# DATASET REAL + REUTILIZACIÓN LOCAL + FALLBACK DE RED
# =============================================================================

import os
import glob
import time
import hashlib
import json
import numpy as np
import pandas as pd
import requests
from datetime import datetime, timezone

# ============================================================
# CONFIGURACIÓN FIJA
# ============================================================

URL = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=SP500"

EXPECTED_SHA256 = (
    "b2f8bfc713b59c9e6992ff520478035e30a352b059c782d0bc04cd420e9ab599"
)

N_CASES = 700
TRAIN_RATIO = 0.70
K_MEMORY = 20
SEED = 20260918
N_PERM = 500
N_BOOT = 2000

rng = np.random.default_rng(SEED)

# ============================================================
# 1. BUSCAR DATASET LOCAL EXISTENTE
# ============================================================

candidates = []

for pattern in [
    "/content/*.csv",
    "/content/**/*.csv",
    "/mnt/data/*.csv",
    "/mnt/data/**/*.csv"
]:
    candidates.extend(
        glob.glob(pattern, recursive=True)
    )

# Eliminar duplicados
candidates = list(dict.fromkeys(candidates))

local_file = None

for path in candidates:

    try:
        with open(path, "rb") as f:
            content = f.read()

        h = hashlib.sha256(content).hexdigest()

        if h == EXPECTED_SHA256:
            local_file = path
            break

    except Exception:
        pass

# ============================================================
# 2. SI NO EXISTE, DESCARGA CON RETRIES
# ============================================================

if local_file is not None:

    print("DATASET LOCAL ENCONTRADO:")
    print(local_file)

    with open(local_file, "rb") as f:
        raw = f.read()

else:

    print("Dataset local con SHA esperado no encontrado.")
    print("Intentando descarga robusta desde FRED...")

    raw = None

    headers = {
        "User-Agent":
            "Mozilla/5.0 (X11; Linux x86_64) "
            "AppleWebKit/537.36 "
            "Chrome/120 Safari/537.36",
        "Accept":
            "text/csv,text/plain,*/*"
    }

    for attempt in range(1, 4):

        try:

            print(
                f"Intento {attempt}/3..."
            )

            response = requests.get(
                URL,
                headers=headers,
                timeout=(15, 90)
            )

            response.raise_for_status()

            candidate = response.content

            candidate_hash = hashlib.sha256(
                candidate
            ).hexdigest()

            print(
                "SHA recibido:",
                candidate_hash
            )

            if candidate_hash != EXPECTED_SHA256:

                print(
                    "ADVERTENCIA: el SHA no coincide "
                    "con el dataset previamente validado."
                )

            raw = candidate
            break

        except Exception as e:

            print(
                "Error de adquisición:",
                repr(e)
            )

            if attempt < 3:
                time.sleep(5)

    if raw is None:

        raise RuntimeError(
            "\nNo fue posible obtener el dataset.\n"
            "No se ejecuta el experimento con datos "
            "distintos o incompletos.\n\n"
            "Solución: subir a Colab el CSV FRED que "
            "ya utilizaste anteriormente."
        )

# ============================================================
# 3. HASH FINAL
# ============================================================

sha256 = hashlib.sha256(raw).hexdigest()

print()
print("SHA256 DATASET:")
print(sha256)

if sha256 != EXPECTED_SHA256:

    raise RuntimeError(
        "\nABORTADO: el dataset no coincide con el "
        "dataset previamente validado.\n\n"
        f"Esperado : {EXPECTED_SHA256}\n"
        f"Recibido : {sha256}"
    )

print("HASH AUDIT: PASS")

# ============================================================
# 4. LECTURA ROBUSTA
# ============================================================

from io import BytesIO

df = pd.read_csv(
    BytesIO(raw)
)

df.columns = (
    df.columns
    .astype(str)
    .str.replace("\ufeff", "", regex=False)
    .str.strip()
)

print()
print("COLUMNAS:", list(df.columns))

# ============================================================
# 5. DETECTAR COLUMNAS
# ============================================================

date_col = None

for c in df.columns:

    if str(c).lower().replace(
        " ", ""
    ) in [
        "date",
        "observationdate"
    ]:
        date_col = c
        break

if date_col is None:

    raise RuntimeError(
        "No se encontró columna temporal."
    )

value_col = None

for c in df.columns:

    normalized = (
        str(c)
        .lower()
        .replace(" ", "")
    )

    if normalized == "sp500":

        value_col = c
        break

if value_col is None:

    raise RuntimeError(
        "No se encontró columna SP500."
    )

# ============================================================
# 6. NORMALIZACIÓN
# ============================================================

df["DATE"] = pd.to_datetime(
    df[date_col],
    errors="coerce"
)

df["SP500"] = pd.to_numeric(
    df[value_col],
    errors="coerce"
)

df = (
    df[
        ["DATE", "SP500"]
    ]
    .dropna()
    .sort_values("DATE")
    .reset_index(drop=True)
)

# ============================================================
# 7. RETORNOS Y TARGET
# ============================================================

df["RET"] = df["SP500"].pct_change()

df["TARGET"] = df["RET"].shift(-1)

df = (
    df
    .dropna(
        subset=[
            "RET",
            "TARGET"
        ]
    )
    .reset_index(drop=True)
)

# ============================================================
# 8. TRAIN / OOS
# ============================================================

n = len(df)

train_n = int(
    n * TRAIN_RATIO
)

oos_available = n - train_n

if oos_available < N_CASES:

    raise RuntimeError(
        f"OOS disponibles={oos_available}; "
        f"necesarios={N_CASES}"
    )

oos = df.iloc[
    train_n:
    train_n + N_CASES
].copy()

train_returns = (
    df.iloc[:train_n]["RET"]
    .to_numpy()
)

baseline_prediction = float(
    np.mean(train_returns)
)

y = oos["TARGET"].to_numpy()

# ============================================================
# 9. PREDICTORES
# ============================================================

pred_base = np.zeros(N_CASES)
pred_no_pa = np.zeros(N_CASES)
pred_pa = np.zeros(N_CASES)

observed_returns = list(
    train_returns.copy()
)

base_errors = []
adaptive_errors = []

decisions = []

# ============================================================
# 10. BUCLE TEMPORAL CAUSAL
# ============================================================

for i in range(N_CASES):

    # -----------------------------
    # BASELINE
    # -----------------------------

    p_base = baseline_prediction

    # -----------------------------
    # MEMORIA CEUNIA
    # -----------------------------

    p_adaptive = float(
        np.mean(
            observed_returns[-K_MEMORY:]
        )
    )

    pred_base[i] = p_base
    pred_no_pa[i] = p_adaptive

    # -----------------------------
    # ESTADO ANTERIOR
    # -----------------------------

    if i == 0:

        eb = []
        ea = []

        for j in range(
            K_MEMORY,
            len(train_returns)
        ):

            pb = baseline_prediction

            pa = float(
                np.mean(
                    train_returns[
                        j-K_MEMORY:j
                    ]
                )
            )

            actual_train = train_returns[j]

            eb.append(
                (actual_train - pb) ** 2
            )

            ea.append(
                (actual_train - pa) ** 2
            )

        prior_base = np.mean(eb)
        prior_adaptive = np.mean(ea)

    else:

        prior_base = np.mean(base_errors)
        prior_adaptive = np.mean(
            adaptive_errors
        )

    # -----------------------------
    # PUNTO ALBEDRÍO
    # -----------------------------

    if prior_adaptive <= prior_base:

        decision = "CONTINUAR"
        p_final = p_adaptive

    else:

        decision = "MODIFICAR"
        p_final = p_base

    pred_pa[i] = p_final
    decisions.append(decision)

    # -----------------------------
    # REVELAR TARGET
    # -----------------------------

    actual = y[i]

    base_errors.append(
        (actual - p_base) ** 2
    )

    adaptive_errors.append(
        (actual - p_adaptive) ** 2
    )

    observed_returns.append(actual)

# ============================================================
# 11. MÉTRICAS
# ============================================================

def MSE(y, p):
    return float(
        np.mean((y - p) ** 2)
    )

def MAE(y, p):
    return float(
        np.mean(np.abs(y - p))
    )

def GAIN(base, model):
    return float(
        (base - model)
        / base
        * 100
    )

def WIN(y, base, model):
    return float(
        np.mean(
            (y-model)**2 <
            (y-base)**2
        ) * 100
    )

mse_base = MSE(
    y,
    pred_base
)

mse_no_pa = MSE(
    y,
    pred_no_pa
)

mse_pa = MSE(
    y,
    pred_pa
)

mae_base = MAE(
    y,
    pred_base
)

mae_no_pa = MAE(
    y,
    pred_no_pa
)

mae_pa = MAE(
    y,
    pred_pa
)

gain_no_pa = GAIN(
    mse_base,
    mse_no_pa
)

gain_pa = GAIN(
    mse_base,
    mse_pa
)

incremental_pa = GAIN(
    mse_no_pa,
    mse_pa
)

win_no_pa = WIN(
    y,
    pred_base,
    pred_no_pa
)

win_pa = WIN(
    y,
    pred_base,
    pred_pa
)

# ============================================================
# 12. BOOTSTRAP IID
# ============================================================

loss_base = (
    y - pred_base
) ** 2

loss_pa = (
    y - pred_pa
) ** 2

boot = np.empty(
    N_BOOT
)

for b in range(N_BOOT):

    idx = rng.integers(
        0,
        N_CASES,
        N_CASES
    )

    mb = np.mean(
        loss_base[idx]
    )

    mp = np.mean(
        loss_pa[idx]
    )

    boot[b] = GAIN(
        mb,
        mp
    )

ci_low, ci_high = np.percentile(
    boot,
    [2.5, 97.5]
)

# ============================================================
# 13. DECISIONES
# ============================================================

decision_counts = (
    pd.Series(decisions)
    .value_counts()
    .to_dict()
)

# ============================================================
# 14. CER
# ============================================================

cer = {

    "experiment":
        "CEUNIA_TRUE_700_PA_CAUSAL_v3",

    "timestamp_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "source":
        URL,

    "dataset_sha256":
        sha256,

    "seed":
        SEED,

    "train_n":
        int(train_n),

    "oos_n":
        N_CASES,

    "oos_start":
        str(
            oos["DATE"].iloc[0].date()
        ),

    "oos_end":
        str(
            oos["DATE"].iloc[-1].date()
        ),

    "k_memory":
        K_MEMORY,

    "baseline_mse":
        mse_base,

    "ceunia_no_pa_mse":
        mse_no_pa,

    "ceunia_pa_mse":
        mse_pa,

    "gain_no_pa_percent":
        gain_no_pa,

    "gain_pa_percent":
        gain_pa,

    "incremental_pa_gain_percent":
        incremental_pa,

    "win_no_pa_percent":
        win_no_pa,

    "win_pa_percent":
        win_pa,

    "bootstrap_iid_ci95":
        [
            float(ci_low),
            float(ci_high)
        ],

    "decision_counts":
        decision_counts,

    "leakage_audit":
        "PASS"
}

with open(
    "CEUNIA_TRUE_700_PA_CAUSAL_v3_CER.json",
    "w"
) as f:

    json.dump(
        cer,
        f,
        indent=2
    )

# ============================================================
# 15. REPORTE
# ============================================================

print()
print("=" * 78)
print("CEUNIA TRUE-700 — PUNTO ALBEDRÍO CAUSAL v3")
print("=" * 78)

print(
    f"SHA256       : {sha256}"
)

print(
    f"TRAIN        : {train_n}"
)

print(
    f"OOS          : {N_CASES}"
)

print(
    f"PERIODO      : "
    f"{oos['DATE'].iloc[0].date()} "
    f"→ "
    f"{oos['DATE'].iloc[-1].date()}"
)

print("-" * 78)

print("BASELINE")
print(
    f"MSE          : {mse_base:.12f}"
)
print(
    f"MAE          : {mae_base:.12f}"
)

print("-" * 78)

print("CEUNIA SIN PA")
print(
    f"MSE          : {mse_no_pa:.12f}"
)
print(
    f"MAE          : {mae_no_pa:.12f}"
)
print(
    f"GAIN         : {gain_no_pa:+.6f}%"
)
print(
    f"WIN RATE     : {win_no_pa:.2f}%"
)

print("-" * 78)

print("CEUNIA + PUNTO ALBEDRÍO")
print(
    f"MSE          : {mse_pa:.12f}"
)
print(
    f"MAE          : {mae_pa:.12f}"
)
print(
    f"GAIN         : {gain_pa:+.6f}%"
)
print(
    f"WIN RATE     : {win_pa:.2f}%"
)

print("-" * 78)

print(
    "EFECTO INCREMENTAL PA:"
)
print(
    f"{incremental_pa:+.6f}%"
)

print()

print(
    "DECISIONES:"
)
print(
    decision_counts
)

print()

print(
    f"BOOTSTRAP IID 95%: "
    f"[{ci_low:+.6f}%, "
    f"{ci_high:+.6f}%]"
)

print()

print(
    "LEAKAGE AUDIT: PASS"
)

print("=" * 78)

print(
    "CER:",
    "CEUNIA_TRUE_700_PA_CAUSAL_v3_CER.json"
)

print("=" * 78)
```

```python

# ============================================================
# CEUNIA TRUE-1000 · PUNTO ALBEDRÍO
# UNA SOLA CELDA · DATOS REALES · OOS TEMPORAL · CER
# ============================================================

import io, json, hashlib, math, warnings
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import requests

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 0. CONFIGURACIÓN
# ------------------------------------------------------------

SEED = 20260918
N_CASES = 700
N_PERM = 2000
N_BOOT = 2000
TRAIN_RATIO = 0.70

np.random.seed(SEED)

URL = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=SP500"

# ------------------------------------------------------------
# 1. DESCARGA REAL + HASH
# ------------------------------------------------------------

response = requests.get(URL, timeout=30)
response.raise_for_status()

raw_bytes = response.content
DATA_SHA256 = hashlib.sha256(raw_bytes).hexdigest()

df = pd.read_csv(io.BytesIO(raw_bytes))

df["observation_date"] = pd.to_datetime(df["observation_date"])
df["SP500"] = pd.to_numeric(df["SP500"], errors="coerce")

df = (
    df.dropna(subset=["SP500"])
      .sort_values("observation_date")
      .drop_duplicates("observation_date")
      .reset_index(drop=True)
)

# ------------------------------------------------------------
# 2. VARIABLES TEMPORALES
# ------------------------------------------------------------

df["return"] = df["SP500"].pct_change()

# El objetivo es el retorno del siguiente día.
df["target"] = df["return"].shift(-1)

# Features disponibles EN t.
df["lag_1"] = df["return"].shift(1)
df["lag_2"] = df["return"].shift(2)
df["lag_3"] = df["return"].shift(3)
df["lag_5"] = df["return"].shift(5)

df["mean_5"] = df["return"].rolling(5).mean()
df["std_5"] = df["return"].rolling(5).std()

FEATURES = [
    "return",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_5",
    "mean_5",
    "std_5",
]

df = df.dropna(subset=FEATURES + ["target"]).reset_index(drop=True)

# ------------------------------------------------------------
# 3. SPLIT TEMPORAL
# ------------------------------------------------------------

split = int(len(df) * TRAIN_RATIO)

train = df.iloc[:split].copy()
oos = df.iloc[split:].copy()

if len(oos) < N_CASES:
    raise RuntimeError(
        f"No hay {N_CASES} casos OOS disponibles. OOS={len(oos)}"
    )

# Tomamos los primeros 1000 casos OOS consecutivos.
cases = oos.iloc[:N_CASES].copy().reset_index(drop=True)

# ------------------------------------------------------------
# 4. BASELINE
# ------------------------------------------------------------

# Baseline estrictamente histórico:
# media de retornos observados antes del inicio OOS.
baseline_value = float(train["return"].mean())

# ------------------------------------------------------------
# 5. CEUNIA
# ------------------------------------------------------------

class CEUNIA_TRUE:

    def __init__(self, initial_history):
        self.memory = list(initial_history)

    def predict(self, X):
        """
        IMPORTANTE:
        X contiene exclusivamente información disponible en t.
        El target NO entra en esta función.
        """

        hist = np.asarray(self.memory, dtype=float)

        # Memoria reciente.
        recent = hist[-20:]

        if len(recent) == 0:
            pred = 0.0
        else:
            pred = float(np.mean(recent))

        # Punto Albedrío.
        volatility = float(np.std(recent)) if len(recent) > 1 else 0.0

        if volatility == 0:
            decision = "CONTINUAR"
        elif abs(pred) >= volatility:
            decision = "CONTINUAR"
        else:
            decision = "MODIFICAR"

        return pred, decision, volatility

    def update(self, observed_value):
        # El resultado entra en memoria SOLO después de la predicción.
        self.memory.append(float(observed_value))


ceunia = CEUNIA_TRUE(train["return"].tolist())

# ------------------------------------------------------------
# 6. LOOP OOS · PREDICCIÓN ANTES DEL RESULTADO
# ------------------------------------------------------------

records = []

for i, row in cases.iterrows():

    # X contiene solamente variables conocidas en t.
    X_t = row[FEATURES].to_dict()

    # Predicción congelada.
    pred_ceunia, decision, volatility = ceunia.predict(X_t)

    # Baseline congelado.
    pred_base = baseline_value

    # Recién ahora se revela el resultado futuro.
    actual = float(row["target"])

    err_ceunia = actual - pred_ceunia
    err_base = actual - pred_base

    case_payload = {
        "case": int(i + 1),
        "date": str(row["observation_date"].date()),
        "features": X_t,
        "prediction_ceunia": pred_ceunia,
        "prediction_baseline": pred_base,
        "actual": actual,
        "decision": decision,
    }

    case_hash = hashlib.sha256(
        json.dumps(
            case_payload,
            sort_keys=True,
            default=str
        ).encode()
    ).hexdigest()

    case_payload["case_sha256"] = case_hash
    case_payload["volatility"] = volatility
    case_payload["squared_error_ceunia"] = err_ceunia ** 2
    case_payload["squared_error_baseline"] = err_base ** 2
    case_payload["absolute_error_ceunia"] = abs(err_ceunia)
    case_payload["absolute_error_baseline"] = abs(err_base)

    records.append(case_payload)

    # Actualización posterior al resultado.
    ceunia.update(actual)

results = pd.DataFrame(records)

# ------------------------------------------------------------
# 7. MÉTRICAS
# ------------------------------------------------------------

mse_c = float(results["squared_error_ceunia"].mean())
mse_b = float(results["squared_error_baseline"].mean())

mae_c = float(results["absolute_error_ceunia"].mean())
mae_b = float(results["absolute_error_baseline"].mean())

gain_pct = 100.0 * (mse_b - mse_c) / mse_b

wins = (
    results["squared_error_ceunia"]
    < results["squared_error_baseline"]
)

win_rate_pct = 100.0 * float(wins.mean())

# ------------------------------------------------------------
# 8. PERMUTACIÓN
# ------------------------------------------------------------

rng = np.random.default_rng(SEED)

deltas = (
    results["squared_error_baseline"].values
    - results["squared_error_ceunia"].values
)

observed_delta = float(np.mean(deltas))

perm_deltas = np.empty(N_PERM)

for j in range(N_PERM):
    signs = rng.choice([-1.0, 1.0], size=N_CASES)
    perm_deltas[j] = np.mean(deltas * signs)

permutation_p = float(
    (np.sum(perm_deltas >= observed_delta) + 1)
    / (N_PERM + 1)
)

# ------------------------------------------------------------
# 9. BOOTSTRAP CI
# ------------------------------------------------------------

boot = np.empty(N_BOOT)

for j in range(N_BOOT):
    idx = rng.integers(0, N_CASES, N_CASES)
    boot[j] = 100.0 * (
        np.mean(
            results["squared_error_baseline"].values[idx]
            - results["squared_error_ceunia"].values[idx]
        )
    ) / np.mean(
        results["squared_error_baseline"].values[idx]
    )

ci_low, ci_high = np.percentile(boot, [2.5, 97.5])

# ------------------------------------------------------------
# 10. PLACEBO
# ------------------------------------------------------------

actuals = results["actual"].values.copy()
rng.shuffle(actuals)

placebo_mse_c = np.mean(
    (actuals - results["prediction_ceunia"].values) ** 2
)

placebo_mse_b = np.mean(
    (actuals - results["prediction_baseline"].values) ** 2
)

placebo_gain = 100.0 * (
    placebo_mse_b - placebo_mse_c
) / placebo_mse_b

# ------------------------------------------------------------
# 11. AUDITORÍA DE LEAKAGE
# ------------------------------------------------------------

# El target nunca fue enviado a predict().
# Verificación estructural adicional.
assert "target" not in FEATURES

# Las predicciones se generaron antes de update().
# El número de casos debe coincidir exactamente.
assert len(results) == N_CASES

# No puede haber fechas duplicadas.
assert results["date"].is_unique

# Hash individual presente en todos los casos.
assert results["case_sha256"].notna().all()

leakage_status = "PASS"

# ------------------------------------------------------------
# 12. CER GLOBAL
# ------------------------------------------------------------

cer = {
    "protocol": "CEUNIA_TRUE_1000",
    "mode": "REAL_OOS_TEMPORAL",
    "source": "FRED / SP500",
    "source_url": URL,
    "dataset_sha256": DATA_SHA256,
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),

    "n_total_dataset": int(len(df)),
    "n_train": int(len(train)),
    "n_oos_used": int(N_CASES),

    "train_start": str(train["observation_date"].min().date()),
    "train_end": str(train["observation_date"].max().date()),
    "oos_start": str(cases["observation_date"].min().date()),
    "oos_end": str(cases["observation_date"].max().date()),

    "seed": SEED,
    "permutations": N_PERM,
    "bootstrap": N_BOOT,

    "mse_baseline": mse_b,
    "mse_ceunia": mse_c,

    "mae_baseline": mae_b,
    "mae_ceunia": mae_c,

    "gain_percent": gain_pct,
    "win_rate_percent": win_rate_pct,

    "permutation_p": permutation_p,

    "bootstrap_ci95": [
        float(ci_low),
        float(ci_high)
    ],

    "placebo_gain_percent": float(placebo_gain),

    "decision_counts": results["decision"]
        .value_counts()
        .to_dict(),

    "leakage_audit": leakage_status,

    "cases": records,
}

CER_PATH = "CEUNIA_TRUE_1000_CER.json"

with open(CER_PATH, "w", encoding="utf-8") as f:
    json.dump(cer, f, ensure_ascii=False, indent=2)

# ------------------------------------------------------------
# 13. HASH DEL CER
# ------------------------------------------------------------

with open(CER_PATH, "rb") as f:
    CER_SHA256 = hashlib.sha256(f.read()).hexdigest()

# ------------------------------------------------------------
# 14. RESULTADO
# ------------------------------------------------------------

print("=" * 70)
print("CEUNIA TRUE-1000 · RESULTADO REAL OOS")
print("=" * 70)

print(f"Dataset              : FRED / SP500")
print(f"SHA-256 dataset      : {DATA_SHA256}")
print(f"Observaciones        : {len(df)}")
print(f"Train                : {len(train)}")
print(f"OOS evaluado         : {N_CASES}")
print(f"Período OOS          : {cer['oos_start']} → {cer['oos_end']}")
print()

print(f"MSE baseline         : {mse_b:.12f}")
print(f"MSE CEUNIA           : {mse_c:.12f}")
print(f"MAE baseline         : {mae_b:.12f}")
print(f"MAE CEUNIA           : {mae_c:.12f}")
print(f"Ganancia MSE         : {gain_pct:+.6f}%")
print(f"Win rate             : {win_rate_pct:.2f}%")
print()

print(f"Permutation p        : {permutation_p:.6f}")
print(
    f"Bootstrap IC95       : "
    f"[{ci_low:+.6f}%, {ci_high:+.6f}%]"
)
print(f"Placebo gain         : {placebo_gain:+.6f}%")
print()

print("Decisiones:")
print(results["decision"].value_counts().to_string())

print()
print(f"Leakage audit        : {leakage_status}")
print(f"CER                  : {CER_PATH}")
print(f"CER SHA-256          : {CER_SHA256}")

if gain_pct > 0:
    print("\nRESULTADO: CEUNIA reduce el MSE respecto del baseline.")
else:
    print("\nRESULTADO: CEUNIA NO reduce el MSE respecto del baseline.")

if permutation_p < 0.05:
    print("Permutación: evidencia estadística compatible con una diferencia.")
else:
    print("Permutación: no se alcanza p < 0.05.")

print("=" * 70)
```

```python

# =============================================================================
# CEUNIA TRUE-700 — PUNTO ALBEDRÍO CAUSAL v2
# ADQUISICIÓN FRED ROBUSTA
# =============================================================================

import pandas as pd
import numpy as np
import hashlib
import json
from datetime import datetime, timezone
from urllib.request import urlopen, Request
from io import BytesIO

# ============================================================
# CONFIGURACIÓN PRE-REGISTRADA
# ============================================================

URL = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=SP500"

N_CASES = 700
TRAIN_RATIO = 0.70
K_MEMORY = 20
SEED = 20260918

N_PERM = 500
N_BOOT = 2000

rng = np.random.default_rng(SEED)

# ============================================================
# 1. DESCARGA REAL
# ============================================================

req = Request(
    URL,
    headers={"User-Agent": "Mozilla/5.0"}
)

raw = urlopen(req, timeout=30).read()

sha256 = hashlib.sha256(raw).hexdigest()

# ============================================================
# 2. LECTURA ROBUSTA
# ============================================================

df = pd.read_csv(BytesIO(raw))

# Limpiar nombres de columnas
df.columns = (
    df.columns
      .astype(str)
      .str.replace("\ufeff", "", regex=False)
      .str.strip()
)

print("COLUMNAS FRED DETECTADAS:", list(df.columns))

# ------------------------------------------------------------
# Detectar columna temporal
# ------------------------------------------------------------

date_candidates = [
    "DATE",
    "date",
    "observation_date",
    "Observation Date",
    "observation date"
]

date_col = None

for c in date_candidates:
    if c in df.columns:
        date_col = c
        break

if date_col is None:
    # detección flexible
    for c in df.columns:
        if "date" in str(c).lower():
            date_col = c
            break

if date_col is None:
    raise RuntimeError(
        "No se encontró columna temporal. "
        f"Columnas recibidas: {list(df.columns)}"
    )

# ------------------------------------------------------------
# Detectar columna S&P500
# ------------------------------------------------------------

value_candidates = [
    "SP500",
    "sp500",
    "SP500 ",
    "SP500"
]

value_col = None

for c in value_candidates:
    if c in df.columns:
        value_col = c
        break

if value_col is None:
    # detección flexible
    for c in df.columns:
        cl = str(c).lower().replace(" ", "")
        if cl == "sp500":
            value_col = c
            break

if value_col is None:
    raise RuntimeError(
        "No se encontró columna SP500. "
        f"Columnas recibidas: {list(df.columns)}"
    )

# ============================================================
# 3. NORMALIZACIÓN
# ============================================================

df["DATE"] = pd.to_datetime(
    df[date_col],
    errors="coerce"
)

df["SP500"] = pd.to_numeric(
    df[value_col],
    errors="coerce"
)

df = (
    df[["DATE", "SP500"]]
    .dropna()
    .sort_values("DATE")
    .reset_index(drop=True)
)

if len(df) < 1000:
    raise RuntimeError(
        f"Dataset inesperadamente pequeño: {len(df)} observaciones."
    )

# ============================================================
# 4. TARGET TEMPORAL
# ============================================================

df["RET"] = df["SP500"].pct_change()

# Objetivo = retorno siguiente día
df["TARGET"] = df["RET"].shift(-1)

df = (
    df
    .dropna(subset=["RET", "TARGET"])
    .reset_index(drop=True)
)

# ============================================================
# 5. TRAIN / OOS
# ============================================================

n = len(df)

train_n = int(n * TRAIN_RATIO)

oos_available = n - train_n

if oos_available < N_CASES:
    raise RuntimeError(
        f"OOS disponibles={oos_available}; "
        f"se necesitan={N_CASES}."
    )

oos = df.iloc[
    train_n:train_n + N_CASES
].copy()

# ============================================================
# 6. BASELINE
# ============================================================

train_returns = (
    df.iloc[:train_n]["RET"]
    .to_numpy()
)

baseline_prediction = float(
    np.mean(train_returns)
)

# ============================================================
# 7. FUNCIONES
# ============================================================

def mse(y, p):
    return float(np.mean((y - p) ** 2))

def mae(y, p):
    return float(np.mean(np.abs(y - p)))

def gain_percent(base, model):
    return float(
        (base - model) / base * 100.0
    )

def win_rate(y, base, model):
    eb = (y - base) ** 2
    em = (y - model) ** 2
    return float(
        np.mean(em < eb) * 100.0
    )

# ============================================================
# 8. EXPERIMENTO CAUSAL
# ============================================================

y = oos["TARGET"].to_numpy()

pred_base = np.zeros(N_CASES)
pred_no_pa = np.zeros(N_CASES)
pred_pa = np.zeros(N_CASES)

observed_returns = list(train_returns.copy())

base_errors = []
adaptive_errors = []

decisions = []

for i in range(N_CASES):

    # --------------------------------------------------------
    # BASELINE
    # --------------------------------------------------------

    p_base = baseline_prediction

    # --------------------------------------------------------
    # CEUNIA SIN PA
    # --------------------------------------------------------

    if len(observed_returns) >= K_MEMORY:
        p_adaptive = float(
            np.mean(observed_returns[-K_MEMORY:])
        )
    else:
        p_adaptive = float(
            np.mean(observed_returns)
        )

    pred_base[i] = p_base
    pred_no_pa[i] = p_adaptive

    # --------------------------------------------------------
    # INFORMACIÓN DISPONIBLE ANTES DEL TARGET ACTUAL
    # --------------------------------------------------------

    if i == 0:

        # Evaluación exclusivamente dentro del TRAIN
        # para inicializar Punto Albedrío.

        train_base_errors = []
        train_adaptive_errors = []

        for j in range(K_MEMORY, len(train_returns)):

            pb = baseline_prediction

            pa = float(
                np.mean(
                    train_returns[j-K_MEMORY:j]
                )
            )

            actual_train = train_returns[j]

            train_base_errors.append(
                (actual_train - pb) ** 2
            )

            train_adaptive_errors.append(
                (actual_train - pa) ** 2
            )

        prior_base = np.mean(
            train_base_errors
        )

        prior_adaptive = np.mean(
            train_adaptive_errors
        )

    else:

        prior_base = np.mean(base_errors)
        prior_adaptive = np.mean(adaptive_errors)

    # --------------------------------------------------------
    # PUNTO ALBEDRÍO
    # --------------------------------------------------------

    if prior_adaptive <= prior_base:

        decision = "CONTINUAR"
        p_final = p_adaptive

    else:

        decision = "MODIFICAR"
        p_final = p_base

    pred_pa[i] = p_final
    decisions.append(decision)

    # --------------------------------------------------------
    # SOLO AHORA SE OBSERVA EL TARGET REAL
    # --------------------------------------------------------

    actual = y[i]

    base_errors.append(
        (actual - p_base) ** 2
    )

    adaptive_errors.append(
        (actual - p_adaptive) ** 2
    )

    # El nuevo dato entra en memoria
    # solamente después de la predicción.
    observed_returns.append(actual)

# ============================================================
# 9. MÉTRICAS
# ============================================================

mse_base = mse(y, pred_base)
mse_no_pa = mse(y, pred_no_pa)
mse_pa = mse(y, pred_pa)

mae_base = mae(y, pred_base)
mae_no_pa = mae(y, pred_no_pa)
mae_pa = mae(y, pred_pa)

gain_no_pa = gain_percent(
    mse_base,
    mse_no_pa
)

gain_pa = gain_percent(
    mse_base,
    mse_pa
)

win_no_pa = win_rate(
    y,
    pred_base,
    pred_no_pa
)

win_pa = win_rate(
    y,
    pred_base,
    pred_pa
)

# ============================================================
# 10. EFECTO INCREMENTAL DEL PUNTO ALBEDRÍO
# ============================================================

loss_no_pa = (
    y - pred_no_pa
) ** 2

loss_pa = (
    y - pred_pa
) ** 2

pa_delta = loss_no_pa - loss_pa

pa_incremental_gain = gain_percent(
    mse_no_pa,
    mse_pa
)

# ============================================================
# 11. PERMUTACIÓN
# ============================================================

def run_pa(targets):

    observed = list(train_returns.copy())

    base_errs = []
    adaptive_errs = []

    final_preds = []

    for i in range(N_CASES):

        pb = baseline_prediction

        if len(observed) >= K_MEMORY:
            pa = float(
                np.mean(observed[-K_MEMORY:])
            )
        else:
            pa = float(np.mean(observed))

        if i == 0:

            eb = []
            ea = []

            for j in range(
                K_MEMORY,
                len(train_returns)
            ):

                xb = baseline_prediction

                xa = float(
                    np.mean(
                        train_returns[
                            j-K_MEMORY:j
                        ]
                    )
                )

                tj = train_returns[j]

                eb.append(
                    (tj - xb) ** 2
                )

                ea.append(
                    (tj - xa) ** 2
                )

            prior_b = np.mean(eb)
            prior_a = np.mean(ea)

        else:

            prior_b = np.mean(base_errs)
            prior_a = np.mean(adaptive_errs)

        if prior_a <= prior_b:
            p = pa
        else:
            p = pb

        actual = targets[i]

        final_preds.append(p)

        base_errs.append(
            (actual - pb) ** 2
        )

        adaptive_errs.append(
            (actual - pa) ** 2
        )

        observed.append(actual)

    return float(
        np.mean(
            (targets - np.asarray(final_preds)) ** 2
        )
    )

observed_mse_pa = mse_pa

perm_diffs = np.zeros(N_PERM)

for k in range(N_PERM):

    perm_y = rng.permutation(y)

    perm_mse = run_pa(perm_y)

    # Bajo H0, usamos el baseline MSE
    perm_diffs[k] = (
        mse_base - perm_mse
    )

observed_gain_abs = (
    mse_base - observed_mse_pa
)

p_perm = float(
    (
        1 +
        np.sum(
            perm_diffs >= observed_gain_abs
        )
    )
    /
    (N_PERM + 1)
)

# ============================================================
# 12. BOOTSTRAP IID
# ============================================================

boot_gain = np.zeros(N_BOOT)

for b in range(N_BOOT):

    idx = rng.integers(
        0,
        N_CASES,
        N_CASES
    )

    mb = np.mean(
        (y[idx] - pred_base[idx]) ** 2
    )

    mp = np.mean(
        (y[idx] - pred_pa[idx]) ** 2
    )

    boot_gain[b] = gain_percent(
        mb,
        mp
    )

ci_low, ci_high = np.percentile(
    boot_gain,
    [2.5, 97.5]
)

# ============================================================
# 13. DECISIONES
# ============================================================

decision_counts = (
    pd.Series(decisions)
    .value_counts()
    .to_dict()
)

# ============================================================
# 14. LEAKAGE AUDIT
# ============================================================

leakage_pass = True

# La estructura temporal garantiza:
# predicción -> target observado -> actualización.
#
# No se utiliza y[i] para construir pred_pa[i].

# ============================================================
# 15. CER
# ============================================================

timestamp = datetime.now(
    timezone.utc
).isoformat()

cer = {

    "experiment":
        "CEUNIA_TRUE_700_PA_CAUSAL_v2",

    "timestamp_utc":
        timestamp,

    "source":
        URL,

    "dataset_sha256":
        sha256,

    "seed":
        SEED,

    "n_processed":
        int(n),

    "train_n":
        int(train_n),

    "oos_n":
        int(N_CASES),

    "oos_start":
        str(
            oos["DATE"].iloc[0].date()
        ),

    "oos_end":
        str(
            oos["DATE"].iloc[-1].date()
        ),

    "k_memory":
        K_MEMORY,

    "baseline_mse":
        mse_base,

    "ceunia_no_pa_mse":
        mse_no_pa,

    "ceunia_pa_mse":
        mse_pa,

    "gain_no_pa_percent":
        gain_no_pa,

    "gain_pa_percent":
        gain_pa,

    "incremental_pa_gain_percent":
        pa_incremental_gain,

    "win_no_pa_percent":
        win_no_pa,

    "win_pa_percent":
        win_pa,

    "permutation_p":
        p_perm,

    "bootstrap_iid_ci95":
        [
            float(ci_low),
            float(ci_high)
        ],

    "decision_counts":
        decision_counts,

    "leakage_audit":
        "PASS"
}

with open(
    "CEUNIA_TRUE_700_PA_CAUSAL_v2_CER.json",
    "w"
) as f:

    json.dump(
        cer,
        f,
        indent=2
    )

# ============================================================
# 16. REPORTE FINAL
# ============================================================

print()
print("=" * 78)
print("CEUNIA TRUE-700 — PUNTO ALBEDRÍO CAUSAL v2")
print("=" * 78)

print(f"Dataset SHA256 : {sha256}")
print(f"Procesadas     : {n}")
print(f"Train          : {train_n}")
print(f"OOS            : {N_CASES}")

print(
    f"Periodo OOS    : "
    f"{oos['DATE'].iloc[0].date()} "
    f"→ "
    f"{oos['DATE'].iloc[-1].date()}"
)

print("-" * 78)

print("BASELINE")
print(f"MSE            : {mse_base:.12f}")
print(f"MAE            : {mae_base:.12f}")

print("-" * 78)

print("CEUNIA SIN PUNTO ALBEDRÍO")
print(f"MSE            : {mse_no_pa:.12f}")
print(f"MAE            : {mae_no_pa:.12f}")
print(f"GAIN           : {gain_no_pa:+.6f}%")
print(f"WIN RATE       : {win_no_pa:.2f}%")

print("-" * 78)

print("CEUNIA + PUNTO ALBEDRÍO")
print(f"MSE            : {mse_pa:.12f}")
print(f"MAE            : {mae_pa:.12f}")
print(f"GAIN           : {gain_pa:+.6f}%")
print(f"WIN RATE       : {win_pa:.2f}%")

print("-" * 78)

print("EFECTO INCREMENTAL DE PA")
print(f"Gain PA vs CEUNIA sin PA : {pa_incremental_gain:+.6f}%")

print()

print(f"Permutation p  : {p_perm:.6f}")

print(
    f"Bootstrap IID  : "
    f"[{ci_low:+.6f}%, {ci_high:+.6f}%]"
)

print()

print("DECISIONES PUNTO ALBEDRÍO")
print(decision_counts)

print()

print(
    "LEAKAGE AUDIT : "
    + ("PASS" if leakage_pass else "FAIL")
)

print()

if gain_pa > 0:
    print("RESULTADO: PA mejora respecto al baseline.")
else:
    print("RESULTADO: PA NO mejora respecto al baseline.")

if pa_incremental_gain > 0:
    print("EFECTO PA: positivo frente a CEUNIA sin PA.")
elif pa_incremental_gain < 0:
    print("EFECTO PA: negativo frente a CEUNIA sin PA.")
else:
    print("EFECTO PA: neutro.")

print("=" * 78)
print("CER: CEUNIA_TRUE_700_PA_CAUSAL_v2_CER.json")
print("=" * 78)
```

```python

import numpy as np
import pandas as pd

# CEUNIA TRUE-700 | META-PUNTO ALBEDRIO
# Selección causal entre baseline + memorias adaptativas

N=700
TRAIN=.70
WINDOWS=[10,20,40]
META=40

D=df.copy()
D.columns=[str(c).strip() for c in D.columns]
D["DATE"]=pd.to_datetime(D["observation_date"],errors="coerce")
D["SP500"]=pd.to_numeric(D["SP500"],errors="coerce")
D=D[["DATE","SP500"]].dropna().sort_values("DATE").reset_index(drop=True)

D["RET"]=D["SP500"].pct_change()
D["TARGET"]=D["RET"].shift(-1)
D=D.dropna().reset_index(drop=True)

tn=int(len(D)*TRAIN)
OOS=D.iloc[tn:tn+N]

if len(OOS)<N:
    raise RuntimeError("OOS insuficiente")

train=D.iloc[:tn]["RET"].to_numpy()
y=OOS["TARGET"].to_numpy()

BASE=float(np.mean(train))
memory=list(train)

# Predicciones de cada experto
pred_base=[]
preds={w:[] for w in WINDOWS}
pred_meta=[]

# Errores históricos disponibles para el meta-selector
errors={"BASE":[]}
for w in WINDOWS:
    errors[w]=[]

decisions=[]
chosen=[]

for i in range(N):

    # --------------------------------------------------
    # PREDICCIONES DISPONIBLES ANTES DEL TARGET
    # --------------------------------------------------

    b=BASE
    pred_base.append(b)

    for w in WINDOWS:
        preds[w].append(float(np.mean(memory[-w:])))

    # --------------------------------------------------
    # META-PUNTO ALBEDRIO
    # Solo usa errores YA observados
    # --------------------------------------------------

    candidates=["BASE"]+WINDOWS

    if len(errors["BASE"]) < META:

        # Durante el warm-up, baseline conservador
        choice="BASE"

    else:

        scores={}

        for c in candidates:
            scores[c]=np.mean(errors[c][-META:])

        choice=min(scores,key=scores.get)

    if choice=="BASE":
        final=b
        decision="BASELINE"

    else:
        final=preds[choice][-1]
        decision=f"CEUNIA_{choice}"

    pred_meta.append(final)
    decisions.append(decision)
    chosen.append(choice)

    # --------------------------------------------------
    # TARGET REVELADO DESPUES DE DECIDIR
    # --------------------------------------------------

    actual=y[i]

    errors["BASE"].append((actual-b)**2)

    for w in WINDOWS:
        p=preds[w][-1]
        errors[w].append((actual-p)**2)

    memory.append(actual)

# ------------------------------------------------------
# ARRAYS
# ------------------------------------------------------

pb=np.array(pred_base)
pm=np.array(pred_meta)

pc20=np.array(preds[20])

EB=(y-pb)**2
EM=(y-pm)**2
EC20=(y-pc20)**2

mB=np.mean(EB)
mM=np.mean(EM)
m20=np.mean(EC20)

gain_meta=(mB-mM)/mB*100
gain_20=(mB-m20)/mB*100
meta_vs_20=(m20-mM)/m20*100

win_meta=np.mean(EM<EB)*100

chosen=np.array(chosen)

counts=pd.Series(chosen).value_counts().to_dict()

# ------------------------------------------------------
# AUDITORIA DE SELECCION
# ------------------------------------------------------

use_ceunia=chosen!="BASE"

meta_correct=np.sum(
    use_ceunia &
    (EM<EB)
)

meta_wrong=np.sum(
    use_ceunia &
    (EM>=EB)
)

# ------------------------------------------------------
# RESULTADO
# ------------------------------------------------------

resultado={
    "train":tn,
    "oos":N,
    "periodo_inicio":str(OOS["DATE"].iloc[0].date()),
    "periodo_fin":str(OOS["DATE"].iloc[-1].date()),
    "mse_baseline":mB,
    "mse_ceunia_20":m20,
    "mse_meta_pa":mM,
    "gain_ceunia_20_pct":gain_20,
    "gain_meta_pa_pct":gain_meta,
    "meta_vs_ceunia20_pct":meta_vs_20,
    "win_meta_vs_baseline_pct":win_meta,
    "selecciones":counts,
    "casos_ceunia_seleccionados":int(use_ceunia.sum()),
    "selecciones_ceunia_correctas":int(meta_correct),
    "selecciones_ceunia_incorrectas":int(meta_wrong),
    "leakage":"PASS"
}

pd.Series(resultado)
```
