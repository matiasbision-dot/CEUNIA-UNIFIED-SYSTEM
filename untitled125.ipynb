```python

# ============================================================
# CEUNIA CORE v1.1.0 — BUILD / TEST / IMPORT ROBUSTO
# ============================================================

import sys
import subprocess
import shutil
import json
import hashlib
import zipfile
import importlib
from pathlib import Path
from datetime import datetime, timezone
import importlib.util

# ============================================================
# 1. DEPENDENCIAS
# ============================================================

subprocess.run(
    [
        sys.executable, "-m", "pip", "install", "-q",
        "fastapi>=0.115",
        "uvicorn>=0.30",
        "pydantic>=2.0",
        "pytest>=8.0"
    ],
    check=True
)

# ============================================================
# 2. PROYECTO
# ============================================================

PROJECT = Path("/content/CEUNIA_Core_v1_1")
PACKAGE = PROJECT / "ceunia"
TESTS = PROJECT / "tests"
ZIP_PATH = Path("/content/CEUNIA_Core_v1_1_MVP.zip")

if PROJECT.exists():
    shutil.rmtree(PROJECT)

if ZIP_PATH.exists():
    ZIP_PATH.unlink()

PACKAGE.mkdir(parents=True)
TESTS.mkdir(parents=True)

# ============================================================
# 3. CORE.PY
# ============================================================

core_code = r'''
from dataclasses import dataclass, field
from datetime import datetime, timezone
import hashlib
import json


@dataclass
class CEUNIA:

    version: str = "1.1.0"
    cycle: int = 0
    memory: list = field(default_factory=list)

    state: dict = field(default_factory=lambda: {
        "coherence": 0.5,
        "potercia": 0.5,
        "meta_potercia": 0.0,
        "learning": 0.0
    })

    def _score(self, text: str) -> float:

        if not text:
            return 0.0

        length_score = min(len(text) / 100.0, 1.0)
        word_score = min(len(text.split()) / 20.0, 1.0)

        return round(
            0.5 * length_score +
            0.5 * word_score,
            6
        )

    def _hash_record(self, record: dict) -> str:

        payload = json.dumps(
            record,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":")
        )

        return hashlib.sha256(
            payload.encode("utf-8")
        ).hexdigest()

    def run(self, input_text: str, candidates=None) -> dict:

        self.cycle += 1

        state_before = dict(self.state)

        score = self._score(input_text)

        alternatives = (
            list(candidates)
            if candidates
            else ["observe", "evaluate", "act"]
        )

        decision = "act" if score >= 0.30 else "observe"

        coherence = round(
            min(1.0, 0.5 + score * 0.5),
            6
        )

        potercia = round(
            min(1.0, 0.5 + score * 0.4),
            6
        )

        meta_potercia = round(
            max(0.0, potercia - coherence * 0.9),
            6
        )

        learning = round(
            min(
                1.0,
                state_before["learning"] + score * 0.1
            ),
            6
        )

        self.state = {
            "coherence": coherence,
            "potercia": potercia,
            "meta_potercia": meta_potercia,
            "learning": learning
        }

        record = {
            "system": "CEUNIA",
            "version": self.version,
            "cycle": self.cycle,
            "timestamp": datetime.now(timezone.utc).isoformat(),
            "state_before": state_before,
            "context": {
                "memory_size": len(self.memory)
            },
            "input": input_text,
            "decision": decision,
            "action": decision,
            "result": {
                "accepted": True,
                "score": score,
                "message": input_text
            },
            "evidence": {
                "input_length": len(input_text),
                "score": score,
                "cycle": self.cycle
            },
            "evaluation": {
                "coherence": coherence,
                "potercia": potercia,
                "meta_potercia": meta_potercia
            },
            "learning": {
                "learning_delta": round(
                    learning - state_before["learning"],
                    6
                )
            },
            "modification": {
                "state_updated": True
            },
            "state_after": dict(self.state),
            "potercia": potercia,
            "meta_potercia": meta_potercia,
            "alternatives": alternatives
        }

        record["hash"] = self._hash_record(record)

        self.memory.append(record)

        return record

    def status(self) -> dict:

        return {
            "system": "CEUNIA",
            "version": self.version,
            "cycle": self.cycle,
            "memory": len(self.memory),
            "state": dict(self.state)
        }
'''

(PACKAGE / "core.py").write_text(
    core_code,
    encoding="utf-8"
)

# ============================================================
# 4. INIT.PY
# ============================================================

init_code = r'''
from .core import CEUNIA

__all__ = ["CEUNIA"]
'''

(PACKAGE / "__init__.py").write_text(
    init_code,
    encoding="utf-8"
)

# ============================================================
# 5. API
# ============================================================

api_code = r'''
from fastapi import FastAPI
from pydantic import BaseModel
from ceunia import CEUNIA

app = FastAPI(
    title="CEUNIA CORE",
    version="1.1.0"
)

system = CEUNIA()


class RunRequest(BaseModel):

    input_text: str
    candidates: list[str] | None = None


@app.post("/run")
def run(request: RunRequest):

    return system.run(
        request.input_text,
        request.candidates
    )


@app.get("/status")
def status():

    return system.status()
'''

(PACKAGE / "api.py").write_text(
    api_code,
    encoding="utf-8"
)

# ============================================================
# 6. TESTS
# ============================================================

tests_code = r'''
import sys
from pathlib import Path

PROJECT = Path(__file__).resolve().parents[1]

sys.path.insert(0, str(PROJECT))

from ceunia import CEUNIA


def test_import():

    system = CEUNIA()

    assert system.version == "1.1.0"


def test_run():

    system = CEUNIA()

    result = system.run("CEUNIA evolution")

    assert result["system"] == "CEUNIA"
    assert result["version"] == "1.1.0"
    assert "hash" in result


def test_hash():

    system = CEUNIA()

    result = system.run("test")

    assert len(result["hash"]) == 64


def test_cycle():

    system = CEUNIA()

    system.run("one")
    system.run("two")

    assert system.cycle == 2
    assert len(system.memory) == 2


def test_status():

    system = CEUNIA()

    system.run("status")

    status = system.status()

    assert status["cycle"] == 1
    assert status["memory"] == 1
'''

(TESTS / "test_core.py").write_text(
    tests_code,
    encoding="utf-8"
)

# ============================================================
# 7. DOCUMENTACIÓN
# ============================================================

(PROJECT / "README.md").write_text(
"""# CEUNIA CORE v1.1.0

CEUNIA CORE is an evolutionary orchestration engine.

Cycle:

Observation -> Evaluation -> Decision -> Action ->
Evidence -> Learning -> State Update

Components:

- CEUNIA Core
- Memory
- Potercia
- MetaPotercia
- Evolutionary state
- CER-compatible execution record
- SHA-256 integrity hash
- FastAPI interface

Author:
Néstor Matías de la Peña
""",
    encoding="utf-8"
)

(PROJECT / "LICENSE").write_text(
"""MIT License

Copyright (c) 2026 Néstor Matías de la Peña

Permission is hereby granted, free of charge, to any person
obtaining a copy of this software and associated documentation
files, to deal in the Software without restriction.
""",
    encoding="utf-8"
)

(PROJECT / "requirements.txt").write_text(
"""fastapi>=0.115
uvicorn>=0.30
pydantic>=2.0
""",
    encoding="utf-8"
)

(PROJECT / ".gitignore").write_text(
"""__pycache__/
*.pyc
.pytest_cache/
.ipynb_checkpoints/
""",
    encoding="utf-8"
)

# ============================================================
# 8. IMPORT DIRECTO DESDE ARCHIVO
#    EVITA COMPLETAMENTE EL PROBLEMA DE COLAB
# ============================================================

spec = importlib.util.spec_from_file_location(
    "ceunia_core_direct",
    PACKAGE / "core.py"
)

ceunia_core_direct = importlib.util.module_from_spec(spec)

spec.loader.exec_module(ceunia_core_direct)

CEUNIA = ceunia_core_direct.CEUNIA

# ============================================================
# 9. VERIFICACIÓN
# ============================================================

assert callable(CEUNIA)

system = CEUNIA()

assert system.version == "1.1.0"

print("DIRECT IMPORT : PASS")

# ============================================================
# 10. EJECUCIÓN
# ============================================================

demo = system.run(
    "CEUNIA CORE v1.1.0 evolutionary validation"
)

assert demo["system"] == "CEUNIA"
assert demo["version"] == "1.1.0"
assert len(demo["hash"]) == 64

print("CORE EXECUTION : PASS")

# ============================================================
# 11. PYTEST
# ============================================================

result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        "-q",
        str(TESTS)
    ],
    cwd=str(PROJECT),
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:

    print(result.stderr)

    raise RuntimeError(
        "Los tests no pasaron."
    )

# ============================================================
# 12. VERIFICACIÓN DEL PACKAGE NORMAL
# ============================================================

# Limpiar módulos CEUNIA que pudieran existir
for name in list(sys.modules):

    if (
        name == "ceunia"
        or name.startswith("ceunia.")
    ):

        del sys.modules[name]

importlib.invalidate_caches()

# Proyecto primero
if str(PROJECT) in sys.path:
    sys.path.remove(str(PROJECT))

sys.path.insert(0, str(PROJECT))

try:

    import ceunia

    assert hasattr(
        ceunia,
        "CEUNIA"
    )

    assert ceunia.CEUNIA.version == "1.1.0" \
        or ceunia.CEUNIA().version == "1.1.0"

    PACKAGE_IMPORT = "PASS"

except Exception as exc:

    PACKAGE_IMPORT = f"FAIL: {repr(exc)}"

print("PACKAGE IMPORT :", PACKAGE_IMPORT)

if not PACKAGE_IMPORT == "PASS":

    raise RuntimeError(
        "El package CEUNIA no pudo importarse correctamente."
    )

# ============================================================
# 13. MANIFEST
# ============================================================

manifest = {

    "system": "CEUNIA CORE",

    "version": "1.1.0",

    "author": "Néstor Matías de la Peña",

    "generated_at":
        datetime.now(timezone.utc).isoformat(),

    "build": "PASS",

    "direct_import": "PASS",

    "package_import": "PASS",

    "tests": "5 passed",

    "files": sorted(
        str(p.relative_to(PROJECT))
        for p in PROJECT.rglob("*")
        if p.is_file()
    )
}

(PROJECT / "MANIFEST.json").write_text(
    json.dumps(
        manifest,
        ensure_ascii=False,
        indent=2
    ),
    encoding="utf-8"
)

# ============================================================
# 14. EXECUTION STATUS
# ============================================================

execution_status = {

    "build": "PASS",

    "direct_import": "PASS",

    "package_import": "PASS",

    "tests": "5 passed",

    "cycle": demo["cycle"],

    "memory": len(system.memory),

    "coherence":
        demo["evaluation"]["coherence"],

    "potercia":
        demo["potercia"],

    "meta_potercia":
        demo["meta_potercia"],

    "cer_hash":
        demo["hash"]
}

(PROJECT / "execution_status.json").write_text(
    json.dumps(
        execution_status,
        ensure_ascii=False,
        indent=2
    ),
    encoding="utf-8"
)

# ============================================================
# 15. SHA256
# ============================================================

sha256sums = {}

for path in PROJECT.rglob("*"):

    if path.is_file():

        sha256sums[
            str(path.relative_to(PROJECT))
        ] = hashlib.sha256(
            path.read_bytes()
        ).hexdigest()

(PROJECT / "SHA256SUMS.json").write_text(
    json.dumps(
        sha256sums,
        ensure_ascii=False,
        indent=2
    ),
    encoding="utf-8"
)

# ============================================================
# 16. ZIP
# ============================================================

with zipfile.ZipFile(
    ZIP_PATH,
    "w",
    zipfile.ZIP_DEFLATED
) as z:

    for path in PROJECT.rglob("*"):

        if path.is_file():

            z.write(
                path,
                arcname=path.relative_to(
                    PROJECT.parent
                )
            )

# ============================================================
# 17. RESULTADO
# ============================================================

print()
print("=" * 60)
print("CEUNIA CORE v1.1.0 — FINAL")
print("=" * 60)
print("BUILD          : PASS")
print("DIRECT IMPORT  : PASS")
print("PACKAGE IMPORT : PASS")
print("TESTS          : 5 passed")
print("VERSION        :", system.version)
print("CYCLE          :", system.cycle)
print("MEMORY         :", len(system.memory))
print("COHERENCE      :", demo["evaluation"]["coherence"])
print("POTERCIA       :", demo["potercia"])
print("META-POTERCIA  :", demo["meta_potercia"])
print("CER SHA256     :", demo["hash"])
print("PROJECT        :", PROJECT)
print("ZIP            :", ZIP_PATH)
print("=" * 60)

# ============================================================
# 18. DESCARGA COLAB
# ============================================================

from google.colab import files

files.download(
    str(ZIP_PATH)
)
```
