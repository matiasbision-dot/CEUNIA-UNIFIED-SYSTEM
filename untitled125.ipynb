```python

# ================================================================
# CEUNIA CORE v1.1 — RECONSTRUCCIÓN FUNCIONAL PORTABLE
# ================================================================

import os
import shutil
import hashlib
import zipfile
import json
import subprocess
import sys
from datetime import datetime, timezone
from google.colab import files

ROOT = "/content"
PKG = os.path.join(ROOT, "CEUNIA_Core_v1_1_RECONSTRUCTED")
ZIP = os.path.join(ROOT, "CEUNIA_Core_v1_1_RECONSTRUCTED.zip")

CER_HISTORICAL = (
    "b82a41c23163f985116733ae04e5cf8d238c277e092e9c14610be10c0cf27432"
)

print("=" * 78)
print("CEUNIA CORE v1.1 — RECONSTRUCCIÓN FUNCIONAL")
print("=" * 78)

# ------------------------------------------------
# Limpieza
# ------------------------------------------------

if os.path.exists(PKG):
    shutil.rmtree(PKG)

os.makedirs(PKG)

# ------------------------------------------------
# 1. Código fuente
# ------------------------------------------------

core_code = r'''
import hashlib
import json
from datetime import datetime, timezone


class CeuniaCore:

    def __init__(self):
        self.version = "1.1.0"
        self.cycle = 0

        self.state = {
            "coherence": 0.715,
            "potercia": 0.6808,
            "meta_potercia": 0.0928
        }

        self.memory = []

    def _timestamp(self):
        return datetime.now(timezone.utc).isoformat()

    def _hash(self, data):
        payload = json.dumps(
            data,
            sort_keys=True,
            ensure_ascii=False
        ).encode("utf-8")

        return hashlib.sha256(payload).hexdigest()

    def evaluate(self, text):
        score = 0.85 if text else 0.0

        decision = "integrar" if score >= 0.5 else "observar"

        return {
            "score": score,
            "decision": decision
        }

    def update(self, text):

        before = dict(self.state)

        evaluation = self.evaluate(text)

        score = evaluation["score"]

        # Actualización funcional del estado.
        # Esta implementación reconstruye el comportamiento
        # observable del MVP, no afirma ser el código original.

        self.state["coherence"] = round(
            min(
                1.0,
                self.state["coherence"] +
                0.0675 * score
            ),
            6
        )

        self.state["potercia"] = round(
            min(
                1.0,
                self.state["potercia"] +
                0.0796 * score
            ),
            6
        )

        self.state["meta_potercia"] = round(
            max(
                0.0,
                self.state["meta_potercia"] -
                0.02955 * score
            ),
            6
        )

        self.cycle += 1

        record = {
            "system": "CEUNIA",
            "version": self.version,
            "cycle": self.cycle,
            "timestamp": self._timestamp(),
            "state_before": before,
            "context": {
                "memory_size": len(self.memory)
            },
            "input": text,
            "decision": evaluation["decision"],
            "action": "actualizar_estado",
            "result": evaluation,
            "evidence": {
                "input_length": len(text)
            },
            "evaluation": {
                "coherence": self.state["coherence"]
            },
            "learning": {
                "score": score
            },
            "modification": {
                "state_updated": True
            },
            "state_after": dict(self.state),
            "potercia": self.state["potercia"],
            "meta_potercia": self.state["meta_potercia"],
            "alternatives": []
        }

        record["hash"] = self._hash(record)

        self.memory.append(record)

        return record

    def status(self):

        return {
            "system": "CEUNIA",
            "version": self.version,
            "cycle": self.cycle,
            "state": dict(self.state),
            "memory_size": len(self.memory)
        }

    def last_record(self):

        if not self.memory:
            return None

        return self.memory[-1]
'''

with open(
    os.path.join(PKG, "ceunia_core.py"),
    "w",
    encoding="utf-8"
) as f:
    f.write(core_code)

print("[1/7] ceunia_core.py creado")

# ------------------------------------------------
# 2. Tests
# ------------------------------------------------

test_code = r'''
from ceunia_core import CeuniaCore


def test_initial_state():

    core = CeuniaCore()

    assert core.version == "1.1.0"
    assert core.cycle == 0
    assert core.state["coherence"] == 0.715
    assert core.state["potercia"] == 0.6808
    assert core.state["meta_potercia"] == 0.0928


def test_learning_cycle():

    core = CeuniaCore()

    record = core.update(
        "CEUNIA aprende del resultado y modifica su estado."
    )

    assert core.cycle == 1
    assert len(core.memory) == 1
    assert record["decision"] == "integrar"
    assert record["modification"]["state_updated"] is True
    assert len(record["hash"]) == 64


def test_three_cycles():

    core = CeuniaCore()

    inputs = [
        "Observar el estado del sistema.",
        "Evaluar el resultado.",
        "CEUNIA aprende del resultado y modifica su estado."
    ]

    for x in inputs:
        core.update(x)

    status = core.status()

    assert status["cycle"] == 3
    assert status["memory_size"] == 3
    assert 0.0 <= status["state"]["coherence"] <= 1.0
    assert 0.0 <= status["state"]["potercia"] <= 1.0
    assert 0.0 <= status["state"]["meta_potercia"] <= 1.0
'''

with open(
    os.path.join(PKG, "test_ceunia_core.py"),
    "w",
    encoding="utf-8"
) as f:
    f.write(test_code)

print("[2/7] test_ceunia_core.py creado")

# ------------------------------------------------
# 3. README
# ------------------------------------------------

readme = f"""# CEUNIA Core v1.1
## Reconstrucción funcional portable

Autor:
Néstor Matías de la Peña

Versión:
1.1.0

Estado:
RECONSTRUCCIÓN FUNCIONAL

Fecha:
{datetime.now(timezone.utc).isoformat()}

## Importante

Este paquete NO afirma recuperar el código fuente original de
CEUNIA Core v1.1.

Es una reconstrucción funcional basada en el comportamiento observable,
los estados registrados y la arquitectura documentada previamente.

## Registro histórico

CER SHA-256:

`{CER_HISTORICAL}`

Estado histórico observado:

- ciclos: 3
- memoria: 3
- coherencia: 0.7825
- Potercia: 0.74848
- MetaPotercia: 0.06768

## Componentes

- ceunia_core.py
- test_ceunia_core.py
- STATE.json
- MANIFEST.sha256

## Principios

Observación
Comprensión
Memoria
Evaluación
Decisión
Acción
Aprendizaje
Evolución

## Integridad

Cada artefacto del paquete posee SHA-256 en MANIFEST.sha256.

El SHA del paquete ZIP se registra externamente al finalizar
la construcción.

## Limitación

La equivalencia exacta con el código histórico original no está
demostrada. Esta versión debe considerarse una implementación
reconstruida y auditable.
"""

with open(
    os.path.join(PKG, "README.md"),
    "w",
    encoding="utf-8"
) as f:
    f.write(readme)

print("[3/7] README.md creado")

# ------------------------------------------------
# 4. Estado histórico
# ------------------------------------------------

state = {
    "system": "CEUNIA",
    "component": "Core",
    "version": "1.1.0",
    "reconstruction": True,
    "historical_cer_sha256": CER_HISTORICAL,
    "historical_state": {
        "cycles": 3,
        "memory": 3,
        "coherence": 0.7825,
        "potercia": 0.74848,
        "meta_potercia": 0.06768
    },
    "created_at": datetime.now(
        timezone.utc
    ).isoformat()
}

with open(
    os.path.join(PKG, "STATE.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        state,
        f,
        indent=2,
        ensure_ascii=False
    )

print("[4/7] STATE.json creado")

# ------------------------------------------------
# 5. Ejecutar tests
# ------------------------------------------------

print("\nEJECUTANDO TESTS...")
print("-" * 60)

result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        "-q"
    ],
    cwd=PKG,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError(
        "Los tests fallaron. El paquete NO será presentado "
        "como reconstrucción validada."
    )

print("[5/7] TESTS: PASS")

# ------------------------------------------------
# 6. MANIFEST SHA-256
# ------------------------------------------------

manifest = []

for filename in sorted(os.listdir(PKG)):

    path = os.path.join(PKG, filename)

    if not os.path.isfile(path):
        continue

    sha = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            sha.update(chunk)

    manifest.append(
        "{}  {}".format(
            sha.hexdigest(),
            filename
        )
    )

manifest_path = os.path.join(
    PKG,
    "MANIFEST.sha256"
)

with open(
    manifest_path,
    "w",
    encoding="utf-8"
) as f:
    f.write(
        "\n".join(manifest) + "\n"
    )

print("[6/7] MANIFEST.sha256 creado")

# ------------------------------------------------
# 7. ZIP final
# ------------------------------------------------

if os.path.exists(ZIP):
    os.remove(ZIP)

with zipfile.ZipFile(
    ZIP,
    "w",
    zipfile.ZIP_DEFLATED
) as z:

    for filename in os.listdir(PKG):

        path = os.path.join(
            PKG,
            filename
        )

        if os.path.isfile(path):
            z.write(
                path,
                arcname=filename
            )

# SHA ZIP
sha_zip = hashlib.sha256()

with open(ZIP, "rb") as f:
    for chunk in iter(
        lambda: f.read(1024 * 1024),
        b""
    ):
        sha_zip.update(chunk)

ZIP_SHA = sha_zip.hexdigest()

SHA_FILE = os.path.join(
    ROOT,
    "CEUNIA_Core_v1_1_RECONSTRUCTED_SHA256.txt"
)

with open(
    SHA_FILE,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "CEUNIA_Core_v1_1_RECONSTRUCTED.zip\n"
        "SHA-256: " + ZIP_SHA + "\n"
    )

print("[7/7] ZIP creado")
print("=" * 78)
print("CEUNIA CORE v1.1 — RECONSTRUCCIÓN COMPLETADA")
print("=" * 78)
print("ZIP:", ZIP)
print("Tamaño:", round(os.path.getsize(ZIP) / 1024, 2), "KB")
print("SHA-256:", ZIP_SHA)
print()
print("CER HISTÓRICO:")
print(CER_HISTORICAL)
print()
print("ESTADO:")
print("RECONSTRUCCIÓN FUNCIONAL AUDITABLE")
print("TESTS: PASS")
print()
print("DESCARGANDO...")

files.download(ZIP)

print("DESCARGA SOLICITADA.")
```
