```python

# ================================================================
# CEUNIA Ω-META v2.0 — CEUNIA SOBRE CEUNIA
# TRUE META-EVOLUTION / SINGLE CELL / OOS / CONTROL / META-Ω
# ================================================================

import numpy as np, pandas as pd, hashlib, json, time
from copy import deepcopy

# -------------------- CONFIGURACIÓN --------------------

SEEDS = [11,22,33,44,55,66,77,88]
CYCLES = 5

# -------------------- DATOS --------------------

def make_data(n, seed):
    r = np.random.default_rng(seed)
    X = r.normal(0,1,(n,6))
    y = (
        .45*X[:,0] -.30*X[:,1] +
        .20*X[:,2]*X[:,3] +
        .15*np.sin(X[:,4]) +
        .10*(X[:,5]**2) +
        r.normal(0,.12,n)
    )
    return X,y

XTR,YTR = make_data(2500,101)
XVA,YVA = make_data(1200,202)
XOOS,YOOS = make_data(1200,303)

# -------------------- MODELO --------------------

class Core:
    def __init__(self,seed):
        r=np.random.default_rng(seed)
        self.w=r.normal(0,.05,6)
        self.b=0.

    def predict(self,X):
        return X@self.w+self.b

    def clone(self):
        z=Core(0)
        z.w=self.w.copy()
        z.b=self.b
        return z

    def fit(self,X,y,lr=.025,epochs=100):
        for _ in range(epochs):
            e=self.predict(X)-y
            self.w-=lr*(X.T@e)/len(X)
            self.b-=lr*e.mean()

def mse(model,X,y):
    return float(np.mean((model.predict(X)-y)**2))

# -------------------- ESTRATEGIAS --------------------

STRATEGIES = {
    "conservadora":(.010,50),
    "normal":(.025,100),
    "agresiva":(.060,180)
}

# -------------------- CEUNIA SOBRE CEUNIA --------------------

class MetaCEUNIA:

    def __init__(self,seed):
        self.model=Core(seed)
        self.history=[]
        self.strategies=[]

    # PRESENCIA
    def observe(self):
        return {
            "train":mse(self.model,XTR,YTR),
            "val":mse(self.model,XVA,YVA),
            "oos":mse(self.model,XOOS,YOOS)
        }

    # CONTRAFACTUAL
    def simulate(self):
        out={}
        for name,(lr,epochs) in STRATEGIES.items():
            m=self.model.clone()
            m.fit(XTR,YTR,lr,epochs)
            out[name]={
                "model":m,
                "val":mse(m,XVA,YVA),
                "oos":mse(m,XOOS,YOOS)
            }
        return out

    # ATENCIÓN + DECISIÓN
    # OOS permanece invisible durante la decisión.
    def choose(self,sims):
        return min(sims,key=lambda k:sims[k]["val"])

    # META-PREDICCIÓN
    def predict_gain(self,before,sims,choice):
        return max(0,before["val"]-sims[choice]["val"])

    # ACCIÓN
    def execute(self,choice):
        lr,epochs=STRATEGIES[choice]
        self.model.fit(XTR,YTR,lr,epochs)

    # CICLO EVOLUTIVO
    def cycle(self):

        before=self.observe()
        sims=self.simulate()

        choice=self.choose(sims)

        predicted=self.predict_gain(
            before,sims,choice
        )

        self.execute(choice)

        after=self.observe()

        real=max(
            0,
            before["val"]-after["val"]
        )

        self.history.append({
            "before_val":before["val"],
            "after_val":after["val"],
            "before_oos":before["oos"],
            "after_oos":after["oos"],
            "predicted_gain":predicted,
            "real_gain":real,
            "meta_error":abs(predicted-real),
            "strategy":choice
        })

        self.strategies.append(choice)

# -------------------- EXPERIMENTO --------------------

def run(seed):

    # BASELINE
    baseline=Core(seed)
    baseline.fit(XTR,YTR)

    baseline_oos=mse(
        baseline,XOOS,YOOS
    )

    # CEUNIA
    c=MetaCEUNIA(seed)

    initial=c.observe()

    for _ in range(CYCLES):
        c.cycle()

    evolved=c.observe()

    # ---------------- TRANSFERENCIA ----------------

    transferred=c.model.clone()

    transfer_oos=mse(
        transferred,XOOS,YOOS
    )

    # ---------------- AUTONOMÍA ----------------

    frozen=c.model.clone()

    autonomy_oos=mse(
        frozen,XOOS,YOOS
    )

    # ---------------- CONTRAFACTUAL REAL ----------------

    control=c.model.clone()

    control.fit(
        XTR,YTR,
        STRATEGIES["conservadora"][0],
        STRATEGIES["conservadora"][1]
    )

    control_oos=mse(
        control,XOOS,YOOS
    )

    actual_oos=evolved["oos"]

    cf_gain=max(
        0,
        control_oos-actual_oos
    )

    counterfactual=cf_gain/(abs(control_oos)+1e-12)

    # ---------------- COMPONENTES ----------------

    performance=1/(1+actual_oos)

    coherence=1/(1+abs(
        evolved["val"]-evolved["oos"]
    ))

    generalization=coherence

    transfer=1/(1+transfer_oos)

    autonomy=1/(1+autonomy_oos)

    memory=1.0 if len(c.strategies)>0 else 0.

    # ---------------- META-Ω ----------------

    h=pd.DataFrame(c.history)

    meta_error=h.meta_error.mean()

    meta_omega=1/(1+meta_error)

    # ---------------- Ω ----------------

    omega=(
        .20*performance+
        .10*memory+
        .15*coherence+
        .15*transfer+
        .15*autonomy+
        .15*min(1,counterfactual)+
        .10*generalization
    )

    initial_performance=1/(1+initial["oos"])

    delta_omega=omega-initial_performance

    return {
        "seed":seed,
        "Ω":omega,
        "ΔΩ":delta_omega,
        "Meta-Ω":meta_omega,
        "baseline_OOS":baseline_oos,
        "CEUNIA_OOS":actual_oos,
        "OOS_gain":(
            baseline_oos-actual_oos
        )/(baseline_oos+1e-12),
        "counterfactual":counterfactual,
        "transfer":transfer,
        "autonomy":autonomy,
        "coherence":coherence,
        "generalization":generalization,
        "performance":performance,
        "memory":memory,
        "meta_error":meta_error,
        "strategies":" → ".join(c.strategies)
    }

# -------------------- V1 REPRODUCIBILIDAD --------------------

results=[
    run(seed)
    for seed in SEEDS
]

df=pd.DataFrame(results)

# -------------------- ESTADÍSTICAS --------------------

omega=df["Ω"].mean()
omega_std=df["Ω"].std()

delta=df["ΔΩ"].mean()
meta=df["Meta-Ω"].mean()

baseline=df["baseline_OOS"].mean()
ceunia=df["CEUNIA_OOS"].mean()

gain=(baseline-ceunia)/(baseline+1e-12)

cf=df["counterfactual"].mean()
transfer=df["transfer"].mean()
autonomy=df["autonomy"].mean()

# -------------------- HASH --------------------

config={
    "name":"CEUNIA Ω-META v2.0",
    "seeds":SEEDS,
    "cycles":CYCLES,
    "strategies":STRATEGIES,
    "OOS_hidden_during_decision":True,
    "counterfactual":True,
    "meta_prediction":True,
    "timestamp":time.time()
}

sha=hashlib.sha256(
    json.dumps(
        config,
        sort_keys=True,
        default=str
    ).encode()
).hexdigest()

# -------------------- INFORME --------------------

print("\n"+"="*72)
print(" CEUNIA Ω-META v2.0")
print(" CEUNIA SOBRE CEUNIA — TRUE META-EVOLUTION")
print("="*72)

print(f"\nΩ medio                 : {omega:.6f}")
print(f"Ω desviación            : {omega_std:.6f}")
print(f"ΔΩ medio                : {delta:+.6f}")
print(f"Meta-Ω                  : {meta:.6f}")

print("\n--- OOS ---")
print(f"Baseline OOS MSE        : {baseline:.8f}")
print(f"CEUNIA OOS MSE          : {ceunia:.8f}")
print(f"Mejora OOS              : {gain*100:+.4f}%")

print("\n--- METAOBSERVACIÓN ---")
print(f"Error meta medio        : {df.meta_error.mean():.8f}")
print(f"Meta-Ω                  : {meta:.6f}")

print("\n--- CONTRAFACTUAL ---")
print(f"Influencia contrafactual: {cf:.6f}")

print("\n--- TRANSFERENCIA ---")
print(f"Transferencia           : {transfer:.6f}")

print("\n--- AUTONOMÍA ---")
print(f"Autonomía               : {autonomy:.6f}")

print("\n--- GENERALIZACIÓN ---")
print(
    f"Generalización          : "
    f"{df.generalization.mean():.6f}"
)

print("\n--- REPRODUCIBILIDAD ---")
print(f"Semillas                : {len(SEEDS)}")
print(
    f"CV de Ω                 : "
    f"{omega_std/(abs(omega)+1e-12):.6f}"
)

print("\n--- ESTRATEGIAS ---")
print(df["strategies"].value_counts())

print("\n--- SHA256 ---")
print(sha)

# -------------------- DIAGNÓSTICO --------------------

print("\n"+"="*72)
print(" DIAGNÓSTICO CEUNIA")
print("="*72)

tests=[]

tests.append(("OOS",gain>0))
tests.append(("Contrafactual",cf>0))
tests.append(("Metaobservación",meta>.5))
tests.append(("Transferencia",transfer>1/(1+baseline)))
tests.append(("Autonomía",autonomy>1/(1+baseline)))

for name,status in tests:
    print(("✓" if status else "•"),name)

score=sum(x[1] for x in tests)/len(tests)

print(f"\nÍNDICE DE VALIDACIÓN: {score:.2%}")

if score>=.8:
    print("ESTADO: SEÑAL META-EVOLUTIVA FUERTE")
elif score>=.6:
    print("ESTADO: SEÑAL META-EVOLUTIVA PARCIAL")
elif score>=.4:
    print("ESTADO: EVIDENCIA INICIAL")
else:
    print("ESTADO: EVIDENCIA INSUFICIENTE")

print("\nNOTA CIENTÍFICA:")
print("No demuestra conciencia fenomenológica.")
print("Evalúa una hipótesis operacional de autoobservación,")
print("decisión adaptativa, contrafactualidad, persistencia")
print("y metaevaluación.")

display(
    df[
        [
            "seed","Ω","ΔΩ","Meta-Ω",
            "OOS_gain","counterfactual",
            "transfer","autonomy",
            "coherence","generalization",
            "meta_error","strategies"
        ]
    ].round(6)
)
```
