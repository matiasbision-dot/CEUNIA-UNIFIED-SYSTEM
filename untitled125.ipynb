```python

# ================================================================
# CEUNIA Ω-META v3.0
# META-PREDICCIÓN INDEPENDIENTE + CONTRAFACTUAL REAL + MEMORIA
# OOS BLOQUEADO | MÚLTIPLES EPISODIOS | VALIDACIÓN CAUSAL
# SINGLE CELL — GOOGLE COLAB
# ================================================================

import numpy as np
import pandas as pd
import hashlib, json, warnings
warnings.filterwarnings("ignore")

SEEDS = [11,22,33,44,55,66,77,88,101,202,303,404]
N_EPISODES = 6

# ------------------------------------------------
# 1. ENTORNO SINTÉTICO NO ESTACIONARIO
# ------------------------------------------------

def make_episode(seed, n_train=700, n_val=350, n_oos=350):
    rng = np.random.default_rng(seed)

    def make(n, regime):
        X = rng.normal(size=(n,8))

        if regime == "train":
            y = (
                1.20*X[:,0]
                -0.85*X[:,1]
                +0.55*X[:,2]*X[:,3]
                +0.30*np.sin(X[:,4])
                +0.18*X[:,5]
            )
        elif regime == "val":
            y = (
                1.15*X[:,0]
                -0.80*X[:,1]
                +0.60*X[:,2]*X[:,3]
                +0.28*np.sin(X[:,4])
                +0.20*X[:,5]
            )
        else:
            # cambio moderado de régimen
            y = (
                1.05*X[:,0]
                -0.72*X[:,1]
                +0.72*X[:,2]*X[:,3]
                +0.24*np.sin(X[:,4])
                +0.25*X[:,5]
                +0.12*X[:,6]
            )

        y += rng.normal(0, 0.35, n)
        return X, y

    return make(n_train,"train"), make(n_val,"val"), make(n_oos,"oos")


# ------------------------------------------------
# 2. MODELO ESTOCÁSTICO
# ------------------------------------------------

def predict(X,w):
    return X @ w

def mse(X,y,w):
    return np.mean((predict(X,w)-y)**2)

def train_model(X,y,lr,epochs,l2,seed):
    rng = np.random.default_rng(seed)
    w = rng.normal(0,0.08,X.shape[1])

    n = len(X)

    for _ in range(epochs):
        # mini-batch estocástico
        idx = rng.choice(n, size=min(96,n), replace=False)
        xb, yb = X[idx], y[idx]

        grad = (2/len(idx))*xb.T@(xb@w-yb)
        grad += 2*l2*w

        w -= lr*grad

    return w


# ------------------------------------------------
# 3. ESTRATEGIAS
# ------------------------------------------------

STRATEGIES = {
    "conservadora": {"lr":0.008, "epochs":35,  "l2":0.030},
    "normal":       {"lr":0.025, "epochs":70,  "l2":0.012},
    "agresiva":     {"lr":0.065, "epochs":120, "l2":0.003},
}


# ------------------------------------------------
# 4. BASELINE
# ------------------------------------------------

def baseline_model(X,y,seed):
    return train_model(
        X,y,
        lr=0.025,
        epochs=70,
        l2=0.012,
        seed=seed
    )


# ------------------------------------------------
# 5. META-PREDICTOR
#    IMPORTANTE:
#    NO observa el resultado de la estrategia.
#    Solo observa estado actual + propiedades de estrategia.
# ------------------------------------------------

def state_features(X,y,w):
    residual = y-predict(X,w)

    return np.array([
        np.mean(residual**2),
        np.std(residual),
        np.linalg.norm(w),
        np.mean(np.abs(w)),
        np.max(np.abs(w)),
        np.mean(np.abs(residual)),
    ])

def strategy_features(cfg):
    return np.array([
        cfg["lr"],
        cfg["epochs"]/120,
        cfg["l2"],
    ])


# Meta-predictor simple y explícitamente separado
# Aprende de episodios anteriores.
class MetaPredictor:

    def __init__(self):
        self.history=[]

    def add(self,state,strategy,actual_gain):
        self.history.append(
            (state.copy(),strategy,actual_gain)
        )

    def predict_gain(self,state,strategy):

        if len(self.history) < 3:
            # prior neutro
            return 0.0

        vals=[]
        for s,st,g in self.history:
            ds=np.linalg.norm(
                state/(np.linalg.norm(state)+1e-9)
                -
                s/(np.linalg.norm(s)+1e-9)
            )

            # similitud de estado
            sim=np.exp(-ds)

            # coincidencia estratégica
            same = 1.0 if st==strategy else 0.35

            vals.append((sim*same,g))

        weights=np.array([v[0] for v in vals])
        gains=np.array([v[1] for v in vals])

        if weights.sum()==0:
            return float(np.mean(gains))

        return float(np.sum(weights*gains)/weights.sum())


# ------------------------------------------------
# 6. EJECUCIÓN CEUNIA
# ------------------------------------------------

rows=[]
memory_global=[]

for seed in SEEDS:

    rng=np.random.default_rng(seed)

    meta=MetaPredictor()

    episode_memory=[]

    for ep in range(N_EPISODES):

        ep_seed = seed*100 + ep

        (Xtr,ytr),(Xv,yv),(Xo,yo)=make_episode(ep_seed)

        # baseline
        wb=baseline_model(Xtr,ytr,ep_seed)

        base_val=mse(Xv,yv,wb)
        base_oos=mse(Xo,yo,wb)

        state=state_features(Xv,yv,wb)

        # -----------------------------------------
        # META-PREDICCIÓN DE TODAS LAS OPCIONES
        # -----------------------------------------

        predictions={
            s:meta.predict_gain(
                state,
                s
            )
            for s in STRATEGIES
        }

        # CEUNIA decide ANTES de mirar OOS
        selected=max(
            predictions,
            key=predictions.get
        )

        # -----------------------------------------
        # CONTRAFACTUALES
        # Se entrenan todos, pero sus OOS
        # permanecen fuera de la decisión.
        # -----------------------------------------

        results={}

        for strategy,cfg in STRATEGIES.items():

            w=train_model(
                Xtr,ytr,
                cfg["lr"],
                cfg["epochs"],
                cfg["l2"],
                ep_seed + hash(strategy)%10000
            )

            val_error=mse(Xv,yv,w)
            oos_error=mse(Xo,yo,w)

            val_gain=(base_val-val_error)/(abs(base_val)+1e-12)
            oos_gain=(base_oos-oos_error)/(abs(base_oos)+1e-12)

            results[strategy]={
                "val_gain":val_gain,
                "oos_gain":oos_gain,
                "w":w
            }

        # resultado REAL de la decisión
        actual_gain=results[selected]["oos_gain"]

        # mejor contrafactual real
        best_strategy=max(
            results,
            key=lambda s: results[s]["oos_gain"]
        )

        best_gain=results[best_strategy]["oos_gain"]

        # -----------------------------------------
        # META-ERROR:
        # diferencia entre predicción y realidad
        # -----------------------------------------

        predicted=predictions[selected]

        meta_error=abs(
            predicted-actual_gain
        )

        meta.add(
            state,
            selected,
            actual_gain
        )

        episode_memory.append({
            "episode":ep,
            "selected":selected,
            "predicted_gain":predicted,
            "actual_gain":actual_gain,
            "best_counterfactual":best_strategy,
            "best_counterfactual_gain":best_gain,
            "meta_error":meta_error,
            "base_oos":base_oos,
            "selected_oos":results[selected]["oos_gain"]
        })

        memory_global.append({
            "seed":seed,
            **episode_memory[-1]
        })


# ------------------------------------------------
# 7. MÉTRICAS GLOBALES
# ------------------------------------------------

df=pd.DataFrame(memory_global)

# Meta-Ω basado en error REAL de predicción
scale=max(
    np.std(df.actual_gain.values),
    0.01
)

meta_omega=float(
    np.exp(-np.mean(df.meta_error)/scale)
)

# precisión de selección
selection_accuracy=np.mean(
    df.selected.values == df.best_counterfactual.values
)

# ganancia real promedio
real_gain=float(
    np.mean(df.actual_gain)
)

# ventaja contra baseline
counterfactual_advantage=float(
    np.mean(
        df.actual_gain -
        df.best_counterfactual_gain
    )
)

# memoria:
# medimos si el error de meta-predicción
# disminuye entre primera y última mitad
mid=len(df)//2

meta_error_first=float(
    np.mean(df.meta_error.iloc[:mid])
)

meta_error_last=float(
    np.mean(df.meta_error.iloc[mid:])
)

memory_improvement=float(
    meta_error_first-meta_error_last
)

memory_score=float(
    np.clip(
        0.5 + memory_improvement*5,
        0,1
    )
)

# estabilidad entre semillas
seed_omega=df.groupby("seed").actual_gain.mean()

reproducibility=float(
    np.clip(
        1/(1+np.std(seed_omega)),
        0,1
    )
)

# Generalización: resultado OOS positivo
generalization=float(
    np.clip(
        np.mean(df.actual_gain>0),
        0,1
    )
)

# ------------------------------------------------
# 8. Ω SECUNDARIO
# pesos FIJOS, no ajustados al resultado
# ------------------------------------------------

performance=float(
    np.clip(
        (real_gain+0.10)/0.20,
        0,1
    )
)

counterfactual_score=float(
    np.clip(
        0.5 + counterfactual_advantage*5,
        0,1
    )
)

omega=np.mean([
    performance,
    meta_omega,
    selection_accuracy,
    memory_score,
    generalization,
    reproducibility
])

# ------------------------------------------------
# 9. BOOTSTRAP CI PARA LA GANANCIA
# ------------------------------------------------

rng=np.random.default_rng(999)

boot=[]

vals=df.actual_gain.values

for _ in range(5000):
    sample=rng.choice(
        vals,
        size=len(vals),
        replace=True
    )
    boot.append(np.mean(sample))

ci_low,ci_high=np.percentile(
    boot,[2.5,97.5]
)

# ------------------------------------------------
# 10. HASH REPRODUCIBLE
# ------------------------------------------------

payload={
    "version":"CEUNIA_OMEGA_META_v3.0",
    "seeds":SEEDS,
    "episodes":N_EPISODES,
    "strategies":STRATEGIES,
    "mean_gain":real_gain,
    "meta_omega":meta_omega,
    "selection_accuracy":selection_accuracy,
    "memory_improvement":memory_improvement,
    "omega":omega
}

digest=hashlib.sha256(
    json.dumps(
        payload,
        sort_keys=True
    ).encode()
).hexdigest()

# ------------------------------------------------
# 11. REPORTE
# ------------------------------------------------

print("\n"+"="*72)
print("CEUNIA Ω-META v3.0")
print("META-PREDICCIÓN INDEPENDIENTE + CONTRAFACTUAL + MEMORIA")
print("="*72)

print(f"\nΩ global                  : {omega:.6f}")
print(f"Meta-Ω                    : {meta_omega:.6f}")
print(f"Ganancia OOS real         : {real_gain*100:+.4f}%")
print(f"IC95% ganancia OOS        : [{ci_low*100:+.4f}%, {ci_high*100:+.4f}%]")
print(f"Precisión selección       : {selection_accuracy*100:.2f}%")
print(f"Ventaja vs mejor contra   : {counterfactual_advantage*100:+.4f}%")
print(f"Error meta inicial        : {meta_error_first:.6f}")
print(f"Error meta final          : {meta_error_last:.6f}")
print(f"Mejora por memoria        : {memory_improvement:+.6f}")
print(f"Generalización OOS        : {generalization:.4f}")
print(f"Reproducibilidad          : {reproducibility:.4f}")

print("\nESTRATEGIAS ELEGIDAS")
print(df.selected.value_counts())

print("\nDIAGNÓSTICO")

if ci_low > 0:
    print("✓ GANANCIA OOS ESTADÍSTICAMENTE POSITIVA")
else:
    print("• GANANCIA OOS NO CONCLUYENTE")

if selection_accuracy > 0.50:
    print("✓ SELECCIÓN META > AZAR")
else:
    print("• SELECCIÓN META NO SUPERA AZAR")

if memory_improvement > 0:
    print("✓ MEMORIA REDUCE ERROR META")
else:
    print("• MEMORIA NO DEMUESTRA MEJORA")

if meta_omega > 0.70:
    print("✓ META-PREDICCIÓN FUNCIONAL")
else:
    print("• META-PREDICCIÓN DÉBIL")

print("\nHASH SHA256")
print(digest)

print("\n"+"="*72)

# ------------------------------------------------
# 12. TABLA DE EPISODIOS
# ------------------------------------------------

display(
    df[[
        "seed",
        "episode",
        "selected",
        "predicted_gain",
        "actual_gain",
        "best_counterfactual",
        "best_counterfactual_gain",
        "meta_error"
    ]].round(6)
)
```
