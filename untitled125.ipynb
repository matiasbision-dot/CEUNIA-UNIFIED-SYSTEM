```python

# ================================================================
# CEUNIA Ω-META v3.1
# META-TRAIN / META-TEST BLOQUEADOS
# MEMORIA vs NO-MEMORIA | CONTRAFACTUAL REAL
# OOS CIEGO PARA LA DECISIÓN | SEMILLAS DETERMINISTAS
# SINGLE CELL — GOOGLE COLAB
# ================================================================

import numpy as np
import pandas as pd
import hashlib, json, warnings
warnings.filterwarnings("ignore")

SEEDS = [11,22,33,44,55,66,77,88,101,202,303,404]
META_TRAIN = 30
META_TEST = 30

STRATEGIES = {
    "conservadora": {"lr":0.008, "epochs":35,  "l2":0.030},
    "normal":       {"lr":0.025, "epochs":70,  "l2":0.012},
    "agresiva":     {"lr":0.045, "epochs":95,  "l2":0.006},
}

STR_ID = {
    "conservadora":1,
    "normal":2,
    "agresiva":3
}

# ------------------------------------------------
# 1. ENTORNO
# ------------------------------------------------

def make_episode(seed, n_train=700, n_val=350, n_oos=350):
    rng=np.random.default_rng(seed)

    def make(n,regime):
        X=rng.normal(size=(n,8))

        if regime=="train":
            co=np.array([1.20,-0.85,0.55,0.30,0.18,0,0,0])
        elif regime=="val":
            co=np.array([1.15,-0.80,0.60,0.28,0.20,0,0,0])
        else:
            co=np.array([1.05,-0.72,0.72,0.24,0.25,0.12,0,0])

        y=(
            co[0]*X[:,0]
            +co[1]*X[:,1]
            +co[2]*X[:,2]*X[:,3]
            +co[3]*np.sin(X[:,4])
            +co[4]*X[:,5]
            +co[5]*X[:,6]
        )

        y += rng.normal(0,0.35,n)
        return X,y

    return (
        make(n_train,"train"),
        make(n_val,"val"),
        make(n_oos,"oos")
    )

# ------------------------------------------------
# 2. MODELO
# ------------------------------------------------

def mse(X,y,w):
    return np.mean((X@w-y)**2)

def train_model(X,y,cfg,seed):
    rng=np.random.default_rng(seed)
    w=rng.normal(0,0.04,X.shape[1])

    for _ in range(cfg["epochs"]):
        idx=rng.choice(
            len(X),
            size=min(96,len(X)),
            replace=False
        )

        xb=X[idx]
        yb=y[idx]

        grad=(2/len(idx))*xb.T@(xb@w-yb)
        grad += 2*cfg["l2"]*w

        w-=cfg["lr"]*grad

    return w

# ------------------------------------------------
# 3. ESTADO
# ------------------------------------------------

def state_features(X,y,w):
    r=y-X@w

    return np.array([
        np.mean(r**2),
        np.std(r),
        np.mean(np.abs(r)),
        np.linalg.norm(w),
        np.mean(np.abs(w)),
        np.max(np.abs(w)),
        np.std(w),
    ])

# ------------------------------------------------
# 4. META-MODELO
#    ENTRENADO SOLO EN META-TRAIN
# ------------------------------------------------

class MetaLearner:

    def __init__(self):
        self.X=[]
        self.y=[]

    def add(self,state,strategy,gain):
        sf=np.r_[state,
                 STRATEGIES[strategy]["lr"],
                 STRATEGIES[strategy]["epochs"]/100,
                 STRATEGIES[strategy]["l2"],
                 STR_ID[strategy]]

        self.X.append(sf)
        self.y.append(gain)

    def fit(self):

        if len(self.X)<5:
            return

        X=np.asarray(self.X)
        y=np.asarray(self.y)

        # estandarización
        self.mu=X.mean(axis=0)
        self.sd=X.std(axis=0)+1e-8

        Z=(X-self.mu)/self.sd

        # ridge cerrado
        lam=0.20

        A=Z.T@Z + lam*np.eye(Z.shape[1])

        self.beta=np.linalg.solve(
            A,
            Z.T@y
        )

    def predict(self,state,strategy):

        if not hasattr(self,"beta"):
            return 0.0

        sf=np.r_[state,
                 STRATEGIES[strategy]["lr"],
                 STRATEGIES[strategy]["epochs"]/100,
                 STRATEGIES[strategy]["l2"],
                 STR_ID[strategy]]

        z=(sf-self.mu)/self.sd

        return float(z@self.beta)


# ------------------------------------------------
# 5. META-TRAIN
# ------------------------------------------------

meta=MetaLearner()

train_rows=[]

for i in range(META_TRAIN):

    seed=10000+i

    (Xtr,ytr),(Xv,yv),(Xo,yo)=make_episode(seed)

    wb=train_model(
        Xtr,ytr,
        STRATEGIES["normal"],
        seed
    )

    base_oos=mse(Xo,yo,wb)
    state=state_features(Xv,yv,wb)

    for j,(name,cfg) in enumerate(STRATEGIES.items()):

        w=train_model(
            Xtr,ytr,
            cfg,
            seed+100+j
        )

        gain=(
            base_oos-mse(Xo,yo,w)
        )/(abs(base_oos)+1e-8)

        meta.add(
            state,
            name,
            gain
        )

        train_rows.append({
            "episode":i,
            "strategy":name,
            "gain":gain
        })

meta.fit()

# ------------------------------------------------
# 6. META-TEST
# ------------------------------------------------

rows=[]

for i in range(META_TEST):

    seed=20000+i

    (Xtr,ytr),(Xv,yv),(Xo,yo)=make_episode(seed)

    wb=train_model(
        Xtr,ytr,
        STRATEGIES["normal"],
        seed
    )

    base_val=mse(Xv,yv,wb)
    base_oos=mse(Xo,yo,wb)

    state=state_features(Xv,yv,wb)

    # ==================================================
    # CEUNIA CON MEMORIA
    # ==================================================

    predictions={
        s:meta.predict(state,s)
        for s in STRATEGIES
    }

    ceunia_selected=max(
        predictions,
        key=predictions.get
    )

    # ==================================================
    # CEUNIA SIN MEMORIA
    # prior neutral basado SOLO en validación
    # ==================================================

    val_scores={}

    for j,(name,cfg) in enumerate(STRATEGIES.items()):

        w=train_model(
            Xtr,ytr,
            cfg,
            seed+500+j
        )

        val_scores[name]=mse(Xv,yv,w)

    no_memory_selected=min(
        val_scores,
        key=val_scores.get
    )

    # ==================================================
    # AZAR
    # ==================================================

    rng=np.random.default_rng(seed+9999)

    random_selected=rng.choice(
        list(STRATEGIES.keys())
    )

    # ==================================================
    # CONTRAFACTUALES
    # ==================================================

    actual={}

    for j,(name,cfg) in enumerate(STRATEGIES.items()):

        w=train_model(
            Xtr,ytr,
            cfg,
            seed+1000+j
        )

        oos=mse(Xo,yo,w)

        actual[name]=(
            base_oos-oos
        )/(abs(base_oos)+1e-8)

    best=max(
        actual,
        key=actual.get
    )

    # ==================================================
    # REGISTRO
    # ==================================================

    rows.append({
        "episode":i,
        "ceunia":ceunia_selected,
        "no_memory":no_memory_selected,
        "random":random_selected,
        "best":best,

        "ceunia_gain":actual[ceunia_selected],
        "no_memory_gain":actual[no_memory_selected],
        "random_gain":actual[random_selected],
        "oracle_gain":actual[best],

        "ceunia_pred":predictions[ceunia_selected],
        "ceunia_meta_error":
            abs(
                predictions[ceunia_selected]
                -actual[ceunia_selected]
            )
    })

df=pd.DataFrame(rows)

# ------------------------------------------------
# 7. MÉTRICAS
# ------------------------------------------------

ce=df.ceunia_gain.mean()
nm=df.no_memory_gain.mean()
rd=df.random_gain.mean()
oracle=df.oracle_gain.mean()

ce_acc=np.mean(
    df.ceunia==df.best
)

nm_acc=np.mean(
    df.no_memory==df.best
)

random_acc=np.mean(
    df.random==df.best
)

meta_error=df.ceunia_meta_error.mean()

meta_omega=float(
    np.exp(
        -meta_error/
        max(np.std(df.oracle_gain),0.02)
    )
)

memory_delta=ce-nm

# estabilidad
seed_blocks=np.array_split(
    df.ceunia_gain,
    6
)

reproducibility=float(
    np.clip(
        1/(1+np.std([
            x.mean() for x in seed_blocks
        ])),
        0,1
    )
)

generalization=float(
    np.mean(df.ceunia_gain>0)
)

performance=float(
    np.clip(
        (ce+0.10)/0.20,
        0,1
    )
)

selection_score=float(
    np.clip(
        ce_acc,
        0,1
    )
)

memory_score=float(
    np.clip(
        0.5+memory_delta*5,
        0,1
    )
)

omega=float(
    np.mean([
        performance,
        meta_omega,
        selection_score,
        memory_score,
        generalization,
        reproducibility
    ])
)

# ------------------------------------------------
# 8. BOOTSTRAP DIFERENCIA CEUNIA - NO MEMORIA
# ------------------------------------------------

rng=np.random.default_rng(777)

diffs=df.ceunia_gain.values-df.no_memory_gain.values

boot=[]

for _ in range(5000):
    sample=rng.choice(
        diffs,
        len(diffs),
        replace=True
    )
    boot.append(sample.mean())

ci_low,ci_high=np.percentile(
    boot,[2.5,97.5]
)

# ------------------------------------------------
# 9. HASH
# ------------------------------------------------

payload={
    "version":"CEUNIA_OMEGA_META_v3.1",
    "meta_train":META_TRAIN,
    "meta_test":META_TEST,
    "strategies":STRATEGIES,
    "ceunia_gain":float(ce),
    "no_memory_gain":float(nm),
    "random_gain":float(rd),
    "oracle_gain":float(oracle),
    "meta_omega":meta_omega,
    "omega":omega
}

digest=hashlib.sha256(
    json.dumps(
        payload,
        sort_keys=True
    ).encode()
).hexdigest()

# ------------------------------------------------
# 10. REPORTE
# ------------------------------------------------

print("\n"+"="*72)
print("CEUNIA Ω-META v3.1")
print("META-TRAIN / META-TEST BLOQUEADOS")
print("="*72)

print(f"\nΩ global                     : {omega:.6f}")
print(f"Meta-Ω                       : {meta_omega:.6f}")

print(f"\nGanancia CEUNIA OOS          : {ce*100:+.4f}%")
print(f"Ganancia SIN memoria         : {nm*100:+.4f}%")
print(f"Ganancia AZAR               : {rd*100:+.4f}%")
print(f"Ganancia ORÁCULO             : {oracle*100:+.4f}%")

print(f"\nCEUNIA precisión selección   : {ce_acc*100:.2f}%")
print(f"Sin memoria precisión        : {nm_acc*100:.2f}%")
print(f"Azar precisión               : {random_acc*100:.2f}%")

print(f"\nError meta medio             : {meta_error:.6f}")
print(f"Δ memoria                    : {memory_delta*100:+.4f}%")
print(f"IC95% CEUNIA - sin memoria  : "
      f"[{ci_low*100:+.4f}%, {ci_high*100:+.4f}%]")

print(f"\nGeneralización OOS           : {generalization:.4f}")
print(f"Reproducibilidad             : {reproducibility:.4f}")

print("\nDIAGNÓSTICO")

if ci_low>0:
    print("✓ MEMORIA MEJORA OOS DE FORMA ESTADÍSTICAMENTE POSITIVA")
elif ci_high<0:
    print("✗ MEMORIA EMPEORA OOS DE FORMA ESTADÍSTICAMENTE POSITIVA")
else:
    print("• DIFERENCIA MEMORIA/NO-MEMORIA NO CONCLUYENTE")

if ce_acc>random_acc:
    print("✓ CEUNIA SUPERA AZAR EN SELECCIÓN")
else:
    print("• CEUNIA NO SUPERA AZAR")

if meta_omega>0.70:
    print("✓ META-PREDICCIÓN FUNCIONAL")
else:
    print("• META-PREDICCIÓN TODAVÍA DÉBIL")

if ce>0:
    print("✓ GANANCIA OOS MEDIA POSITIVA")
else:
    print("✗ GANANCIA OOS MEDIA NEGATIVA")

print("\nHASH SHA256")
print(digest)

print("\n"+"="*72)

display(
    df.round(6)
)
```
