```python

# ================================================================
# CEUNIA MASTER ARCHIVE
# Consolidación, auditoría y preservación verificable
# ================================================================

import os
import sqlite3
import hashlib
import zipfile
import json
from datetime import datetime, timezone
from pathlib import Path
from google.colab import files

ROOT = "/content"
WORK = os.path.join(ROOT, "CEUNIA_MASTER_ARCHIVE")
DB = os.path.join(ROOT, "ceunia_integral.db")
OLD_ZIP = os.path.join(ROOT, "CEUNIA_Core_v1_1_MVP.zip")

MASTER_ZIP = os.path.join(
    ROOT,
    "CEUNIA_MASTER_ARCHIVE_2026-09.zip"
)

CER_V11 = (
    "b82a41c23163f985116733ae04e5cf8d238c277e092e9c14610be10c0cf27432"
)

OLD_ZIP_SHA = (
    "778f792ea926f927fc537a49142cfb10eaaccef360d9bd40217a9148fe9af0ba"
)

print("=" * 78)
print("CEUNIA MASTER ARCHIVE — CONSOLIDACIÓN VERIFICABLE")
print("=" * 78)

# ------------------------------------------------
# 1. Preparar estructura
# ------------------------------------------------

if os.path.exists(WORK):
    import shutil
    shutil.rmtree(WORK)

os.makedirs(WORK)
os.makedirs(os.path.join(WORK, "CORE_v1.1"))
os.makedirs(os.path.join(WORK, "VALIDATION"))
os.makedirs(os.path.join(WORK, "DATABASE"))
os.makedirs(os.path.join(WORK, "HASHES"))
os.makedirs(os.path.join(WORK, "INVENTORY"))

# ------------------------------------------------
# 2. Verificar DB
# ------------------------------------------------

if not os.path.exists(DB):
    raise FileNotFoundError(
        "No existe /content/ceunia_integral.db"
    )

print("\n[1/8] DB encontrada")
print("      ", DB)
print("      tamaño:", round(os.path.getsize(DB) / 1024, 2), "KB")

# Copia exacta de la DB
import shutil
DB_COPY = os.path.join(
    WORK,
    "DATABASE",
    "ceunia_integral.db"
)

shutil.copy2(DB, DB_COPY)

# ------------------------------------------------
# 3. Auditoría SQLite
# ------------------------------------------------

conn = sqlite3.connect(DB)

tables = conn.execute(
    "SELECT name FROM sqlite_master "
    "WHERE type='table' ORDER BY name"
).fetchall()

audit = []

for row in tables:
    table = row[0]

    try:
        count = conn.execute(
            "SELECT COUNT(*) FROM " + '"' + table + '"'
        ).fetchone()[0]
    except Exception:
        count = None

    audit.append({
        "table": table,
        "rows": count
    })

conn.close()

print("\n[2/8] TABLAS DETECTADAS")

for item in audit:
    print(
        "      {:30s} {:>8}".format(
            item["table"],
            str(item["rows"])
        )
    )

# ------------------------------------------------
# 4. Registro histórico de Core v1.1
# ------------------------------------------------

core_record = {
    "system": "CEUNIA",
    "component": "CEUNIA Core",
    "version": "1.1.0",
    "status": "historical_execution_record",
    "author": "Néstor Matías de la Peña",

    "cycles_executed": 3,
    "memory_records": 3,

    "coherence": 0.7825,
    "potercia": 0.74848,
    "meta_potercia": 0.06768,

    "CER_SHA256": CER_V11,

    "note": (
        "Registro histórico recuperado de la ejecución previa. "
        "La tabla memory del Core v1.1 no está presente en la DB "
        "sobreviviente de la sesión actual."
    )
}

CORE_JSON = os.path.join(
    WORK,
    "CORE_v1.1",
    "core_v1_1_historical_state.json"
)

with open(CORE_JSON, "w", encoding="utf-8") as f:
    json.dump(
        core_record,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------
# 5. Estado actual
# ------------------------------------------------

now = datetime.now(timezone.utc).isoformat()

state_md = """# CEUNIA MASTER ARCHIVE

## Identidad

**Sistema:** CEUNIA
**Autor:** Néstor Matías de la Peña
**Fecha de consolidación:** {date}

## Core v1.1

Registro histórico de ejecución:

- Ciclos: 3
- Registros de memoria: 3
- Coherencia: 0.7825
- Potercia: 0.74848
- MetaPotercia: 0.06768
- CER SHA-256: `{cer}`

Este registro corresponde a la ejecución histórica de CEUNIA Core
v1.1. No se afirma que la DB sobreviviente contenga esos registros.

## Base de datos sobreviviente

Archivo:

`DATABASE/ceunia_integral.db`

Tablas detectadas:

{tables}

## Artefacto reconstruido anterior

`CEUNIA_Core_v1_1_MVP.zip`

SHA-256 registrado:

`{oldzip}`

## Distinción criptográfica

El SHA del CER y el SHA del ZIP son identificadores diferentes:

- CER Core v1.1: `{cer}`
- ZIP reconstruido: `{oldzip}`

El primero identifica el registro CER histórico.
El segundo identifica el archivo ZIP.

## Validación

La DB sobreviviente contiene resultados de:

- validación por régimen;
- meta-selector;
- CEUNIA v2.0;
- CEUNIA v2.1.

Los resultados experimentales no deben interpretarse automáticamente
como validación científica general. Constituyen evidencia experimental
del comportamiento de los prototipos bajo las condiciones de cada
experimento.

## Limitación de preservación

El código fuente original completo de Core v1.1 no está presente en
la sesión actual. Por eso este archivo maestro preserva el estado
histórico conocido y los artefactos sobrevivientes, pero no afirma
recuperar archivos que ya no existen.

## Principio de auditoría

No se sobrescribe la evidencia existente.
Cada artefacto se identifica mediante SHA-256.

""".format(
    date=now,
    cer=CER_V11,
    oldzip=OLD_ZIP_SHA,
    tables="\n".join(
        "- `{}`: {} registros".format(
            x["table"],
            x["rows"]
        )
        for x in audit
    )
)

STATE_FILE = os.path.join(
    WORK,
    "CORE_v1.1",
    "ESTADO_ACTUAL.md"
)

with open(STATE_FILE, "w", encoding="utf-8") as f:
    f.write(state_md)

# ------------------------------------------------
# 6. Inventario de /content
# ------------------------------------------------

inventory = []

for root, dirs, filenames in os.walk(ROOT):

    # No incluir el propio directorio maestro
    if root.startswith(WORK):
        continue

    for filename in filenames:
        path = os.path.join(root, filename)

        try:
            size = os.path.getsize(path)

            sha = hashlib.sha256()

            with open(path, "rb") as f:
                for chunk in iter(
                    lambda: f.read(1024 * 1024),
                    b""
                ):
                    sha.update(chunk)

            inventory.append({
                "path": path,
                "size_bytes": size,
                "sha256": sha.hexdigest()
            })

        except Exception as e:
            inventory.append({
                "path": path,
                "error": str(e)
            })

INV_JSON = os.path.join(
    WORK,
    "INVENTORY",
    "content_inventory.json"
)

with open(INV_JSON, "w", encoding="utf-8") as f:
    json.dump(
        inventory,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\n[3/8] Inventario creado")
print("      archivos inspeccionados:", len(inventory))

# ------------------------------------------------
# 7. Copiar ZIP reconstruido si existe
# ------------------------------------------------

if os.path.exists(OLD_ZIP):

    shutil.copy2(
        OLD_ZIP,
        os.path.join(
            WORK,
            "CORE_v1.1",
            "CEUNIA_Core_v1_1_MVP_reconstructed.zip"
        )
    )

    print("\n[4/8] ZIP reconstruido incluido")

else:
    print("\n[4/8] ZIP reconstruido no encontrado")

# ------------------------------------------------
# 8. Manifiesto SHA-256
# ------------------------------------------------

manifest_lines = []

for root, dirs, filenames in os.walk(WORK):

    for filename in filenames:

        path = os.path.join(root, filename)

        if path == os.path.join(
            WORK,
            "HASHES",
            "MANIFEST.sha256"
        ):
            continue

        rel = os.path.relpath(
            path,
            WORK
        )

        sha = hashlib.sha256()

        with open(path, "rb") as f:
            for chunk in iter(
                lambda: f.read(1024 * 1024),
                b""
            ):
                sha.update(chunk)

        manifest_lines.append(
            "{}  {}".format(
                sha.hexdigest(),
                rel
            )
        )

manifest_lines.sort()

MANIFEST = os.path.join(
    WORK,
    "HASHES",
    "MANIFEST.sha256"
)

with open(MANIFEST, "w", encoding="utf-8") as f:
    f.write(
        "\n".join(manifest_lines) + "\n"
    )

print("\n[5/8] MANIFEST.sha256 creado")
print("      entradas:", len(manifest_lines))

# ------------------------------------------------
# 9. Crear ZIP maestro
# ------------------------------------------------

if os.path.exists(MASTER_ZIP):
    os.remove(MASTER_ZIP)

with zipfile.ZipFile(
    MASTER_ZIP,
    "w",
    zipfile.ZIP_DEFLATED
) as z:

    for root, dirs, filenames in os.walk(WORK):

        for filename in filenames:

            path = os.path.join(
                root,
                filename
            )

            arcname = os.path.relpath(
                path,
                WORK
            )

            z.write(
                path,
                arcname=arcname
            )

print("\n[6/8] ARCHIVO MAESTRO CREADO")

# ------------------------------------------------
# 10. SHA del archivo maestro
# ------------------------------------------------

master_sha = hashlib.sha256()

with open(MASTER_ZIP, "rb") as f:
    for chunk in iter(
        lambda: f.read(1024 * 1024),
        b""
    ):
        master_sha.update(chunk)

MASTER_SHA = master_sha.hexdigest()

# Registrar SHA externo dentro de la salida
MASTER_SHA_FILE = os.path.join(
    ROOT,
    "CEUNIA_MASTER_ARCHIVE_SHA256.txt"
)

with open(
    MASTER_SHA_FILE,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "CEUNIA_MASTER_ARCHIVE_2026-09.zip\n"
        "SHA-256: " + MASTER_SHA + "\n"
    )

print("\n[7/8] SHA-256 MAESTRO")
print("      ", MASTER_SHA)

print("\n" + "=" * 78)
print("CEUNIA MASTER ARCHIVE — COMPLETADO")
print("=" * 78)

print("ARCHIVO:")
print(MASTER_ZIP)

print("\nTAMAÑO:")
print(round(os.path.getsize(MASTER_ZIP) / 1024, 2), "KB")

print("\nSHA-256 MASTER:")
print(MASTER_SHA)

print("\nCONTENIDO:")
for line in manifest_lines:
    print(" ", line)

# ------------------------------------------------
# 11. Descargar
# ------------------------------------------------

print("\n[8/8] INICIANDO DESCARGA...")
files.download(MASTER_ZIP)

print("\nDESCARGA SOLICITADA CORRECTAMENTE.")
print("CEUNIA MASTER ARCHIVE PRESERVADO.")
```
