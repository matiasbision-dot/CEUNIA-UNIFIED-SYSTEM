```python

# ================================================================
# CEUNIA Ω-META v3.2 MAX
# ================================================================
# META-EVOLUTIONARY BENCHMARK
#
# META-TRAIN / META-VAL / META-TEST / OOD
# REGÍMENES LATENTES
# ESTRATEGIA ÓPTIMA VARIABLE
# MEMORIA ADAPTATIVA
# CONTRAFACTUALES
# ABLATIONS
# BOOTSTRAP CI
# PERMUTATION TEST
# CALIBRACIÓN META
# DIVERSIDAD DE DECISIONES
# REPRODUCIBILIDAD
# HASH CRIPTOGRÁFICO
#
# SINGLE CELL — GOOGLE COLAB
# ================================================================

import numpy as np
import pandas as pd
import hashlib, json, time, warnings
from collections import Counter
warnings.filterwarnings("ignore")

T0=time.time()

# ================================================================
# 0. CONFIGURACIÓN
# ================================================================

MASTER_SEED=20260911

N_META_TRAIN=180
N_META_VAL=60
N_META_TEST=120
N_OOD=120

N_TRAIN=900
N_VAL=450
N_OOS=450

STRATEGIES=[
    "conservadora",
    "normal",
    "agresiva",
    "exploratoria"
]

STR_ID={
    "conservadora":0,
    "normal":1,
    "agresiva":2,
    "exploratoria":3
}

# estrategias suficientemente diferentes
STRATEGY_CFG={

    "conservadora":{
        "lr":0.006,
        "epochs":35,
        "l2":0.040
    },

    "normal":{
        "lr":0.022,
        "epochs":65,
        "l2":0.014
    },

    "agresiva":{
        "lr":0.055,
        "epochs":105,
        "l2":0.005
    },

    "exploratoria":{
        "lr":0.090,
        "epochs":150,
        "l2":0.001
    }
}

# ================================================================
# 1. REGÍMENES LATENTES
# ================================================================

REGIMES={

    "A":{
        "shift":0.00,
        "noise":0.35,
        "nonlinear":0.35
    },

    "B":{
        "shift":0.18,
        "noise":0.38,
        "nonlinear":0.55
    },

    "C":{
        "shift":-0.18,
        "noise":0.42,
        "nonlinear":0.75
    },

    "D":{
        "shift":0.35,
        "noise":0.48,
        "nonlinear":0.95
    },

    "E":{
        "shift":-0.35,
        "noise":0.30,
        "nonlinear":0.60
    },

    "F":{
        "shift":0.08,
        "noise":0.55,
        "nonlinear":1.15
    }
}

REGIME_LIST=list(REGIMES.keys())

# ================================================================
# 2. GENERADOR DE TAREAS
# ================================================================

def generate_task(seed,regime,ood=False):

    rng=np.random.default_rng(seed)

    p=REGIMES[regime]

    n_features=10

    Xtr=rng.normal(
        p["shift"],
        1.0,
        (N_TRAIN,n_features)
    )

    Xv=rng.normal(
        p["shift"]*1.15,
        1.0,
        (N_VAL,n_features)
    )

    # OOS tiene shift adicional
    extra_shift=0.0

    if ood:
        extra_shift=(
            0.65
            if regime in ["A","C","E"]
            else -0.65
        )

    Xo=rng.normal(
        p["shift"]+extra_shift,
        1.0,
        (N_OOS,n_features)
    )

    def target(X):

        y=(
            1.10*X[:,0]
            -0.90*X[:,1]
            +0.55*X[:,2]*X[:,3]
            +0.35*np.sin(X[:,4])
            +0.25*X[:,5]
        )

        # régimen modifica estructura
        y += p["nonlinear"]*0.18*(X[:,6]**2)

        if regime in ["C","D"]:
            y += 0.28*X[:,7]*X[:,8]

        if regime in ["E","F"]:
            y -= 0.22*np.sin(X[:,7])

        return y

    ytr=target(Xtr)+rng.normal(
        0,p["noise"],len(Xtr)
    )

    yv=target(Xv)+rng.normal(
        0,p["noise"],len(Xv)
    )

    yo=target(Xo)+rng.normal(
        0,p["noise"],len(Xo)
    )

    return (Xtr,ytr),(Xv,yv),(Xo,yo)


# ================================================================
# 3. MODELO ESTOCÁSTICO
# ================================================================

def train_model(X,y,cfg,seed):

    rng=np.random.default_rng(seed)

    w=rng.normal(
        0,
        0.08,
        X.shape[1]
    )

    for _ in range(cfg["epochs"]):

        bs=min(96,len(X))

        idx=rng.choice(
            len(X),
            bs,
            replace=False
        )

        xb=X[idx]
        yb=y[idx]

        pred=xb@w

        grad=(
            2/bs
        )*xb.T@(pred-yb)

        grad+=2*cfg["l2"]*w

        # ruido exploratorio controlado
        if cfg["lr"]>=0.08:

            grad+=rng.normal(
                0,
                0.003,
                len(w)
            )

        w-=cfg["lr"]*grad

        # seguridad numérica
        w=np.clip(
            w,
            -10,
            10
        )

    return w


def mse(X,y,w):

    pred=X@w

    return float(
        np.mean((pred-y)**2)
    )


# ================================================================
# 4. DIAGNÓSTICO DEL ESTADO
# ================================================================

def state_features(X,y,w):

    residual=y-X@w

    return np.array([

        np.mean(residual**2),

        np.std(residual),

        np.mean(np.abs(residual)),

        np.linalg.norm(w),

        np.mean(np.abs(w)),

        np.max(np.abs(w)),

        np.std(w),

        np.mean(X,axis=0).mean(),

        np.std(X),

        np.mean(np.abs(X)).mean(),

    ],dtype=float)


# ================================================================
# 5. NORMALIZADOR META
# ================================================================

class Standardizer:

    def fit(self,X):

        self.mu=np.mean(X,axis=0)

        self.sd=np.std(X,axis=0)+1e-8

    def transform(self,X):

        return (
            X-self.mu
        )/self.sd


# ================================================================
# 6. META-LEARNER MULTIESTRATEGIA
# ================================================================

class MetaLearner:

    def __init__(self):

        self.X=[]

        self.y=[]

    def encode(
        self,
        state,
        strategy
    ):

        cfg=STRATEGY_CFG[strategy]

        return np.r_[

            state,

            cfg["lr"],

            cfg["epochs"]/150,

            cfg["l2"],

            STR_ID[strategy],

            STR_ID[strategy]**2

        ]

    def add(
        self,
        state,
        strategy,
        gain
    ):

        self.X.append(
            self.encode(
                state,
                strategy
            )
        )

        self.y.append(gain)

    def fit(self):

        X=np.asarray(self.X)

        y=np.asarray(self.y)

        self.scaler=Standardizer()

        self.scaler.fit(X)

        Z=self.scaler.transform(X)

        # ridge
        lam=0.35

        A=(
            Z.T@Z
            +
            lam*np.eye(Z.shape[1])
        )

        self.beta=np.linalg.solve(
            A,
            Z.T@y
        )

    def predict(
        self,
        state,
        strategy
    ):

        z=self.scaler.transform(
            self.encode(
                state,
                strategy
            )[None,:]
        )[0]

        return float(
            z@self.beta
        )


# ================================================================
# 7. META-TRAIN
# ================================================================

meta=MetaLearner()

train_records=[]

for i in range(N_META_TRAIN):

    seed=100000+i

    regime=REGIME_LIST[
        i%len(REGIME_LIST)
    ]

    (Xtr,ytr),(Xv,yv),(Xo,yo)=generate_task(
        seed,
        regime,
        ood=False
    )

    wb=train_model(
        Xtr,
        ytr,
        STRATEGY_CFG["normal"],
        seed
    )

    base_oos=mse(
        Xo,
        yo,
        wb
    )

    state=state_features(
        Xv,
        yv,
        wb
    )

    for j,s in enumerate(STRATEGIES):

        w=train_model(
            Xtr,
            ytr,
            STRATEGY_CFG[s],
            seed+1000+j
        )

        gain=(
            base_oos
            -
            mse(Xo,yo,w)
        )/(
            abs(base_oos)+1e-8
        )

        meta.add(
            state,
            s,
            gain
        )

        train_records.append({

            "episode":i,
            "regime":regime,
            "strategy":s,
            "gain":gain
        })


meta.fit()


# ================================================================
# 8. META-VALIDACIÓN
# ================================================================

validation_records=[]

for i in range(N_META_VAL):

    seed=200000+i

    regime=REGIME_LIST[
        (i*3)%len(REGIME_LIST)
    ]

    (Xtr,ytr),(Xv,yv),(Xo,yo)=generate_task(
        seed,
        regime,
        ood=False
    )

    wb=train_model(
        Xtr,
        ytr,
        STRATEGY_CFG["normal"],
        seed
    )

    base_oos=mse(Xo,yo,wb)

    state=state_features(
        Xv,
        yv,
        wb
    )

    preds={
        s:meta.predict(state,s)
        for s in STRATEGIES
    }

    selected=max(
        preds,
        key=preds.get
    )

    validation_records.append({
        "regime":regime,
        "selected":selected,
        "predicted":preds[selected]
    })


# ================================================================
# 9. META-TEST PRINCIPAL
# ================================================================

def evaluate_suite(
    n_tasks,
    offset,
    ood=False,
    use_memory=True
):

    local_meta=meta

    rows=[]

    memory=[]

    for i in range(n_tasks):

        seed=offset+i

        # distribución diferente en OOD
        if ood:

            regime=REGIME_LIST[
                (i*5+2)%len(REGIME_LIST)
            ]

        else:

            regime=REGIME_LIST[
                (i*7+1)%len(REGIME_LIST)
            ]

        (Xtr,ytr),(Xv,yv),(Xo,yo)=generate_task(
            seed,
            regime,
            ood=ood
        )

        wb=train_model(
            Xtr,
            ytr,
            STRATEGY_CFG["normal"],
            seed
        )

        base_val=mse(
            Xv,
            yv,
            wb
        )

        base_oos=mse(
            Xo,
            yo,
            wb
        )

        state=state_features(
            Xv,
            yv,
            wb
        )

        # --------------------------------------------------------
        # PREDICCIÓN META
        # --------------------------------------------------------

        if use_memory:

            preds={
                s:local_meta.predict(
                    state,
                    s
                )
                for s in STRATEGIES
            }

        else:

            # sin memoria:
            # decisión exclusivamente local por validación
            preds={
                s:0.0
                for s in STRATEGIES
            }

        # --------------------------------------------------------
        # ELECCIÓN
        # --------------------------------------------------------

        if use_memory:

            selected=max(
                preds,
                key=preds.get
            )

        else:

            val_results={}

            for j,s in enumerate(STRATEGIES):

                w=train_model(
                    Xtr,
                    ytr,
                    STRATEGY_CFG[s],
                    seed+5000+j
                )

                val_results[s]=mse(
                    Xv,
                    yv,
                    w
                )

            selected=min(
                val_results,
                key=val_results.get
            )

        # --------------------------------------------------------
        # TODOS LOS CONTRAFACTUALES
        # --------------------------------------------------------

        actual={}

        for j,s in enumerate(STRATEGIES):

            w=train_model(
                Xtr,
                ytr,
                STRATEGY_CFG[s],
                seed+10000+j
            )

            oos_error=mse(
                Xo,
                yo,
                w
            )

            actual[s]=(
                base_oos-oos_error
            )/(
                abs(base_oos)+1e-8
            )

        oracle=max(
            actual,
            key=actual.get
        )

        # --------------------------------------------------------
        # MEMORIA DE ERROR
        # --------------------------------------------------------

        if use_memory:

            pred=preds[selected]

            meta_error=abs(
                pred-actual[selected]
            )

            memory.append({
                "regime":regime,
                "strategy":selected,
                "gain":actual[selected],
                "error":meta_error
            })

        else:

            pred=0.0

            meta_error=abs(
                actual[selected]
            )

        rows.append({

            "task":i,
            "regime":regime,

            "selected":selected,

            "oracle":oracle,

            "selected_gain":
                actual[selected],

            "oracle_gain":
                actual[oracle],

            "base_gain":
                0.0,

            "prediction":pred,

            "meta_error":
                meta_error,

            "optimal":
                int(selected==oracle)

        })

    return (
        pd.DataFrame(rows),
        pd.DataFrame(memory)
    )


# ================================================================
# 10. TEST IID
# ================================================================

df_mem,mem_hist=evaluate_suite(
    N_META_TEST,
    300000,
    ood=False,
    use_memory=True
)

df_nomem,_=evaluate_suite(
    N_META_TEST,
    300000,
    ood=False,
    use_memory=False
)


# ================================================================
# 11. TEST OOD
# ================================================================

df_ood,_=evaluate_suite(
    N_OOD,
    400000,
    ood=True,
    use_memory=True
)


# ================================================================
# 12. RANDOM BASELINE
# ================================================================

rng=np.random.default_rng(
    MASTER_SEED
)

random_gain=[]

for i in range(N_META_TEST):

    seed=300000+i

    regime=REGIME_LIST[
        (i*7+1)%len(REGIME_LIST)
    ]

    (Xtr,ytr),(Xv,yv),(Xo,yo)=generate_task(
        seed,
        regime,
        False
    )

    wb=train_model(
        Xtr,
        ytr,
        STRATEGY_CFG["normal"],
        seed
    )

    base=mse(
        Xo,
        yo,
        wb
    )

    s=rng.choice(
        STRATEGIES
    )

    w=train_model(
        Xtr,
        ytr,
        STRATEGY_CFG[s],
        seed+15000+STR_ID[s]
    )

    g=(
        base-mse(Xo,yo,w)
    )/(
        abs(base)+1e-8
    )

    random_gain.append(g)

random_gain=np.array(
    random_gain
)


# ================================================================
# 13. MÉTRICAS
# ================================================================

ce_gain=df_mem.selected_gain.values
nm_gain=df_nomem.selected_gain.values
oracle_gain=df_mem.oracle_gain.values
ood_gain=df_ood.selected_gain.values

selection_accuracy=df_mem.optimal.mean()

ood_accuracy=df_ood.optimal.mean()

random_accuracy=np.mean(
    rng.choice(STRATEGIES,N_META_TEST)
    ==
    df_mem.oracle.values
)

memory_delta=np.mean(
    ce_gain-nm_gain
)

meta_error=np.mean(
    df_mem.meta_error
)

meta_error_median=np.median(
    df_mem.meta_error
)

# ================================================================
# 14. DIVERSIDAD DE DECISIÓN
# ================================================================

decision_counts=Counter(
    df_mem.selected
)

decision_entropy=0.0

total=sum(
    decision_counts.values()
)

for n in decision_counts.values():

    p=n/total

    if p>0:

        decision_entropy-=p*np.log2(p)

decision_entropy_norm=(
    decision_entropy/
    np.log2(len(STRATEGIES))
)


# ================================================================
# 15. BOOTSTRAP
# ================================================================

rng_boot=np.random.default_rng(
    123456
)

def bootstrap_ci(values,n=10000):

    values=np.asarray(values)

    means=np.empty(n)

    for i in range(n):

        sample=rng_boot.choice(
            values,
            size=len(values),
            replace=True
        )

        means[i]=np.mean(sample)

    return np.percentile(
        means,
        [2.5,97.5]
    )


ci_gain=bootstrap_ci(
    ce_gain
)

ci_memory=bootstrap_ci(
    ce_gain-nm_gain
)

ci_ood=bootstrap_ci(
    ood_gain
)


# ================================================================
# 16. PERMUTATION TEST MEMORIA
# ================================================================

observed=np.mean(
    ce_gain-nm_gain
)

perm=np.empty(10000)

for i in range(10000):

    signs=rng_boot.choice(
        [-1,1],
        size=len(ce_gain)
    )

    perm[i]=np.mean(
        (ce_gain-nm_gain)*signs
    )

p_value=np.mean(
    np.abs(perm)>=abs(observed)
)


# ================================================================
# 17. META-Ω
# ================================================================

meta_scale=max(
    np.std(oracle_gain),
    0.02
)

meta_omega=np.exp(
    -meta_error/meta_scale
)

# calibración relativa:
# cuanto menor error respecto a dispersión
calibration_ratio=(
    meta_error/
    meta_scale
)


# ================================================================
# 18. COMPONENTES Ω
# ================================================================

performance=np.clip(
    (np.mean(ce_gain)+0.05)/0.10,
    0,
    1
)

selection_score=np.clip(
    selection_accuracy,
    0,
    1
)

memory_score=np.clip(
    0.5+memory_delta*5,
    0,
    1
)

generalization=np.mean(
    ce_gain>0
)

ood_generalization=np.mean(
    ood_gain>0
)

reproducibility=np.clip(
    1/(1+np.std(ce_gain)),
    0,
    1
)

counterfactual_score=np.clip(
    np.mean(
        ce_gain/oracle_gain.clip(
            min=1e-6
        )
    ),
    0,
    1
)


# Ω principal
omega=np.mean([

    performance,

    selection_score,

    memory_score,

    generalization,

    ood_generalization,

    reproducibility,

    meta_omega,

])


# ================================================================
# 19. TEST DE SUPERACIÓN DEL AZAR
# ================================================================

random_gain_mean=np.mean(
    random_gain
)

random_ci=bootstrap_ci(
    random_gain
)

delta_random=(
    np.mean(ce_gain)
    -
    random_gain_mean
)

ci_delta_random=bootstrap_ci(
    ce_gain-random_gain
)


# ================================================================
# 20. EVOLUCIÓN DE MEMORIA
# ================================================================

if len(mem_hist)>10:

    half=len(mem_hist)//2

    first_error=np.mean(
        mem_hist.error.iloc[:half]
    )

    last_error=np.mean(
        mem_hist.error.iloc[half:]
    )

    memory_learning=(
        first_error-last_error
    )

else:

    first_error=np.nan
    last_error=np.nan
    memory_learning=np.nan


# ================================================================
# 21. HASH
# ================================================================

payload={

    "version":
        "CEUNIA_OMEGA_META_v3.2_MAX",

    "master_seed":
        MASTER_SEED,

    "meta_train":
        N_META_TRAIN,

    "meta_val":
        N_META_VAL,

    "meta_test":
        N_META_TEST,

    "ood":
        N_OOD,

    "strategies":
        STRATEGY_CFG,

    "selection_accuracy":
        float(selection_accuracy),

    "ood_accuracy":
        float(ood_accuracy),

    "gain":
        float(np.mean(ce_gain)),

    "memory_delta":
        float(memory_delta),

    "meta_omega":
        float(meta_omega),

    "omega":
        float(omega)
}

digest=hashlib.sha256(
    json.dumps(
        payload,
        sort_keys=True
    ).encode()
).hexdigest()


# ================================================================
# 22. REPORTE FINAL
# ================================================================

print("\n")
print("="*78)
print("CEUNIA Ω-META v3.2 MAX")
print("META-EVOLUTIONARY GENERALIZATION BENCHMARK")
print("="*78)

print("\nCONFIGURACIÓN")
print(f"Meta-train       : {N_META_TRAIN}")
print(f"Meta-validation  : {N_META_VAL}")
print(f"Meta-test        : {N_META_TEST}")
print(f"OOD test         : {N_OOD}")
print(f"Estrategias      : {len(STRATEGIES)}")
print(f"Regímenes        : {len(REGIMES)}")

print("\nRESULTADOS IID")

print(
    f"Ganancia CEUNIA OOS       : "
    f"{np.mean(ce_gain)*100:+.4f}%"
)

print(
    f"IC95%                     : "
    f"[{ci_gain[0]*100:+.4f}%, "
    f"{ci_gain[1]*100:+.4f}%]"
)

print(
    f"Selección correcta        : "
    f"{selection_accuracy*100:.2f}%"
)

print(
    f"Ganancia sin memoria      : "
    f"{np.mean(nm_gain)*100:+.4f}%"
)

print(
    f"Ganancia aleatoria       : "
    f"{random_gain_mean*100:+.4f}%"
)

print(
    f"Δ CEUNIA - azar           : "
    f"{delta_random*100:+.4f}%"
)

print(
    f"IC95% Δ vs azar           : "
    f"[{ci_delta_random[0]*100:+.4f}%, "
    f"{ci_delta_random[1]*100:+.4f}%]"
)

print("\nMEMORIA")

print(
    f"Δ memoria                 : "
    f"{memory_delta*100:+.4f}%"
)

print(
    f"IC95% Δ memoria           : "
    f"[{ci_memory[0]*100:+.4f}%, "
    f"{ci_memory[1]*100:+.4f}%]"
)

print(
    f"p permutation             : "
    f"{p_value:.6f}"
)

print(
    f"Error meta inicial        : "
    f"{first_error:.6f}"
)

print(
    f"Error meta final          : "
    f"{last_error:.6f}"
)

print(
    f"Aprendizaje memoria      : "
    f"{memory_learning:+.6f}"
)

print("\nMETA-COGNICIÓN OPERACIONAL")

print(
    f"Meta-Ω                    : "
    f"{meta_omega:.6f}"
)

print(
    f"Error meta medio          : "
    f"{meta_error:.6f}"
)

print(
    f"Ratio calibración         : "
    f"{calibration_ratio:.6f}"
)

print("\nCONTRAFACTUAL")

print(
    f"Ganancia oráculo          : "
    f"{np.mean(oracle_gain)*100:+.4f}%"
)

print(
    f"Distancia a oráculo       : "
    f"{(np.mean(oracle_gain)-np.mean(ce_gain))*100:+.4f}%"
)

print("\nOOD")

print(
    f"Ganancia OOD              : "
    f"{np.mean(ood_gain)*100:+.4f}%"
)

print(
    f"IC95% OOD                 : "
    f"[{ci_ood[0]*100:+.4f}%, "
    f"{ci_ood[1]*100:+.4f}%]"
)

print(
    f"Selección correcta OOD    : "
    f"{ood_accuracy*100:.2f}%"
)

print("\nDIVERSIDAD")

print(
    f"Entropía decisiones       : "
    f"{decision_entropy_norm:.4f}"
)

print(
    "Distribución:"
)

for s in STRATEGIES:

    print(
        f"  {s:16s}: "
        f"{decision_counts.get(s,0)}"
    )

print("\nCOMPONENTES Ω")

print(
    f"Performance               : {performance:.6f}"
)

print(
    f"Selection                 : {selection_score:.6f}"
)

print(
    f"Memory                    : {memory_score:.6f}"
)

print(
    f"Generalización IID        : {generalization:.6f}"
)

print(
    f"Generalización OOD        : {ood_generalization:.6f}"
)

print(
    f"Reproducibilidad          : {reproducibility:.6f}"
)

print(
    f"Meta-Ω                    : {meta_omega:.6f}"
)

print(
    f"\nΩ GLOBAL                  : {omega:.6f}"
)

# ================================================================
# 23. DIAGNÓSTICO AUTOMÁTICO
# ================================================================

print("\n")
print("="*78)
print("DIAGNÓSTICO CEUNIA")
print("="*78)

if selection_accuracy>0.50:
    print("✓ Selección superior a azar")
else:
    print("✗ Selección no supera azar")

if ci_delta_random[0]>0:
    print("✓ Ventaja OOS frente a azar estadísticamente positiva")
else:
    print("• Ventaja frente a azar no concluyente")

if ci_memory[0]>0:
    print("✓ Memoria mejora significativamente")
elif ci_memory[1]<0:
    print("✗ Memoria perjudica significativamente")
else:
    print("• Memoria sin efecto concluyente")

if memory_learning>0:
    print("✓ Error meta disminuye con experiencia")
else:
    print("• No se observa aprendizaje meta temporal")

if ood_accuracy>0.50:
    print("✓ Transferencia OOD superior a azar")
else:
    print("• Transferencia OOD débil")

if meta_omega>0.70:
    print("✓ Meta-calibración fuerte")
elif meta_omega>0.40:
    print("• Meta-calibración intermedia")
else:
    print("• Meta-calibración débil")

if decision_entropy_norm>0.50:
    print("✓ CEUNIA utiliza diversidad estratégica")
else:
    print("• Posible colapso hacia una estrategia")

if (
    selection_accuracy>0.70
    and
    ci_delta_random[0]>0
    and
    ci_memory[0]>0
):
    print("\n>>> SEÑAL META-EVOLUTIVA ROBUSTA <<<")
else:
    print("\n>>> EVIDENCIA INCOMPLETA: CONTINUAR VALIDACIÓN <<<")

print("\nSHA256")
print(digest)

print(
    f"\nTiempo ejecución: "
    f"{time.time()-T0:.2f} s"
)

print("="*78)


# ================================================================
# 24. TABLAS
# ================================================================

print("\nTABLA META-TEST")
display(
    df_mem.round(6)
)

print("\nTABLA OOD")
display(
    df_ood.round(6)
)

print("\nHISTORIAL DE MEMORIA")
display(
    mem_hist.round(6).tail(50)
)
```
