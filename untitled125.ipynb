```python

# =====================================================================
# CEUNIA Ω CORE v3.4
# TRUE CONTEXTUAL STRATEGY SELECTION
# STRUCTURAL REGIME ADAPTATION + OOS + BEST-FIXED + RANDOM + ORACLE
# SINGLE CELL | COLAB
# =====================================================================

import numpy as np, pandas as pd, hashlib, json, time, warnings
from sklearn.linear_model import Ridge, LogisticRegression
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline

warnings.filterwarnings("ignore")
T0 = time.time()

# ============================== CONFIG ===============================

MASTER_SEED = 7341
rng = np.random.default_rng(MASTER_SEED)

N_META_TRAIN = 320
N_META_VAL   = 100
N_META_TEST  = 160
N_OOD        = 160

N_TRAIN = 320
N_VAL   = 220
N_OOS   = 420

N_FEATURES = 12
N_SIGNAL   = 5
N_STRATEGIES = 4
N_REGIMES = 4

BOOT = 2000

STRATEGIES = [
    "conservadora",
    "normal",
    "agresiva",
    "exploratoria"
]

# ========================== STRATEGIES ===============================
# Ahora NO son simples velocidades de entrenamiento.
# Cada estrategia tiene distinta capacidad/regularización.

STRATEGY_CFG = {
    "conservadora": dict(kind="linear",  alpha=25.0),
    "normal":       dict(kind="linear",  alpha=2.0),
    "agresiva":     dict(kind="poly",    alpha=5.0),
    "exploratoria": dict(kind="poly",    alpha=.35)
}

# ============================= REGIMES ===============================
# Diseñados para favorecer estrategias distintas.
#
# 0: poco dato + mucho ruido + baja no-linealidad -> conservadora
# 1: señal limpia + lineal -> normal
# 2: no-linealidad fuerte + ruido moderado -> agresiva
# 3: no-linealidad fuerte + señal suficiente -> exploratoria

REGIMES = [
    dict(noise=.85, nonlin=.08, shift=.04, nmult=.55, scale=1.00),
    dict(noise=.22, nonlin=.08, shift=.05, nmult=1.15, scale=1.00),
    dict(noise=.35, nonlin=.65, shift=.08, nmult=1.00, scale=1.00),
    dict(noise=.16, nonlin=.92, shift=.06, nmult=1.25, scale=1.00),
]

# ============================== DATA =================================

def make_task(seed, regime_id, ood=False):
    rr = np.random.default_rng(seed)
    R = REGIMES[regime_id]

    if ood:
        noise = R["noise"] * rr.uniform(1.10, 1.30)
        nonlin = np.clip(R["nonlin"] + rr.uniform(.05, .15), 0, 1)
        shift = R["shift"] + rr.uniform(.10, .22)
        nmult = R["nmult"] * rr.uniform(.78, .92)
        scale = R["scale"] * rr.uniform(1.05, 1.22)
    else:
        noise = R["noise"] * rr.uniform(.92, 1.08)
        nonlin = np.clip(R["nonlin"] + rr.normal(0,.035), 0, 1)
        shift = max(0, R["shift"] + rr.normal(0,.018))
        nmult = R["nmult"] * rr.uniform(.94,1.06)
        scale = R["scale"] * rr.uniform(.97,1.03)

    ntr = max(100, int(N_TRAIN*nmult))

    beta = rr.normal(0,1,N_SIGNAL)
    beta /= np.linalg.norm(beta)+1e-12

    def sample(n, shifted=False):
        X = rr.normal(0,1,(n,N_FEATURES))*scale

        if shifted:
            X += rr.normal(0,shift,(n,N_FEATURES))

        z = X[:,:N_SIGNAL] @ beta

        nonlinear = nonlin * (
            .70*np.sin(1.7*X[:,0]) +
            .55*(X[:,1]**2 - 1) +
            .40*X[:,2]*X[:,3]
        )

        y = z + nonlinear + rr.normal(0,noise,n)

        return X,y

    return (
        *sample(ntr,False),
        *sample(N_VAL,False),
        *sample(N_OOS,True)
    )

# ============================ TRAINERS ===============================

def train_strategy(Xtr,ytr,Xv,yv,strategy,seed):
    cfg = STRATEGY_CFG[strategy]

    if cfg["kind"] == "linear":
        model = make_pipeline(
            StandardScaler(),
            Ridge(alpha=cfg["alpha"])
        )

    else:
        model = make_pipeline(
            StandardScaler(),
            PolynomialFeatures(
                degree=2,
                include_bias=False
            ),
            Ridge(alpha=cfg["alpha"])
        )

    model.fit(Xtr,ytr)

    pred = model.predict(Xv)
    vmse = np.mean((yv-pred)**2)

    return model,float(vmse)

# ============================== STATE =================================
# Estado NO utiliza OOS ni oracle.

def task_state(Xtr,ytr,Xv,yv):

    base = make_pipeline(
        StandardScaler(),
        Ridge(alpha=2.0)
    )

    base.fit(Xtr,ytr)

    ptr = base.predict(Xtr)
    pv  = base.predict(Xv)

    tr = np.mean((ytr-ptr)**2)
    va = np.mean((yv-pv)**2)

    resid = yv-pv

    gap = va/(tr+1e-9)-1

    mu_shift = np.mean(
        np.abs(Xtr.mean(0)-Xv.mean(0))
    )

    var_shift = np.mean(
        np.abs(Xtr.var(0)-Xv.var(0))
    )

    nonlinear_signal = np.mean([
        abs(np.corrcoef(
            Xv[:,0]**2,
            resid
        )[0,1]),
        abs(np.corrcoef(
            Xv[:,1]**2,
            resid
        )[0,1]),
        abs(np.corrcoef(
            Xv[:,2]*Xv[:,3],
            resid
        )[0,1])
    ])

    return np.array([
        np.log1p(len(Xtr)),
        tr,
        va,
        gap,
        mu_shift,
        var_shift,
        resid.std(),
        nonlinear_signal,
        np.std(Xtr),
        np.std(ytr),
        np.mean(np.abs(ytr)),
        np.mean(np.abs(Xtr))
    ])

# ============================ META TASK ===============================

def build_meta(n_tasks,offset,ood=False):

    rows=[]

    for i in range(n_tasks):

        seed = MASTER_SEED + offset + i*37

        regime = i % N_REGIMES

        Xtr,ytr,Xv,yv,Xo,yo = make_task(
            seed,
            regime,
            ood
        )

        state = task_state(
            Xtr,ytr,Xv,yv
        )

        base = np.mean(
            (yo-np.mean(ytr))**2
        )

        gains={}

        for j,s in enumerate(STRATEGIES):

            model,_ = train_strategy(
                Xtr,ytr,Xv,yv,
                s,
                seed+1000+j
            )

            mse = np.mean(
                (yo-model.predict(Xo))**2
            )

            gains[s] = float(
                (base-mse)/(base+1e-12)
            )

        oracle = max(
            gains,
            key=gains.get
        )

        rows.append({
            "task":i,
            "regime":regime,
            "oracle":oracle,
            "oracle_gain":gains[oracle],
            "state":state,
            "gains":gains,
            "ood":ood
        })

    return rows

# ============================ META DATA ===============================

train_tasks = build_meta(
    N_META_TRAIN,
    10000,
    False
)

val_tasks = build_meta(
    N_META_VAL,
    30000,
    False
)

test_tasks = build_meta(
    N_META_TEST,
    50000,
    False
)

ood_tasks = build_meta(
    N_OOD,
    80000,
    True
)

# ======================= BENCHMARK SANITY CHECK ======================

def oracle_distribution(tasks):

    return pd.Series(
        [t["oracle"] for t in tasks]
    ).value_counts(
        normalize=True
    )

oracle_train_dist = oracle_distribution(
    train_tasks
)

oracle_test_dist = oracle_distribution(
    test_tasks
)

print("="*76)
print("CEUNIA Ω CORE v3.4")
print("TRUE CONTEXTUAL STRATEGY SELECTION")
print("="*76)

print("\nORACLE META-TRAIN")
for s in STRATEGIES:
    print(
        f"{s:15s}: "
        f"{oracle_train_dist.get(s,0)*100:6.2f}%"
    )

# Condición científica:
# cada estrategia debe representar al menos 10% del oracle
# en META-TRAIN. Si no, el benchmark queda invalidado.

MIN_ORACLE_SHARE = .10

if any(
    oracle_train_dist.get(s,0) < MIN_ORACLE_SHARE
    for s in STRATEGIES
):
    print("\n❌ BENCHMARK INVALIDADO")
    print(
        "El oracle no presenta diversidad estratégica suficiente."
    )
    print(
        "No se permite entrenar CEUNIA sobre un espacio "
        "de decisión colapsado."
    )
    print(
        "Revisar generador de regímenes."
    )

    raise RuntimeError(
        "BENCHMARK INVALID: ORACLE STRATEGY COLLAPSE"
    )

# ============================ META MODEL ==============================

X_meta = np.vstack([
    t["state"] for t in train_tasks
])

y_meta = np.array([
    STRATEGIES.index(t["oracle"])
    for t in train_tasks
])

meta_clf = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        C=.35,
        max_iter=4000,
        class_weight="balanced"
    )
)

meta_clf.fit(
    X_meta,
    y_meta
)

# ============================ VALIDATION ==============================

Xv_meta = np.vstack([
    t["state"] for t in val_tasks
])

yv_meta = np.array([
    STRATEGIES.index(t["oracle"])
    for t in val_tasks
])

meta_val_acc = np.mean(
    meta_clf.predict(Xv_meta)==yv_meta
)

# =============================== MEMORY ===============================

scaler_mem = StandardScaler().fit(
    X_meta
)

Xm = scaler_mem.transform(
    X_meta
)

def memory_predict(state,k=19):

    z = scaler_mem.transform(
        state.reshape(1,-1)
    )[0]

    dist = np.linalg.norm(
        Xm-z,
        axis=1
    )

    idx = np.argsort(dist)[:k]

    probs = np.zeros(
        N_STRATEGIES
    )

    weights = 1/(dist[idx]+1e-5)

    for ii,w in zip(idx,weights):
        probs[
            y_meta[ii]
        ] += w

    return probs/(probs.sum()+1e-12)

def predict_ceunia(state,use_memory=True):

    raw = meta_clf.predict_proba(
        state.reshape(1,-1)
    )[0]

    pfull = np.zeros(
        N_STRATEGIES
    )

    for cls,p in zip(
        meta_clf[-1].classes_,
        raw
    ):
        pfull[int(cls)] = p

    if not use_memory:
        return pfull

    pmem = memory_predict(state)

    z = scaler_mem.transform(
        state.reshape(1,-1)
    )[0]

    nearest = np.min(
        np.linalg.norm(
            Xm-z,
            axis=1
        )
    )

    weight = np.clip(
        np.exp(-nearest*.18),
        .10,
        .55
    )

    p = (
        (1-weight)*pfull +
        weight*pmem
    )

    return p/(p.sum()+1e-12)

# ============================== EVALUATE ==============================

def evaluate(tasks,use_memory=True):

    rows=[]

    for t in tasks:

        p = predict_ceunia(
            t["state"],
            use_memory
        )

        choice = STRATEGIES[
            int(np.argmax(p))
        ]

        rows.append({
            "choice":choice,
            "oracle":t["oracle"],
            "gain":t["gains"][choice],
            "oracle_gain":t["oracle_gain"],
            "regret":
                t["oracle_gain"]
                -t["gains"][choice],
            "correct":
                int(choice==t["oracle"]),
            "confidence":
                float(np.max(p)),
            "probs":p
        })

    return pd.DataFrame(rows)

df_mem = evaluate(
    test_tasks,
    True
)

df_nomem = evaluate(
    test_tasks,
    False
)

df_ood = evaluate(
    ood_tasks,
    True
)

# ========================== FIXED BASELINES ===========================

fixed_rows=[]

for s in STRATEGIES:

    g = np.array([
        t["gains"][s]
        for t in test_tasks
    ])

    fixed_rows.append({
        "strategy":s,
        "gain":g.mean()
    })

fixed_table = pd.DataFrame(
    fixed_rows
)

best_fixed_strategy = fixed_table.iloc[
    fixed_table["gain"].argmax()
]["strategy"]

best_fixed_gain = fixed_table["gain"].max()

best_fixed_arr = np.array([
    t["gains"][best_fixed_strategy]
    for t in test_tasks
])

# ============================== RANDOM =================================

rr = np.random.default_rng(
    MASTER_SEED+444
)

random_choices = rr.integers(
    0,
    N_STRATEGIES,
    len(test_tasks)
)

random_gains = np.array([
    t["gains"][
        STRATEGIES[c]
    ]
    for t,c in zip(
        test_tasks,
        random_choices
    )
])

# ============================== BOOTSTRAP ==============================

def bootstrap_ci(x,B=BOOT,seed=123):

    rr=np.random.default_rng(seed)

    x=np.asarray(x)

    vals=[]

    for _ in range(B):

        vals.append(
            np.mean(
                rr.choice(
                    x,
                    len(x),
                    replace=True
                )
            )
        )

    return np.percentile(
        vals,
        [2.5,97.5]
    )

gain = df_mem.gain.values
gain_nm = df_nomem.gain.values
gain_ood = df_ood.gain.values

delta_random = gain-random_gains
delta_fixed = gain-best_fixed_arr
delta_memory = gain-gain_nm

ci_gain = bootstrap_ci(gain)
ci_random = bootstrap_ci(delta_random)
ci_fixed = bootstrap_ci(delta_fixed)
ci_memory = bootstrap_ci(delta_memory)
ci_ood = bootstrap_ci(gain_ood)

# ============================ PERMUTATION ==============================

def permutation_paired(a,b,B=4000,seed=44):

    rr=np.random.default_rng(seed)

    d=np.asarray(a)-np.asarray(b)

    obs=np.mean(d)

    count=0

    for _ in range(B):

        signs=rr.choice(
            [-1,1],
            len(d)
        )

        if np.mean(d*signs)>=obs:
            count+=1

    return (count+1)/(B+1)

p_random = permutation_paired(
    gain,
    random_gains
)

p_fixed = permutation_paired(
    gain,
    best_fixed_arr
)

p_memory = permutation_paired(
    gain,
    gain_nm
)

# ============================= CALIBRATION =============================

conf = df_mem.confidence.values
correct = df_mem.correct.values

bins=np.linspace(0,1,11)

ece=0

for a,b in zip(
    bins[:-1],
    bins[1:]
):

    mask=(
        (conf>=a)
        &
        (conf<=b)
    )

    if mask.sum():

        ece += (
            mask.mean()
            *
            abs(
                correct[mask].mean()
                -
                conf[mask].mean()
            )
        )

probs=np.vstack(
    df_mem.probs.values
)

oracle_idx=np.array([
    STRATEGIES.index(x)
    for x in df_mem.oracle
])

onehot=np.eye(
    N_STRATEGIES
)[oracle_idx]

brier=np.mean(
    np.sum(
        (probs-onehot)**2,
        axis=1
    )
)

meta_omega=max(
    0,
    1-brier/(1-1/N_STRATEGIES)
)

# ============================== DIVERSITY ==============================

counts=df_mem.choice.value_counts()

p_counts=np.array([
    counts.get(s,0)/len(df_mem)
    for s in STRATEGIES
])

entropy=-np.sum(
    np.where(
        p_counts>0,
        p_counts*np.log(
            p_counts+1e-12
        ),
        0
    )
)

diversity=entropy/np.log(
    N_STRATEGIES
)

# ========================== REGIME TABLE ==============================

regime_rows=[]

for r in range(N_REGIMES):

    idx=[
        i for i,t in enumerate(test_tasks)
        if t["regime"]==r
    ]

    choices=[
        df_mem.iloc[i]["choice"]
        for i in idx
    ]

    oracles=[
        test_tasks[i]["oracle"]
        for i in idx
    ]

    regime_rows.append({

        "regime":r,

        "n":len(idx),

        "selection":
            np.mean([
                a==b
                for a,b in zip(
                    choices,
                    oracles
                )
            ]),

        "gain":
            np.mean([
                test_tasks[i]["gains"][
                    choices[j]
                ]
                for j,i in enumerate(idx)
            ]),

        "oracle_gain":
            np.mean([
                test_tasks[i]["oracle_gain"]
                for i in idx
            ]),

        "oracle_dominante":
            max(
                set(oracles),
                key=oracles.count
            )
    })

regime_table=pd.DataFrame(
    regime_rows
)

# ============================= Ω ======================================

performance=np.clip(
    .5+gain.mean()*2,
    0,1
)

selection=df_mem.correct.mean()

memory_score=np.clip(
    .5+delta_memory.mean()*10,
    0,1
)

ood_score=np.clip(
    .5+gain_ood.mean()*2,
    0,1
)

# proxy de estabilidad, no "reproducibilidad" real
stability=1/(1+np.std(gain))

omega=np.mean([
    performance,
    selection,
    memory_score,
    diversity,
    ood_score,
    stability,
    meta_omega
])

# =============================== HASH =================================

summary={
    "version":
        "CEUNIA Ω CORE v3.4",
    "seed":
        MASTER_SEED,
    "meta_train":
        len(train_tasks),
    "meta_validation":
        N_META_VAL,
    "meta_test":
        N_META_TEST,
    "ood":
        N_OOD,
    "selection_accuracy":
        float(selection),
    "iid_gain":
        float(gain.mean()),
    "best_fixed_gain":
        float(best_fixed_gain),
    "delta_vs_best_fixed":
        float(delta_fixed.mean()),
    "delta_vs_random":
        float(delta_random.mean()),
    "memory_delta":
        float(delta_memory.mean()),
    "ood_gain":
        float(gain_ood.mean()),
    "meta_omega":
        float(meta_omega),
    "diversity":
        float(diversity),
    "omega":
        float(omega)
}

digest=hashlib.sha256(
    json.dumps(
        summary,
        sort_keys=True
    ).encode()
).hexdigest()

# ============================== REPORT =================================

print("\n"+"="*76)
print("CEUNIA Ω CORE v3.4")
print("TRUE CONTEXTUAL STRATEGY SELECTION")
print("="*76)

print("\nCONFIG")
print("Meta-train      :",len(train_tasks))
print("Meta-validation :",N_META_VAL)
print("Meta-test IID   :",N_META_TEST)
print("OOD             :",N_OOD)
print("Regímenes       :",N_REGIMES)

print("\nMETA-TRAIN ORACLE")
for s in STRATEGIES:
    print(
        f"{s:15s}: "
        f"{oracle_train_dist.get(s,0)*100:6.2f}%"
    )

print("\nMETA-VALIDATION")
print(
    f"accuracy                : "
    f"{meta_val_acc:.4f}"
)

print("\nIID META-TEST")
print(
    f"CEUNIA gain             : "
    f"{gain.mean()*100:+.4f}%"
)
print(
    f"IC95                    : "
    f"[{ci_gain[0]*100:+.4f}%, "
    f"{ci_gain[1]*100:+.4f}%]"
)
print(
    f"selection accuracy      : "
    f"{selection:.4f}"
)

print("\nFIXED STRATEGIES")
print(
    fixed_table.to_string(
        index=False
    )
)
print(
    f"\nMejor fija              : "
    f"{best_fixed_strategy}"
)
print(
    f"Mejor fija gain         : "
    f"{best_fixed_gain*100:+.4f}%"
)
print(
    f"CEUNIA - mejor fija     : "
    f"{delta_fixed.mean()*100:+.4f} pp"
)
print(
    f"IC95                    : "
    f"[{ci_fixed[0]*100:+.4f}, "
    f"{ci_fixed[1]*100:+.4f}] pp"
)
print(
    f"permutation p           : "
    f"{p_fixed:.6f}"
)

print("\nRANDOM")
print(
    f"random gain             : "
    f"{random_gains.mean()*100:+.4f}%"
)
print(
    f"CEUNIA - random         : "
    f"{delta_random.mean()*100:+.4f} pp"
)
print(
    f"IC95                    : "
    f"[{ci_random[0]*100:+.4f}, "
    f"{ci_random[1]*100:+.4f}] pp"
)
print(
    f"permutation p           : "
    f"{p_random:.6f}"
)

print("\nMEMORY ABLATION")
print(
    f"no-memory gain          : "
    f"{gain_nm.mean()*100:+.4f}%"
)
print(
    f"memory Δ                : "
    f"{delta_memory.mean()*100:+.4f} pp"
)
print(
    f"IC95                    : "
    f"[{ci_memory[0]*100:+.4f}, "
    f"{ci_memory[1]*100:+.4f}] pp"
)
print(
    f"permutation p           : "
    f"{p_memory:.6f}"
)

print("\nMETA-CALIBRATION")
print(
    f"ECE                     : "
    f"{ece:.6f}"
)
print(
    f"Brier                   : "
    f"{brier:.6f}"
)
print(
    f"Meta-Ω                  : "
    f"{meta_omega:.6f}"
)

print("\nSTRATEGY DIVERSITY")
for s in STRATEGIES:
    print(
        f"{s:15s}: "
        f"{counts.get(s,0):4d} "
        f"({counts.get(s,0)/len(df_mem)*100:6.2f}%)"
    )

print(
    f"Normalized entropy      : "
    f"{diversity:.6f}"
)

print("\nOOD")
print(
    f"OOD gain                : "
    f"{gain_ood.mean()*100:+.4f}%"
)
print(
    f"IC95                    : "
    f"[{ci_ood[0]*100:+.4f}%, "
    f"{ci_ood[1]*100:+.4f}%]"
)
print(
    f"OOD selection           : "
    f"{df_ood.correct.mean():.4f}"
)

print("\nΩ COMPONENTS")
print(
    f"Performance             : "
    f"{performance:.6f}"
)
print(
    f"Selection               : "
    f"{selection:.6f}"
)
print(
    f"Memory                  : "
    f"{memory_score:.6f}"
)
print(
    f"Diversity               : "
    f"{diversity:.6f}"
)
print(
    f"OOD                     : "
    f"{ood_score:.6f}"
)
print(
    f"Stability               : "
    f"{stability:.6f}"
)
print(
    f"Meta-Ω                  : "
    f"{meta_omega:.6f}"
)
print(
    f"\nΩ GLOBAL                : "
    f"{omega:.6f}"
)

print("\nREGIME ADAPTATION")
print(
    regime_table.to_string(
        index=False
    )
)

print("\nDIAGNÓSTICO")

if diversity >= .45:
    print(
        "✓ diversidad estratégica real"
    )
else:
    print(
        "⚠ diversidad estratégica insuficiente"
    )

if delta_fixed.mean()>0 and p_fixed<.05:
    print(
        "✓ CEUNIA supera a la mejor estrategia fija"
    )
else:
    print(
        "⚠ no supera significativamente a la mejor fija"
    )

if delta_random.mean()>0 and p_random<.05:
    print(
        "✓ ventaja contra random significativa"
    )
else:
    print(
        "⚠ ventaja contra random no demostrada"
    )

if delta_memory.mean()>0 and p_memory<.05:
    print(
        "✓ memoria con efecto significativo"
    )
else:
    print(
        "⚠ memoria sin evidencia significativa"
    )

if gain_ood.mean()>0:
    print(
        "✓ transferencia OOD positiva"
    )
else:
    print(
        "⚠ transferencia OOD negativa"
    )

if meta_omega>.70:
    print(
        "✓ meta-calibración fuerte"
    )
else:
    print(
        "⚠ meta-calibración débil"
    )

print("\nSHA256")
print(digest)

print(
    f"\nTiempo total: "
    f"{time.time()-T0:.2f} s"
)

print("="*76)

# OBJETOS DISPONIBLES:
# df_mem
# df_nomem
# df_ood
# fixed_table
# regime_table
# summary
# digest
```
