```python

import os
import sqlite3
import hashlib
import zipfile
from datetime import datetime, timezone
from google.colab import files

DB = "/content/ceunia_integral.db"
ZIP = "/content/CEUNIA_Core_v1_1_MVP.zip"
README = "/content/CEUNIA_Core_v1_1_README.txt"

print("=" * 72)
print("CEUNIA CORE v1.1 — RECONSTRUCCIÓN DEL PAQUETE")
print("=" * 72)

# ------------------------------------------------------------
# 1. Verificar DB
# ------------------------------------------------------------
if not os.path.exists(DB):
    raise FileNotFoundError("No existe /content/ceunia_integral.db")

print("DB encontrada:", DB)
print("Tamaño DB:", round(os.path.getsize(DB) / 1024, 2), "KB")

# ------------------------------------------------------------
# 2. Inspeccionar SQLite
# ------------------------------------------------------------
conn = sqlite3.connect(DB)
cur = conn.cursor()

tables = cur.execute(
    "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name"
).fetchall()

table_names = [x[0] for x in tables]

print("\nTABLAS:")
for t in table_names:
    print(" -", t)

memory_count = 0

if "memory" in table_names:
    memory_count = cur.execute(
        "SELECT COUNT(*) FROM memory"
    ).fetchone()[0]

print("\nREGISTROS MEMORY:", memory_count)

# ------------------------------------------------------------
# 3. Obtener último estado disponible
# ------------------------------------------------------------
latest = None

if "memory" in table_names and memory_count > 0:
    cols = [x[1] for x in cur.execute("PRAGMA table_info(memory)").fetchall()]
    rows = cur.execute("SELECT * FROM memory ORDER BY rowid DESC LIMIT 1").fetchone()

    latest = dict(zip(cols, rows))

conn.close()

# ------------------------------------------------------------
# 4. README de reconstrucción
# ------------------------------------------------------------
timestamp = datetime.now(timezone.utc).isoformat()

readme = """
CEUNIA CORE v1.1
MVP RECONSTRUIDO DESDE ESTADO PERSISTIDO

Autor:
Néstor Matías de la Peña

Sistema:
CEUNIA

Versión:
1.1.0

Fecha de reconstrucción:
{timestamp}

Base de datos:
ceunia_integral.db

Registros de memoria:
{memory_count}

Estado:
El paquete fue reconstruido a partir de la base de datos persistida
en la sesión de Google Colab.

NOTA:
La ausencia del ZIP original no implica pérdida de la base de datos.
Este paquete es una reconstrucción verificable del estado disponible.

Último registro disponible:
{latest}
""".format(
    timestamp=timestamp,
    memory_count=memory_count,
    latest=latest
)

with open(README, "w", encoding="utf-8") as f:
    f.write(readme)

# ------------------------------------------------------------
# 5. Crear ZIP nuevo
# ------------------------------------------------------------
if os.path.exists(ZIP):
    os.remove(ZIP)

with zipfile.ZipFile(ZIP, "w", zipfile.ZIP_DEFLATED) as z:
    z.write(DB, arcname="ceunia_integral.db")
    z.write(README, arcname="CEUNIA_Core_v1_1_README.txt")

# ------------------------------------------------------------
# 6. SHA-256 del paquete
# ------------------------------------------------------------
sha = hashlib.sha256()

with open(ZIP, "rb") as f:
    for chunk in iter(lambda: f.read(1024 * 1024), b""):
        sha.update(chunk)

ZIP_SHA = sha.hexdigest()

print("\n" + "=" * 72)
print("PAQUETE RECONSTRUIDO")
print("=" * 72)
print("ZIP:", ZIP)
print("Tamaño:", round(os.path.getsize(ZIP) / 1024, 2), "KB")
print("SHA-256:", ZIP_SHA)

# ------------------------------------------------------------
# 7. Descargar
# ------------------------------------------------------------
print("\nINICIANDO DESCARGA...")
files.download(ZIP)

print("\nDESCARGA SOLICITADA CORRECTAMENTE.")
print("CEUNIA CORE v1.1 — PAQUETE PRESERVADO")
```
