```python
# ============================================================
# CEUNIA CORE v1.1 — EJECUCIÓN COMPLETA EN UNA CELDA DE COLAB
# Autor: Néstor Matías de la Peña
# ============================================================

!pip -q install fastapi uvicorn pydantic pytest

from pathlib import Path
import sys, json, hashlib, shutil, zipfile
from datetime import datetime, timezone

ROOT = Path("/content/ceunia_core_v1_1")
if ROOT.exists():
    shutil.rmtree(ROOT)

(ROOT / "ceunia").mkdir(parents=True)
(ROOT / "tests").mkdir(parents=True)

# ------------------------------------------------------------
# 1. CORE
# ------------------------------------------------------------

core_code = r'''
from dat
aclasses import dataclass, field
from datetime import datetime, timezone
import hashlib
import json

@dataclass
class CEUNIA:

    version: str = "1.1.0"

    memory: list = field(default_factory=list)

    state: dict = field(default_factory=lambda: {
        "coherence": 0.5,
        "potercia": 0.5,
        "meta_potercia": 0.0
    })

    cycle: int = 0

    def _score(self, text: str) -> float:
        """
        Evaluador determinista MVP.
        No pretende representar inteligencia general.
        """
        if not text:
            return 0.0

        n = len(text.strip())

        return min(
            1.0,
            0.35 + (n % 67) / 100
        )

    def _hash_record(self, record):
        """
        SHA-256 del registro CER.
        Garantiza integridad digital del registro,
        no la verdad del contenido.
        """

        payload = json.dumps(
            record,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":")
        )

        return hashlib.sha256(
            payload.encode("utf-8")
        ).hexdigest()

    def run(self, input_text: str, candidates=None):

        self.cycle += 1

        state_before = dict(self.state)

        # OBSERVAR / EVALUAR
        score = self._score(input_text)

        # DECIDIR
        decision = (
            "integrar"
            if score >= 0.55
            else
            "observar"
        )

        # ACTUAR
        action = (
            "actualizar_estado"
            if decision == "integrar"
            else
            "conservar_y_observar"
        )

        result = {
            "score": round(score, 6),
            "decision": decision
        }

        # APRENDER / EVOLUCIONAR

        self.state["coherence"] = round(
            (state_before["coherence"] + score) / 2,
            6
        )

        self.state["potercia"] = round(
            min(
                1.0,
                0.6 * state_before["potercia"]
                +
                0.4 * score
            ),
            6
        )

        self.state["meta_potercia"] = round(
            self.state["potercia"]
            -
            state_before["potercia"],
            6
        )

        # CER — CEUNIA EVOLUTION RECORD

        record = {

            "system": "CEUNIA",

            "version": self.version,

            "cycle": self.cycle,

            "timestamp":
                datetime.now(timezone.utc).isoformat(),

            "state_before":
                state_before,

            "context": {
                "memory_size":
                    len(self.memory)
            },

            "input":
                input_text,

            "decision":
                decision,

            "action":
                action,

            "result":
                result,

            "evidence": {
                "input_length":
                    len(input_text)
            },

            "evaluation": {
                "coherence":
                    self.state["coherence"]
            },

            "learning": {
                "score":
                    score
            },

            "modification": {
                "state_updated":
                    True
            },

            "state_after":
                dict(self.state),

            "potercia":
                self.state["potercia"],

            "meta_potercia":
                self.state["meta_potercia"],

            "alternatives":
                candidates or []
        }

        # INTEGRIDAD DEL REGISTRO

        record["hash"] = self._hash_record(record)

        # MEMORIA

        self.memory.append(record)

        return record

    def status(self):

        return {

            "system":
                "CEUNIA",

            "version":
                self.version,

            "cycle":
                self.cycle,

            "state":
                dict(self.state),

            "memory_size":
                len(self.memory)
        }
'''

(ROOT / "ceunia" / "core.py").write_text(core_code)

# ------------------------------------------------------------
# 2. API FASTAPI
# ------------------------------------------------------------

api_code = r'''
from fastapi import FastAPI
from pydantic import BaseModel

from ceunia.core import CEUNIA

app = FastAPI(
    title="CEUNIA Core",
    version="1.1.0"
)

system = CEUNIA()


class RunRequest(BaseModel):

    input_text: str

    candidates: list[str] = []


@app.post("/run")
def run(req: RunRequest):

    return system.run(
        req.input_text,
        req.candidates
    )


@app.get("/status")
def status():

    return system.status()
'''

(ROOT / "ceunia" / "api.py").write_text(api_code)

# ------------------------------------------------------------
# 3. TESTS
# ------------------------------------------------------------

tests_code = r'''
import sys
from pathlib import Path

sys.path.insert(
    0,
    str(Path(__file__).resolve().parents[1])
)

from ceunia.core import CEUNIA


def test_run_and_hash():

    c = CEUNIA()

    r = c.run(
        "CEUNIA execution test"
    )

    assert r["system"] == "CEUNIA"

    assert r["version"] == "1.1.0"

    assert r["cycle"] == 1

    assert len(r["hash"]) == 64

    assert r["state_after"]["coherence"] >= 0


def test_memory():

    c = CEUNIA()

    c.run("one")

    c.run("two")

    status = c.status()

    assert status["cycle"] == 2

    assert status["memory_size"] == 2
'''

(ROOT / "tests" / "test_core.py").write_text(tests_code)

# ------------------------------------------------------------
# 4. REQUIREMENTS
# ------------------------------------------------------------

(ROOT / "requirements.txt").write_text(
    "fastapi\nuvicorn\npydantic\npytest\n"
)

# ------------------------------------------------------------
# 5. README
# ------------------------------------------------------------

readme = r'''
# CEUNIA Core v1.1

MVP ejecutable del núcleo CEUNIA.

Componentes:

- Estado cognitivo mínimo
- Evaluación determinista
- Memoria
- Coherencia
- Potercia
- MetaPotercia
- CER
- SHA-256
- API FastAPI
- Tests

El hash acredita integridad digital del registro,
no demuestra la verdad del contenido.

Esta implementación es un MVP experimental.
No implica que CEUNIA sea actualmente un estándar
universal adoptado.
'''

(ROOT / "README.md").write_text(readme)

# ------------------------------------------------------------
# 6. IMPORTAR Y EJECUTAR
# ------------------------------------------------------------

sys.path.insert(0, str(ROOT))

from ceunia.core import CEUNIA

ceunia = CEUNIA()

# ------------------------------------------------------------
# 7. EJECUCIÓN REAL
# ------------------------------------------------------------

inputs = [

    "CEUNIA observa el estado del sistema.",

    "CEUNIA integra información y evalúa coherencia.",

    "CEUNIA aprende del resultado y modifica su estado."
]

results = []

for text in inputs:

    result = ceunia.run(text)

    results.append(result)

# ------------------------------------------------------------
# 8. VALIDACIÓN
# ------------------------------------------------------------

assert ceunia.status()["cycle"] == 3

assert ceunia.status()["memory_size"] == 3

for r in results:

    assert r["system"] == "CEUNIA"

    assert r["version"] == "1.1.0"

    assert len(r["hash"]) == 64

# ------------------------------------------------------------
# 9. EJECUTAR TESTS
# ------------------------------------------------------------

import subprocess

test_result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        "-q",
        str(ROOT / "tests")
    ],
    capture_output=True,
    text=True
)

print("\n==============================")
print("CEUNIA CORE v1.1")
print("==============================")

print(
    test_result.stdout
)

if test_result.returncode != 0:

    print(test_result.stderr)

    raise RuntimeError(
        "Los tests no pasaron."
    )

# ------------------------------------------------------------
# 10. MOSTRAR ESTADO
# ------------------------------------------------------------

print("\nESTADO FINAL:")

print(
    json.dumps(
        ceunia.status(),
        indent=2,
        ensure_ascii=False
    )
)

print("\nÚLTIMO CER:")

print(
    json.dumps(
        results[-1],
        indent=2,
        ensure_ascii=False
    )
)

# ------------------------------------------------------------
# 11. GUARDAR MEMORIA
# ------------------------------------------------------------

memory_file = (
    ROOT /
    "ceunia_memory.json"
)

memory_file.write_text(
    json.dumps(
        ceunia.memory,
        indent=2,
        ensure_ascii=False
    )
)

# ------------------------------------------------------------
# 12. CREAR ZIP
# ------------------------------------------------------------

ZIP = Path(
    "/content/CEUNIA_Core_v1_1_MVP.zip"
)

if ZIP.exists():
    ZIP.unlink()

with zipfile.ZipFile(
    ZIP,
    "w",
    zipfile.ZIP_DEFLATED
) as z:

    for file in ROOT.rglob("*"):

        if file.is_file():

            z.write(
                file,
                file.relative_to(ROOT.parent)
            )

# ------------------------------------------------------------
# 13. RESULTADO
# ------------------------------------------------------------

print("\n==============================")
print("VALIDACIÓN COMPLETADA")
print("==============================")

print(
    "Ciclos ejecutados:",
    ceunia.cycle
)

print(
    "Registros en memoria:",
    len(ceunia.memory)
)

print(
    "Coherencia:",
    ceunia.state["coherence"]
)

print(
    "Potercia:",
    ceunia.state["potercia"]
)

print(
    "MetaPotercia:",
    ceunia.state["meta_potercia"]
)

print(
    "Hash último CER:",
    results[-1]["hash"]
)

print(
    "\nZIP generado:"
)

print(ZIP)

# En Colab muestra enlace de descarga
from google.colab import files

files.dfownload(str(ZIP))
```
